# Lambda Exercises

Write your answer in the empty code cell under each question. Check yourself against `solutions.ipynb` when done.

**Part 1** = warm-up · **Part 2** = practice · **Part 3** = challenge.

## Part 1 — Warm-up

**1. Instant functions.** Write a two-parameter lambda `power` computing `base ** exponent` and print `power(2, 10)`. Then write a ZERO-parameter lambda `celebrate` that simply produces the string `Victory day!` and call it.

**2. The def twin.** Given the named function `triple_def(n)` returning `n * 3`, write its lambda twin. Prove they behave identically for every integer from -3 to 3 (loop and compare). Then print each function's `__name__` — what does the anonymous one call itself?

**3. Sort by length.** Given `words = ["banana", "fig", "cherry", "date"]`, print them sorted shortest-first using `key=len`, then sorted by their LAST letter using a lambda key.

## Part 2 — Practice

**4. Price list sorting.** With `products = [("Notebook", 120), ("Pen", 25), ("Backpack", 1450), ("Calculator", 640)]`, print the list ordered by price ascending, then descending with `reverse=True` — leaving the original untouched both times.

**5. Cheapest and priciest.** Using tuples like the previous exercise, find the cheapest and priciest products with `min` and `max` plus a `key` lambda — remember you get back the ORIGINAL item. Print both. Then, with `players = [("Sarah", 92), ("Rahim", 85), ("Aisha", 95)]`, print just the NAME of the top scorer.

**6. Transform everything.** Apply a VAT of 15% (`p * 1.15`) to every price in `[120, 250, 80]` using `map` with a lambda. First print the raw `map` object itself, then the proper list of results. What kind of thing did `map` hand back?

**7. Filter, then polish.** From `scores = [45, 82, 67, 91, 58]`, keep only the passing scores (60 or more) with `filter`, then add 5 curve points to those survivors with `map`. Print the final list — feeding one straight into the other counts as one line.

## Part 3 — Challenge

**8. The one-shot iterator.** Build `doubled = map(lambda n: n * 2, [1, 2, 3, 4])`, then wrap it in `list()` TWICE and print both results. Add a comment explaining why the second list is empty.

**9. Know the limits.** Try to compile the snippet `f = lambda x: y = x + 1` (an assignment inside a lambda) inside `try`/`except SyntaxError` and print the error message. Then follow PEP 8: rewrite the named lambda `f = lambda x: 2 * x` as a proper `def double(x)` with a docstring, and show `double(21)`.
