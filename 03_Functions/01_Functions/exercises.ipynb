# Functions Exercises

Write your answer in the empty code cell under each question. Check yourself against `solutions.ipynb` when done.

**Part 1** = warm-up · **Part 2** = practice · **Part 3** = challenge.

## Part 1 — Warm-up

**1. Your first function.** Define a function `greet()` that prints `Hello, Python!`, then call it two times. Nothing should run until you call it.

**2. Defining is not running.** Define `double(n)` so it returns `n * 2`. Right after the definition print `Definition ready...` — notice no result appears yet. Only then call `double(4)` and print what comes back.

**3. Parameter or argument.** Define `describe_pet(name, species)` that prints a line like `Tommy is a cat`. Call it with `"Tommy"` and `"cat"`. Add comments labelling which names in your code are parameters and which values are arguments.

## Part 2 — Practice

**4. Built-in manual page.** Write `to_fahrenheit(celsius)` returning `celsius * 9 / 5 + 32`, give it a one-line docstring, then read that docstring back two ways: with `help(to_fahrenheit)` and by printing `to_fahrenheit.__doc__`.

**5. Fruitful vs void.** Write two functions: `add_vat(price)` which **returns** `price * 1.15`, and `show_receipt(price)` which only **prints** a line like `Receipt printed for 1000 taka`. Capture each result in its own variable and print both captured values — one holds a number, the other holds something suspicious.

**6. Every function answers.** Define `mystery()` whose body computes `x = 5 * 5` but never returns anything. Call it, store the outcome, then print both the outcome and `outcome is None`. What did Python quietly hand you?

**7. Objects, not just recipes.** Define `square(n)` returning `n * n`. Bind a second name with `sq = square` (no parentheses!) and call through the new name. Then store `square` and a `cube(n)` function in a dictionary `operations = {...}` and loop over it printing lines like `square of 6 is 36`.

## Part 3 — Challenge

**8. DRY refactor.** These copy-pasted lines compute discounted prices: `1200 - 1200 * 0.10`, `450 - 450 * 0.10`, `2500 - 2500 * 0.25`. Replace all three with ONE function `apply_discount(price, rate)` and print the three results. Then call it again with `0.15` on the first price to preview a deeper sale — notice the rule still lives in one place.

**9. Order of execution.** Put a call to `announce("Show starts soon!")` ABOVE the `def announce(message):` block, wrapped in `try`/`except NameError` so you can see the failure message printed. Below the definition, call it again successfully.
