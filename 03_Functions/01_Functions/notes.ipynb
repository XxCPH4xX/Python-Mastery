# Python Functions

> 📘 **Python Mastery** · Module 03 — Functions · Lesson 1/5

A function is a named box of code you can run whenever you like. This lesson shows why functions matter (write once, reuse everywhere), how to define and call them, and a fact that surprises most beginners: in Python, a function is an object, just like a number or a string.

## 🎯 Learning Objectives

By the end of this lesson you will be able to:

- **Explain** why functions exist and how they enforce the DRY principle
- **Define** functions with `def` and **call** them with arguments
- **Distinguish** parameters (placeholders) from arguments (real values)
- **Document** functions with docstrings and read them back with `help()`
- **Predict** what a function returns — including the invisible `None`
- **Use functions as objects**: rename them, store them in lists and dictionaries

## 1. Why Functions?

Copy-pasting the same five lines ten times works — until the logic changes and you must hunt down ten copies. A function packages logic under one name so you write it **once** and reuse it everywhere. Programmers call this **DRY: Don't Repeat Yourself**. The second win is **abstraction**: once `apply_discount()` exists, callers only need to know *what* it does, never *how*.

**Syntax:**

```python
# Without a function: the same logic, retyped for every case
result_1 = 1200 - 1200 * 0.10     # copy 1
result_2 = 450 - 450 * 0.10       # copy 2   <- one change means many edits

# With a function: define once, call many times
def function_name(parameters):
    body
    return result

function_name(arguments)
```

**Example:** first, feel the pain of repetition.

````python
# Three customers, the same 10% discount -- the formula is typed three times
price_1 = 1200 - 1200 * 0.10   # discounted sari
price_2 = 450 - 450 * 0.10     # discounted shirt
price_3 = 2500 - 2500 * 0.10   # discounted shoes

print(price_1, price_2, price_3)

# If the shop changes to 15%, we must edit THREE lines.
# Imagine 500 products... this is exactly the problem functions solve.
````

## 2. Defining and Calling a Function (`def`)

`def` creates the function: a name, parentheses holding parameter names, a colon, and an indented body. Think of a recipe card — writing it down cooks nothing. Likewise, **defining prepares the code; calling runs it.** A call is simply the function's name followed by parentheses.

**Syntax:**

```python
def function_name(parameter1, parameter2):
    """What this function does."""       # optional docstring
    # body -- indented 4 spaces
    return result                         # hands a value back

function_name(argument1, argument2)       # the call actually runs the body
```

**Example:**

````python
def apply_discount(price, rate):
    """Return the price after taking rate off (0.10 means 10%)."""
    final_price = price - price * rate
    return final_price


# Notice: nothing was printed yet. DEFINING does not EXECUTE the body.
print("Function is ready, but nothing has run...")

# Calling it (three times, three different arguments):
print(apply_discount(1200, 0.10))
print(apply_discount(450, 0.10))
print(apply_discount(2500, 0.25))

# The discount rule now lives in ONE place -- change it once, fix it everywhere.
````

````python
def greet():
    """A function with no parameters and no return value."""
    print("Hello, Sarah! Welcome to Python.")


greet()   # the parentheses mean "run the body now"
greet()   # call it as many times as you like
````

## 3. Parameters vs Arguments

Two words that are easy to mix up:

- **Parameter:** the placeholder name in the definition — an empty labelled box waiting to be filled.
- **Argument:** the actual value you supply in the call — what goes *into* the box.

In `def split_bill(total_taka, friends)`, the names are parameters. In `split_bill(2400, 4)`, the values `2400` and `4` are arguments.

**Syntax:**

```python
def function_name(param_a, param_b):    # <- parameters live here
    ...

function_name(10, 20)                   # <- arguments live here
```

**Example:**

````python
def split_bill(total_taka, friends):
    """Split a restaurant bill evenly between friends."""
    each_pays = total_taka / friends
    return each_pays


# Arguments flow into parameters by position: 2400 -> total_taka, 4 -> friends
share = split_bill(2400, 4)
print(f"Each friend pays {share} taka")

# Same function, different arguments -> different result
print(split_bill(900, 3))
````

## 4. Docstrings and `help()`

A **docstring** is a string in triple quotes placed as the first line of a function body. It is the built-in manual page for your function: editors show it on hover, and `help()` prints it in the console. Triple quotes allow the text to span several lines — convention is a one-line summary, then details about parameters and the return value.

**Syntax:**

```python
def function_name(param):
    """One-line summary.

    Optional longer explanation.

    Args:
        param: what it means.

    Returns:
        What comes back to the caller.
    """
    ...
```

**Example:**

````python
def celsius_to_fahrenheit(celsius):
    """Convert a Celsius temperature to Fahrenheit.

    Args:
        celsius: temperature in degrees C (int or float).

    Returns:
        The equivalent temperature in degrees F.
    """
    return celsius * 9 / 5 + 32


print(celsius_to_fahrenheit(36.5))   # normal human body temperature

# Read the manual back:
help(celsius_to_fahrenheit)

# The docstring is stored on the function object itself:
print(celsius_to_fahrenheit.__doc__)
````

## 5. `return`: Fruitful vs Void Functions

The `return` statement sends a value back to whoever called the function — and it ends the function immediately. Functions that return a useful value are called **fruitful**; functions that only act (like printing a receipt) are **void**.

Here is the surprise: a void function still returns something — the special value `None`, meaning "nothing here". In Python, every function returns exactly one thing.

**Syntax:**

```python
def fruitful(a, b):
    return a + b            # a value goes back to the caller

def void_action(message):
    print(message)          # performs an action, returns nothing (= None)
```

**Example:**

````python
def add_vat(price):
    """Fruitful: returns the price with 15% VAT added."""
    return price * 1.15


def show_receipt(price):
    """Void: prints a line, returns nothing."""
    print(f"Receipt printed for {price} taka")


total = add_vat(1000)            # capture the returned value
print("Total with VAT:", total)

result = show_receipt(1000)      # capture... what exactly?
print("show_receipt gave us:", result)

# A reliable way to detect "nothing":
def no_return():
    x = 1 + 1                    # computed, then thrown away


outcome = no_return()
print(outcome, outcome is None)  # None -- and `is None` is the canonical check
````

## 6. Define Before You Call

Python executes a script top to bottom. A function does not exist until its `def` line has run, so calling it earlier raises a `NameError`. This bites beginners hardest in notebooks: restart the kernel, skip the defining cell, and every later cell breaks.

Subtle bonus: the *body* of a function is also not executed until the first call. So a function may safely mention a name that is defined further down — as long as that name exists by the time the function actually **runs**.

**Syntax:**

```python
say_hello()            # NameError! say_hello does not exist yet

def say_hello():
    print("Hi")        # moving the call BELOW this block fixes it
```

**Example:**

````python
try:
    announce("Class starts soon!")        # called too early...
except NameError as err:
    print("NameError:", err)


def announce(message):                    # ...and only defined afterwards
    print("Announcement:", message)


announce("Now the function exists.")      # works fine below the def
````

````python
def late_greeting():
    print(later_helper())        # mentioned here, defined BELOW


def later_helper():
    return "I was defined after my user!"


late_greeting()   # perfectly legal: the helper exists by the time the body RUNS
````

## 7. Functions Are Objects

> 🔍 **Under the Hood:** when Python runs a `def`, it builds a **function object** in memory — complete with a name, a compiled code object, the docstring, and default values — then binds your variable name to it, exactly like `x = 5` binds `x` to an integer. A function is a value, just a strange-looking one.

Because they are objects, you can assign a function to another variable, store functions in a list or dictionary, pass them into other functions, even `del` the original name — the object survives as long as any name still points to it.

**Syntax:**

```python
def square(n):
    return n * n

alias = square                       # NO parentheses: copy the reference, don't call
operations = {"square": square}      # store functions in a dict
other_function(operations["square"]) # pass a function as an argument
```

**Example:**

````python
def square(n):
    return n * n

def cube(n):
    return n ** 3


print(type(square))              # functions have their very own type

f = square                       # bind a SECOND name to the SAME object
print(id(square) == id(f))       # proof: one object, two names
print(f(6), square(6))           # either name calls the same function

# A "dispatch table": map names to functions, then look up and call
operations = {"square": square, "cube": cube}
for name, fn in operations.items():
    print(name, "of 5 is", fn(5))

# Hand a function AS an argument to another function
def apply_twice(func, value):
    return func(func(value))


print(apply_twice(square, 3))    # square(square(3))
````

````python
def shout(text):
    return text.upper()

def whisper(text):
    return text.lower()


actions = [shout, whisper]               # a LIST of functions
for action in actions:
    print(action("Dhaka"))

# Sort functions alphabetically BY NAME -- key= returns each one's __name__
print(sorted(actions, key=lambda f: f.__name__))
# (That little lambda previews Lesson 5!)
````

## 8. Naming Conventions

Python style (PEP 8) gives functions lowercase names joined by underscores — `snake_case`. Pick names that read like instructions:

- Start with a verb: `calculate_total`, `send_email`, `split_bill`.
- Yes/no questions begin with `is_`, `has_`, `can_` and should return a boolean.
- Never shadow built-ins: naming a variable `list` or `sum` quietly breaks those tools for the rest of the file.

**Syntax:**

```python
# Good snake_case names
def calculate_total(prices): ...
def is_valid_email(address): ...

# Bad
def CalculateTotal(P): ...    # CamelCase, vague single-letter parameter
def list(items): ...          # shadows the built-in list!
```

**Example:**

````python
def calculate_average(scores):
    """Return the mean of a list of exam scores."""
    return sum(scores) / len(scores)


def has_passed(score):
    """Return True when a score reaches the pass mark."""
    return score >= 33


exam_scores = [78, 92, 55, 81]
average = calculate_average(exam_scores)

print(f"Average: {average:.1f}")
print("Sarah passed:", has_passed(exam_scores[0]))

# What NOT to do: shadowing a built-in name
# sum = 100        # <- uncommenting this breaks sum(...) for the rest of the file!
print(sum(exam_scores))   # built-in still intact because we never overwrote it
````

## ⚠️ Common Mistakes & Gotchas

| Mistake | Problem | Fix |
| --- | --- | --- |
| Calling `my_func()` before its `def` has run | `NameError` — the name does not exist yet in this session | Keep definitions above their calls; re-run cells in order after a kernel restart |
| Writing `my_func` without `()` | You get the function *object* (e.g. `<function my_func at 0x...>`), nothing runs | Parentheses mean "call it": `my_func()` |
| Using `print()` where a value is needed | Text appears on screen but the call returns `None`; `total = printer(...)` stores `None` | `return` the value (deep-dive in Lesson 3) |
| Naming variables `list`, `sum`, `str`, `id` | Shadows the built-in; unrelated code crashes later with baffling errors | Descriptive plurals: `scores`, `names`, `items` |
| Leaving the body empty under `def` | `IndentationError` before the program even starts | Write at least `pass` as a placeholder |

Mentioning a function and calling it are worlds apart:

❌

```python
greet    # evaluates to the function object itself; nothing happens
```

✅

```python
greet()  # parentheses mean "run the body"
```

## 💡 Best Practices & Pro Tips

- **One function, one job.** If your docstring needs the word "and", consider splitting the function in two.
- Keep functions short — a few lines to one screen. Short functions are easier to test, name and reuse.
- Document every function someone else will call (including future you) with at least a one-line docstring.
- **Return data, print at the edges.** Let the caller decide whether to display, save, or combine a result.
- Give parameters honest names: `discount_rate` beats `r` six months later.
- 🤖 **AI-engineering relevance:** nearly everything in ML code is a function passed around as an object — loss functions handed to training loops, activation functions stacked inside layers, preprocessing steps composed into pipelines. The dispatch-table pattern above is exactly how plugin registries and callback systems work.

## 📌 Summary

| Tool | What it does | Example |
| --- | --- | --- |
| `def name(params):` | Defines (creates) a function | `def area(w, h):` |
| `name(args)` | Calls (runs) the function | `area(3, 4)` |
| `return value` | Sends a result back and ends the call | `return w * h` |
| `"""Docstring."""` | Built-in documentation, first line of the body | `"""Return rectangle area."""` |
| `help(func)` | Prints the docstring nicely formatted | `help(area)` |
| `func.__doc__` | Raw docstring stored on the function object | `print(area.__doc__)` |
| `alias = func` | Functions are objects: bind extra names | `f = area` |
| dict / list of functions | Dispatch table, batch operations | `ops = {"add": add}` |
| `func(other_func)` | Pass behaviour as an argument | `apply_twice(square, 3)` |

Key takeaways:

- Define once, call many times — DRY turns 500 edits into 1.
- **Parameters** are placeholders in the `def`; **arguments** are the real values in the call.
- Every function returns something; forget `return` and Python quietly gives you `None`.
- A function is an object: store it, pass it, sort it — just like `"hello"` or `42`.
