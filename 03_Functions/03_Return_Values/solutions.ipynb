## Part 1 — Warm-up

**1. Hand something back.** Write `area(width, height)` that RETURNS `width * height` (no printing inside!). Store `area(4, 5)` in a variable called `result`, then print it.

```
def area(width, height):
    """Return rectangle area."""
    return width * height


result = area(4, 5)
print(result)      # 20
```

**2. Two answers at once.** Write `min_max(scores)` that returns the lowest AND highest score together. Unpack them into `low, high` and print both. Then capture a second call's result in a single variable and print it along with its type.

```
def min_max(scores):
    """Return lowest and highest scores as one tuple."""
    return min(scores), max(scores)


low, high = min_max([78, 92, 55, 81])
print(low, high)                     # 55 92

both = min_max([78, 92, 55, 81])
print(both, type(both).__name__)     # (55, 92) tuple
```

**3. Print is not return.** Here is a broken helper: `average_prints(scores)` computes `sum(scores) / len(scores)` but only PRINTS it. Capture its "result" in a variable and print that variable — what came out? Then fix it as `average_returns(scores)` which RETURNS the value, and capture the real number.

```
def average_prints(scores):
    print(sum(scores) / len(scores))    # screen only


def average_returns(scores):
    return sum(scores) / len(scores)    # usable value


captured = average_prints([70, 80, 90])
print("Captured:", captured)            # None!

real = average_returns([70, 80, 90])
print("Real:", real)                    # 80.0
```

## Part 2 — Practice

**4. Dead code patrol.** Write `surprise()` whose body is `return "done"` followed by `print("You made it!")`. Before running, say out loud whether the message can ever appear — then run it and mark the unreachable line with a comment.

```
def surprise():
    return "done"
    print("You made it!")   # dead code: return already exited


print(surprise())           # done -- and nothing else
```

**5. Guard clause first.** Write `safe_divide(a, b)` that returns `None` immediately when `b == 0`, and otherwise returns `a / b` on a flat, unindented happy path. Test it with `10 / 4` and `10 / 0`.

```
def safe_divide(a, b):
    """Return a / b, or None when dividing by zero."""
    if b == 0:
        return None           # guard: leave early
    return a / b              # happy path stays unindented


print(safe_divide(10, 4))     # 2.5
print(safe_divide(10, 0))     # None
```

**6. Every path answers.** Write `grade_letter(score)` mapping 90+ to A, 80+ to B, 70+ to C, 60+ to D, anything lower to F — make sure NO input falls off the end. Print grades for 95, 84, 71, 63 and 40. Bonus: write `risky_grade(score)` that returns `"Pass"` only for 33+ with no `else`, and show what escapes when the score is 20.

```
def grade_letter(score):
    """Map a score out of 100 to a letter grade."""
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    elif score >= 60:
        return "D"
    else:
        return "F"            # catch-all: no path escapes


for s in (95, 84, 71, 63, 40):
    print(s, "->", grade_letter(s))


def risky_grade(score):
    if score >= 33:
        return "Pass"         # lower scores fall off the edge...


print(risky_grade(90))        # Pass
print(risky_grade(20))        # None sneaks out
```

**7. Booleans straight up.** Rewrite this clunky checker WITHOUT any `if`: a function `is_adult(age)` that should be `True` when `age >= 18` and `False` otherwise. Then use it inside a list comprehension to keep only the adults from `[12, 19, 34, 17, 65]`.

```
def is_adult(age):
    return age >= 18          # no if needed!


ages = [12, 19, 34, 17, 65]
adults = [a for a in ages if is_adult(a)]
print(adults)                 # [19, 34, 65]
```

## Part 3 — Challenge

**8. Stop at the first hit.** Write `first_even(numbers)` that loops through the list and returns the FIRST even number the moment it appears; if the loop finishes without finding one, return `None`. Try it on `[7, 3, 8, 10]` and `[1, 3, 5]`.

```
def first_even(numbers):
    """Return the first even number, or None if there isn't one."""
    for n in numbers:
        if n % 2 == 0:
            return n          # early exit: stop scanning
    return None               # loop finished empty-handed


print(first_even([7, 3, 8, 10]))   # 8
print(first_even([1, 3, 5]))       # None
```

**9. Function factory.** Write `make_multiplier(factor)` that defines an inner `multiply(number)` using `factor`, then RETURNS the inner function itself (no parentheses on that return!). Create `double` and `triple` from it and print `double(10)` and `triple(10)`. Finish by calling `make_multiplier(5)(100)` in one breath.

```
def make_multiplier(factor):
    """Return a brand-new function that multiplies by factor."""

    def multiply(number):
        return number * factor

    return multiply            # the FUNCTION itself -- no parentheses


double = make_multiplier(2)
triple = make_multiplier(3)

print(double(10), triple(10))   # 20 30
print(make_multiplier(5)(100))  # 500
```
