## Part 1 — Warm-up

**1. Position means everything.** Define `describe_book(title, pages)` that prints a line like `'Python Basics' has 350 pages.` Call it correctly, then call it again with the two arguments swapped — it still runs! Add a comment describing what went wrong the second time.

```
def describe_book(title, pages):
    """Describe a book by title and page count."""
    print(f"'{title}' has {pages} pages.")


describe_book("Python Basics", 350)   # correct order
describe_book(350, "Python Basics")   # runs fine -- but nonsense
```

**2. Keywords set you free.** Write `book_ticket(movie, seats, show_time)` that prints something like `3 seat(s) for 'Up' at 7:30 PM.` Call it once with all positional arguments and once with all keyword arguments in a DIFFERENT order. Both calls should print identical sentences.

```
def book_ticket(movie, seats, show_time):
    """Print a cinema booking summary."""
    print(f"{seats} seat(s) for '{movie}' at {show_time}.")


book_ticket("Up", 3, "7:30 PM")
book_ticket(show_time="7:30 PM", movie="Up", seats=3)   # same output
```

**3. Sensible defaults.** Write `make_tea(flavor="milk", sugar_level=1)` returning a sentence like `One cup of milk tea with 1 spoon(s) of sugar.` Call it four ways: with no arguments, overriding flavor positionally, overriding both positionally, and overriding only the sugar level by keyword.

```
def make_tea(flavor="milk", sugar_level=1):
    """Prepare tea with sensible defaults."""
    return f"One cup of {flavor} tea with {sugar_level} spoon(s) of sugar."


print(make_tea())                 # both defaults
print(make_tea("green"))          # override flavor only
print(make_tea("lemon", 0))       # override both
print(make_tea(sugar_level=2))    # keyword skips flavor
```

## Part 2 — Practice

**4. The ordering law.** Predict first: is `def brew(cups=1, flavor): ...` legal? Check your prediction safely by compiling that signature inside `try`/`except SyntaxError` and printing the error message. Then write a corrected tea-brewing function whose defaulted parameter comes last, and show one call using the default.

```
source = """
def brew(cups=1, flavor):
    return flavor
"""

try:
    compile(source, "<demo>", "exec")
except SyntaxError as err:
    print("SyntaxError:", err.msg)     # default before required


def brew_tea(flavor, cups=1):          # required FIRST, defaulted AFTER
    return f"{cups} cup(s) of {flavor} tea"


print(brew_tea("green"))               # default kicks in
print(brew_tea("lemon", 3))
```

**5. The haunted cart.** Run this buggy helper three times and watch orders pile up: `add_item_buggy(item, cart=[])` appends `item` to `cart` and returns it. Prove WHY it happens by printing `add_item_buggy.__defaults__`. Then write `add_item_safe(item, cart=None)` with the `None` sentinel and show it starts fresh every call.

```
def add_item_buggy(item, cart=[]):     # ONE list created once, at def time
    cart.append(item)
    return cart


print(add_item_buggy("apples"))        # ['apples']
print(add_item_buggy("milk"))          # ['apples', 'milk'] -- surprise!
print(add_item_buggy.__defaults__)     # proof: stored inside the function


def add_item_safe(item, cart=None):
    """Append item to a fresh list whenever none is given."""
    if cart is None:
        cart = []                      # brand-new list on EVERY call
    cart.append(item)
    return cart


print(add_item_safe("apples"))         # ['apples']
print(add_item_safe("milk"))           # ['milk'] -- clean slate
```

**6. Any number of prices.** Write `total_bill(*prices)` that adds up however many prices arrive and returns the total. Inside, also print what `prices` became and its type. Call it with one, four, and zero arguments.

```
def total_bill(*prices):
    """Add up any number of menu prices."""
    print("Received:", prices, "| type:", type(prices).__name__)
    return sum(prices)


print(total_bill(120))
print(total_bill(120, 45, 300, 80))
print(total_bill())                    # zero items -> sum 0
```

**7. Profile builder.** Write `build_profile(name, **details)` that merges `name` with any number of extra keyword details into one dictionary and returns it. Build Sarah's profile from the keywords `age=21`, `city="Dhaka"`, `course="Python"`, then print the whole dict followed by each `key: value` on its own line.

```
def build_profile(name, **details):
    """Collect any number of keyword details onto a name."""
    profile = {"name": name}
    profile.update(details)
    return profile


sarah = build_profile("Sarah", age=21, city="Dhaka", course="Python")
print(sarah)

for key, value in sarah.items():
    print(f"{key}: {value}")
```

## Part 3 — Challenge

**8. The full signature.** Combine every kind of parameter in one function: `place_order(customer, *items, priority=False, **extras)`. Return a summary string with the item count, add ` [PRIORITY]` when the flag is true, and append ` | extras={...}` when any extras arrived. Make three calls that exercise all corners.

```
def place_order(customer, *items, priority=False, **extras):
    """positional -> *args -> keyword default -> **kwargs."""
    summary = f"Order for {customer}: {len(items)} item(s)"
    if priority:
        summary += " [PRIORITY]"
    if extras:
        summary += f" | extras={extras}"
    return summary


print(place_order("Sarah", "pen", "notebook"))
print(place_order("Rahim", "calculator", priority=True))
print(place_order("Aisha", "bag", gift_wrap=True, note="Happy birthday"))
```

**9. Unpacking at the call site.** Given `menu_prices = [120, 45, 300]` and `order = {"amount": 35, "currency": "dollars"}`, feed the list into a `total_bill(*prices)` function using `*`, and the dict into `charge(amount, currency="taka")` using `**`.

```
def total_bill(*prices):
    return sum(prices)


def charge(amount, currency="taka"):
    return f"Charged {amount} {currency}"


menu_prices = [120, 45, 300]
order = {"amount": 35, "currency": "dollars"}

print(total_bill(*menu_prices))    # unpacks the list into arguments
print(charge(**order))             # unpacks the dict into keywords
```
