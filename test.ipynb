```python
def status_message(code):
    match code:
        case 200:
            return "OK"
        case 301:
            return "Moved Permanently"
        case 404:
            return "Not Found"
        case 500:
            return "Internal Server Error"
        case _:
            return f"Unknown status code: {code}"


for code in [301, 404, 765]:
    print(code, "->", status_message(code))
```

```python
def describe(x):
    match x:
        case 0:
            return "zero"
        case 1 | -1:
            return "plus or minus one"
        case other:  # captures anything left over
            return f"something else: {other!r}"


for sample in (0, -1, "banana", 3.14):
    print(describe(sample))
```

```python
score = [31, 64, 12, 86, 45, 34, 98]
score.sort(reverse=True)

match score:
    case []:
        print("No score")
    case [top]:
        print(f"Top score: {top}")
    case [second]:
        print(f"Second place: {second}")
    case [third]:
        print(f"Third place: {third}")
    case [top, second, third, *others]:
        print(
            f"Top 3 -> First :{top}, Second :{second}, Third :{third}, and {len(others)} others"
        )
```

```python
secret = 7
guesses = input("Enter your guesses: ")  # e.g., player types "13579"

i = 0
while i < len(guesses):
    guess = int(guesses[i])  # Convert string digit to integer
    print("Guess:", guess)
    if guess == secret:
        print("Correct! Unlocked in", i + 1, "tries.")
        break  # stop searching - no point continuing
    i += 1

print("Game finished.")
```

```python
secret = 7

# Get maximum allowed attempts from the user
max_attempts = int(input("How many attempts do you want? "))

attempts = 0
while attempts < max_attempts:
    guess_input = input(f"Attempt {attempts + 1}/{max_attempts} - Enter your guess: ")

    # Check if the input is a valid number to prevent crashing
    if not guess_input.isdigit():
        print("Please enter a valid number!")
        continue

    guess = int(guess_input)
    attempts += 1

    if guess == secret:
        print(f"Correct! Unlocked in {attempts} try/tries.")
        break
    else:
        print("Wrong guess, try again!")

if guess != secret:
    print(f"Game over! You ran out of attempts. The secret was {secret}.")

print("Game finished.")
```

```python
# BUGGY: removing while looping - an even number sneaks through
buggy = [1, 2, 4, 5, 6]
for n in buggy:
    if n % 2 == 0:
        buggy.remove(n)
print("After buggy removal:", buggy)  # 4 survived - it was SKIPPED

# SAFE approach 1: build a NEW list (usually clearest)
nums = [1, 2, 4, 5, 6]
odds = []
for n in nums:
    if n % 2 != 0:
        odds.append(n)
nums = odds
print("Rebuilt list:", nums)

cart = ["milk", "eggs", "butter"]
for item in cart[:]:
    if item == "eggs":
        cart.remove(item)
print("Cart now:", cart)
```

```python
def celsius_to_fahrenheit(celsius):
    """Convert a Celsius temperature to Fahrenheit.

    Args:
        celsius: temperature in degrees C (int or float).

    Returns:
        The equivalent temperature in degrees F.
    """
    return celsius * 9 / 5 + 32


print(celsius_to_fahrenheit(36.5))  # normal human body temperature

# Read the manual back:
help(celsius_to_fahrenheit)

# The docstring is stored on the function object itself:
print(celsius_to_fahrenheit.__doc__)
```

```python
def square(n):
    return n * n


def cube(n):
    return n * n * n


print(type(square))


print(type(square))

f = square
print(id(square) == id(f))
print(f(6), square(6))

operations = {"square": square, "cube": cube}
for name, fn in operations.items():
    print(name, "of 5 is", fn(5))


def apply_twice(func, value):
    return func(func(value))


print(apply_twice(square, 3))
```
