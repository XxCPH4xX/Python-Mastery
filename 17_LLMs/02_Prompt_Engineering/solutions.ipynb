**1. Zero-Shot Sentiment Analysis.** Write a zero-shot prompt that classifies customer reviews into Positive, Negative, or Neutral.

```python
def create_zero_shot_prompt(review_text):
    """Create a zero-shot sentiment classification prompt."""
    return f"""You are a sentiment classifier. Classify the following customer review as exactly one of: Positive, Negative, Neutral.

Review: {review_text}

Sentiment:"""

# Test with 3 sample reviews
reviews = [
    "This product exceeded my expectations! Amazing quality and fast shipping.",
    "Terrible experience. The item broke after one day and support was unhelpful.",
    "The product is okay. Nothing special but does what it's supposed to do."
]

for review in reviews:
    prompt = create_zero_shot_prompt(review)
    print(f"Review: {review[:50]}...")
    print(f"Prompt:\n{prompt}\n")
```

**2. Few-Shot Conversion.** Convert your zero-shot prompt into a few-shot prompt with 3 examples.

```python
def create_few_shot_prompt(review_text, examples=None):
    """Create a few-shot sentiment classification prompt."""
    if examples is None:
        examples = [
            ("Love this! Best purchase ever.", "Positive"),
            ("Waste of money. Don't buy.", "Negative"),
            ("It works as described. Nothing more.", "Neutral"),
        ]
    
    example_str = "\n".join([f"Review: {r}\nSentiment: {s}" for r, s in examples])
    
    return f"""You are a sentiment classifier. Classify reviews as Positive, Negative, or Neutral.

{example_str}

Review: {review_text}
Sentiment:"""

prompt = create_few_shot_prompt("Great value for the price!")
print(prompt)
```

**4. Chain-of-Thought for Math.** Create a chain-of-thought prompt for solving math word problems.

```python
def create_cot_math_prompt(problem):
    """Create a chain-of-thought math word problem prompt."""
    return f"""Solve this step by step.

Problem: {problem}

Let's break this down:
1. First, identify what we know
2. Then, determine what we're solving for
3. Set up the equation
4. Solve step by step
5. State the final answer clearly

Step by step:"""

problem = "A bakery sells cupcakes for $3 each and cookies for $2 each. If Sarah buys 5 cupcakes and spends a total of $23, how many cookies did she buy?"

print(create_cot_math_prompt(problem))
```

**7. Multi-Turn Conversation Structure.** Design a multi-turn conversation for a technical support bot.

```python
conversation = [
    {"role": "system", "content": "You are a Python debugging assistant. Ask diagnostic questions before suggesting fixes."},
    {"role": "user", "content": "I'm getting a TypeError: can't multiply sequence by non-int of type 'float'"},
    {"role": "assistant", "content": "I can help with that. Can you share the line of code causing the error? Also, what are the types of the variables involved?"},
    {"role": "user", "content": "The line is: result = my_string * my_float. my_string is 'hello' and my_float is 2.5"},
    {"role": "assistant", "content": "The issue is clear: Python strings can only be multiplied by integers, not floats. You need to convert the float to int first. Use: result = my_string * int(my_float) or result = my_string * round(my_float)."},
]

print("Multi-turn conversation flow:")
for msg in conversation:
    print(f"\n{msg['role'].upper()}:")
    print(f"  {msg['content']}")
```

**8. Complex Chain-of-Thought Reasoning.** Analyze a business decision with step-by-step reasoning.

```python
def create_business_analysis_prompt(scenario):
    """Create a structured business analysis prompt."""
    return f"""Analyze this business decision step by step.

Scenario: {scenario}

Analyze in this order:
1. Calculate current monthly revenue
2. Calculate projected revenue after price increase
3. Estimate customer loss based on churn research
4. Calculate new revenue with reduced customer base
5. Compare net revenue change
6. Consider other factors (CLV, market position, competitive landscape)
7. Provide a clear recommendation with confidence level

Show your math at each step."""

scenario = "A SaaS company with 1,000 customers at $50/month is considering raising prices to $75/month. Their churn rate is currently 5% per month. Market research suggests 20% of customers would cancel if prices increase. Should they raise prices?"

print(create_business_analysis_prompt(scenario))
```

**9. Handling Edge Cases.** Create a robust prompt that handles edge cases for date extraction.

```python
import json

def create_date_extraction_prompt(text):
    """Create a robust date extraction prompt handling edge cases."""
    return f"""Extract dates from this text. Handle relative dates, vague dates, and missing dates.

Text: {text}

Return ONLY valid JSON:
{{
  "dates": [
    {{
      "original_text": "the phrase containing the date",
      "parsed_date": "YYYY-MM-DD or null if unparseable",
      "confidence": "high|medium|low",
      "note": "explanation if ambiguous"
    }}
  ],
  "has_date": true/false
}}

Rules:
- "next Tuesday" -> calculate actual date, medium confidence
- "end of day" -> today's date, low confidence
- "Q2" -> 2024-04-01 to 2024-06-30, low confidence
- "yesterday" -> calculate actual date, high confidence
- No date mentioned -> has_date: false

JSON:"""

test_cases = [
    "The meeting is scheduled for next Tuesday",
    "I'll send it by end of day",
    "Let's meet sometime in Q2",
    "The deadline was yesterday",
    "No specific date mentioned"
]

for text in test_cases:
    print(f"\nInput: {text}")
    prompt = create_date_extraction_prompt(text)
    print(f"Prompt length: {len(prompt)} chars")
```
