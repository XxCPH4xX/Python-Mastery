## Part 1 — Warm-up

**1. Three stamps, one mold.** Every construction yields an independent object; mutating one never leaks into another.

```
class Lamp:
    def __init__(self, room):
        self.room = room
        self.on = False


kitchen = Lamp("kitchen")
office = Lamp("office")

kitchen.on = True               # writes ONLY into the kitchen object

print(kitchen.room, kitchen.on)
print(office.room, office.on)
```

**2. Dot-operator workout.** Attributes hold real values, so whatever a dot returns can immediately be used further.

```
class Profile:
    def __init__(self, username):
        self.username = username


me = Profile("sarah_a")

print(me.username)              # read
print(me.username.upper())      # chain onto the returned string

me.username = "Sarah_Ahmed"     # write
print(me.username)
```

**3. Identity vs equality.** `b` is a second name for one object; `c` merely looks identical — and custom classes get identity-based `==` by default.

```
class Tag:
    def __init__(self, label):
        self.label = label


a = Tag("box-7")
b = a                   # second name, SAME object
c = Tag("box-7")        # equal-looking, genuinely different

print(a is b)           # True  - literally one object
print(a is c)           # False - two objects, one coincidence
print(a == c)           # False - no __eq__, so == falls back to identity
print(id(a) == id(b))   # True
```

## Part 2 — Practice

**4. Mutation leaks through aliases.** Assignment copies the reference, not the object — both names steer the very same thing.

```
class Config:
    def __init__(self, mode):
        self.mode = mode


cfg = Config("training")
alias = cfg                     # another NAME, same object

alias.mode = "eval"

print(cfg.mode)                 # changed via alias, visible through cfg
print(cfg is alias)             # proof: one object, two names
# Assignment copied the REFERENCE; the object itself was never duplicated.
```

**5. The dynamic trio.** Attribute names are strings underneath, so they can be probed, read and created at runtime.

```
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price


item = Product("USB Microphone", 74.5)

print(hasattr(item, "price"), hasattr(item, "discount"))
print(getattr(item, "discount", 0.0))     # missing -> default, no crash

setattr(item, "discount", 0.15)           # created on the fly
print(item.discount)

for field in ("name", "price", "discount"):
    print(field, "=", getattr(item, field))
```

**6. Settings from a dictionary.** Config-driven scripts build attributes straight from data — `vars()` shows the result.

```
settings = {"theme": "dark", "font_size": 14, "autoplay": True}


class AppSettings:
    pass


app = AppSettings()
for key, value in settings.items():
    setattr(app, key, value)            # attribute names come from data

print(app.theme, app.font_size, app.autoplay)
print(vars(app))
```

**7. Shadow detective.** Instance assignment shadows the class default for that one object; deleting the shadow lets the class value shine through again.

```
class Vehicle:
    wheels = 4                      # class-level default


car = Vehicle()
rickshaw = Vehicle()

rickshaw.wheels = 3                 # INSTANCE copy shadows the class one
print(rickshaw.wheels)              # 3 - the instance copy wins
print(car.wheels)                   # 4 - untouched sibling
print(Vehicle.wheels)               # 4 - the class itself

del rickshaw.wheels                 # drop the shadow...
print(rickshaw.wheels)              # 4 - class default visible again
```

## Part 3 — Challenge

**8. Mini contact book.** Objects in dictionaries beat parallel lists — one record travels with all of its fields.

```
class Contact:
    """One address-book entry."""

    def __init__(self, name, phone, city):
        self.name = name
        self.phone = phone
        self.city = city

    def __str__(self):
        return f"{self.name} ({self.phone}) - {self.city}"


contacts = {
    "sarah": Contact("Sarah Ahmed", "01711-223344", "Dhaka"),
    "rafi": Contact("Rafi Chowdhury", "01822-556677", "Chattogram"),
    "amina": Contact("Amina Karim", "01933-889900", "Sylhet"),
}

print(contacts["rafi"])                 # fetch the object, then use it
print(contacts["amina"].city.upper())   # chain straight into attributes

for handle, entry in contacts.items():
    print(f"@{handle}: {entry}")
```

**9. Deleting names vs deleting attributes.** `del obj.attr` removes one attribute; `del obj` removes only the name — memory stays Python's business.

```
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price


item = Product("USB-C Cable", 12.0)
item.temp_note = "clearance"

del item.temp_note
print(hasattr(item, "temp_note"))    # False - the attribute is gone

del item                             # removes the NAME binding
try:
    print(item)
except NameError as err:
    print("NameError:", err)
    # First deletion touched the object's contents;
    # the second only unbound a variable name.
```
