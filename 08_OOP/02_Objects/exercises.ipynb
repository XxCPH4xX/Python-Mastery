# Objects Exercises

Write your answer in the empty code cell under each question. Check yourself against `solutions.ipynb` when done.

**Part 1** = warm-up · **Part 2** = practice · **Part 3** = challenge.

## Part 1 — Warm-up

**1. Three stamps, one mold.** Define a `Lamp` class whose `__init__` takes `room` and starts every lamp switched off (`on = False`). Create lamps for `"kitchen"` and `"office"`, switch the kitchen one on by direct assignment, then print both states — only one may have changed.

**2. Dot-operator workout.** With a `Profile` class holding `username`: read the username, call `.upper()` on what comes back (dot-chaining), then overwrite the attribute with a corrected spelling and print it again.

**3. Identity vs equality.** Define a tiny `Tag` class storing `label`. Make `a = Tag("box-7")`, `b = a`, and `c = Tag("box-7")`. BEFORE running, predict the output of `a is b`, `a is c` and `a == c` — then print all three and grade yourself.

## Part 2 — Practice

**4. Mutation leaks through aliases.** Define a `Config` object holding `mode = "training"`; bind `alias = cfg`; change the mode THROUGH the alias, then print `cfg.mode`. Explain in a comment what the assignment copied — and what it did not.

**5. The dynamic trio.** For a `Product("USB Microphone", 74.5)`: check with `hasattr` whether `price` and `discount` exist; read `discount` safely with `getattr(item, "discount", 0.0)`; CREATE a `discount` of `0.15` with `setattr`; finally loop over `("name", "price", "discount")` printing each field.

**6. Settings from a dictionary.** Turn `settings = {"theme": "dark", "font_size": 14, "autoplay": True}` into attributes of an empty `AppSettings` object using a `setattr` loop, then print the attributes and peek with `vars(app)`.

**7. Shadow detective.** Give a `Vehicle` class a class attribute `wheels = 4`. Set `rickshaw.wheels = 3` on one instance. Predict what `rickshaw.wheels`, a second vehicle's `wheels`, and `Vehicle.wheels` each print. Then `del rickshaw.wheels` and predict again. Explain shadowing in a comment.

## Part 3 — Challenge

**8. Mini contact book.** Build a `Contact` class with `name`, `phone`, `city` and a readable `__str__`. Store three contacts in a dict keyed by handle (`"sarah"`, `"rafi"`, `"amina"`), then: print one entry directly, print another contact's city via chaining, and loop over `items()` printing `@handle: contact` lines.

**9. Deleting names vs deleting attributes.** Create a `Product` and give it a throwaway attribute `temp_note = "clearance"`. Delete that attribute with `del item.temp_note` and prove it is gone using `hasattr`. Then `del item` (the NAME) and show that using it now raises `NameError` — catch it. In comments, state the difference between the two deletions.
