## Part 1 — Warm-up

**1. Your first class.** `__init__` fires automatically once per construction, attaching data to each new object.

```
class Dog:
    """A pet dog identified by name and age."""

    def __init__(self, name, age):
        self.name = name
        self.age = age


rex = Dog("Rex", 3)
bella = Dog("Bella", 5)

print(rex.name, rex.age)
print(bella.name, bella.age)
```

**2. The empty blueprint.** Even an empty class is a working type — and a class is itself an object, of type `type`.

```
class Stub:
    """Placeholder blueprint -- filled in later."""

    pass


thing = Stub()
print(Stub)
print(type(Stub))
print(thing)
```

**3. Hello, `self`.** `obj.greet()` is sugar for `Greeter.greet(obj)` — Python quietly hands the instance in as `self`.

```
class Greeter:
    def __init__(self, name):
        self.name = name

    def greet(self):
        return f"Hi, I'm {self.name}"


g = Greeter("Sarah")

print(g.greet())
print(Greeter.greet(g))
print(g.greet() == Greeter.greet(g))
```

## Part 2 — Practice

**4. Bank account basics.** State lives in `self.balance`; methods return strings so callers stay in control.

```
class BankAccount:
    """State + behaviour bundled together."""

    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        return f"{self.owner}: deposited {amount}, balance {self.balance}"

    def withdraw(self, amount):
        if amount > self.balance:
            return f"{self.owner}: insufficient funds (have {self.balance})"
        self.balance -= amount
        return f"{self.owner}: withdrew {amount}, balance {self.balance}"


acc = BankAccount("Sarah", 100)
print(acc.deposit(50))
print(acc.withdraw(30))
print(acc.withdraw(1000))
```

**5. Shared scoreboard.** Class attributes live on the class, so every instance reads the very same number.

```
class Player:
    player_count = 0                     # shared by ALL players

    def __init__(self, name):
        self.name = name
        Player.player_count += 1


players = [Player(n) for n in ("Sarah", "Rafi", "Amina", "Omar")]

print(Player.player_count)      # 4 - read straight from the class
print(players[0].player_count)  # 4 - instances see the same value
```

**6. Two costumes for one object.** `print()` prefers `__str__`; `repr()` and containers always use `__repr__`.

```
class Movie:
    def __init__(self, title, year):
        self.title = title
        self.year = year

    def __str__(self):                          # friendly: for users
        return f"'{self.title}' ({self.year})"

    def __repr__(self):                         # precise: for developers
        return f"Movie(title={self.title!r}, year={self.year})"


film = Movie("Inception", 2010)

print(film)             # __str__
print(repr(film))       # __repr__
print([film, film])     # containers ALWAYS display items via __repr__
```

**7. Missing-self detective.** `r.status()` becomes `Robot.status(r)` — a method with no parameters refuses the instance Python tries to hand it.

```
class Robot:
    def status():                     # forgot self!
        return "all systems go"


r = Robot()
try:
    r.status()
except TypeError as e:
    print("TypeError:", e)
    # r.status() is really Robot.status(r): Python passes the instance,
    # but a method declared with zero parameters cannot receive it.


class FixedRobot:
    def status(self):                 # FIXED: room made for the instance
        return "all systems go"


print(FixedRobot().status())
```

## Part 3 — Challenge

**8. Playlist planner.** One shared class attribute, per-object state, and methods that RETURN their answers.

```
class Playlist:
    platform = "PyMusic"                       # shared by every playlist

    def __init__(self, owner, songs=None):
        self.owner = owner
        self.songs = list(songs) if songs else []   # avoid the shared-default trap

    def add_song(self, title):
        self.songs.append(title)
        return len(self.songs)

    def summary(self):
        return f"{self.owner} - {len(self.songs)} songs on {Playlist.platform}"


night = Playlist("Sarah")
print(night.add_song("Monpura"))
print(night.add_song("Tumi Jake Bhalobasho"))

road = Playlist("Rafi", ["Ekhon Onek Raat"])
print(road.summary())
print(night.summary())
```

**9. From parallel lists to a class.** Data and the operations on it travel together — no indexes left to keep aligned.

```
names = ["Sarah", "Rafi"]
grades = [[88, 92, 79], [55, 61, 47]]


def average_of(grade_list):
    return sum(grade_list) / len(grade_list)


for n, g in zip(names, grades):        # the old way: discipline required
    print(f"{n}: avg={average_of(g)}, passed={average_of(g) >= 60}")


class Student:
    """One learner: their grades AND the maths on them."""

    def __init__(self, name, grades):
        self.name = name
        self.grades = grades

    def average(self):
        return sum(self.grades) / len(self.grades)

    def has_passed(self):
        return self.average() >= 60


for s in (Student("Sarah", [88, 92, 79]), Student("Rafi", [55, 61, 47])):
    print(f"{s.name}: avg={s.average()}, passed={s.has_passed()}")
```
