## Part 1 — Warm-up

**1. Left and right slots.** `plt.subplot(rows, cols, n)` activates slot n; every `plt.*` call lands there until another slot is chosen.

```
import matplotlib.pyplot as plt

week  = ["Sat", "Sun", "Mon", "Tue", "Wed"]
study = [2, 3, 5, 4, 6]
sleep = [7, 8, 6, 9, 8]

plt.figure(figsize=(10, 4))     # wide canvas for two panels

plt.subplot(1, 2, 1)            # slot 1: LEFT
plt.plot(week, study, marker="o")
plt.title("Study hours")
plt.xlabel("Day")

plt.subplot(1, 2, 2)            # slot 2: RIGHT
plt.plot(week, sleep, marker="o", color="#2a9d8f")
plt.title("Sleep hours")
plt.xlabel("Day")

plt.show()
```

**2. Panels count from one.** Slots run 1-based, left-to-right then top-to-bottom - so slot 3 sits BOTTOM-left (and `subplot(223)` is the same thing).

```
import matplotlib.pyplot as plt

for n in (1, 2, 3, 4):
    plt.subplot(2, 2, n)              # NOT zero-based!
    plt.plot([1, 2, 3], [n, 2 * n, 3 * n], marker="o")
    plt.title(f"Slot {n}")

plt.show()

# Slot 3 is the bottom-LEFT panel: rows fill left-to-right,
# top-to-bottom, starting the count at 1.
```

**3. Objects, not globals.** The OO interface hands you the Axes object: `plt.title` becomes `ax.set_title`, `plt.xlim` becomes `ax.set_xlim`, and so on.

```
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, 4))

months  = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
temp_c  = [19, 22, 27, 30, 30, 29]

ax.plot(months, temp_c, color="#2a6f97", marker="o")
ax.set_title("Dhaka Average Temperature (Jan-Jun)")
ax.set_xlabel("Month")
ax.set_ylabel("Temperature (°C)")
ax.grid(True, linestyle="--", alpha=0.5)

plt.show()

# Explicit handles: no "whatever is current" surprises, and ax can be
# passed to functions or loops later.
```

## Part 2 — Practice

**4. Same ruler, both panels.** Without shared limits each panel auto-scales independently - heights stop meaning the same thing.

```
import matplotlib.pyplot as plt

days  = ["Sat", "Sun", "Mon", "Tue", "Wed", "Thu"]
karim = [35, 42, 38, 51, 47, 55]
priya = [40, 38, 48, 55, 50, 62]

fig, axes = plt.subplots(2, 1, sharex=True, sharey=True, figsize=(8, 6))

axes[0].plot(days, karim, marker="o", color="#2a6f97")
axes[0].set_title("Karim's earnings")

axes[1].plot(days, priya, marker="s", color="#e76f51")
axes[1].set_title("Priya's earnings")

for ax in axes:
    ax.set_ylabel("Earnings")
    ax.grid(True, linestyle="--", alpha=0.5)

plt.show()

# Identical y-limits: a glance now honestly compares the two heights.
```

**5. One loop fills any grid.** `.ravel()` flattens the grid to a sequence, so the same loop fits ANY layout - change (1, 5) to (5, 1) and nothing breaks.

```
import matplotlib.pyplot as plt

subjects = ["Math", "Physics", "Chemistry", "Biology", "English"]
marks    = [88, 74, 91, 79, 84]

fig, axes = plt.subplots(1, 5, figsize=(14, 3), sharey=True)

for ax, subj, mk in zip(axes.ravel(), subjects, marks):
    ax.bar(subj, mk, color="#2a6f97" if mk >= 80 else "#adb5bd")
    ax.set_title(subj, fontsize=10)
    ax.set_ylim(0, 100)

axes[0].set_ylabel("Marks")
plt.show()

# Swap the grid to (5, 1) or (3, 2)... the SAME loop still fills it.
```

**6. Untangle the titles.** `fig.tight_layout()` measures every title and label and nudges the panels apart - call it right before show().

```
import matplotlib.pyplot as plt

quizzes = {
    "Math":      [88, 92, 79, 85, 90, 84],
    "Physics":   [74, 71, 80, 68, 77, 72],
    "Chemistry": [81, 78, 85, 74, 83, 80],
    "Biology":   [90, 86, 94, 88, 91, 87],
}
positions = [(0, 0), (0, 1), (1, 0), (1, 1)]

fig, axes = plt.subplots(2, 2, figsize=(12, 8))

for (subject, scores), pos in zip(quizzes.items(), positions):
    axes[pos].hist(scores, bins=4, edgecolor="white", color="#457b9d")
    axes[pos].set_title(f"{subject} quiz scores - detailed breakdown")
    axes[pos].set_xlabel("Score")
    axes[pos].set_ylabel("Students")

fig.tight_layout()      # <- the missing line: without it the long
                        # titles overlap the row above
plt.show()
```

**7. Give the main act more stage.** `gridspec_kw` forwards layout options - `[2, 1]` gives row one twice the height of row two.

```
import matplotlib.pyplot as plt

months  = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
revenue = [420, 380, 510, 480, 590, 650]
profit  = [42, 30, 65, 52, 80, 95]

fig, axes = plt.subplots(
    2, 1,
    figsize=(10, 6),
    sharex=True,
    gridspec_kw={"height_ratios": [2, 1]},   # revenue gets 2/3 of the height
)

axes[0].plot(months, revenue, marker="o", linewidth=2.5, color="#2a6f97")
axes[0].set_title("Revenue (main story)")
axes[0].set_ylabel("Thousand BDT")

axes[1].bar(months, profit, color="#e9c46a")
axes[1].set_title("Profit (companion)")
axes[1].set_ylabel("Thousand BDT")

fig.tight_layout()
plt.show()
```

## Part 3 — Challenge

**8. The dashboard.** One dataset, four questions, one figure: suptitle is the headline, each set_title is a caption.

```
import matplotlib.pyplot as plt

weeks      = list(range(1, 9))
study_avg  = [3.0, 3.5, 4.0, 3.5, 5.0, 5.5, 6.0, 6.5]
subjects   = ["Math", "Physics", "Chemistry", "Biology", "English"]
subj_marks = [88, 74, 91, 79, 84]
all_marks  = [72, 68, 88, 54, 91, 77, 63, 85, 79, 94, 58, 81]
stu_hours  = [2, 3, 4, 5, 6, 7, 8, 4, 5, 6, 3, 9]
stu_marks  = [55, 62, 68, 74, 79, 85, 90, 66, 73, 78, 60, 92]

fig, axes = plt.subplots(2, 2, figsize=(12, 8))

# [0,0] line - trend
axes[0, 0].plot(weeks, study_avg, marker="o", color="#2a6f97")
axes[0, 0].set_title("Weekly Study Hours (class average)")
axes[0, 0].set_xlabel("Week")
axes[0, 0].set_ylabel("Hours / week")

# [0,1] bar - comparison
axes[0, 1].bar(subjects, subj_marks, color="#adb5bd")
axes[0, 1].bar(["Chemistry"], [91], color="#e76f51")   # highlight the winner
axes[0, 1].set_title("Average Mark per Subject")
axes[0, 1].set_ylim(0, 100)
axes[0, 1].tick_params(axis="x", labelrotation=30)

# [1,0] histogram - distribution
axes[1, 0].hist(all_marks, bins=6, edgecolor="white", color="#457b9d")
axes[1, 0].set_title("Distribution of Exam Marks")
axes[1, 0].set_xlabel("Mark")
axes[1, 0].set_ylabel("Number of exams")

# [1,1] scatter - relationship
axes[1, 1].scatter(stu_hours, stu_marks, s=60, alpha=0.8,
                   color="#2a9d8f", edgecolor="black")
axes[1, 1].set_title("Study Hours vs Marks")
axes[1, 1].set_xlabel("Hours studied / week")
axes[1, 1].set_ylabel("Exam mark")

fig.suptitle("Student Analytics Dashboard", fontsize=16, fontweight="bold")
fig.tight_layout(rect=[0, 0, 1, 0.96])   # headroom for the suptitle

plt.show()
```

**9. Layout by name.** Repeating a label across cells makes that panel span them; you then address panels by NAME, not by counting.

```
import matplotlib.pyplot as plt

months  = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
revenue = [420, 380, 510, 480, 590, 650]
profit  = [42, 30, 65, 52, 80, 95]
daily   = [480, 520, 610, 450, 700, 650, 720, 560, 690, 730, 610, 760]

fig, axd = plt.subplot_mosaic(
    [["main", "side"],
     ["main", "bottom"]],      # "main" spans both rows of the left column
    figsize=(11, 6),
)

axd["main"].plot(months, revenue, marker="o", color="#2a6f97", linewidth=2.5)
axd["main"].set_title("Revenue - spans both rows")

axd["side"].bar(months, profit, color="#e9c46a")
axd["side"].set_title("Profit")
axd["side"].tick_params(axis="x", labelrotation=90, labelsize=8)

axd["bottom"].hist(daily, bins=6, edgecolor="white", color="#457b9d")
axd["bottom"].set_title("Daily revenue spread")
axd["bottom"].set_xlabel("Thousand BDT")

fig.tight_layout()
plt.show()

# Names beat numbers: rearranging the layout means editing the map,
# never renumbering slots - and axd["main"] reads like English.
```
