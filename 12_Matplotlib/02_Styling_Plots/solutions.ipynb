## Part 1 — Warm-up

**1. Three ways to say a color.** Names, hex codes and RGB tuples are all accepted - internally each becomes RGBA floats between 0 and 1.

```
import matplotlib.pyplot as plt

days    = ["Sat", "Sun", "Mon", "Tue", "Wed"]
walking = [42, 51, 38, 55, 60]
running = [20, 25, 15, 30, 28]
cycling = [33, 21, 40, 26, 35]

plt.plot(days, walking, color="seagreen", label="Walking")          # CSS name
plt.plot(days, running, color="#d1495b", label="Running")           # hex code
plt.plot(days, cycling, color=(0.1, 0.3, 0.55), label="Cycling")    # RGB tuple

plt.title("Daily Activity Minutes")
plt.xlabel("Day")
plt.ylabel("Minutes")
plt.legend()

plt.show()
```

**2. Four linestyles, one story.** The codes are `-` solid, `--` dashed, `-.` dash-dot, `:` dotted; `linewidth=` is measured in points.

```
import matplotlib.pyplot as plt

months   = [1, 2, 3, 4, 5, 6]
revenue  = [42, 45, 51, 49, 58, 63]
plan     = [r * 0.85 for r in revenue]
last_year = [r * 0.7 for r in revenue]
break_even = [40] * 6

plt.plot(months, revenue, linestyle="-", linewidth=2.5, label="Actual (-)")
plt.plot(months, plan, linestyle="--", linewidth=1.8, label="Plan (--)")
plt.plot(months, last_year, linestyle="-.", linewidth=1.8, label="Last year (-.)")
plt.plot(months, break_even, linestyle=":", linewidth=1.8, label="Break-even (:)")

plt.title("Revenue Against Its References")
plt.xlabel("Month")
plt.ylabel("Thousand BDT")
plt.legend(loc="upper left")
plt.grid(True, linestyle="--", alpha=0.4)

plt.show()
```

**3. Decode the shorthand.** A fmt string packs color + marker + linestyle in any order; the letters are r g b c m y k w.

```
import matplotlib.pyplot as plt

# "ro--" -> color="red",    marker="o", linestyle="--"
# "g^-"  -> color="green",  marker="^", linestyle="-"
# "k*"   -> color="black",  marker="*", linestyle=""   (dots only)
# "yo:"  -> color="yellow", marker="o", linestyle=":"

day    = [1, 2, 3, 4, 5, 6]
city_a = [31, 33, 32, 34, 33, 35]
city_b = [24, 25, 23, 26, 27, 25]
events = [2, 3, 1, 4, 2, 5]

plt.plot(day, city_a, "ro--", label="City A highs")
plt.plot(day, city_b, "g^-", label="City B highs")
plt.plot(day, events, "k*", label="Outdoor events")

plt.title("Format Strings in Action")
plt.xlabel("Day")
plt.legend()
plt.grid(True, linestyle=":", alpha=0.6)

plt.show()
```

## Part 2 — Practice

**4. Marker gallery.** `linestyle=""` suppresses the connecting line; `markerfacecolor`/`markeredgecolor` dress each dot.

```
import matplotlib.pyplot as plt

students = ["Sarah", "Rakib", "Aisha", "Tanvir", "Nadia"]
marks    = [91, 74, 84, 67, 79]
styles   = [("o", "#1f77b4"), ("s", "#ff7f0e"), ("^", "#2ca02c"),
            ("*", "#d62728"), ("D", "#9467bd")]

for (mk, col), name, score in zip(styles, students, marks):
    plt.plot(name, score, marker=mk, markersize=12,
             markerfacecolor=col, markeredgecolor="black",
             linestyle="", label=name)

plt.title("Final Exam Scores")
plt.ylabel("Marks (out of 100)")
plt.ylim(0, 100)
plt.grid(True, axis="y", linestyle="--", alpha=0.5)

plt.show()
```

**5. Legend, placed and frameless.** Every plot call needs `label=`, then ONE `plt.legend()` - loc pins the corner, frameon=False drops the box.

```
import matplotlib.pyplot as plt

days  = ["Sat", "Sun", "Mon", "Tue", "Wed", "Thu"]
karim = [35, 42, 38, 51, 47, 55]
juthi = [30, 36, 44, 40, 52, 49]
hasan = [25, 31, 29, 38, 41, 44]
priya = [40, 38, 48, 55, 50, 62]

plt.plot(days, karim, marker="o", label="Karim")
plt.plot(days, juthi, marker="s", label="Juthi")
plt.plot(days, hasan, marker="^", label="Hasan")
plt.plot(days, priya, marker="*", markersize=11, label="Priya")

plt.title("Weekly Delivery Earnings")
plt.xlabel("Day")
plt.ylabel("Earnings (BDT hundreds)")
plt.legend(loc="upper left", frameon=False, fontsize=10)
plt.grid(True, linestyle="--", alpha=0.5)

plt.show()
```

**6. See through the tangle.** alpha is opacity from 0.0 to 1.0; low values are reserved for overlapping layers, never main data alone.

```
import matplotlib.pyplot as plt

epochs = [1, 2, 3, 4, 5, 6, 7, 8]
runs = [
    [3.2, 2.9, 2.7, 2.5, 2.3, 2.2, 2.1, 2.0],
    [3.5, 3.4, 3.0, 2.8, 2.6, 2.5, 2.4, 2.3],
    [4.1, 3.6, 3.3, 3.0, 2.9, 2.7, 2.6, 2.6],
    [2.9, 2.8, 2.6, 2.4, 2.2, 2.1, 2.0, 1.9],
]

for path in runs:
    plt.plot(epochs, path, alpha=0.35, color="#1f77b4")

plt.title("alpha = 0.35 - overlap stays readable")
plt.xlabel("Epoch")
plt.ylabel("Training loss")
plt.grid(True, linestyle="--", alpha=0.4)

plt.show()

# At alpha=1.0 the four lines would paint over each other;
# translucent strokes reveal where they cluster and cross.
```

## Part 3 — Challenge

**7. Theme makeover.** `plt.style.use(name)` flips a GLOBAL switch that persists until you call `plt.style.use("default")`.

```
import numpy as np
import matplotlib.pyplot as plt

print("ggplot installed:", "ggplot" in plt.style.available)

plt.style.use("ggplot")              # flip the global theme switch

x = np.linspace(0, 10, 100)

plt.plot(x, np.sin(x), linewidth=2.5, label="sin(x)")
plt.plot(x, np.cos(x), linewidth=2.5, label="cos(x)")

plt.title("AFTER - ggplot Style")
plt.xlabel("x")
plt.ylabel("value")
plt.legend()

plt.show()

plt.style.use("default")             # reset - otherwise EVERY later plot
# in the notebook silently inherits the grey ggplot look.
```

**8. Dressed for the report.** Build a text hierarchy: big bold headline, medium labels, small ticks.

```
import matplotlib.pyplot as plt

months  = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
revenue = [42, 45, 51, 49, 58, 63]      # thousand BDT

plt.plot(months, revenue, color="#2a6f97", linewidth=2.5, marker="o")

plt.title("Monthly Revenue - First Half", fontsize=16, fontweight="bold")
plt.xlabel("Month", fontsize=12)
plt.ylabel("Revenue (thousand BDT)", fontsize=12)
plt.tick_params(axis="both", labelsize=10)
plt.grid(True, linestyle="--", alpha=0.5)

plt.show()
```
