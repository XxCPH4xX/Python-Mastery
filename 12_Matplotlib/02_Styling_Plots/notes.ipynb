# Matplotlib Styling

> 📘 **Python Mastery** · Module 12 — Matplotlib · Lesson 2/5

A default chart works; a *styled* chart persuades. In this lesson you take
control of color, line style, markers, transparency, legends, fonts, and even
whole built-in themes — the difference between "it runs" and "it communicates".

## 🎯 Learning Objectives

By the end of this lesson you will be able to:

- **Apply** colors using names, hex codes, and RGB tuples
- **Choose** linestyle codes (`-`, `--`, `-.`, `:`) and marker codes (`o`, `s`, `^`, `*`)
- **Write** compact format strings like `'ro--'` and expand them back to keyword arguments
- **Control** overlap readability with `alpha`
- **Build** legends with `label=` + `legend(loc=...)`
- **Restyle** entire charts instantly with style templates such as `'ggplot'`

## 1. Color — Three Ways to Say It

Every colored element accepts a `color=` argument, and Matplotlib happily
accepts three common formats:

| Format | Example | When to use it |
|---|---|---|
| CSS name | `"steelblue"` | Quick experiments |
| Hex code | `"#FF5733"` | Matching brand/report colors |
| RGB tuple | `(0.2, 0.4, 0.8)` | Programmatic colors; floats from 0 to 1 |

> 🔍 **Under the Hood:** internally every color becomes an RGBA tuple of floats
> between 0 and 1 — `"red"` is literally converted to `(1.0, 0.0, 0.0, 1.0)`.
> And when you *don't* pass a color, the Axes pulls the next entry from its
> property cycle (`axes.prop_cycle`), the familiar sequence nicknamed **C0**
> blue `#1f77b4`, **C1** orange `#ff7f0e`, **C2** green `#2ca02c`, ... up to C9.
> That is why consecutive unstyled lines always come out blue, orange, green.

**Syntax:**
```python
plt.plot(x, y, color="steelblue")       # 1. CSS-style name
plt.plot(x, y, color="#FF5733")         # 2. hex code (like web design)
plt.plot(x, y, color=(0.2, 0.4, 0.8))   # 3. RGB tuple, floats 0-1
```

**Example:** one dataset, three product lines, each colored a different way.
(`label=` feeds the legend — dissected properly in section 6.)

````python
import matplotlib.pyplot as plt

months   = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
tea      = [120, 132, 128, 145, 160, 155]     # kg of tea sold
coffee   = [90, 96, 110, 105, 122, 130]
softdrink = [60, 55, 70, 88, 95, 101]

plt.plot(months, tea, color="steelblue", label="Tea")          # name
plt.plot(months, coffee, color="#8B4513", label="Coffee")      # hex (saddle brown)
plt.plot(months, softdrink, color=(0.85, 0.33, 0.10), label="Soft drinks")  # RGB

plt.title("Cafe Sales by Product")
plt.xlabel("Month")
plt.ylabel("Units sold")
plt.legend()      # shows the labels; more on this soon

plt.show()
````

## 2. Linestyles — Four Characters to Remember

The `linestyle=` argument has four single-character codes worth memorizing:

| Code | Name | Typical use |
|---|---|---|
| `'-'` | solid (default) | Real/measured data |
| `'--'` | dashed | Predictions, targets, references |
| `'-.'` | dash-dot | Secondary reference lines |
| `':'` | dotted | Guides, thresholds, guesses |

Pair them with `linewidth=` (often written `lw=`), measured in points.

**Syntax:**
```python
plt.plot(x, y, linestyle="--", linewidth=2)
```

**Example:** actual monthly visitors against three kinds of reference line.

````python
import matplotlib.pyplot as plt

months  = [1, 2, 3, 4, 5, 6]
actual  = [820, 940, 1010, 1180, 1240, 1410]

plt.plot(months, actual, linestyle="-",  linewidth=2.5, label="Actual (-)")
plt.plot([m * 0.75 for m in months], linestyle="--", linewidth=1.8, label="Forecast (--)")
plt.plot([m * 0.5 for m in months],  linestyle="-.", linewidth=1.8, label="Last year (-.)")
plt.plot([600] * len(months),        linestyle=":",  linewidth=1.8, label="Break-even (:)")

plt.title("Linestyle Cheat Sheet")
plt.xlabel("Month")
plt.ylabel("Visitors")
plt.legend(loc="upper left")
plt.grid(True, linestyle="--", alpha=0.4)

plt.show()
````

## 3. Markers — Putting Dots on Data

Lines hide individual points. Markers expose them — vital for small datasets,
lab measurements, or exam scores where each dot IS a data point.

| Code | Marker | Vibe |
|---|---|---|
| `'o'` | circle | The all-purpose default |
| `'s'` | square | Blocky, technical |
| `'^'` | triangle-up | Points upward, good for growth |
| `'*'` | star | Highlight series |
| `'D'` | diamond | Rare but readable |

Useful companions: `markersize=` (`ms=`), `markerfacecolor=` (`mfc=`),
`markeredgecolor=` (`mec=`).

**Syntax:**
```python
plt.plot(x, y, marker="o", markersize=10)
```

**Example:** five students' final-exam scores, one bold marker each.

````python
import matplotlib.pyplot as plt

students = ["Sarah", "Rakib", "Aisha", "Tanvir", "Nadia"]
marks    = [88, 74, 91, 67, 83]

styles = [  # (marker code, fill color) - one row per student
    ("o", "#1f77b4"),
    ("s", "#ff7f0e"),
    ("^", "#2ca02c"),
    ("*", "#d62728"),
    ("D", "#9467bd"),
]

for (mk, col), name, score in zip(styles, students, marks):
    plt.plot(name, score, marker=mk, markersize=12,
             markerfacecolor=col, markeredgecolor="black",
             linestyle="", label=name)   # linestyle="" -> dots only, no line

plt.title("Final Exam Scores")
plt.ylabel("Marks (out of 100)")
plt.ylim(0, 100)
plt.grid(True, axis="y", linestyle="--", alpha=0.5)

plt.show()
````

## 4. The Compact Format String — `'ro--'`

Matplotlib's oldest shortcut packs **color + marker + linestyle** into one tiny
string: `'ro--'` reads as **r**ed **o**circles joined by **--**dashed lines.

| Format string | Means |
|---|---|
| `'ro'` | red circles, no connecting line |
| `'g--'` | green dashed line |
| `'b^-'` | blue triangles, solid line |
| `'k*'` | black stars, no line |
| `'yo:'` | yellow dots, dotted line |
| `'ro--'` | red circles + dashed red line |

Order inside the string does not matter. These single-letter colors are available:
`r`ed, `g`reen, `b`lue, `c`yan, `m`agenta, `y`ellow, `k` black, `w`hite.

**Syntax:**
```python
plt.plot(x, y, "ro--")                        # compact...
plt.plot(x, y, color="red", marker="o", linestyle="--")   # ...exactly equivalent
```

**Example:** both spellings side by side on the same data.

````python
import matplotlib.pyplot as plt

week    = [1, 2, 3, 4, 5, 6]
battery = [100, 87, 76, 58, 40, 22]     # phone battery % at day's end

plt.plot(week, battery, "ro--", label="'ro--' (compact)")
plt.plot([w + 0.15 for w in week], [b - 6 for b in battery],
         color="red", marker="o", linestyle="--", label="keyword version")

plt.title("Battery Drain Through the Week")
plt.xlabel("Day of week")
plt.ylabel("Battery (%)")
plt.legend(loc="upper right")
plt.grid(True, linestyle=":", alpha=0.6)

plt.show()

# Same shape, same color, same dashes - the fmt string is pure shorthand.
````

## 5. Transparency — the `alpha` Knob

`alpha` sets opacity from 0.0 (invisible) to 1.0 (solid). It is THE tool when
many elements overlap: drop it to ~0.3 and dense regions turn dark while
individual marks stay visible.

**Syntax:**
```python
plt.plot(x, y, alpha=0.3)    # ghostly - great for many overlapping series
```

**Example:** 30 random-walk study sessions plotted twice — first solid (a mess),
then with alpha. Seeded NumPy keeps the data identical every run.

````python
import numpy as np
import matplotlib.pyplot as plt

rng_data = []                       # 30 deterministic random walks
np.random.seed(42)
for session in range(30):
    rng_data.append(np.cumsum(np.random.randn(60)))

plt.figure(figsize=(11, 4))

plt.subplot(1, 2, 1)                # left: everything fully opaque
for path in rng_data:
    plt.plot(path)
plt.title("alpha = 1.0 - unreadable tangle")

plt.subplot(1, 2, 2)                # right: same paths, translucent
for path in rng_data:
    plt.plot(path, alpha=0.25, color="#1f77b4")
plt.title("alpha = 0.25 - density emerges")

plt.tight_layout()
plt.show()
````

## 6. Legends — Naming Your Lines

A legend maps each line to its name. The recipe never changes:

1. Add `label="..."` to **every** `plot` call.
2. Call `plt.legend()` once before `show()`.

Handy options:

| Option | Effect |
|---|---|
| `loc="best"` | Matplotlib picks the emptiest corner (this is the default) |
| `loc="upper left"` / `"lower right"` / ... | Pin it to a specific corner |
| `ncol=2` | Split entries into columns (good for dashboards) |
| `frameon=False` | Drop the box — cleaner look |
| `fontsize=11` | Control text size |

**Syntax:**
```python
plt.plot(x, y1, label="Revenue")
plt.plot(x, y2, label="Cost")
plt.legend(loc="best")
```

**Example:** four delivery riders' weekly earnings compared in one chart.

````python
import matplotlib.pyplot as plt

days    = ["Sat", "Sun", "Mon", "Tue", "Wed", "Thu"]
karim   = [35, 42, 38, 51, 47, 55]
juthi   = [30, 36, 44, 40, 52, 49]
hasan   = [25, 31, 29, 38, 41, 44]
priya   = [40, 38, 48, 55, 50, 62]

plt.plot(days, karim, marker="o", label="Karim")
plt.plot(days, juthi, marker="s", label="Juthi")
plt.plot(days, hasan, marker="^", label="Hasan")
plt.plot(days, priya, marker="*", markersize=11, label="Priya")

plt.title("Weekly Delivery Earnings (BDT hundreds)")
plt.xlabel("Day")
plt.ylabel("Earnings")
plt.legend(loc="upper left", frameon=False, fontsize=10)
plt.grid(True, linestyle="--", alpha=0.5)

plt.show()
````

## 7. Text That Talks — Titles, Labels, Font Sizes

Any text-bearing call accepts `fontsize=` (points) and `fontweight=`
(`"normal"`, `"bold"`). Set up a clear hierarchy: big bold title, medium axis
labels, small ticks.

**Syntax:**
```python
plt.title("Headline", fontsize=16, fontweight="bold")
plt.xlabel("X meaning", fontsize=12)
plt.tick_params(axis="both", labelsize=10)   # tick number/label size
```

**Example:** the same cafe-sales chart, now dressed for a report.

````python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
revenue = [42, 45, 51, 49, 58, 63]      # thousand BDT

plt.plot(months, revenue, color="#2a6f97", linewidth=2.5, marker="o")

plt.title("Monthly Revenue - First Half", fontsize=16, fontweight="bold")
plt.xlabel("Month", fontsize=12)
plt.ylabel("Revenue (thousand BDT)", fontsize=12)
plt.tick_params(axis="both", labelsize=10)
plt.grid(True, linestyle="--", alpha=0.5)

plt.show()
````

## 8. Style Templates — One Line, Whole New Look

Matplotlib ships ready-made themes called **style sheets**. Two calls matter:

- `plt.style.available` — list every installed theme
- `plt.style.use("name")` — activate one for ALL subsequent plots

Popular choices: `"ggplot"` (grey panel, white gridlines — mimics R's ggplot2),
`"seaborn-v0_8"` (soft statistical look), `"dark_background"` (slides!), and
`"fivethirtyeight"` (news-graphics look).

⚠️ `use()` flips a **global switch**: every later cell inherits the theme until
you call `plt.style.use("default")`.

**Syntax:**
```python
print(plt.style.available)   # peek at all installed themes
plt.style.use("ggplot")      # activate one (global!)
plt.style.use("default")     # ...and how to undo it
```

**Example:** first the default look...

````python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)

plt.plot(x, np.sin(x), linewidth=2.5, label="sin(x)")
plt.plot(x, np.cos(x), linewidth=2.5, label="cos(x)")

plt.title("BEFORE - Default Style")
plt.xlabel("x")
plt.ylabel("value")
plt.legend()

plt.show()
````

````python
import matplotlib.pyplot as plt
import numpy as np

print("Installed styles:", plt.style.available)

plt.style.use("ggplot")              # flip the global theme switch

x = np.linspace(0, 10, 100)

plt.plot(x, np.sin(x), linewidth=2.5, label="sin(x)")
plt.plot(x, np.cos(x), linewidth=2.5, label="cos(x)")

plt.title("AFTER - ggplot Style")
plt.xlabel("x")
plt.ylabel("value")
plt.legend()

plt.show()

plt.style.use("default")             # reset so later lessons stay standard
````

## 9. Teaser — Annotating the Maximum Point

Two ideas combine here: find the peak **in the data** (never hard-code it!) with
`max()` / `np.argmax()`, then place text there with `annotate`. Full arrow-drawn
callouts arrive in Lesson 5 — for now, a labelled point.

**Syntax:**
```python
peak_index = int(np.argmax(y))            # WHERE the maximum sits
plt.annotate("text", xy=(x[peak_index], y[peak_index]))   # what & where
```

**Example:** mark the busiest month on a website-traffic chart.

````python
import numpy as np
import matplotlib.pyplot as plt

months  = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug"]
visits  = [4200, 4800, 5100, 6300, 5900, 7200, 6800, 7500]

plt.plot(months, visits, color="#d62728", linewidth=2.5, marker="o")

peak_i = int(np.argmax(visits))                     # index of the biggest value
plt.annotate(f"Peak: {visits[peak_i]:,} visits",
             xy=(peak_i, visits[peak_i]),
             xytext=(peak_i - 2.8, visits[peak_i] + 250),
             fontweight="bold", color="#7f1d1d")

plt.title("Monthly Website Visits")
plt.xlabel("Month")
plt.ylabel("Visits")
plt.grid(True, linestyle="--", alpha=0.5)

plt.show()
````

## ⚠️ Common Mistakes & Gotchas

| Mistake | Problem | Fix |
|---|---|---|
| Putting a hex code inside a fmt string (`'#ff0000o--'`) | Invalid format-string error | Use basic letters in fmt, or pass `color=` separately: `plt.plot(x, y, "o--", color="#ff0000")` |
| Calling `plt.legend()` with no `label=` anywhere | Empty/warning: *"No artists with labels found"* | Add `label=` to each `plot` call first |
| Forgetting that `style.use` is global | Every later plot silently wears the old theme | Reset with `plt.style.use("default")` after demos |
| Red next to green for two key series | Unreadable for ~1 in 12 men with color-vision deficiency | Prefer blue/orange pairs (Matplotlib defaults already do this) |
| Stacking decoration on decoration | Chart noise hides the data | ≤ 4 series, one accent color, restraint |
| `alpha` below 0.3 on main data | Washed-out, hard-to-print charts | Reserve low alpha for background/overlap layers |

## 💡 Best Practices & Pro Tips

- **Color follows the entity.** If Tea is steelblue in chart 1, it stays steelblue
  in chart 5 — readers learn your palette like a language.
- Reach for perceptually uniform colormaps (`viridis`, `magma`) whenever color
  encodes a *value* — they are colorblind-safe by design.
- `frameon=False` plus a corner `loc` declutters a legend more than any other tweak.
- Use `with plt.style.context("ggplot"):` to borrow a theme for exactly one figure
  without flipping the global switch.
- 🤖 **AI-engineering relevance:** training dashboards live and die by styling —
  train vs validation loss must be distinguishable at a glance, and paper-ready
  figures need deliberate fonts, weights, and colors long before submission.

## 📌 Summary

| Tool | What it does | Example |
|---|---|---|
| `color=` | Any name / hex / RGB-tuple color | `color="#FF5733"` |
| `linestyle=` | `-`, `--`, `-.`, `:` | `linestyle="--"` |
| `linewidth=` | Stroke thickness in points | `linewidth=2.5` |
| `marker=` + `markersize=` | Dots on data: `o s ^ * D` | `marker="o", ms=10` |
| `"ro--"` | Compact color+marker+linestyle | `plt.plot(x, y, "ro--")` |
| `alpha=` | Opacity 0.0–1.0 for overlaps | `alpha=0.3` |
| `label=` + `plt.legend(loc=...)` | Name and place the legend | `legend(loc="upper left")` |
| `fontsize=` / `fontweight=` | Text hierarchy | `fontweight="bold"` |
| `plt.style.use(...)` | Apply a whole theme globally | `use("ggplot")` |

**Key takeaways**

- Format strings are shorthand; keywords are self-documenting — mix freely.
- Legends require `label=` on the plots, then one `legend()` call.
- `alpha` turns overlapping clutter into readable density.
- Style sheets restyle everything at once — remember they persist until reset.
