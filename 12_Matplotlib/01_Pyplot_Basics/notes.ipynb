## 2. Anatomy of a Matplotlib Figure

Every Matplotlib image has three layers. Learn these words now — the entire
library speaks this language:

| Layer | What it is | Think of it as |
|---|---|---|
| **Figure** | The whole window/image — the outer canvas that can hold many plots | The photo frame |
| **Axes** | ONE individual plot: its own data area, its own title, its own x and y | The photograph inside the frame |
| **Axis** | The number lines themselves — `xaxis` and `yaxis`, with ticks and tick labels | The rulers glued to the photo |

⚠️ Naming trap: an **Axes** (with an "e") is a *whole plot*, while an **Axis**
is one ruler inside it. Almost everyone trips over this exactly once.

> 🔍 **Under the Hood:** `pyplot` is a *state machine*. When you call
> `plt.plot(...)`, Matplotlib silently asks "is there a *current* Figure?" — that
> is `plt.gcf()`, *get current figure* — and creates one if none exists. It then
> grabs the *current Axes* (`plt.gca()`) and attaches a `Line2D` **artist** to it.
> Every visible thing — lines, ticks, titles, the grid — is an Artist in a tree
> owned by the Figure, and `plt.show()` walks that tree and paints it to the screen.
> This is also why calling `plt.plot` twice stacks a *second* line onto the *same*
> plot instead of erasing the first.

**Syntax:** the general shape of every pyplot program.

```python
import matplotlib.pyplot as plt

plt.plot(x_values, y_values)   # draw data into the current Axes
plt.title("...")               # decorations attach to the current Axes
plt.show()                     # render the finished figure — always last
```

**Example:** Sarah logs her weekly study hours while preparing for finals. Two
plain lists are all a line plot needs.

````python
# Sarah's weekly study hours, logged by hand
import matplotlib.pyplot as plt

weeks = [1, 2, 3, 4, 5, 6]       # x-axis: week number
hours = [4, 7, 6, 9, 11, 14]     # y-axis: hours studied that week

plt.plot(weeks, hours)           # connects the dots, left to right
plt.show()
````

## 3. Passing Only One List

Give `plt.plot` a single list and it uses the list *positions* (0, 1, 2, ...) as
the x-values automatically. Handy whenever the x-spacing is uniform anyway.

**Syntax:**
```python
plt.plot(y_values)   # x is implied: 0, 1, 2, ...
```

**Example:** a player's score after each level of a mobile game — the level
numbers are implied by position.

````python
import matplotlib.pyplot as plt

scores = [0, 50, 120, 180, 260, 310, 420, 500]   # cumulative score per level

plt.plot(scores, marker="s")     # x is implied: 0, 1, ..., 7

plt.title("Score After Each Level")
plt.xlabel("Level")              # the implied positions 0..7 ARE the levels
plt.ylabel("Total score")
plt.show()
````

## 4. Titles and Axis Labels

An unlabelled chart is a mystery box: rising *what*, measured in *what*?
Three one-line calls fix that forever.

**Syntax:**
```python
plt.title("Chart headline")      # what the chart shows
plt.xlabel("x-axis meaning")     # meaning/unit of the horizontal ruler
plt.ylabel("y-axis meaning")     # meaning/unit of the vertical ruler
```

**Example:** label Sarah's study chart so anyone can read it without asking.

````python
import matplotlib.pyplot as plt

weeks = [1, 2, 3, 4, 5, 6]
hours = [4, 7, 6, 9, 11, 14]

plt.plot(weeks, hours, marker="o")

plt.title("Sarah's Weekly Study Hours")   # the headline
plt.xlabel("Week number")                 # horizontal ruler
plt.ylabel("Hours studied")               # vertical ruler

plt.show()
````

## 5. Gridlines

Gridlines are faint reference lines that let a reader trace any point back to
its value on the ruler — essential when people read exact numbers off a chart.

**Syntax:**
```python
plt.grid(True)                    # plain grid
plt.grid(True, linestyle="--")    # dashed grid — softer, less dominant
```

**Example:** monthly sales for a small bookshop in Dhaka, drawn over a dashed grid.

````python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales  = [320, 290, 410, 380, 450, 520]       # books sold per month

plt.plot(months, sales, marker="o")            # marker="o" = dot on each point

plt.title("Bookshop Sales - First Half of the Year")
plt.xlabel("Month")
plt.ylabel("Books sold")
plt.grid(True, linestyle="--", alpha=0.7)      # alpha softens the grid further

plt.show()
````

## 6. Multiple Lines in One Figure

Call `plt.plot` again *before* `plt.show()` and the new line lands on the same
pair of Axes — exactly what you want when comparing series. Notice you did not
pick any colors: Matplotlib cycles through its default palette automatically
(blue → orange → green → red ...), so consecutive lines stay distinguishable.

**Syntax:**
```python
plt.plot(x1, y1)   # line 1 — automatic blue
plt.plot(x2, y2)   # line 2 — automatic orange, SAME figure
plt.plot(x3, y3)   # line 3 — automatic green
plt.show()         # one figure, three lines
```

**Example:** three friends compare daily study hours during exam week.

````python
import matplotlib.pyplot as plt

days  = ["Sat", "Sun", "Mon", "Tue", "Wed"]
sarah = [2, 3, 5, 4, 6]
rakib = [1, 2, 2, 4, 5]
aisha = [3, 1, 4, 3, 7]

plt.plot(days, sarah)    # automatic blue
plt.plot(days, rakib)    # automatic orange
plt.plot(days, aisha)    # automatic green

plt.title("Exam-Week Study Hours")
plt.xlabel("Day")
plt.ylabel("Hours studied")
plt.grid(True, linestyle="--")

plt.show()

# But which color belongs to WHO? That is the job of the legend — next lesson.
````

## 7. Plotting Math Functions with NumPy

Real curves — sound waves, activation functions, probability distributions —
come from formulas, not lists. Sample the formula at many x-values and plot;
the smooth "curve" is really hundreds of tiny straight segments.

**Syntax:**
```python
import numpy as np

x = np.linspace(start, stop, num_points)   # evenly spaced x values
y = formula(x)                             # evaluated for ALL points at once
plt.plot(x, y)
```

**Example:** the sine wave — the "hello world" of scientific plotting.

````python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi, 200)   # 200 evenly spaced points from 0 to 2π
y = np.sin(x)                        # computed for every point in one shot

plt.plot(x, y)

plt.title("y = sin(x)")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.grid(True, linestyle="--")

plt.show()
````

## 8. `plt.show()` — the Full Stop of Plotting

`plt.show()` renders everything accumulated since the last `show()`, then clears
the deck. Rules of thumb:

- Call it **once, at the end** of each chart.
- In **Jupyter**, it draws the figure inline right below the cell.
- In a plain **`.py` script**, it opens an interactive window — skip it and the
  script may finish having shown *nothing*.
- After `show()`, the next `plt.plot` starts a brand-new figure.

**Syntax:**
```python
plt.plot(...)        # build
plt.title(...)       # decorate
plt.show()           # THEN display — nothing plot-related after this
```

**Example:** two charts produced by one cell — the first `show()` separates them.

````python
import matplotlib.pyplot as plt

week  = [1, 2, 3, 4, 5, 6]
study = [4, 7, 6, 9, 11, 14]
sleep = [6, 7, 8, 6, 7, 8]

plt.plot(week, study, marker="o")
plt.title("Study Hours per Week")
plt.show()                     # figure 1 rendered ...

plt.plot(week, sleep, marker="o", color="green")
plt.title("Sleep Hours per Week")
plt.show()                     # ... figure 2 rendered separately
````

````python
# Proof that show() clears the deck: after it, plotting starts FRESH
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]

plt.plot(x, [2, 4, 6, 8, 10])
plt.title("Chart A")
plt.show()                      # canvas emptied after rendering

plt.plot(x, [10, 8, 6, 4, 2])   # this begins a brand-new figure
plt.title("Chart B - a fresh canvas")
plt.show()
````

## 9. 📌 Recipe Card — the Standard 6-Line Plot

Ninety percent of your quick plots are this exact template. Memorize it once,
reuse it forever.

| Line | Code | Job |
|---|---|---|
| 1 | `import matplotlib.pyplot as plt` | bring the toolbox |
| 2 | `plt.plot(x, y)` | draw the data |
| 3 | `plt.title("...")` | say what the chart is |
| 4 | `plt.xlabel("...")` + `plt.ylabel("...")` | label the rulers |
| 5 | `plt.grid(True, linestyle="--")` | make it readable |
| 6 | `plt.show()` | render it |

**Example:** the recipe applied end-to-end — average monthly temperature in Dhaka.

````python
import matplotlib.pyplot as plt                        # 1. toolbox

month  = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]    # the data
temp_c = [19, 22, 27, 30, 30, 29]

plt.plot(month, temp_c, marker="o")                    # 2. data
plt.title("Dhaka Average Temperature (Jan-Jun)")       # 3. headline
plt.xlabel("Month")                                    # 4. rulers
plt.ylabel("Temperature (°C)")
plt.grid(True, linestyle="--")                         # 5. grid
plt.show()                                             # 6. render
````

## ⚠️ Common Mistakes & Gotchas

| Mistake | Problem | Fix |
|---|---|---|
| `x` and `y` of different lengths | `ValueError: x and y must have the same size` | Check `len(x) == len(y)` before plotting |
| Forgetting `plt.show()` in a `.py` script | Script ends, no window ever appears | One `plt.show()` at the end |
| Calling `plt.show()` mid-chart (after `title`, before `grid`) | Chart gets cut in half; stray empty figures appear | Decorate first, `show()` last |
| Expecting a second `plt.plot` to replace the first | Lines pile up on one figure | Start a clean slate with `plt.figure()` |
| Confusing **Axes** with **Axis** | Misreading docs and error messages | Axes = the plot; Axis = one ruler |
| Publishing charts without labels | Nobody — including future you — knows the units | `title` + `xlabel` + `ylabel`, always |

## 💡 Best Practices & Pro Tips

- Stick to the canonical import `import matplotlib.pyplot as plt`. Never
  `from matplotlib.pyplot import *` — it dumps 100+ functions into your namespace.
- One message per chart. If your title needs the word "and", you probably want two charts.
- Label first, beautify later: a labelled ugly chart beats a beautiful mystery.
- Notebook cells share one Python session — after restarting the kernel, re-run
  cells top-to-bottom so imports and variables exist again.
- 🤖 **AI-engineering relevance:** the first move after loading any dataset — or
  after any training epoch — is a plot. Loss curves, prediction-vs-truth scatters,
  feature histograms: Matplotlib is how you *see* whether a model misbehaves
  before you trust its metrics.

## 📌 Summary

| Method | What it does | Example |
|---|---|---|
| `plt.plot(x, y)` | Draws a line through the points | `plt.plot(weeks, hours)` |
| `plt.plot(y)` | Uses positions 0, 1, 2, ... as x | `plt.plot(scores)` |
| `plt.title(t)` | Adds the chart headline | `plt.title("Sales")` |
| `plt.xlabel(t)` / `plt.ylabel(t)` | Labels the x / y ruler | `plt.xlabel("Month")` |
| `plt.grid(True, linestyle="--")` | Adds a dashed reference grid | `plt.grid(True, linestyle="--")` |
| `plt.plot(...)` again | Stacks another auto-colored line | see §6 |
| `plt.show()` | Renders the figure — always last | `plt.show()` |

**Key takeaways**

- Figure = canvas, Axes = the plot, Axis = a ruler — this vocabulary unlocks the docs.
- `pyplot` draws into the *current* Axes; repeated `plot` calls stack rather than replace.
- Every chart deserves a title, both axis labels, and `plt.show()` at the very end.
- The 6-line recipe is your default skeleton — everything in the next lessons hangs off it.
