## Part 1 — Warm-up

**1. Plumbing check.** Everyone imports pyplot as `plt`; the version lives on the parent `matplotlib` module.

```
import matplotlib
import matplotlib.pyplot as plt   # the convention everyone uses

print("matplotlib version:", matplotlib.__version__)
```

**2. Sarah's study log.** `plt.plot(x, y)` connects the points left to right; title and labels decorate the current Axes.

```
import matplotlib.pyplot as plt

weeks = [1, 2, 3, 4, 5, 6]       # x-axis: week number
hours = [4, 7, 6, 9, 11, 14]     # y-axis: hours studied

plt.plot(weeks, hours, marker="o")

plt.title("Sarah's Weekly Study Hours")
plt.xlabel("Week number")
plt.ylabel("Hours studied")

plt.show()
```

**3. Positions as x-values.** With a single list, plot uses positions 0, 1, 2, ... as the x-values automatically.

```
import matplotlib.pyplot as plt

scores = [0, 50, 120, 180, 260, 310]

plt.plot(scores, marker="s")     # x is implied: 0, 1, ..., 5

plt.title("Score After Each Level")
plt.xlabel("Level")
plt.ylabel("Total score")

plt.show()
```

## Part 2 — Practice

**4. A softer background.** `plt.grid(True, linestyle="--", alpha=0.7)` draws faint dashed rulers readers can trace values along.

```
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales  = [320, 290, 410, 380, 450, 520]

plt.plot(months, sales, marker="o")

plt.title("Bookshop Sales - First Half of the Year")
plt.xlabel("Month")
plt.ylabel("Books sold")
plt.grid(True, linestyle="--", alpha=0.7)

plt.show()
```

**5. Three friends, one canvas.** Repeated `plot` calls stack onto the same Axes; the default palette cycles blue, orange, green, ...

```
import matplotlib.pyplot as plt

days  = ["Sat", "Sun", "Mon", "Tue", "Wed"]
sarah = [2, 3, 5, 4, 6]
rakib = [1, 2, 2, 4, 5]
aisha = [3, 1, 4, 3, 7]

plt.plot(days, sarah)    # automatic blue
plt.plot(days, rakib)    # automatic orange
plt.plot(days, aisha)    # automatic green

plt.title("Exam-Week Study Hours")
plt.xlabel("Day")
plt.ylabel("Hours studied")
plt.grid(True, linestyle="--")

plt.show()

# Colors were never chosen: the default cycle made them blue/orange/green.
# (Which line is WHOSE? That needs labels + a legend - Lesson 2.)
```

**6. A formula, not a list.** linspace samples the formula at many points; the smooth curve is really hundreds of tiny segments.

```
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi, 200)
y = np.sin(x)

plt.plot(x, y)

plt.title("y = sin(x)")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.grid(True, linestyle="--")

plt.show()
```

## Part 3 — Challenge

**7. show() separates.** `plt.show()` renders everything built so far and clears the deck - the next plot starts a brand-new figure.

```
import matplotlib.pyplot as plt

week  = [1, 2, 3, 4]
study = [4, 7, 6, 9]
sleep = [6, 7, 8, 6]

plt.plot(week, study, marker="o")
plt.title("Study Hours")
plt.show()                     # figure 1 rendered, canvas emptied ...

plt.plot(week, sleep, marker="o", color="green")
plt.title("Sleep Hours - a fresh canvas")
plt.show()                     # ... so this is a brand-new figure
```

**8. The six-line recipe.** toolbox, data, headline, rulers, grid, render - ninety percent of quick plots are exactly this skeleton.

```
import matplotlib.pyplot as plt                        # 1. toolbox

month  = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]    # the data
temp_c = [19, 22, 27, 30, 30, 29]

plt.plot(month, temp_c, marker="o")                    # 2. data
plt.title("Dhaka Average Temperature (Jan-Jun)")       # 3. headline
plt.xlabel("Month")                                    # 4. rulers
plt.ylabel("Temperature (°C)")
plt.grid(True, linestyle="--")                         # 5. grid
plt.show()                                             # 6. render
```
