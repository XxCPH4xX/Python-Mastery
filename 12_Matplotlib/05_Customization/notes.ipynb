## 1. Axis Limits — Zooming with Intent

`xlim(low, high)` / `ylim(low, high)` crop the view to where the story lives.
Two legitimate uses: focusing on the interesting window, and giving bars their
honest zero baseline. One illegitimate use: cropping to *fake* a trend.

**Syntax:**
```python
plt.xlim(start_x, end_x)
plt.ylim(bottom, top)
```

**Example:** a full year of sales first, then zoomed into the spring surge.

````python
import matplotlib.pyplot as plt

month_num = list(range(1, 13))
month_lbl = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
             "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
sales = [310, 295, 340, 520, 610, 640, 430, 415, 450, 480, 700, 780]

plt.figure(figsize=(11, 4))

plt.subplot(1, 2, 1)
plt.plot(month_num, sales, marker="o", color="#2a6f97")
plt.title("Full year")
plt.xticks(month_num, month_lbl, rotation=45)
plt.grid(True, linestyle="--", alpha=0.5)

plt.subplot(1, 2, 2)
plt.plot(month_num, sales, marker="o", color="#2a6f97")
plt.xlim(2.5, 6.5)                 # zoom: March..June only
plt.ylim(250, 700)                 # tighten y too
plt.title("Zoomed - the spring surge")
plt.xticks(range(3, 7), ["Mar", "Apr", "May", "Jun"])
plt.grid(True, linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
# Same data! Always say WHEN you zoom - cropped axes can mislead.
````

## 2. Tick Control — Rotation for Crowded Labels

Long category or date labels overlap into an unreadable smear.
`xticks(rotation=45)` turns them; adding `ha="right"` anchors each label to its
tick instead of drifting sideways. Works for dates, product names, anything wide.

**Syntax:**
```python
plt.xticks(rotation=45, ha="right")     # classic readable combo
plt.tick_params(axis="y", labelsize=9)  # shrink tick text if needed
```

**Example:** six product names that refuse to fit horizontally.

````python
import matplotlib.pyplot as plt

products = ["Instant Noodles", "Milk Chocolate", "Green Tea",
            "Basmati Rice 5kg", "Olive Oil 1L", "Protein Bars"]
units    = [820, 640, 590, 750, 310, 275]

plt.figure(figsize=(9, 4))
plt.bar(products, units, color="#2a9d8f")

plt.title("Units Sold by Product")
plt.ylabel("Units")

plt.xticks(rotation=45, ha="right")      # rotate AND anchor right
plt.tick_params(axis="y", labelsize=9)
plt.grid(True, axis="y", linestyle="--", alpha=0.5)

plt.show()

# Try deleting rotation=45 mentally... six overlapping smudges.
````

## 3. Annotations & Threshold Lines

Three tools turn a chart into commentary:

| Tool | Job |
|---|---|
| `axhline(y)` | Full-width **horizontal** line — targets, averages, limits |
| `axvline(x)` | Full-width **vertical** line — launch dates, cut-offs |
| `annotate(text, xy=..., xytext=..., arrowprops=...)` | Arrow calling out ONE point |

Find the point to annotate **from the data** (`np.argmax`), never hard-coded —
so the annotation survives new data.

**Syntax:**
```python
peak_i = int(np.argmax(values))
plt.annotate("Best month!",
             xy=(x[peak_i], values[peak_i]),        # arrow TIP
             xytext=(x[peak_i] - 2, values[peak_i] + 60),   # text home
             arrowprops=dict(arrowstyle="->", color="gray"))
plt.axhline(target, linestyle="--", linewidth=1.5, label="Target")
```

**Example:** monthly users against a growth target, best month flagged.

````python
import numpy as np
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
users  = [1200, 1500, 1800, 2600, 2400, 3100,
          2900, 3400, 3300, 4100, 4800, 5200]
target = 4000

plt.figure(figsize=(10, 5))
plt.plot(months, users, marker="o", linewidth=2.5, color="#2a6f97", label="Monthly users")
plt.axhline(target, linestyle="--", linewidth=1.5, color="#e76f51", label=f"Target ({target:,})")

peak_i = int(np.argmax(users))            # WHERE is the peak? ask the data!
plt.annotate(f"Peak: {users[peak_i]:,}",
             xy=(peak_i, users[peak_i]),
             xytext=(peak_i - 3.2, users[peak_i] - 900),
             fontsize=11, fontweight="bold", color="#264653",
             arrowprops=dict(arrowstyle="->", color="gray"))

plt.title("App Users vs Growth Target")
plt.ylabel("Active users")
plt.legend(loc="upper left")
plt.grid(True, linestyle="--", alpha=0.4)

plt.show()
````

````python
# Vertical reference lines: mark WHEN something happened.
import matplotlib.pyplot as plt

weeks        = ["W1", "W2", "W3", "W4", "W5", "W6", "W7", "W8"]
study_hours  = [3, 4, 3, 7, 8, 6, 9, 10]

plt.figure(figsize=(9, 4))
plt.plot(weeks, study_hours, marker="o", linewidth=2.5, color="#2a6f97",
         label="Study hours")

exam_week = weeks.index("W4")                 # WHERE does finals season start?
plt.axvline(exam_week, linestyle="-.", linewidth=1.6, color="#e76f51",
            label="Finals begin - week " + weeks[exam_week])

plt.title("Vertical Line = Event Marker")
plt.ylabel("Hours studied")
plt.legend(loc="upper left")
plt.grid(True, linestyle="--", alpha=0.4)

plt.show()
````

## 4. Minimal Charts — Removing Spines

Spines are the four border lines around every Axes (top, right, bottom, left).
Charts read cleaner when only the two working rulers remain:

```python
fig, ax = plt.subplots()
...
ax.spines[["top", "right"]].set_visible(False)
```

(`ax.spines[["top", "right"]]` uses fancy indexing to grab both at once.)
The philosophy: **ink belongs to data** — borders and heavy grids are noise.
This is the look of The Economist / FiveThirtyEight style graphics.

**Syntax:**
```python
ax.spines[["top", "right"]].set_visible(False)   # keep bottom + left
ax.grid(axis="y", linestyle=":", alpha=0.5)      # recessive grid too
```

**Example:** before-and-after in one figure.

````python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 12, 100)
y = 50 + 18 * np.sin(x / 1.9)

fig, axes = plt.subplots(1, 2, figsize=(11, 4), sharey=True)

axes[0].plot(x, y, color="#2a6f97", linewidth=2)
axes[0].set_title("Before - four spines + full grid")
axes[0].grid(True)

axes[1].plot(x, y, color="#2a6f97", linewidth=2.5)
axes[1].set_title("After - minimal, ink on data")
axes[1].spines[["top", "right"]].set_visible(False)   # drop the dead borders
axes[1].grid(axis="y", linestyle=":", alpha=0.5)

plt.show()
````

## 5. Log Scale — When Data Spans Orders of Magnitude

Some data grows *multiplicatively*: viral posts, bacteria colonies, compound
interest, model parameter counts over the years. A linear axis flattens the
early history into the floor; a **log axis** gives every power of ten equal room.

- `plt.semilogy()` — log y-axis, linear x
- Also exist: `semilogx()`, `loglog()`, and OO `ax.set_yscale("log")`
- Diagnostic bonus: exponential growth becomes a STRAIGHT LINE on a log scale

⚠️ Logs are undefined for 0 and negatives — filter those out first.

**Syntax:**
```python
plt.semilogy(x, y)              # y jumps powers of 10: 1, 10, 100, ...
```

**Example:** a video doubling its views daily — plotted both ways.

````python
import numpy as np
import matplotlib.pyplot as plt

day   = np.arange(0, 10)
views = 250 * 2 ** day          # doubles every day: 250 ... 128,000

plt.figure(figsize=(11, 4))

plt.subplot(1, 2, 1)
plt.plot(day, views, marker="o", color="#e76f51")
plt.title("Linear scale - early days vanish")
plt.xlabel("Day")
plt.ylabel("Views")

plt.subplot(1, 2, 2)
plt.semilogy(day, views, marker="o", linewidth=2.5, color="#2a9d8f")
plt.title("Log scale - constant % growth = straight line")
plt.xlabel("Day")
plt.ylabel("Views (log)")

plt.tight_layout()
plt.show()
````

## 6. `imshow` — Your First Heatmap

A 2-D array of numbers becomes a picture: every cell gets a color from a
colormap (`cmap=`). Heatmaps show matrices — activity calendars, correlation
tables, confusion matrices, and literally images.

> 🔍 **Under the Hood:** this is deeper than it looks — digital images ARE 2-D
> arrays (grayscale) or 3-D arrays of shape `(height, width, 3)` for RGB.
> `imshow` maps each number through a normalization into RGBA pixels. When you
> later do computer vision and call `plt.imshow(image_array)`, it will be this
> exact function. Row 0 draws at the TOP (array order), which is why photos
> appear upright while math-style matrices look flipped.

Handy arguments: `cmap="viridis"` (colorblind-safe default choice), `aspect="auto"`
(so non-square matrices fill the panel), plus `xticks`/`yticks` to label rows/columns.

**Syntax:**
```python
plt.imshow(matrix_2d, cmap="viridis", aspect="auto")
plt.colorbar(label="what the colors measure")
```

**Example:** focus scores across a week — days × time blocks.

````python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(21)
days   = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
blocks = ["06-09", "09-12", "12-15", "15-18", "18-21", "21-24"]

base = np.array([[55, 85, 70, 75, 60, 20],    # hand-shaped weekly rhythm
                 [60, 88, 68, 72, 62, 25],
                 [58, 82, 66, 74, 64, 22],
                 [52, 86, 69, 70, 58, 30],
                 [50, 78, 60, 55, 48, 35],
                 [35, 42, 50, 62, 70, 45],
                 [30, 38, 46, 55, 60, 40]])
focus = np.clip(base + np.random.randint(-6, 7, base.shape), 0, 100)

plt.figure(figsize=(8, 5))
plt.imshow(focus, cmap="viridis", aspect="auto")

plt.colorbar(label="Focus score")
plt.xticks(range(len(blocks)), blocks)
plt.yticks(range(len(days)), days)
plt.title("Weekly Focus Heatmap - darker green = deep work")

plt.show()
# Read it like a calendar: weekday mornings glow; Sunday evening glows least.
````

````python
# The bridge to computer vision: an IMAGE is just an array.
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(5)
rgb_image = np.random.randint(0, 256, (12, 16, 3))   # height x width x RGB

plt.figure(figsize=(9, 4))

plt.subplot(1, 2, 1)
plt.imshow(rgb_image)                    # full color - shape (12, 16, 3)
plt.title("A random RGB 'image'")

plt.subplot(1, 2, 2)
plt.imshow(rgb_image[:, :, 0], cmap="magma")   # just the RED channel = 2-D
plt.title("Red channel alone - a plain heatmap")
plt.colorbar(label="Red intensity")

plt.show()
# Same function, same idea: numbers in, colors out.
````

## 7. Global Defaults — `rcParams`

Repeating `figsize`, fontsizes, and colors on every call gets old. Matplotlib
keeps every default in one dictionary-like config: `plt.rcParams`. Set it once
at the top of a notebook and every subsequent figure obeys.
(`plt.rcdefaults()` restores factory settings.)

**Syntax:**
```python
plt.rcParams["figure.figsize"] = [10, 5]     # units are inches
plt.rcParams["font.size"] = 12
plt.rcParams["axes.titlesize"] = 15
```

**Example:** restyle EVERY chart in three lines, then prove it.

````python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams["figure.figsize"] = [9, 4]      # every new figure now 9x4
plt.rcParams["font.size"] = 11
plt.rcParams["axes.titleweight"] = "bold"
plt.rcParams["lines.linewidth"] = 2.5

x = np.linspace(0, 10, 100)

plt.plot(x, np.sin(x), label="sin(x)")       # no styling args needed!
plt.plot(x, np.sin(2 * x), label="sin(2x)")
plt.title("All defaults applied automatically")
plt.legend()

plt.show()

plt.rcdefaults()      # be polite: restore factory settings for later lessons
````

## 8. 🏁 FINALE — One Publication-Quality Chart

Every technique from five lessons, assembled deliberately:

| Element | Where it came from |
|---|---|
| `fig, ax = plt.subplots(figsize=...)` | Lesson 4 — OO convention |
| Custom hex colors, `linewidth`, markers | Lesson 2 — styling |
| Dashed target via `axhline` + legend entry | Lesson 5 §3 |
| Arrow annotation computed from data | Lessons 2 & 5 |
| Rotated ticks, `ylim` focus, subtle grid | Lessons 1 & 5 |
| Spines removed for a minimal frame | Lesson 5 §4 |

**Example:** monthly revenue for a cafe chain — report-ready.

````python
import numpy as np
import matplotlib.pyplot as plt

# ---- data ---------------------------------------------------------
months  = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
           "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
revenue = [420, 380, 510, 480, 590, 650,
           610, 640, 720, 760, 910, 1050]         # thousand BDT
target  = 800                                    # annual run-rate goal

# ---- canvas (OO convention) ---------------------------------------
fig, ax = plt.subplots(figsize=(11, 5.5))

# ---- the data line (styled) ---------------------------------------
ax.plot(months, revenue, color="#1d3557", linewidth=2.8,
        marker="o", markersize=7,
        markerfacecolor="#e63946", markeredgecolor="white",
        label="Monthly revenue")

# ---- threshold line -------------------------------------------------
ax.axhline(target, linestyle="--", linewidth=1.6, color="#e76f51",
           label=f"Annual target ({target}k)")

# ---- annotation, computed FROM the data -----------------------------
peak_i = int(np.argmax(revenue))
ax.annotate(f"Record month: {months[peak_i]} ({revenue[peak_i]}k)",
            xy=(peak_i, revenue[peak_i]),
            xytext=(peak_i - 4.4, revenue[peak_i] - 95),
            fontsize=11, fontweight="bold", color="#1d3557",
            arrowprops=dict(arrowstyle="->", color="#457b9d"))

# ---- axes, ticks, zoom ----------------------------------------------
ax.set_ylim(300, 1150)                       # focus on the action
ax.set_title("Cafe Chain Revenue 2025 - Report Figure", fontsize=15,
             fontweight="bold", pad=14)
ax.set_xlabel("Month", fontsize=12)
ax.set_ylabel("Revenue (thousand BDT)", fontsize=12)
ax.tick_params(axis="both", labelsize=10)
ax.set_xticks(range(len(months)))
ax.set_xticklabels(months, rotation=45, ha="right")

# ---- minimal furniture ----------------------------------------------
ax.spines[["top", "right"]].set_visible(False)
ax.grid(axis="y", linestyle=":", alpha=0.5)

# ---- legend, then render --------------------------------------------
ax.legend(loc="upper left", frameon=False, fontsize=11)

fig.tight_layout()
plt.show()
````

## ⚠️ Common Mistakes & Gotchas

| Mistake | Problem | Fix |
|---|---|---|
| Cropping `ylim` on BAR charts to exaggerate change | Dishonest chart — differences look huge | Bars start at 0; reserve tight `ylim` for lines/points |
| `rotation=45` without `ha="right"` | Labels drift away from their ticks | Use them together: `xticks(rotation=45, ha="right")` |
| Hard-coding annotation coordinates | Breaks the moment data changes | Compute positions: `np.argmax`, `max`, index arithmetic |
| Log scale with zeros/negatives present | Points silently vanish; warnings appear | Filter to positives before `semilogy`/`set_yscale("log")` |
| Removing ALL spines and ticks | Chart loses every reference frame | Keep bottom + left spines; drop only top/right |
| Forgetting `aspect="auto"` on tall/wide matrices | `imshow` squeezes the array into a tiny square strip | Pass `aspect="auto"` for matrix heatmaps |

## 💡 Best Practices & Pro Tips

- Set global defaults once per notebook instead of repeating kwargs:
  `plt.rcParams["figure.figsize"] = [10, 5]` and friends.
- Annotate sparingly — ONE arrow per message. Three arrows compete and none win.
- Pick one accent color for "the point" and gray everything contextual.
- Outside notebooks, export figures for reports with
  `fig.savefig("chart.png", dpi=300)` — high DPI for print, PNG/SVG for slides.
- 🤖 **AI-engineering relevance:** these are daily tools — log-scale loss curves
  (losses span decades), attention-matrix heatmaps via `imshow` (transformers!),
  confusion matrices, and annotated validation curves in stakeholder decks.

## 📌 Summary

| Method | What it does | Example |
|---|---|---|
| `plt.xlim(a, b)` / `plt.ylim(a, b)` | Zoom the visible range | `xlim(2.5, 6.5)` |
| `plt.xticks(rotation=45, ha="right")` | Fit long/dated tick labels | see §2 |
| `plt.annotate(text, xy=, xytext=, arrowprops=)` | Arrow callout on one point | peak labelling |
| `plt.axhline(v, ls="--")` / `axvline(v)` | Horizontal / vertical reference line | target line |
| `ax.spines[["top", "right"]].set_visible(False)` | Minimal-chart borders | clean frames |
| `plt.semilogy(x, y)` | Log y-axis for multiplicative data | viral growth |
| `plt.imshow(m, cmap="viridis", aspect="auto")` + `colorbar` | 2-D array → heatmap | focus calendar |
| `plt.rcParams[...]` | Global defaults for the session | `figure.figsize` |

**Key takeaways**

- Zoom to reveal, never to deceive — and always announce cropped axes.
- Annotate from data (`np.argmax`), so insights survive updated numbers.
- Less ink, more signal: drop top/right spines, use recessive grids.
- Log scales make multiplicative growth readable — and straight lines diagnostic.
- You now own the full pipeline: data → styled, annotated, minimal, dashboard-grade figures.
