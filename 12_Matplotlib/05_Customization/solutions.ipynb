## Part 1 — Warm-up

**1. Crop to the story.** `xlim`/`ylim` crop the view to where the story lives - legitimate for focus, dishonest if used to fake a trend.

```
import matplotlib.pyplot as plt

month_num = list(range(1, 13))
sales = [310, 295, 340, 520, 610, 640, 430, 415, 450, 480, 700, 780]

plt.figure(figsize=(9, 4))
plt.plot(month_num, sales, marker="o", color="#2a6f97")

plt.xlim(2.5, 6.5)                 # zoom: March..June only
plt.ylim(250, 700)                 # tighten y too
plt.xticks(range(3, 7), ["Mar", "Apr", "May", "Jun"])

plt.title("Zoomed - the spring surge")
plt.grid(True, linestyle="--", alpha=0.5)

plt.show()

# Same data as the full year, but cropped axes can mislead -
# always SAY when the view has been narrowed.
```

**2. Rotate the smear away.** Rotation turns the labels; `ha="right"` anchors each one to ITS tick instead of drifting sideways.

```
import matplotlib.pyplot as plt

products = ["Instant Noodles", "Milk Chocolate", "Green Tea",
            "Basmati Rice 5kg", "Olive Oil 1L", "Protein Bars"]
units    = [820, 640, 590, 750, 310, 275]

plt.figure(figsize=(9, 4))
plt.bar(products, units, color="#2a9d8f")

plt.title("Units Sold by Product")
plt.ylabel("Units")

plt.xticks(rotation=45, ha="right")      # rotate AND anchor right
plt.tick_params(axis="y", labelsize=9)
plt.grid(True, axis="y", linestyle="--", alpha=0.5)

plt.show()

# rotation=45 alone leaves labels drifting away from their ticks;
# ha="right" pins the label's right end under its own bar.
```

**3. Draw the finish line.** `axhline(y)` draws a full-width horizontal reference - perfect for targets and averages.

```
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
users  = [1200, 1500, 1800, 2600, 2400, 3100,
          2900, 3400, 3300, 4100, 4800, 5200]
target = 4000

plt.figure(figsize=(10, 5))
plt.plot(months, users, marker="o", linewidth=2.5,
         color="#2a6f97", label="Monthly users")
plt.axhline(target, linestyle="--", linewidth=1.5, color="#e76f51",
            label=f"Target ({target:,})")

plt.title("App Users vs Growth Target")
plt.ylabel("Active users")
plt.legend(loc="upper left")
plt.grid(True, linestyle="--", alpha=0.4)

plt.show()
```

## Part 2 — Practice

**4. Mark the moment.** `axvline(x)` flags a moment in time; computing the position from the data keeps the chart honest when data changes.

```
import matplotlib.pyplot as plt

weeks       = ["W1", "W2", "W3", "W4", "W5", "W6", "W7", "W8"]
study_hours = [3, 4, 3, 7, 8, 6, 9, 10]

plt.figure(figsize=(9, 4))
plt.plot(weeks, study_hours, marker="o", linewidth=2.5, color="#2a6f97",
         label="Study hours")

exam_week = weeks.index("W4")                 # WHERE does finals season start?
plt.axvline(exam_week, linestyle="-.", linewidth=1.6, color="#e76f51",
            label="Finals begin - week " + weeks[exam_week])

plt.title("Vertical Line = Event Marker")
plt.ylabel("Hours studied")
plt.legend(loc="upper left")
plt.grid(True, linestyle="--", alpha=0.4)

plt.show()
```

**5. Point at the peak.** Annotate FROM the data (`np.argmax`), so the callout survives tomorrow's updated numbers.

```
import numpy as np
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug"]
visits = [4200, 4800, 5100, 6300, 5900, 7200, 6800, 7500]

plt.figure(figsize=(9, 5))
plt.plot(months, visits, marker="o", linewidth=2.5, color="#d62728")

peak_i = int(np.argmax(visits))               # ask the DATA where the peak is
plt.annotate(f"Peak: {visits[peak_i]:,}",
             xy=(peak_i, visits[peak_i]),
             xytext=(peak_i - 2.5, visits[peak_i] - 900),
             fontsize=11, fontweight="bold", color="#264653",
             arrowprops=dict(arrowstyle="->", color="gray"))

plt.title("Monthly Website Visits")
plt.ylabel("Visits")
plt.grid(True, linestyle="--", alpha=0.5)

plt.show()

# Hard-coding (7, 7500) dies the moment new data arrives;
# argmax recomputes for free.
```

**6. Ink belongs to data.** Drop only top and right; bottom + left remain as the working rulers - borders and heavy grids are noise.

```
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(9, 4))

months  = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
revenue = [420, 380, 510, 480, 590, 650]

ax.plot(months, revenue, color="#2a6f97", linewidth=2.5, marker="o")
ax.set_title("Minimal - ink belongs to data")
ax.set_ylabel("Thousand BDT")

ax.spines[["top", "right"]].set_visible(False)   # drop the dead borders
ax.grid(axis="y", linestyle=":", alpha=0.5)      # recessive grid

fig.tight_layout()
plt.show()

# Bottom + left stay: they are the actual rulers being read.
```

**7. Straighten the growth.** `semilogy` gives every power of ten equal room; constant-percentage growth becomes a straight LINE.

```
import matplotlib.pyplot as plt

day   = list(range(10))
views = [250, 500, 1000, 2000, 4000,
         8000, 16000, 32000, 64000, 128000]

plt.figure(figsize=(11, 4))

plt.subplot(1, 2, 1)
plt.plot(day, views, marker="o", color="#e76f51")
plt.title("Linear scale - early days vanish")
plt.xlabel("Day")
plt.ylabel("Views")

plt.subplot(1, 2, 2)
plt.semilogy(day, views, marker="o", linewidth=2.5, color="#2a9d8f")
plt.title("Log scale - doubling reads as a straight line")
plt.xlabel("Day")
plt.ylabel("Views (log)")

plt.tight_layout()
plt.show()

# Doubling is constant-PERCENTAGE growth: on a log axis every x2 step
# is the same distance, so the curve straightens out. (Zeros/negatives
# would vanish silently - logs need positive values.)
```

## Part 3 — Challenge

**8. Calendar in colors.** `imshow` maps a 2-D array through a colormap cell by cell; `aspect="auto"` lets non-square matrices fill the panel.

```
import matplotlib.pyplot as plt

days   = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
blocks = ["06-09", "09-12", "12-15", "15-18", "18-21", "21-24"]
focus  = [
    [55, 85, 70, 75, 60, 20],
    [60, 88, 68, 72, 62, 25],
    [58, 82, 66, 74, 64, 22],
    [52, 86, 69, 70, 58, 30],
    [50, 78, 60, 55, 48, 35],
    [35, 42, 50, 62, 70, 45],
    [30, 38, 46, 55, 60, 40],
]

plt.figure(figsize=(8, 5))
plt.imshow(focus, cmap="viridis", aspect="auto")

plt.colorbar(label="Focus score")
plt.xticks(range(len(blocks)), blocks)
plt.yticks(range(len(days)), days)
plt.title("Weekly Focus Heatmap - brighter = deep work")

plt.show()

# Read it like a calendar: weekday mornings glow brightest;
# late-night weekend cells are the darkest.
```

**9. Set it once.** rcParams restyle every later figure session-wide; `rcdefaults()` politely hands factory settings back.

```
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams["figure.figsize"] = [9, 4]      # every new figure now 9x4
plt.rcParams["font.size"] = 11
plt.rcParams["axes.titleweight"] = "bold"
plt.rcParams["lines.linewidth"] = 2.5

x = np.linspace(0, 10, 100)

plt.plot(x, np.sin(x), label="sin(x)")       # no styling args needed!
plt.plot(x, np.sin(2 * x), label="sin(2x)")
plt.title("All defaults applied automatically")
plt.legend()

plt.show()

plt.rcdefaults()      # restore factory settings so later cells stay standard

# Outside notebooks, reports want files: fig.savefig("chart.png", dpi=300)
```
