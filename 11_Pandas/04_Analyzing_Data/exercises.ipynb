## Part 2 — Practice

**4. The `numeric_only` law.** Call `students.mean()` on the full mixed table wrapped in `try`/`except TypeError` and print the error message. Then print the fix: `students.mean(numeric_only=True)`.

**6. Podium, no full sort.** From the seven-student table print the top 3 by score (names and scores only) using `nlargest`, then the bottom 2 with `nsmallest`.
