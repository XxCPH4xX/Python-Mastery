**2. Counting categories.** On the same seven-student table, print: `city.value_counts()`, the same counts as proportions rounded to 2 decimals, then `city.unique()` and `city.nunique()`. How many distinct cities are there?

```
import pandas as pd

students = pd.DataFrame({
    "name":        ["Sarah", "Rafi", "Nabila", "Karim", "Mim", "Tanvir", "Jony"],
    "age":         [22, 25, 21, 23, 20, 24, 22],
    "city":        ["Dhaka", "Sylhet", "Dhaka", "Chattogram", "Khulna", "Dhaka", "Sylhet"],
    "study_hours": [2, 4, 1, 5, 3, 4, 2],
    "score":       [88, 92, 79, 95, 84, 90, 74],
})
print(students["city"].value_counts())                          # Dhaka wins: 3
print()
print(students["city"].value_counts(normalize=True).round(2))   # fractions of 7
print()
print(students["city"].unique(), "| distinct:", students["city"].nunique())
```

**3. Typical numbers.** From the students table print: the median score, the score standard deviation rounded to 2, the total weekly study hours, and the 75th percentile of scores.

```
import pandas as pd

students = pd.DataFrame({
    "name":        ["Sarah", "Rafi", "Nabila", "Karim", "Mim", "Tanvir", "Jony"],
    "age":         [22, 25, 21, 23, 20, 24, 22],
    "city":        ["Dhaka", "Sylhet", "Dhaka", "Chattogram", "Khulna", "Dhaka", "Sylhet"],
    "study_hours": [2, 4, 1, 5, 3, 4, 2],
    "score":       [88, 92, 79, 95, 84, 90, 74],
})
print("median score:", students["score"].median())            # 88
print("std:", round(students["score"].std(), 2))
print("total study hours:", students["study_hours"].sum())    # 21
print("75th percentile:", students["score"].quantile(0.75))   # 91.0
```

## Part 2 — Practice

**4. The `numeric_only` law.** Call `students.mean()` on the full mixed table wrapped in `try`/`except TypeError` and print the error message. Then print the fix: `students.mean(numeric_only=True)`.

```
import pandas as pd

students = pd.DataFrame({
    "name":        ["Sarah", "Rafi", "Nabila", "Karim", "Mim", "Tanvir", "Jony"],
    "age":         [22, 25, 21, 23, 20, 24, 22],
    "city":        ["Dhaka", "Sylhet", "Dhaka", "Chattogram", "Khulna", "Dhaka", "Sylhet"],
    "study_hours": [2, 4, 1, 5, 3, 4, 2],
    "score":       [88, 92, 79, 95, 84, 90, 74],
})
try:
    students.mean()
except TypeError as e:
    print("TypeError:", e)
print()
print(students.mean(numeric_only=True))    # the fix: skip non-numeric columns
```

**5. count vs size.** Build `messy = pd.DataFrame({"name": ["Sarah", "Rafi", None, "Karim"], "score": [88, None, 79, 95]})`. Print the table, `messy.count().to_dict()`, `messy.size`, and the number of missing cells computed as `size - count().sum()`.

```
import pandas as pd

messy = pd.DataFrame({
    "name":  ["Sarah", "Rafi", None,     "Karim"],
    "score": [88,      None,  79,        95],
})
print(messy)
print()
print("non-null per column:", messy.count().to_dict())    # present values only
print("total cells:", messy.size)                          # rows x columns
print("missing cells:", messy.size - messy.count().sum())  # 2
```

**6. Podium, no full sort.** From the seven-student table print the top 3 by score (names and scores only) using `nlargest`, then the bottom 2 with `nsmallest`.

**7. Sort within groups.** Sort the students by city A→Z and, WITHIN each city, score high→low — pass lists to both `by=` and `ascending=`. Print just the `city`, `name` and `score` columns of the sorted copy.

```
import pandas as pd

students = pd.DataFrame({
    "name":        ["Sarah", "Rafi", "Nabila", "Karim", "Mim", "Tanvir", "Jony"],
    "age":         [22, 25, 21, 23, 20, 24, 22],
    "city":        ["Dhaka", "Sylhet", "Dhaka", "Chattogram", "Khulna", "Dhaka", "Sylhet"],
    "study_hours": [2, 4, 1, 5, 3, 4, 2],
    "score":       [88, 92, 79, 95, 84, 90, 74],
})
ranked = students.sort_values(by=["city", "score"], ascending=[True, False])
print(ranked[["city", "name", "score"]])   # Chattogram first; Dhaka trio best->worst
```

## Part 3 — Challenge

**8. Do hours move scores.** Compute `students.corr(numeric_only=True).round(2)` on the seven-student table and print the whole matrix. Then pull the SINGLE number relating `study_hours` to `score` out of it with `.loc` and print that. Comment: what do its sign and its closeness to 1 suggest — and why is it still only a hint?

```
import pandas as pd

students = pd.DataFrame({
    "name":        ["Sarah", "Rafi", "Nabila", "Karim", "Mim", "Tanvir", "Jony"],
    "age":         [22, 25, 21, 23, 20, 24, 22],
    "city":        ["Dhaka", "Sylhet", "Dhaka", "Chattogram", "Khulna", "Dhaka", "Sylhet"],
    "study_hours": [2, 4, 1, 5, 3, 4, 2],
    "score":       [88, 92, 79, 95, 84, 90, 74],
})
corr = students.corr(numeric_only=True).round(2)
print(corr)
print()
r = corr.loc["study_hours", "score"]
print("study_hours vs score:", r)
# Positive and close to +1: more study tends to mean higher scores.
# Still a hint, not causation.
```

**9. Query like English, chain like a recipe.** Using the students table: (a) print everyone aged 21+ living in Dhaka with `.query` (mind the quote styles!); (b) store `min_score = 90` in a variable and query `score >= @min_score`; (c) chain four steps in one parenthesised expression: keep Dhaka students → select `name`/`study_hours`/`score` → sort by score descending → reset the index.

```
import pandas as pd

students = pd.DataFrame({
    "name":        ["Sarah", "Rafi", "Nabila", "Karim", "Mim", "Tanvir", "Jony"],
    "age":         [22, 25, 21, 23, 20, 24, 22],
    "city":        ["Dhaka", "Sylhet", "Dhaka", "Chattogram", "Khulna", "Dhaka", "Sylhet"],
    "study_hours": [2, 4, 1, 5, 3, 4, 2],
    "score":       [88, 92, 79, 95, 84, 90, 74],
})
print(students.query("age >= 21 and city == 'Dhaka'"))   # double out, single in
print()
min_score = 90                                            # ordinary variable
print(students.query("score >= @min_score")[["name", "score"]])
print()
report = (
    students
    .query("city == 'Dhaka'")                     # 1. keep Dhaka students
    .loc[:, ["name", "study_hours", "score"]]     # 2. useful columns only
    .sort_values("score", ascending=False)        # 3. best first
    .reset_index(drop=True)                       # 4. tidy labels 0..n-1
)
print(report)
```
