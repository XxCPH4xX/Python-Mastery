## Part 2 — Practice

**4. One bracket or two.** From a small table with `name` and `score` columns: store `df["score"]` in `one` and `df[["name", "score"]]` in `many`. Print BOTH objects and their types. Which came back as a Series and which as a smaller DataFrame?

```
import pandas as pd

students = pd.DataFrame({
    "name":  ["Sarah", "Rafi", "Nabila"],
    "score": [88, 92, 79],
})

one = students["score"]              # single column -> Series
many = students[["name", "score"]]   # list of columns -> DataFrame

print(one)
print()
print(many)
print()
print(type(one).__name__, "|", type(many).__name__)
```

**5. Columns from columns.** Start from a `cart` table: `item` pen/notebook/backpack, `price` 25/120/1450, `qty` 4/2/1. Derive a `total` column (`price * qty`) and a boolean `in_budget` column (`total <= 500`). Print the finished four-column table.

```
import pandas as pd

cart = pd.DataFrame({
    "item":  ["pen", "notebook", "backpack"],
    "price": [25, 120, 1450],
    "qty":   [4, 2, 1],
})

cart["total"] = cart["price"] * cart["qty"]     # derived column
cart["in_budget"] = cart["total"] <= 500        # derived BOOLEAN column
print(cart)
```

**6. Drop and rename.** Still holding Q5's `cart` (all four columns): drop the `in_budget` column, then rename `qty` to `quantity`. Print the result, then prove the ORIGINAL `cart` still has all its columns — what does that tell you about `drop` and `rename`?

```
import pandas as pd

cart = pd.DataFrame({
    "item":  ["pen", "notebook", "backpack"],
    "price": [25, 120, 1450],
    "qty":   [4, 2, 1],
})
cart["total"] = cart["price"] * cart["qty"]
cart["in_budget"] = cart["total"] <= 500

tidy = cart.drop("in_budget", axis=1)           # axis=1 -> a COLUMN
tidy = tidy.rename(columns={"qty": "quantity"})
print(tidy)
print()
# Original untouched: drop/rename hand back copies
print(list(cart.columns))
```

**7. Slice surprise.** Build a `weather` table with `city` Dhaka/Sylhet/Khulna/Barishal and `temp` 33/29/31/30, indexed by `["Sat", "Sun", "Mon", "Tue"]`. Print `weather.loc["Sat":"Mon"]` and `weather.iloc[0:2]`. Count the rows of each in a comment — why does the LABEL slice grab three rows while the POSITION slice grabs two?

```
import pandas as pd

weather = pd.DataFrame({
    "city": ["Dhaka", "Sylhet", "Khulna", "Barishal"],
    "temp": [33, 29, 31, 30],
}, index=["Sat", "Sun", "Mon", "Tue"])

print(weather.loc["Sat":"Mon"])   # END INCLUDED -> Sat, Sun, Mon (3 rows)
print()
print(weather.iloc[0:2])          # end EXCLUDED -> rows 0 and 1 (2 rows)
```

## Part 3 — Challenge

**8. Write only through `.loc`.** Build `students` with `name` Sarah/Rafi/Nabila/Karim, `grade` B/A/C/A and `score` 88/92/79/95. Using ONE `.loc` line, give everyone scoring 90 or more an `"A+"`. Then fix Nabila's score to 81 by addressing that exact cell. Finally copy the frame into `broken` and attempt the chained shortcut `broken[broken["score"] > 90]["grade"] = "X"`; print the grades afterwards — where did the X go?

```
import pandas as pd

students = pd.DataFrame({
    "name":  ["Sarah", "Rafi", "Nabila", "Karim"],
    "grade": ["B", "A", "C", "A"],
    "score": [88, 92, 79, 95],
})

students.loc[students["score"] >= 90, "grade"] = "A+"   # conditional write
students.loc[2, "score"] = 81                            # one exact cell
print(students)

broken = students.copy()
broken[broken["score"] > 90]["grade"] = "X"   # chained assignment: no effect
print()
print(broken["grade"].tolist(), "<- unchanged: the X landed on a throwaway copy")
```

**9. A meaningful index.** Take a students table keyed by ID: `student_id` S001/S002/S003, `name` Sarah/Rafi/Nabila, `score` 88/92/79. Promote `student_id` to be the row labels with `set_index`, then fetch Rafi's whole row BY ID. Finish with `reset_index()` and note what happened to the ID.

```
import pandas as pd

students = pd.DataFrame({
    "student_id": ["S001", "S002", "S003"],
    "name":       ["Sarah", "Rafi", "Nabila"],
    "score":      [88, 92, 79],
})

indexed = students.set_index("student_id")
print(indexed)
print()
print(indexed.loc["S002"])        # fetch by ID -- no position counting
print()
print(indexed.reset_index())      # undo: ID back to being a column
```
