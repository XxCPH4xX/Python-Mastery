## Part 1 — Warm-up

**1. Your first Series.** Create `scores = pd.Series([88, 92, 79, 95], name="final_score")` and print the whole Series, then its `shape` and its `size`. In a comment: which numbers are the labels and which are the values?

```
import pandas as pd

scores = pd.Series([88, 92, 79, 95], name="final_score")
print(scores)
print()
# Labels are 0..3 on the left; the scores are the values on the right.
print("shape:", scores.shape, "| size:", scores.size)   # (4,) | 4
```

**2. Labels for free.** Build a Series named `population_millions` from the dictionary `{"Dhaka": 22.5, "Chattogram": 5.4, "Khulna": 2.0, "Sylhet": 3.9}`. Print the Series, then print just the labels as a plain list using `.index`. What did pandas take the labels from?

```
import pandas as pd

population = pd.Series(
    {"Dhaka": 22.5, "Chattogram": 5.4, "Khulna": 2.0, "Sylhet": 3.9},
    name="population_millions",
)
print(population)
print()
print(list(population.index))   # keys became the labels
```

**3. Position or label.** Build `temps = pd.Series([31.2, 33.8, 30.5, 29.9], index=["Sat", "Sun", "Mon", "Tue"], name="temp_c")`. Then print: the first value by POSITION, the last value by position (negative!), the value LABELED `"Mon"`, and the shorthand form for the value labeled `"Tue"`.

```
import pandas as pd

temps = pd.Series([31.2, 33.8, 30.5, 29.9],
                  index=["Sat", "Sun", "Mon", "Tue"], name="temp_c")

print(temps.iloc[0])      # first POSITION
print(temps.iloc[-1])     # negative positions work too
print(temps.loc["Mon"])   # by LABEL
print(temps["Tue"])       # shorthand for temps.loc["Tue"]
```

## Part 2 — Practice

**4. Look before you compute.** Build `stock = pd.Series({"pen": 120, "book": 45, "bag": 18}, name="units_in_stock")` and print, one per line: `.index`, `.values`, `.dtype`, `.name`, then `.shape` and `.size` together.

```
import pandas as pd

stock = pd.Series({"pen": 120, "book": 45, "bag": 18}, name="units_in_stock")

print(stock.index)             # the labels
print(stock.values)            # plain NumPy array underneath
print(stock.dtype)             # int64
print(stock.name)              # units_in_stock
print(stock.shape, stock.size) # (3,) 3
```

**5. Vectorized math.** With `prices = pd.Series([150, 80, 1200], index=["book", "pen", "bag"], name="price")`: put `prices * 2` into `doubled` and print it. Then subtract a discount Series of `[50, 50, 50]` carrying the SAME index and print that result. No loops allowed!

```
import pandas as pd

prices = pd.Series([150, 80, 1200], index=["book", "pen", "bag"], name="price")

doubled = prices * 2                       # every element at once
print(doubled)
print()

discounted = prices - pd.Series([50, 50, 50], index=["book", "pen", "bag"])
print(discounted)                          # labels matched -> clean result
```

**6. The alignment surprise.** Write down your prediction FIRST, then run: `mon = pd.Series([100, 200, 300], index=["Dhaka", "Sylhet", "Khulna"])`, `tue = pd.Series([10, 20], index=["Sylhet", "Barishal"])`, `total = mon + tue`. Print `total` AND `total.dtype`. Comment on why Sylhet got 210 while Dhaka shows NaN, and why an int+int sum reports float64.

```
import pandas as pd

mon = pd.Series([100, 200, 300], index=["Dhaka", "Sylhet", "Khulna"])
tue = pd.Series([10, 20],         index=["Sylhet", "Barishal"])

total = mon + tue                 # aligned BY LABEL, not by position
print(total)
# Sylhet matched: 200 + 10 = 210. Dhaka/Khulna exist only in mon,
# Barishal only in tue -> nothing to pair with -> NaN.
print(total.dtype)                # float64 -- NaN is a float, ints upcast
```

**7. Ask, then filter.** Given `scores = pd.Series([88, 92, 79, 95, 61], index=["Sarah", "Rafi", "Nabila", "Karim", "Mim"])`: store the mask `scores >= 80` in a variable and PRINT THE MASK ITSELF, then use it to filter. Finish with a single line printing who scored strictly above 90.

```
import pandas as pd

scores = pd.Series([88, 92, 79, 95, 61],
                   index=["Sarah", "Rafi", "Nabila", "Karim", "Mim"])

passed = scores >= 80          # step 1: ask a True/False question
print(passed)
print()
print(scores[passed])          # step 2: keep only the True rows
print()
print(scores[scores > 90])     # all in one line: Rafi and Karim
```

## Part 3 — Challenge

**8. Integer-index detective.** A lottery stores prizes under NUMBER labels: `lottery = pd.Series(["toy", "car", "trip"], index=[10, 20, 30])`. BEFORE running anything, write your predictions for `lottery.loc[10]`, `lottery.iloc[0]` and `lottery.iloc[1]` as comments. Run and check — why do `.loc[10]` and `.iloc[0]` agree only by luck here?

```
import pandas as pd

lottery = pd.Series(["toy", "car", "trip"], index=[10, 20, 30])

# Predictions:
#   loc[10]  -> whatever has LABEL 10  -> "toy"
#   iloc[0]  -> whatever sits at POSITION 0 -> "toy"
#   iloc[1]  -> POSITION 1 -> "car"  (NOT label 20!)
print(lottery.loc[10])
print(lottery.iloc[0])
print(lottery.iloc[1])
# They agree only because label 10 happens to sit at position 0.
# Bare lottery[10] would be ambiguous -- always write .loc or .iloc.
```

**9. Missing-value detective.** You'll need both imports: build `visits = pd.Series([3, np.nan, 7, None, 2], index=["Mon", "Tue", "Wed", "Thu", "Fri"])`. Print how many values are MISSING and how many are PRESENT, then print only the real observations. Finally print `visits.mean()` — did you need to remove the holes first?

```
import numpy as np
import pandas as pd

visits = pd.Series([3, np.nan, 7, None, 2], index=["Mon", "Tue", "Wed", "Thu", "Fri"])

print("missing:", visits.isna().sum())      # 2 -- nan AND None count
print("present:", visits.notna().sum())     # 3
print()
print(visits[visits.notna()])               # keep only real observations
print("mean ignoring NaN:", visits.mean())  # 4.0 -- pandas skips NaN for you
```
