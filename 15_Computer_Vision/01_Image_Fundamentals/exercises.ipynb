## Part 1 — Warm-up

**1. Your first canvas.** Create a blank grayscale image of **100 rows x 160 columns** with `np.zeros` (`dtype=np.uint8`). Print its shape, dtype, min and max. Then paint the single pixel at column 100, row 50 pure white — remembering the `[y, x]` index order — and print that pixel's value.

**2. Stripe orientation.** On a fresh dark canvas of **80 rows x 200 columns**, use ONE slice assignment to make columns 90-109 fully white. BEFORE running: will the stripe look vertical or horizontal? Then run it and print how many white pixels you created to confirm your mental model of rows vs columns.

**3. The 200 + 100 = 44 trap.** Fill a small canvas with the value 200 and produce three brightened versions: plain `+ 100` on the `uint8` array, `cv2.add(canvas, 100)`, and the widen-clip-cast idiom (`int16 -> clip -> uint8`). Print the resulting pixel value from each and explain the difference in a comment.

## Part 2 — Practice

**4. Paint a scene from nothing.** Build a 240-row x 320-column BGR canvas with `np.zeros`: fill the bottom third dark green `(30, 90, 40)` with a filled rectangle, draw a filled yellow sun `(0, 215, 255)` of radius 25 near the top-right, and write a label in white with `cv2.putText`. Display it through matplotlib correctly (convert BGR -> RGB first!).

**5. Channel surgery.** Draw a pure RED square and a pure BLUE square side by side on a black BGR canvas (remember: red is `(0, 0, 255)`!). Print the full B-G-R vector of one pixel inside each square. Then extract channel 0 and channel 2 into separate 2-D planes and print their shapes plus each plane's max value.

**6. Crop with confidence.** Fill a 200x300 canvas with 40, then paint a white block over rows 30-149, columns 30-129. Crop rows 20:100, columns 100:180 into `patch`, print its shape, then set `patch[:] = 77` and check what happened to the block's interior in the ORIGINAL. Finally redo the crop with `.copy()`, edit the copy, and show the original survives. Add a comment naming the view-vs-copy rule.

**7. Resize, width first.** Build a 120-row x 240-column brightness gradient (hint: `np.linspace` across 240 values, then `np.tile` down). Shrink it with `cv2.resize(..., interpolation=cv2.INTER_AREA)` to a quarter of the area, then blow it back up to full size twice - once with `INTER_NEAREST`, once with `INTER_LANCZOS4`. Print every shape and comment on which flag suits shrinking vs growing. Watch out: `resize` takes `(width, height)`!

## Part 3 — Challenge

**8. HSV fruit picker.** Build a 220-row x 340-column basket: start from `np.full(..., 35)` (dark tray), then add three filled balls - blue `(255, 0, 0)` at (70, 110), green `(0, 180, 0)` at (170, 130), red `(0, 0, 255)` at (270, 105), all radius 46. Convert to HSV, build a GREEN mask with `cv2.inRange(hsv, (35, 80, 80), (85, 255, 255))`, isolate the matching pixels with `cv2.bitwise_and`, and print what percentage of the image the mask covers. Which ball survives?

**9. Round trip: PNG vs JPEG.** Draw a logo: 160x200 black canvas, filled orange `(0, 140, 255)` circle at (100, 70) radius 45, white "PyMastery" text underneath. Save it twice under `sample_data/` - once as PNG, once as JPEG quality 40. Reload both (checking for `None`!), report whether each matches the original exactly, and print the JPEG's worst per-pixel drift. Comment on which format belongs in pixel-exact tests.
