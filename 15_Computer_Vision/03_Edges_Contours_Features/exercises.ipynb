## Part 1 — Warm-up

**1. Your first edges.** Build an inspection image: gray background (value 90), a filled light rectangle rows 45-154 / cols 35-134 (value 210) and a filled circle at (225, 100) radius 55 (value 205). Gaussian-blur it `(5, 5)`, run `cv2.Canny(smoothed, 60, 180)`, and print how many edge pixels appeared. Add a comment listing the four stages Canny bundles into that one call.

**2. Threshold tuning.** Rebuild the same part, then fake a noisy camera frame: a seeded rng adds Gaussian hiss (sigma 8) via int16 + clip before the same `(5, 5)` blur. Run `Canny` with `(20, 40)`, `(60, 180)` and `(150, 400)`. Print the edge-pixel count for each setting. In comments, connect the numbers to the failure modes: which ratio invites hairline noise, which erases real boundaries, and what t2/t1 ratio is the practical sweet spot?

**3. Meet the contours.** Threshold the same fixture at 127, then run `cv2.findContours(bw, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)`. Print HOW MANY contours were found, the shape and dtype of the first contour array, and its first three points.

## Part 2 — Practice

**4. Measure everything.** Draw two shapes on a black canvas: a filled rectangle rows 50-169 / cols 30-129 and a triangle through (190, 40), (310, 100), (205, 190) via `cv2.fillPoly`. For EVERY external contour print: area, perimeter, centroid (via moments), bounding box, and the vertex count of `approxPolyDP(c, 0.02 * perimeter, True)`. Sort contours by area, biggest first.

**5. Holes need RETR_TREE.** Build a bracket: gray background 90, filled rectangle rows 45-174 / cols 35-139 at value 210, a smaller rectangle rows 80-129 / cols 70-104 punched back to 90 (the hole), plus a circle at (255, 110) radius 58 (value 205). Threshold at 127 and compare `RETR_EXTERNAL` vs `RETR_TREE` contour counts. Then use the hierarchy table to find the hole's contour index (its Parent entry is not -1) and print it.

**6. Sort and filter blobs.** On a black 220x360 canvas draw three filled circles: radius 15 at (60, 110), radius 32 at (170, 80), radius 52 at (290, 140). Sprinkle junk with `default_rng(3)`: every pixel where `rng.random(...) < 0.005` becomes white. Find ALL external contours, sort by area descending, keep only those with `contourArea >= 500`, and print the raw vs filtered counts plus each survivor's area and bounding box.
