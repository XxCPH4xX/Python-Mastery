## Part 1 — Warm-up

**1. Brightness rescue.** Build a synthetic page: `np.full((160, 300), 190, np.uint8)` with a dark band (value 60) over rows 60-99, columns 40-259. Darken it with `cv2.add(page, -70)` and rescue it again with `cv2.add(dim, 70)`. Print min/max of all three images and check whether `rescued == page` everywhere — what happened to the darkest pixels?

**2. Read a histogram.** Make a 120x200 image: background 230, dark object (value 40) over rows 30-89, columns 60-139. Compute its histogram with `np.histogram(doc, bins=256, range=(0, 256))` and report how many pixels sit exactly at 40 and at 230, plus the histogram total. Comment on why two tall spikes like these are good news for Otsu thresholding.

**3. Odd kernels only.** On a small two-level test image, call `cv2.GaussianBlur(img, (6, 6), 0)` inside `try`/`except cv2.error` and print the complaint. Then blur successfully with `(5, 5)`. In a comment, explain WHY kernel sizes must be odd.

## Part 2 — Practice

**4. Gamma lookup table.** Write `gamma_lut(g)` that builds the 256-entry table `(i / 255) ** g * 255` (clip, then cast to `uint8`). Darken a synthetic photo by 90 with `cv2.add(photo, -90)`, then rescue it with `cv2.LUT(dark, gamma_lut(0.45))`. Print what input value 60 maps to under gamma 0.45 and under gamma 2.0, plus the mean brightness before and after the rescue. Comment on which direction each gamma pushes the midtones.

**5. Three ways to binarize.** Make a synthetic page (background 225) with four lines of dark text via `cv2.putText`. Produce: `THRESH_BINARY` at 127, `THRESH_BINARY_INV` at 127, and `THRESH_BINARY_INV + THRESH_OTSU` (let OpenCV pick the cut). Print the Otsu threshold it chose and the percentage of white pixels in each output. Note in a comment which version OCR pipelines prefer and why.

**6. Salt, pepper and the right filter.** Start from a two-level synthetic image (background 150, block 90). With a seeded `np.random.default_rng(42)`, sprinkle salt-and-pepper noise (`mask < 0.03` -> 0, `mask > 0.97` -> 255). Then apply box blur `(5, 5)`, GaussianBlur `(5, 5)` and `medianBlur 5`, reporting for each: (a) pixels left that are EXACTLY 0 or 255, and (b) mean absolute error against the clean base. Which filter deletes the dots rather than smearing them?

**7. Unsharp masking.** Draw a crisp pattern (two bright blocks on black) and blur it with `cv2.GaussianBlur(pattern, (7, 7), 2.5)` to fake an out-of-focus scan. Sharpen it two ways: the classic kernel (centre 5, orthogonal neighbours -1) via `cv2.filter2D`, and the explicit form `blurry + 1.0*(blurry - more_blur)` via `cv2.addWeighted`. Compare all four images by printing each one's standard deviation (`im.std()`). Comment on what pushing the centre weight to 9 would do.

## Part 3 — Challenge

**8. Morphology cleanup crew.** Build a binary scene: black 180x320 canvas, a filled rectangle rows 40-139 / cols 30-119 and a filled circle at (230, 90) radius 55. Corrupt it with a seeded rng: `noise < 0.02` becomes white specks, and existing white pixels where `noise > 0.995` become pinholes (keep them sparse!). Using a 7x7 elliptical kernel, apply MORPH_OPEN, then MORPH_CLOSE on the opened result, and finally MORPH_GRADIENT for outlines. Print the white-pixel count at each stage.

**9. The full scan-cleaning pipeline.** Simulate a bad scan yourself: draw a clean page (light background, three `putText` lines), subtract a horizontal brightness gradient (`np.tile(np.linspace(210, 175, w), ...)` after `-40`), add Gaussian hiss (sigma 14, via int16 + clip), and sprinkle dust (`rng < 0.015` -> 0, `rng > 0.985` -> 255). Then run the pipeline: medianBlur 5 -> `THRESH_BINARY_INV + THRESH_OTSU` -> MORPH_CLOSE with a 3x3 rect kernel -> MORPH_OPEN with the same kernel. Report ink coverage (%) right after thresholding and after the final cleanup, and state the pipeline's ordering rule.
