## Part 1 — Warm-up

**1. Name that crash.** Four accidents, four named handlers — each `except` names exactly the disease it treats.

```
# 1) right type, unacceptable value
try:
    int("twelve")
except ValueError as e:
    print("ValueError:", e)

# 2) wrong type altogether
try:
    "hello" - 1
except TypeError as e:
    print("TypeError:", e)

# 3) index past the end of the list
try:
    [90, 85, 70][99]
except IndexError as e:
    print("IndexError:", e)

# 4) dictionary key does not exist
try:
    {"name": "Sam"}["age"]
except KeyError as e:
    print("KeyError:", e)
```

**2. Divide with a safety net.** The loop survives its worst member: the zero is caught, reported and skipped.

```
readings = [400, 0, 80]

for r in readings:
    try:
        share = 1000 / r
        print(f"1000 / {r} = {share}")
    except ZeroDivisionError:
        print(f"1000 / {r} -> impossible, skipping")

print("Done - the zero never sank the run.")
```

**3. Accept or reject.** `int()` raises `ValueError` for `"12.5"` and `"octet"` alike; `else:` speaks only when nothing failed.

```
attempts = ["128", "12.5", "octet"]

for text in attempts:
    try:
        value = int(text)
    except ValueError:
        print(f"'{text}' rejected - digits only")
    else:
        print(f"'{text}' accepted as {value}")
```

## Part 2 — Practice

**4. One try, two failure modes.** Python checks the handlers top to bottom and runs the first that matches.

```
submissions = ["15", "0", "five"]

for raw in submissions:
    try:
        number = int(raw)                               # ValueError for junk text
        print(f"300 / {number} = {300 / number:.1f}")   # ZeroDivisionError for 0
    except ValueError:
        print(f"'{raw}' is not a number at all")
    except ZeroDivisionError:
        print("'0' is a number, but 300 / 0 is undefined")
```

**5. Handler order detective.** `except Exception` matches first because `ValueError` is one of its children — specific handlers belong on top.

```
# BROKEN ORDER: except Exception sits on top and matches EVERYTHING,
# including ValueError, so its branch always wins and prints
# "something went wrong" - the specific handler below is unreachable.

# FIXED: most specific handler first.
try:
    price = int("not a number")
except ValueError:
    print("that was not a number")
except Exception:
    print("some other surprise")
```

**6. Capture the evidence.** The bound object knows its own class name and carries the message — log both, then talk like a human.

```
try:
    distance = int("24 km")
except ValueError as e:
    print("technical type :", type(e).__name__)
    print("raw message    :", e)
    print("human version  : digits only, please - no units")
```

**7. Refuse nonsense loudly.** Raising with the offending value included turns the next debugging session into a five-second read.

```
def withdraw(balance, amount):
    """Return the new balance, refusing nonsense loudly."""
    if amount <= 0:
        raise ValueError(f"amount must be positive, got {amount}")
    if amount > balance:
        raise ValueError(f"cannot withdraw {amount}, balance is only {balance}")
    return balance - amount


balance = 200
for amount in (50, 0, 500):
    try:
        balance = withdraw(balance, amount)
        print(f"withdrew {amount}, balance now {balance}")
    except ValueError as e:
        print(f"refused {amount}:", e)
```

## Part 3 — Challenge

**8. Age-gate validator.** Decimals, junk text and non-positive numbers each fail fast with their own reason; `else` records only genuine success.

```
attempts = ["abc", "-5", "17"]


def parse_positive_int(text):
    """Return a positive whole number, or raise ValueError saying exactly why."""
    if "." in text:
        raise ValueError(f"'{text}' looks like a decimal - whole numbers only")
    value = int(text)          # raises ValueError for junk like 'abc'
    if value <= 0:
        raise ValueError(f"{value} is not positive")
    return value


for text in attempts:
    try:
        age = parse_positive_int(text)
    except ValueError as e:
        print(f"x '{text}' rejected -> {e}")
    else:
        print(f"OK '{text}' accepted -> age recorded as {age}")
        break
```

**9. Family-tree detective.** One parent name covers the whole lookup branch — while `KeyboardInterrupt` deliberately bypasses ordinary handlers.

```
print("IndexError        child of LookupError:", issubclass(IndexError, LookupError))
print("KeyError          child of LookupError:", issubclass(KeyError, LookupError))
print("KeyboardInterrupt child of Exception  :", issubclass(KeyboardInterrupt, Exception),
      "<- Ctrl+C bypasses normal handlers")

# One parent name catches both children.
try:
    [10, 20][9]
except LookupError as e:
    print("caught via parent:", type(e).__name__, "-", e)

try:
    {"page": 1}["chapter"]
except LookupError as e:
    print("caught via parent:", type(e).__name__, "-", e)
```
