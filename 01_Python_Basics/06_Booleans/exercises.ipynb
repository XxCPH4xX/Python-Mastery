**8. De Morgan's law.** Pick at least three different pairs of values for `a` and `b`. Print `not (a and b)` next to `(not a) or (not b)` for each pair and confirm they always match. State the rule in a comment.

```python
a, b = True, False
print(not (a and b), "|", (not a) or (not b))  # True | True

a, b = True, True
print(not (a and b), "|", (not a) or (not b))  # False | False

a, b = False, False
print(not (a and b), "|", (not a) or (not b))  # True | True
```

**9. Truthiness shortcut.** Start with `items = []` and check it two ways: `bool(items)` and `len(items) > 0`. Then set `items = ["x"]` and check both again. Which style would you rather read?

```python
items = []
print(bool(items), len(items) > 0)  # False False

items = ["x"]
print(bool(items), len(items) > 0)  # True True
```
