# Booleans Exercises

Write your answer in the empty code cell under each question. Check yourself against `solutions.ipynb` when done.

**Part 1** = warm-up · **Part 2** = practice · **Part 3** = challenge.

## Part 1 — Warm-up

**1. Predict & check.** Say each answer out loud *before* running: `10 > 5`, `10 == "10"`, `3 != 3`, `"a" < "b"`.

**2. Truthiness lab.** Run `bool()` on each of: `0`, `42`, `-1`, `""`, `"hi"`, `[]`, `None`. List in a comment which ones came back falsy.

**3. Logic gates.** With `a = True` and `b = False`, evaluate `a and b`, `a or b`, `not a` and `not b`.

## Part 2 — Practice

**4. Range test.** Is `age = 24` between 18 and 65 (inclusive)? Do it in one expression using a chained comparison: `18 <= age <= 65`.

**5. Alphabet order.** Compare `"apple" < "banana"` and `"Zebra" < "apple"`. Why does the capital-letter word sort first? (Hint: think ASCII/Unicode order.)

**6. Equal vs identical.** For `a = [1, 2]` and `b = [1, 2]`: compare both `a == b` and `a is b`. You'll get two different answers — explain why in a comment.

**7. Compound conditions.** Define `has_ticket = True` and `age = 16`. Build boolean expressions for:

- `can_enter`: has a ticket AND is 18 or older
- `gets_discount`: under 18 OR over 65

## Part 3 — Challenge

**8. De Morgan's law.** Pick at least three different pairs of values for `a` and `b`. Print `not (a and b)` next to `(not a) or (not b)` for each pair and confirm they always match. State the rule in a comment.

**9. Truthiness shortcut.** Start with `items = []` and check it two ways: `bool(items)` and `len(items) > 0`. Then set `items = ["x"]` and check both again. Which style would you rather read?
