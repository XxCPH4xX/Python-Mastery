## Part 1 — Warm-up

**1. Predict & check.** `10 == "10"` is `False` because an int never equals a str.

```
print(10 > 5)       # True
print(10 == "10")   # False — number vs text
print(3 != 3)       # False
print("a" < "b")    # True — alphabetical
```

**2. Truthiness lab.** The falsy four here: `0`, `""`, `[]`, `None`. (`-1` and any non-empty string are truthy.)

```
print(bool(0))    # False
print(bool(42))   # True
print(bool(-1))   # True  — only zero is falsy among numbers
print(bool(""))   # False
print(bool("hi")) # True
print(bool([]))   # False
print(bool(None)) # False
```

**3. Logic gates.**

```
a = True
b = False

print(a and b)   # False — both must be True
print(a or b)    # True  — at least one is True
print(not a)     # False
print(not b)     # True
```

## Part 2 — Practice

**4. Range test.** Chained comparison reads like plain English and checks both ends at once.

```
age = 24

in_working_range = 18 <= age <= 65
print(in_working_range)   # True
```

**5. Alphabet order.** Uppercase letters have smaller code points (`Z` = 90 vs `a` = 97), so capitals always sort first.

```
print("apple" < "banana")   # True — alphabetical
print("Zebra" < "apple")    # True?! — uppercase Z (90) < lowercase a (97)
```

**6. Equal vs identical.** Two separately built lists can hold equal *values* but still be different *objects*.

```
a = [1, 2]
b = [1, 2]

print(a == b)   # True  — same contents
print(a is b)   # False — two different list objects in memory
```

**7. Compound conditions.** Translate each sentence straight into `and`/`or`.

```
has_ticket = True
age = 16

can_enter = has_ticket and age >= 18
gets_discount = age < 18 or age > 65

print(can_enter)     # False — ticket alone isn't enough
print(gets_discount) # True  — under 18 qualifies
```

## Part 3 — Challenge

**8. De Morgan's law.** The rule: **not (A and B) ≡ (not A) or (not B)** — negating an AND flips it into an OR.

```
a, b = True, False
print(not (a and b), "|", (not a) or (not b))   # True | True

a, b = True, True
print(not (a and b), "|", (not a) or (not b))   # False | False

a, b = False, False
print(not (a and b), "|", (not a) or (not b))   # True | True
```

**9. Truthiness shortcut.** Empty containers are falsy, so `bool(items)` already asks "is there anything in here?".

```
items = []
print(bool(items), len(items) > 0)   # False False

items = ["x"]
print(bool(items), len(items) > 0)   # True True
```
