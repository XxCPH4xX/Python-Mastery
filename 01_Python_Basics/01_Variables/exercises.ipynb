# Variables Exercises

Write your answer in the empty code cell under each question. Check yourself against `solutions.ipynb` when done.

**Part 1** = warm-up · **Part 2** = practice · **Part 3** = challenge.

## Part 1 — Warm-up

**1. Your first variables.** Create three variables: `name` (text), `age` (whole number) and `height` (decimal). Print all three.

**2. Growing score.** Create `score = 0`. Increase it by `10` two times using reassignment (`score = score + 10`). Print the final value.

**3. Swap trick.** Given `a = 1` and `b = 2`, swap their values **without** using a third variable. Hint: Python allows `x, y = y, x`.

## Part 2 — Practice

**4. Dynamic typing.** Assign `x = 5` and print `type(x)`. Then assign `x = "five"` and print `type(x)` again. What changed?

**5. One line, many variables.** Assign the colors `"red"`, `"green"`, `"blue"` to `c1`, `c2`, `c3` in a **single line**.

**6. Shared value.** Run `x = y = 100`, then set `y = 50` and print both. Did `x` change? Explain why or why not in a comment.

**7. Name police.** These names are illegal or bad: `2players`, `my-score`, `class`. Rename each one correctly and add a comment explaining which rule each original broke.

## Part 3 — Challenge

**8. Game state.** A hero starts with `health = 100`, takes `30` damage twice, then drinks a potion worth `15`. Track each event in well-named variables and finish with an f-string status line like `Hero HP: 55`.

**9. Integer caching detective.** Run `a = 200; b = 200; print(a is b)` and then `a = 300; b = 300; print(a is b)`. One prints `True`, the other `False` — connect what you see to *small integer interning* (the −5 to 256 cache) from the notes.
