## 1. What Is a Variable?

A **variable** in Python is a named reference that points to a value stored in memory. Think of it as a sticky note with a name written on it, attached to a value. When you write `x = 5`, you're creating a label called `x` that refers to the integer object `5` somewhere in memory.

Unlike languages like C or Java where you must declare the type (`int x = 5;`), Python determines the type automatically based on the value you assign. This is called **dynamic typing**.

**Syntax:**
```python
variable_name = value
```

Variables are created the moment you assign a value to them using the equal sign (`=`).

````python
x = 5                  # Integer
name = "Alice"         # String
pi = 3.14              # Float
is_active = True       # Boolean

print(x)
print(name)
print(pi)
print(is_active)
````

````python
# Check the type of each variable
print(type(x))          # <class 'int'>
print(type(name))       # <class 'str'>
print(type(pi))         # <class 'float'>
print(type(is_active))  # <class 'bool'>
````

## 2. Dynamic Typing: Variables Can Change Type

Python uses **dynamic typing**, which means a variable can change its type at any point during execution. The same variable can start as an integer, then become a string, then a list — Python doesn't care. The type is attached to the **value**, not the variable name.

This flexibility makes Python easy to learn but requires discipline — accidentally reassigning a variable to a different type is a common source of bugs.

````python
x = 10               # x is an integer
print(x, type(x))

x = "Hello"          # x is now a string
print(x, type(x))

x = [1, 2, 3]        # x is now a list
print(x, type(x))
````

## 3. Storing Different Data Types

Variables can hold any Python object: numbers, text, booleans, lists, dictionaries, and more. The type of the variable is determined by the value you assign to it.

````python
# Storing Text (Strings)
user_name = "Sarah"
country = "Bangladesh"

# Storing Whole Numbers (Integers)
player_score = 1500
items_in_cart = 3

# Storing Decimals (Floats)
product_price = 19.99
temperature_celsius = 36.5

# Storing True/False (Booleans)
is_logged_in = True
has_discount = False

print(user_name)
print(country)
print(player_score)
print(items_in_cart)
print(product_price)
print(temperature_celsius)
print(is_logged_in)
print(has_discount)
````

## 4. Variables Are Labels, Not Boxes

In Python, assignment does **not** copy the value into a box labeled with the variable name. Instead, it creates a **reference** (or label) pointing to the object in memory.

When you write `a = [1, 2, 3]`, Python creates a list object `[1, 2, 3]` and makes `a` point to it. If you then write `b = a`, both `a` and `b` point to the **same list** — changing one affects the other.

````python
# Reassigning a variable
score = 0               # Variable 'score' starts at 0
print("Initial score:", score)

score = score + 10      # Variable 'score' is now 10
print("After adding 10:", score)

score = 25              # Variable 'score' is now 25
print("After reassignment:", score)
````

````python
# Variables storing lists
shopping_list = ["Apples", "Milk", "Bread"]
todo_tasks = ["Study Python", "Go to gym", "Pay bills"]

print(shopping_list)
print(todo_tasks)
````

````python
# Two variables pointing to the same list
list_a = [1, 2, 3]
list_b = list_a          # list_b is an ALIAS for list_a

list_b.append(4)         # Modify through list_b
print(list_a)            # [1, 2, 3, 4] -> list_a changed too!
print(list_b)            # [1, 2, 3, 4]

print(list_a is list_b)  # True -> they point to the same object
````

## 5. Identity and Small-Integer Caching

Every object in Python has a unique identity, which you can inspect with `id()`. Two variables point to the same object when their IDs match.

**Small-Integer Caching (Interning):** Python pre-allocates and caches integer objects in the range **-5 to 256** for performance. When you create an integer in this range, Python reuses the existing cached object instead of creating a new one. Outside this range, Python allocates a new object every time.

This is an implementation detail of CPython — **never** rely on `is` to compare numbers in production code. Use `==` for value comparison.

````python
# Small Integer Caching (Python caches integers between -5 and 256)
x = 100
y = 100
print(x is y)           # True (Both reference the same cached object)
print(id(x), id(y))     # Same memory address

# Large Integers (Outside the cache)
a = 1000
b = 1000
print(a is b)           # False (Different objects with the same value)
print(id(a), id(b))     # Different memory addresses
print(a == b)           # True (Values are equal)
````

````python
# Floats are NOT cached the same way
c = 3.14
d = 3.14
print(c is d)           # Usually False (different objects)
print(c == d)           # True (same value)
````

## 6. Variable Naming Rules

Python has strict rules for variable names:

1. **Must start with a letter (a-z, A-Z) or underscore (`_`)** — never a digit.
2. **Can contain letters, digits, and underscores** — no spaces, hyphens, or special characters.
3. **Case-sensitive** — `score`, `Score`, and `SCORE` are three different variables.
4. **Cannot use reserved keywords** — words like `if`, `for`, `while`, `class`, `def`, `return`, etc.

**Conventions (PEP 8):**
- Use **snake_case** for variable names: `user_name`, `total_price`, `is_active`.
- Use **UPPER_CASE** for constants: `MAX_RETRIES`, `PI`, `API_KEY`.
- Choose **descriptive names** — `user_age` beats `a`, and `total_price` beats `tp`.

````python
# Valid variable names
user_name = "John"
_private_var = 42
score2 = 100
totalPrice = 199.99      # Valid but not Pythonic (use snake_case)

# Invalid variable names (uncommenting these will cause errors)
# 2score = 100           # SyntaxError: cannot start with digit
# user-name = "John"     # SyntaxError: hyphens not allowed
# for = 10               # SyntaxError: 'for' is a reserved keyword
````

````python
# Case sensitivity
score = 100
Score = 200
SCORE = 300

print(score)     # 100
print(Score)     # 200
print(SCORE)     # 300
````

## 7. Multiple Assignment and Unpacking

Python allows you to assign multiple variables in one line, which is cleaner and more readable than separate statements. You can also **unpack** sequences (lists, tuples) directly into variables.

````python
# Assign the same value to multiple variables
x = y = z = 0
print(x, y, z)          # 0 0 0

# Assign different values in one line
a, b, c = 1, 2, 3
print(a, b, c)          # 1 2 3

# Swapping variables (no temp variable needed!)
x, y = 10, 20
print("Before swap:", x, y)

x, y = y, x
print("After swap:", x, y)
````

````python
# Unpacking a list or tuple
coordinates = (10, 20, 30)
x, y, z = coordinates
print(f"x={x}, y={y}, z={z}")

# Unpacking with the rest operator (*)
numbers = [1, 2, 3, 4, 5]
first, *middle, last = numbers
print(f"First: {first}, Middle: {middle}, Last: {last}")
````

## ⚠️ Common Mistakes & Gotchas

| Mistake | Problem | Fix |
| --- | --- | --- |
| Using reserved keywords as variable names | `for = 5` raises `SyntaxError` | Check Python's keyword list with `import keyword; print(keyword.kwlist)` |
| Starting variable names with digits | `2score = 100` is invalid | Start with a letter or underscore: `score_2` |
| Assuming `a = b` copies the value | For mutable objects (lists, dicts), it creates an alias | Use `b = a.copy()` or `b = list(a)` |
| Relying on `is` to compare numbers | Works for small integers by accident, fails elsewhere | Use `==` for value comparison |
| Using ambiguous single-letter names | `a`, `b`, `x` tell the reader nothing | Choose descriptive names: `user_age`, `total_cost` |
| Forgetting Python is case-sensitive | `Score` and `score` are different variables | Stick to snake_case consistently |

## 💡 Best Practices & Pro Tips

- **Use descriptive names.** `user_age` beats `a`, and `total_price` beats `tp`. Code is read far more often than written.
- **Follow PEP 8:** snake_case for variables, UPPER_CASE for constants, avoid single-letter names except in loops or math.
- **Avoid reusing variable names for different purposes** — it confuses readers and makes debugging harder.
- **Use unpacking** to make initialization cleaner: `x, y = 0, 0` instead of two separate lines.
- **Remember that assignment creates references, not copies.** For mutable objects, use `.copy()` when you need independence.
- **Check object identity with `id()`** when debugging aliasing issues — but never rely on `is` for value comparison in production.
- **🤖 AI-engineering relevance:** Variables in ML code often hold tensors, models, datasets, and hyperparameters. Clear naming (`learning_rate`, `train_loader`, `model_weights`) keeps experimental notebooks readable when you revisit them weeks later. Dynamic typing lets you swap a NumPy array for a PyTorch tensor without changing variable declarations — but discipline prevents accidental type mismatches.
