## Part 1 — Warm-up

**1. Your first variables.** Three variables with three different types, printed.

```
name = "Fahim"
age = 21
height = 5.9

print(name)
print(age)
print(height)
```

**2. Growing score.** Each reassignment replaces the old value, ending at `20`.

```
score = 0
score = score + 10   # 10
score = score + 10   # 20
print(score)
```

**3. Swap trick.** Tuple packing/unpacking swaps both names in one line.

```
a = 1
b = 2

a, b = b, a
print(a, b)   # 2 1
```

## Part 2 — Practice

**4. Dynamic typing.** The *name* `x` stays the same — only the object it points to changed.

```
x = 5
print(type(x))   # <class 'int'>

x = "five"
print(type(x))   # <class 'str'>
```

**5. One line, many variables.** Comma-separated targets and values, matched left to right.

```
c1, c2, c3 = "red", "green", "blue"
print(c1, c2, c3)
```

**6. Shared value.** `x = y = 100` makes both names point to the *same* `100`. Assigning `y = 50` only re-points `y`; `x` still looks at `100`.

```
x = y = 100
y = 50          # only y now points to a new object

print(x)        # 100 — unchanged
print(y)        # 50
```

**7. Name police.** Valid replacements, with the broken rule noted.

```
# 2players -> cannot start with a digit
players2 = 2

# my-score -> hyphen reads as minus; use snake_case
my_score = 99

# class -> reserved keyword; avoid or rephrase
class_name = "Python 101"

print(players2, my_score, class_name)
```

## Part 3 — Challenge

**8. Game state.** 100 − 30 − 30 + 15 = **55**.

```
hero_health = 100

dragon_hit_1 = 30
hero_health = hero_health - dragon_hit_1      # 70

dragon_hit_2 = 30
hero_health = hero_health - dragon_hit_2      # 40

healing_potion = 15
hero_health = hero_health + healing_potion    # 55

print(f"Hero HP: {hero_health}")
```

**9. Integer caching detective.** `200` lives in CPython's interned cache (−5…256), so both names point to one shared object → `is` says `True`. `300` is outside the cache, so each assignment builds a separate object → `is` says `False`, even though the *values* are equal (`==`).

```
a = 200
b = 200
print(a is b)   # True  — 200 is in the cached range

a = 300
b = 300
print(a is b)   # False — two separate objects outside the cache
```
