# Variable

#### Variable - A variable in Python is a 'named container' that (contains info) or reference that stores a data value in memory. Unlike many other programming languages, you do not need to declare a variable's type explicitly—Python automatically determines the type based on the assigned value.

x, name, pi, is_active are variable.
Variables are created the moment you assign a value to them using the equal sign (=) , variable = value (or data).

Example : In the expression x = 5, 'x' is the variable. 5 is the value (or data). 
'x' is a variable that holds the value 5. including 'name, pi, is_active' are diffrent types of variables.

Code Example

```python
x = 5  # Integer
name = "Alice"  # String
pi = 3.14  # Float
is_active = True  # Boolean

print(x)
print(name)
print(pi)
print(is_active)
```

```python
# Dynamic Typing: A variable can change its data type at any point during execution
# Example : x is an integer, then becomes a string.

x = 10  # x is an integer
x = "Hello"  # x is now a string

print(x)
```

```python
# Storing Text (Strings) in Variables named user_name and country
user_name = "Sarah"
country = "Bangladesh"

# Storing Whole Numbers (Integers)
player_score = 1500
items_in_cart = 3

# Storing Decimals (Floats)
product_price = 19.99
temperature_celsius = 36.5

# Storing True/False (Booleans)
is_logged_in = True
has_discount = False

print(user_name)
print(country)
print(player_score)
print(items_in_cart)
print(product_price)
print(temperature_celsius)
print(is_logged_in)
print(has_discount)
```

```python
# A game score increasing
score = 0  # Variable 'score' starts at 0
score = score + 10  # Variable 'score' is now 10
score = 25  # Variable 'score' is now 25
```

```python
# shopping_list and todo_tasks are variables that store lists
shopping_list = ["Apples", "Milk", "Bread"]
todo_tasks = ["Study Python", "Go to gym", "Pay bills"]

print(shopping_list)
print(todo_tasks)
```

```python
# Small Integer Interning (Python caches integers between -5 and 256)
x = 100
y = 100
print(x is y)  # True (Both variables (x and y) reference the exact same memory object)

# Large Integers (Outside the intern cache)
a = 1000
b = 1000
print(a is b)  # False (Different memory objects with the same value)
# Python optimizes memory using a technique called Small Integer Caching (Interning). During startup, CPython pre-allocates and caches integer objects in the range -5 to 256.
# For numbers between -5 and 256, Python reuses the existing cached object. For numbers outside this range (like 1000), Python allocates a new memory object every time.

# Floats (Python caches floats between -5.0 and 256.0)
c = 3.14159
d = 3.14159
print(c is d)  # True (Both variables reference the exact same memory object)
```
