**2. Strings sneak in.** Ask for a number, then run `number + number`. Why do you get e.g. `77` instead of `14`? Fix it by casting with `int()`.

## Part 2 — Practice

**4. Mini calculator.** Ask for two numbers as text, cast them, and print their sum, difference and product.

````python
a, b = input("Enter two numbers: ").split(" ")

print(a, b)

a = int(a)
b = int(b)

sum = a + b
difference = a - b
product = a * b
print(sum, difference, product)
````

**5. Age machine.** Ask for a birth year, cast to int, and compute the age that person turns during 2026.

````python
birthday = input("When is your birthday? ").split("/")

age = 2026 - int(birthday[2])

print(age)
````

**6. Temperature converter.** Ask for a temperature in Celsius, convert with `F = C * 9/5 + 32`, and show the result with one decimal place.

````python
celsius = float(input("Temperature in Celsius: "))
fahrenheit = celsius * 9 / 5 + 32

print(f"{celsius}°C is {fahrenheit:.1f}°F")
````

**7. Number check.** Ask for some input and print whether `.isdigit()` is `True` or `False`. No `if` statement needed yet — just print the boolean.

````python
number = input("Type anything: ")
print(number.isdigit())
````

## Part 3 — Challenge

**8. Mad libs.** Collect an adjective, a noun and a verb through three inputs, then insert them into a short story stored in a triple-quoted f-string.

````python
adjective = input("Adjective: ")
noun = input("Noun: ")
verb = input("Verb: ")

story = f"""Once upon a time, a {adjective} {noun}
decided to {verb} all the way to the moon.
The end."""

print(story)
````

**9. Receipt builder.** Ask for an item name, a quantity (cast to int) and a unit price (cast to float). Print a receipt line showing the item, quantity and total cost formatted with `:.2f`.

````python
item_name = input("Item name: ")
item_price = float(input("Item price: "))
item_quantity = int(input("Item quantity: "))

total = item_price * item_quantity

print(f"Item: {item_name}")
print(f"Quantity: {item_quantity}pc")
print(f"Price: {item_price}BDT")
print(f"Total: {total}BDT")
````
