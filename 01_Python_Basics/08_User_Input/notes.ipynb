## 1. What Is `input()`?

`input()` pauses your program, prints an optional prompt message, waits for the user to type something and press **Enter**, then hands you whatever they typed as a **string**.

```python
name = input("What is your name? ")     # the program pauses RIGHT HERE
print("Hello,", name)
```

The text inside the parentheses is the **prompt**. It is optional — but a program that asks nothing feels broken. Try this in a real terminal: save it as `greet.py` and run it with a command like:

```bash
python greet.py
```

One rule before we continue: `input()` is a *conversation*, and conversations need someone on the other end. Where that someone comes from matters — which brings us to notebooks.


## 2. ⚠️ Notebooks Can't Pause: How We Simulate Input

`input()` **blocks execution** until a human types at a keyboard, and Jupyter code cells have no live terminal waiting behind them. Calling it inside a notebook freezes (or interrupts) the kernel — so these notes follow one convention throughout:

- **Grey code fences** (the blocks marked `python`) show the *real* `input()` code, exactly as you would write it in a `.py` file or terminal.
- **Runnable code cells** hardcode the same value with a `# Simulated input:` comment.

When you copy an example into a real script, swap the hardcoded value back for the genuine call.


````python
# Real version (in a .py script):  name = input("What is your name? ")

# Simulated input: the user typed "Sarah"
name = "Sarah"
print("Hello,", name)
````

## 3. Storing and Echoing Input

Capture the typed value into a variable, then treat it like any other string: print it, embed it in an f-string, transform it with string methods.

```python
city = input("Which city do you live in? ")
print("You live in", city)
print(f"{city.upper()} is a beautiful city!")
```

Echoing input back confirms to users that the program heard them — the same trick CLI wizards and chatbots use.


````python
# Simulated input: the user typed "Dhaka"
city = "Dhaka"

print("You live in", city)
print(f"{city.upper()} is a beautiful city!")

nickname = city.lower()            # strings arrive ready with tools like .lower()
print("Local spelling:", nickname)
````

## 4. `input()` ALWAYS Returns a String

Whatever the user types — `42`, `3.14`, even `True` — reaches your program as a **string of characters**. Digits are not special. This is the single most common source of beginner bugs.

```python
age = input("Your age: ")        # typing 42 gives you the STRING "42"
print(age + 1)                   # TypeError: can only concatenate str to str
```

The sneakier failure is `"42" + "1"`: no crash at all — it silently produces `"421"`, because between strings `+` means *concatenate*.


````python
# Simulated input: the user typed 42
age_text = "42"

print(age_text, type(age_text))    # 42 <class 'str'> <- a string, not a number!

print(age_text + "1")              # "421" <- concatenation, NOT addition
# print(age_text + 1)              # TypeError: can only concatenate str (not "int") to str
````

## 5. Converting Input with `int()` and `float()`

To do maths, convert the string to a number first — `int()` for whole numbers, `float()` for decimals. The classic one-liner wraps `input()` directly inside the converter:

```python
age = int(input("Your age: "))           # whole number
height = float(input("Height in m: "))   # decimal
print(age * 2, height * 2)
```

But conversion trusts the user completely. Type `abc` (or just press Enter on an empty line) and the program crashes:

```python
>>> int("abc")
ValueError: invalid literal for int() with base 10: 'abc'
```

That crash is a **`ValueError`** — Python saying "this text isn't a valid number." Handling it gracefully is exactly what the **Error Handling module (07)** teaches with `try/except`. Until then, our defence is *validation*, coming up next.


````python
# Simulated inputs: the user typed 20 and 1.68
age_text = "20"          # int(input("Your age: "))
height_text = "1.68"     # float(input("Height in metres: "))

age = int(age_text)
height = float(height_text)

print(age + 5)                    # 25   -> real arithmetic at last
print(height * 2)                 # 3.36
print(type(age), type(height))    # <class 'int'> <class 'float'>
````

## 6. Validating Numeric Input with `str.isdigit()`

Instead of crashing, **check before you convert**. The string method `.isdigit()` returns `True` only when *every* character is a digit — and conveniently, `"".isdigit()` is `False`, so blank input is rejected too.

Real programs repeat the question until the answer is valid, which needs a `while` loop (next module). Here is the pattern you will grow into:

```python
raw = input("Enter your age: ")
while not raw.isdigit():          # keep asking until it's all digits
    print("Please type digits only!")
    raw = input("Enter your age: ")
age = int(raw)                    # NOW conversion is guaranteed safe
```

And the notebook-friendly simulation:


````python
# Simulated session: the user typed "twenty" (rejected), then "20" (accepted)
first_try = "twenty"
second_try = "20"

print(first_try.isdigit())     # False -> rejected, ask again
print(second_try.isdigit())    # True  -> accepted!

age = int(second_try)          # conversion can no longer crash
print("Age accepted:", age)
````

````python
# .isdigit() is strict: negatives, decimals and spaces all FAIL
print("-5".isdigit())       # False  (the minus sign is not a digit)
print("3.14".isdigit())     # False  (the dot is not a digit)
print(" 7".isdigit())       # False  (a leading space breaks it)
print("".isdigit())         # False  (empty -> nicely rejects a bare Enter)

# So .isdigit() suits ages, quantities, PIN codes -- not prices or temperatures.
# For those, use float() inside try/except (Error Handling module, 07).
````

## 7. Multiple Values from One Line: `.split()` and `map(int, ...)`

Users hate being asked three questions when one will do. `.split()` chops a string at whitespace into a **list**, and unpacking grabs the pieces straight into separate variables:

```python
name, age_text, city = input("Name age city: ").split()
```

To convert every piece at once, `map(int, parts)` applies `int` to each element. It returns a lazy *map object*, so wrap it in `list()` or unpack it immediately:

```python
a, b = map(int, input("Two numbers: ").split())
print(a + b)
```


````python
# Simulated input: the user typed "Sarah 24 Dhaka"
line = "Sarah 24 Dhaka"

parts = line.split()                # splits on ANY run of whitespace
print(parts)                        # ['Sarah', '24', 'Dhaka']

name, age_text, city = parts        # unpack into separate variables
age = int(age_text)                 # convert just the numeric piece

print(f"{name} ({age}) from {city}")

# The all-at-once version:
x, y = map(int, "12 30".split())    # map applies int() to every piece
print(x + y)                        # 42
````

## 8. Friendlier Prompts with f-strings

Prompts can reuse information you already have — a username, a limit, the expected units. Build them with an f-string so each user sees a personalised question:

```python
name = input("Your name: ")
limit = 1000
amount = int(input(f"{name}, how much (max {limit})? "))
print(f"Withdrawing {amount} BDT")
```

A precise prompt ("height in metres, e.g. 1.68") prevents bad input at the source — always cheaper than repairing it afterwards.


````python
# Simulated inputs: the user typed "Sarah", then "750"
name = "Sarah"
limit = 1000

amount = 750                        # int(input(f"{name}, how much (max {limit})? "))

print(f"{name}, withdrawing {amount} BDT (limit {limit})")
print("Within limit:", amount <= limit)
````

## 9. Mini Project 1: Simple Interest Calculator

Time to combine Lessons 1–8: variables, types, casting, arithmetic operators and f-strings. The formula is `interest = principal × rate × years ÷ 100`. In a real script you would read the three values with `input()`:

```python
principal = float(input("Principal (BDT): "))
rate = float(input("Annual rate (%): "))
years = int(input("Years: "))
```

Here we simulate the whole session so the cell runs cleanly:


````python
# Simulated inputs: the user typed 5000, 7.5 and 3
principal = 5000.0     # float(input("Principal (BDT): "))
rate = 7.5             # float(input("Annual rate (%): "))
years = 3              # int(input("Years: "))

interest = principal * rate / 100 * years
total = principal + interest

print("----- Simple Interest Report -----")
print(f"Principal : {principal:.2f} BDT")
print(f"Rate      : {rate}% per year, for {years} year(s)")
print(f"Interest  : {interest:.2f} BDT")
print(f"Total     : {total:.2f} BDT")
````

## ⚠️ Common Mistakes & Gotchas

| Mistake | Problem | Fix |
| --- | --- | --- |
| Doing maths on raw input (`input() + 1`) | `TypeError` — input is always `str`, even `"42"` | Convert immediately: `int(input(...))` |
| Expecting `.isdigit()` to accept `-5` or `3.14` | Minus/dot/space aren't digits, so valid decimals get rejected | Validate with `try: float(raw)` once you reach the Error Handling module |
| Skipping `.strip()` on text input | Invisible spaces make `"yes "` != `"yes"` and break `.isdigit()` | Call `.strip()` before comparing or validating |
| Unpacking blindly (`x, y = line.split()`) | `ValueError: too many/not enough values to unpack` if the user adds or drops a word | State the expected format in the prompt; split defensively |
| Calling `input()` inside a notebook cell | Blocks/interrupts the kernel — no terminal is attached | Hardcode a value with `# Simulated input:` while learning |

```python
# The classic crash, reproduced safely:
bad = "abc"
# age = int(bad)             # ValueError: invalid literal for int() with base 10: 'abc'
print(bad.isdigit())          # False -> our early-warning system
```


## 💡 Best Practices & Pro Tips

- **Convert at the boundary.** Turn text into numbers right after reading it; keep the original string only if you may need to re-prompt.
- **Write prompts that answer questions:** what to enter, in which units, in what format — `"Enter height in metres (e.g. 1.68): "`.
- **Validate before converting**, and reject bad input politely instead of crashing: the `.strip().isdigit()` check covers ages, counts and PINs.
- **Prefer one well-labelled question per line of input** over long interrogations — use `.split()` only when the values genuinely belong together.
- **🤖 AI-engineering relevance:** scripts and agents still ask humans for things — API keys, model choices, hyperparameters. The read → convert → validate pipeline you built here scales directly into `argparse` CLIs, config files and notebook parameterisation later in this course.
