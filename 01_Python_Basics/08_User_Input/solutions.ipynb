# User_Input Solutions

Solutions for `exercises.ipynb`. Try the exercises first — peek here only after you've attempted each question.

Every cell here calls `input()` — run them yourself and type into the prompt.

## Part 1 — Warm-up

**1. Hello, you.** `input()` always returns text, so an f-string can slot it straight in.

```python
name = input("What is your name? ")
print(f"Hello, {name}!")
```

**2. Strings sneak in.** Untreated input is str, so `+` concatenates. Cast with `int()` to do math.

```python
number = input("Pick a number: ")
print(number + number)        # "77" — text glued together!

number = int(number)
print(number + number)        # 14 — real addition after casting
```

**3. Band name generator.** Two inputs, one f-string.

```python
city = input("Which city are you from? ")
pet = input("What is your pet's name? ")

print(f'Your band could be "{city} {pet}s".')
```

## Part 2 — Practice

**4. Mini calculator.** Cast once at the top, then all three results come free.

```python
num_1 = int(input("First number: "))
num_2 = int(input("Second number: "))

print(f"Sum:        {num_1 + num_2}")
print(f"Difference: {num_1 - num_2}")
print(f"Product:    {num_1 * num_2}")
```

**5. Age machine.** Simple subtraction after an int cast.

```python
birth_year = int(input("What year were you born? "))
age_in_2026 = 2026 - birth_year

print(f"You turn {age_in_2026} in 2026.")
```

**6. Temperature converter.** Float cast for decimals, `:.1f` for tidy output.

```python
celsius = float(input("Temperature in Celsius: "))
fahrenheit = celsius * 9 / 5 + 32

print(f"{celsius}°C is {fahrenheit:.1f}°F")
```

**7. Number check.** `.isdigit()` reports whether casting to int would work.

```python
text = input("Type anything: ")
print(text.isdigit())   # True only if every character is a digit
```

## Part 3 — Challenge

**8. Mad libs.** Collect three words, then let the f-string drop them into the story.

```python
adjective = input("Give me an adjective: ")
noun = input("Give me a noun: ")
verb = input("Give me a verb: ")

story = f"""Once upon a time, a {adjective} {noun}
decided to {verb} all the way to the moon.
The end."""

print(story)
```

**9. Receipt builder.** Int cast for quantity, float for price, `:.2f` for money.

```python
item = input("Item name: ")
quantity = int(input("Quantity: "))
unit_price = float(input("Unit price: "))

total = quantity * unit_price
print(f"RECEIPT: {quantity} x {item} @ {unit_price:.2f} = {total:.2f}")
```
