# Numbers Exercises

Write your answer in the empty code cell under each question. Check yourself against `solutions.ipynb` when done.

**Part 1** = warm-up · **Part 2** = practice · **Part 3** = challenge.

## Part 1 — Warm-up

**1. Four operations.** With `a = 17` and `b = 5`, print the result of `+`, `-`, `*` and `/`. Notice what type `/` always returns.

**2. Divide differently.** Predict first, then run: `17 / 5`, `17 // 5`, `17 % 5` and `-17 // 5`. Add a comment explaining what `//` and `%` do.

**3. Powers.** Compute 2 to the power of 10, and the cube of 7, using the `**` operator.

## Part 2 — Practice

**4. Even or odd?** Using `%`, write an expression that is `True` when `n = 42` is even, and another that is `True` when it's odd.

**5. Precedence puzzles.** Write down your guess for each as a comment, then verify:

- `2 + 3 * 4`
- `(2 + 3) * 4`
- `2 ** 3 ** 2` (careful — `**` is *right*-associative!)
- `-3 ** 2`

**6. Handy built-ins.** In one cell (with comments): use `abs(-7.5)`, `max(3, 11, 7)`, `min(3, 11, 7)` and `round(3.14159, 2)`.

**7. Floating trouble.** Run `print(0.1 + 0.2)` and `print(0.1 + 0.2 == 0.3)`. Add a comment explaining why the second line prints `False`.

## Part 3 — Challenge

**8. The number family.** Convert `7` to a float and `7.9` to an int. Then build the complex number `2 + 3j` and print its `.real`, `.imag`, and its distance from zero with `abs()`.

**9. Time splitter.** With `total = 3665` seconds, use only `//` and `%` to work out hours, minutes and remaining seconds. Print it like: `1h 1m 5s`.
