## Part 1 — Warm-up

**1. Four operations.** Only `/` returns a float, even for clean divisions.

```
a = 17
b = 5

print(a + b)   # 22
print(a - b)   # 12
print(a * b)   # 85
print(a / b)   # 3.4  -> always a float
```

**2. Divide differently.** `//` floors toward negative infinity; `%` returns what's left over.

```
print(17 / 5)    # 3.4
print(17 // 5)   # 3    -> floor division (rounds DOWN)
print(17 % 5)    # 2    -> remainder of 17 = 3*5 + 2
print(-17 // 5)  # -4   -> floors, not truncates! (-4 <= -3.4)
```

**3. Powers.**

```
print(2 ** 10)   # 1024
print(7 ** 3)    # 343
```

## Part 2 — Practice

**4. Even or odd.** Remainder zero ⇔ divisible by two.

```
n = 42

is_even = n % 2 == 0
is_odd = n % 2 != 0

print(is_even)   # True
print(is_odd)    # False
```

**5. Precedence puzzles.** `**` binds tighter than unary minus and is right-associative.

```
print(2 + 3 * 4)    # 14   -> multiplication first
print((2 + 3) * 4)  # 20   -> parentheses override
print(2 ** 3 ** 2)  # 512  -> right-assoc: 2 ** (3 ** 2) = 2 ** 9
print(-3 ** 2)      # -9   -> reads as -(3 ** 2), NOT (-3) ** 2
```

**6. Handy built-ins.**

```
print(abs(-7.5))        # 7.5   -> distance from zero
print(max(3, 11, 7))    # 11
print(min(3, 11, 7))    # 3
print(round(3.14159, 2))# 3.14  -> rounded to 2 decimal places
```

**7. Floating trouble.** Binary floating-point can't store `0.1` exactly, so tiny errors accumulate — never compare floats with `==`.

```
print(0.1 + 0.2)          # 0.30000000000000004
print(0.1 + 0.2 == 0.3)   # False!
```

## Part 3 — Challenge

**8. The number family.** `int()` truncates, complex numbers carry `.real`/`.imag`, and `abs()` gives magnitude (√13 ≈ 3.606).

```
print(float(7))     # 7.0
print(int(7.9))     # 7 — chops, no rounding

z = 2 + 3j
print(z.real)       # 2.0
print(z.imag)       # 3.0
print(abs(z))       # 3.605551275463989 — sqrt(2² + 3²)
```

**9. Time splitter.** Peel off the biggest unit first, keep the remainder for the next.

```
total = 3665

hours = total // 3600        # 1
remainder = total % 3600     # 65
minutes = remainder // 60    # 1
seconds = remainder % 60     # 5

print(f"{hours}h {minutes}m {seconds}s")   # 1h 1m 5s
```
