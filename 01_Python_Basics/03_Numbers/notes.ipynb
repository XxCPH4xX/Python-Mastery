## 1. The Three Numeric Types

Python provides three built-in numeric types, each suited to different mathematical tasks:

| Type | What it holds | Example | Use case |
| --- | --- | --- | --- |
| `int` | Whole numbers, unlimited size | `42`, `-1000`, `9999999999999999` | Counting, indexing, IDs |
| `float` | Decimal numbers (IEEE 754) | `3.14`, `-0.001`, `2.5e8` | Measurements, science, money* |
| `complex` | Numbers with real + imaginary parts | `3+5j`, `2j`, `-1-2j` | Engineering, signal processing |

\* For precise financial calculations, use the `decimal` module instead of `float` to avoid rounding errors.

**Syntax:**
```python
age = 25                # int
price = 19.99           # float
impedance = 3 + 5j      # complex
```

Variables become numeric simply by assigning a number — Python infers the type from the syntax.

````python
x = 1  # int
y = 2.8  # float
z = 1j  # complex

print(type(x))
print(type(y))
print(type(z))
````

To verify the type of any object in Python, use the type() function:

````python
print(type(x))
print(type(y))
print(type(z))
````

## 2. Integers (`int`) — Unlimited Precision

An **integer** is a whole number — positive, negative, or zero — with no decimal point. Unlike many languages (C, Java) where integers overflow at a fixed size, Python integers have **unlimited precision**: they grow as large as your memory allows.

**Syntax:**
```python
x = 42
y = -1000
z = 999999999999999999999999999999    # 30 digits, no problem!
```

> 🔍 **Under the Hood:** Python 3 integers use arbitrary precision arithmetic. Small integers (-5 to 256) are cached as singletons for performance — every `42` in your program is literally the same object. Larger integers are dynamically allocated: Python stores them as arrays of "digits" in base 2³⁰ (or 2¹⁵ on 32-bit systems), so a 100-digit number takes more memory but never overflows.

````python
x = 1
y = 35656222554887711
z = -3255522

print(x, type(x))
print(y, type(y))
print(z, type(z))
````

## 3. Floats (`float`) — Decimal Numbers

A **float** (floating-point number) represents decimals — numbers with a fractional part. Python floats follow the IEEE 754 double-precision standard (64 bits), giving you about 15-17 decimal digits of precision.

**Syntax:**
```python
x = 3.14
y = -0.001
z = 2.0          # still a float, even though it looks whole
```

Floats can also use **scientific notation** with `e` or `E` to represent powers of 10:
```python
x = 3.5e2        # 3.5 × 10² = 350.0
y = 1.2e-4       # 1.2 × 10⁻⁴ = 0.00012
```

> 🔍 **Under the Hood:** Floats store numbers in binary scientific notation: sign × mantissa × 2^exponent. This is fast but imprecise — `0.1 + 0.2` does not exactly equal `0.3` due to binary representation limits. For exact decimal arithmetic (finance, accounting), use the `decimal.Decimal` type instead.

````python
x = 1.10
y = 1.0
z = -35.59

print(x, type(x))
print(y, type(y))
print(z, type(z))
````

Float can also be scientific numbers with an "e" to indicate the power of 10.

````python
x = 35e3
y = 12e4
z = -87.7e100

print(x, type(x))
print(y, type(y))
print(z, type(z))
````

## 4. Complex Numbers (`complex`) — Real + Imaginary

A **complex number** has a real part and an imaginary part. In mathematics, the imaginary unit is `i` (where i² = -1); Python uses `j` instead, following engineering convention.

**Syntax:**
```python
z = 3 + 5j       # real part: 3, imaginary part: 5
w = 2j           # real part: 0, imaginary part: 2
x = -1 - 2j      # both parts can be negative
```

You can access the parts with `.real` and `.imag` attributes:
```python
z = 3 + 5j
print(z.real)    # 3.0
print(z.imag)    # 5.0
```

> 🔍 **Under the Hood:** Complex numbers are stored as a pair of floats (real, imaginary). Python provides built-in support for complex arithmetic: `(1+2j) * (3+4j)` works out of the box. This is rare — most languages require a library.

````python
x = 3 + 5j
y = 5j
z = -5j

print(x, type(x))
print(y, type(y))
print(z, type(z))
````

## 5. Type Conversion Between Numbers

You can explicitly convert between numeric types using `int()`, `float()`, and `complex()`. This is useful when you need a specific type — for example, converting user input from string to number, or truncating decimals.

**Syntax:**
```python
int(x)       # converts to integer (truncates decimals, doesn't round)
float(x)     # converts to float
complex(x)   # converts to complex (real part = x, imaginary part = 0)
```

**Conversion rules:**
- `int()` **truncates** toward zero: `int(2.9)` → `2`, `int(-2.9)` → `-2`
- `float()` adds a decimal point: `float(5)` → `5.0`
- `complex()` cannot convert back to `int` or `float` — you must extract `.real` or `.imag` first
- You can create complex from two numbers: `complex(3, 5)` → `3+5j`

> ⚠️ **Watch Out:** `int()` truncates, it does not round. Use `round()` if you want mathematical rounding: `round(2.6)` → `3`.

````python
x = 1  # int
y = 2.8  # float
z = 1j  # complex

# convert from int to float:
a = float(x)

# convert from float to int:
b = int(y)

# convert from int to complex:
c = complex(x)

print(a)
print(b)
print(c)

print(type(a))
print(type(b))
print(type(c))
````

## 6. Mixing Number Types in Arithmetic

When you combine different numeric types in an operation, Python automatically converts to the "wider" type to avoid losing precision:

| Operation | Result type | Reason |
| --- | --- | --- |
| `int + int` | `int` | Both same type |
| `int + float` | `float` | Float is wider (has decimals) |
| `float + float` | `float` | Both same type |
| `int + complex` | `complex` | Complex is widest |
| `float + complex` | `complex` | Complex is widest |

This is called **implicit type coercion** — Python promotes the narrower type automatically.

```python
x = 5 + 2.5        # int + float -> float (7.5)
y = 3 + 2j         # int + complex -> complex (3+2j)
```

## ⚠️ Common Mistakes & Gotchas

| Mistake | Problem | Fix |
| --- | --- | --- |
| Assuming `int()` rounds | `int(2.9)` → `2`, not `3` — it truncates | Use `round(2.9)` → `3` for rounding |
| Expecting exact decimal arithmetic with floats | `0.1 + 0.2` → `0.30000000000000004` (binary representation) | Use `decimal.Decimal("0.1")` for exact decimals |
| Dividing integers expecting an integer | `5 / 2` → `2.5` (float division in Python 3) | Use `5 // 2` → `2` for integer division |
| Using `i` instead of `j` for complex | `3 + 5i` raises `NameError` | Python uses `j`: `3 + 5j` |
| Converting complex directly to int/float | `int(3+5j)` raises `TypeError` | Extract the part: `int(z.real)` |
| Forgetting scientific notation is a float | `1e3` looks like 1000 but is `1000.0` (float) | Cast if you need int: `int(1e3)` |

```python
# The float precision trap:
print(0.1 + 0.2)              # 0.30000000000000004 -- binary representation issue
print(0.1 + 0.2 == 0.3)       # False -- NEVER use == for float comparisons

# Safe float comparison using tolerance:
import math
result = 0.1 + 0.2
print(math.isclose(result, 0.3))   # True -- allows tiny difference

# The division surprise (Python 3):
print(5 / 2)                  # 2.5 (float division, even with ints)
print(5 // 2)                 # 2 (integer division, floors the result)
```

## 💡 Best Practices & Pro Tips

- **Use `int` for counting, indexing, and exact quantities.** List indices, loop counters, database IDs — anything that must be whole.
- **Use `float` for measurements and scientific calculations.** Temperature, distance, probabilities — but remember floats are approximate.
- **For money, use `decimal.Decimal`, not `float`.** `from decimal import Decimal; price = Decimal("19.99")` gives exact arithmetic without rounding errors.
- **Compare floats with tolerance, not `==`.** Use `math.isclose(a, b)` to allow tiny differences from floating-point imprecision.
- **Remember `/` always returns float, even for integers.** Use `//` when you need integer division: `10 // 3` → `3`.
- **Use `round()` for rounding, `int()` for truncation.** They are not the same: `int(2.9)` → `2`, `round(2.9)` → `3`.
- **🤖 AI-engineering relevance:** Neural networks live in float land — weights, gradients, loss values are all floats. Understanding float precision matters when debugging why your model's loss is `0.30000000000000004` instead of `0.3`. NumPy and PyTorch default to float32 (single precision) for speed; use float64 (double) only when precision matters. Complex numbers appear in signal processing (Fourier transforms, audio) and quantum computing libraries.

**Key takeaways:**

- Python has three numeric types: `int` (unlimited precision whole numbers), `float` (approximate decimals), and `complex` (real + imaginary).
- Integers never overflow — they grow as large as memory allows.
- Floats are fast but imprecise — `0.1 + 0.2 != 0.3` due to binary representation. Use `math.isclose()` for comparisons.
- Use `j` (not `i`) for complex numbers: `3+5j`.

- Mix numeric types freely — Python promotes to the wider type automatically.

## 7. Generating Random Numbers

Python does not have a built-in `random()` function, but the standard library includes the `random` module for generating random numbers. This is essential for simulations, games, sampling, and testing.

**Common functions:**
```python
import random

random.randint(a, b)       # random integer between a and b (inclusive)
random.random()            # random float between 0.0 and 1.0
random.uniform(a, b)       # random float between a and b
random.choice(sequence)    # pick a random item from a list/tuple
```

**Example:**
```python
import random

dice_roll = random.randint(1, 6)        # simulate a 6-sided die
probability = random.random()           # 0.0 <= probability < 1.0
temperature = random.uniform(20.0, 30.0)  # random float in range
```

> 💡 **Pro Tip:** For cryptographic applications (passwords, tokens, security keys), use the `secrets` module instead of `random` — it provides cryptographically strong randomness.
