## Part 1 — Warm-up

**1. Arithmetic tour.** All seven in one pass.

```
a = 20
b = 6

print(a + b)    # 26
print(a - b)    # 14
print(a * b)    # 120
print(a / b)    # 3.3333333333333335 -> true division (float)
print(a // b)   # 3                  -> floor division
print(a % b)    # 2                  -> remainder
print(a ** b)   # 64000000           -> power
```

**2. Shortcut chains.** Each augmented assignment folds the step into itself.

```
x = 10
x += 5;  print(x)   # 15
x *= 2;  print(x)   # 30
x -= 3;  print(x)   # 27
x //= 4; print(x)   # 6
x %= 3;  print(x)   # 0
```

**3. Comparison set.** `7 == 7.0` is `True` — Python compares *values* across int/float.

```
print(7 < 7.0)    # False
print(7 <= 7.0)   # True
print(7 > 7.0)    # False
print(7 >= 7.0)   # True
print(7 == 7.0)   # True — same value, different type
print(7 != 7.0)   # False
```

## Part 2 — Practice

**4. Logic combo.**

```
temp = 28
comfortable = 20 <= temp <= 30   # True
hot = temp > 30                  # False

print(comfortable or hot)   # True
print(not hot)              # True
```

**5. Membership tests.** For dicts, `in` checks **keys**, not values.

```
print("py" in "python")          # True — substring
print(3 in [1, 2, 3])            # True — list element
print("k" in {"key": 1})         # True — dict KEY lookup
print(5 not in range(3))         # True — range is 0,1,2
```

**6. is vs ==.** Values built at *runtime* aren't shared, so equal numbers can still be distinct objects. Rule of thumb: use `==` for numbers and strings, reserve `is` for `None` checks.

```
a = 256
b = 256
print(a == b, a is b)   # True True — 256 sits inside the interned cache (-5..256)

c = 257
d = int("257")          # created at runtime, not from the cache
print(c == d, c is d)   # True False — same value, different objects
```

**7. Precedence traps.** `**` binds tighter than unary minus; comparisons bind tighter than `not`.

```
print(-3 ** 2)    # -9  -> -(3 ** 2)
print((-3) ** 2)  # 9   -> parentheses change everything
print(not 5 > 3)  # False -> parses as not (5 > 3)
print(2 * 3 ** 2) # 18  -> exponent first: 2 * 9
```

## Part 3 — Challenge

**8. Expression evaluator.** By hand: `2² + 5×10 − 10//2` → `4 + 50 − 5` = **49**.

```
a, b, c = 2, 5, 10

result = a ** 2 + b * c - c // a
print(result)   # 49 — matches the hand calculation
```

**9. Research bonus — the walrus.** `:=` assigns **and returns the value in the same expression**, so it can live inside a `print()` call — plain `=` is a statement, not an expression, and would be a syntax error there.

```
print(answer := 21 * 2)   # prints 42 AND binds answer
print(answer)             # 42 — usable afterwards
```
