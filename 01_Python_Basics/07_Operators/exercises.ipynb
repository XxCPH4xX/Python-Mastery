## Part 1 — Warm-up

**1. Arithmetic tour.** For `a = 20` and `b = 6`, print the result of all seven arithmetic operators: `+`, `-`, `*`, `/`, `//`, `%`, `**`.

```python
a = 20
b = 6

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a % b)
print(a**b)
print(a // b)
```

**2. Shortcut chains.** Start with `x = 10`. Apply in order: `x += 5`, `x *= 2`, `x -= 3`, `x //= 4`, `x %= 3` — printing `x` after each step.

```python
x = 10
x += 5
print(x)  # 15
x *= 2
print(x)  # 30
x -= 3
print(x)  # 27
x //= 4
print(x)  # 6
x %= 3
print(x)  # 0
```

**3. Comparison set.** Evaluate `<`, `<=`, `>`, `>=`, `==` and `!=` between `7` and `7.0`. Is `7 == 7.0` what you expected?

```python
print(7 < 7.0)  # False
print(7 <= 7.0)  # True
print(7 > 7.0)  # False
print(7 >= 7.0)  # True
print(7 == 7.0)  # True — same value, different type
print(7 != 7.0)  # False
```

## Part 2 — Practice

**4. Logic combo.** With `temp = 28`, define `comfortable = 20 <= temp <= 30` and `hot = temp > 30`. Print `comfortable or hot` and `not hot`.

```python
temp = 28
comfortable = 20 <= temp <= 30
hot = temp > 30

print(comfortable or hot)
print(not hot)
```

**5. Membership tests.** Check each of these and note the boolean result: `"py" in "python"`, `3 in [1, 2, 3]`, `"k" in {"key": 1}`, `5 not in range(3)`.

```python
print("py" in "python")  # True — substring
print(3 in [1, 2, 3])  # True — list element
print("k" in {"key": 1})  # True — dict KEY lookup
print(5 not in range(3))  # True — range is 0,1,2
```

**6. is vs ==.** Compare both `==` and `is` for the pairs `a = 256; b = 256` and `c = 257; d = 257`. Comment on how integer caching explains the difference.

```python
a = 256
b = 256
print(a == b, a is b)  # True True — 256 sits inside the interned cache (-5..256)

b = 257
c = 257
print(b is c, b == c)
```

**7. Precedence traps.** Guess each result first (write it as a comment), then run: `-3 ** 2`, `(-3) ** 2`, `not 5 > 3`, `2 * 3 ** 2`.
