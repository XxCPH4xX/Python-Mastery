## 1. Arithmetic Operators

Arithmetic operators do maths on numbers: `+ - * / // % **`. The subtle one is division, which comes in **two** flavours:

- `/` **true division** — always returns a `float`, even when the answer is whole (`10 / 2` gives `5.0`).
- `//` **floor division** — rounds *down* to the next whole number (`7 // 2` gives `3`; and `-7 // 2` gives `-4`, because `-4` sits below `-3.5`).

Then `%` (**modulo**) yields the remainder after floor division, and `**` raises a number to a power.

**Syntax:**
```python
total = a + b      # addition
diff  = a - b      # subtraction
prod  = a * b      # multiplication
ratio = a / b      # true division -> always float
whole = a // b     # floor division -> rounded down
rem   = a % b      # remainder (modulo)
power = a ** b     # exponentiation
```


````python
price = 250            # BDT
quantity = 4

print(price * quantity)     # 1000 -> total cost
print(price + 50)           # 300  -> price rise
print(price - 30)           # 220  -> discount
print(price / 3)            # 83.33333333333333 -> ALWAYS a float
print(price ** 2)           # 62500 -> price squared
````

````python
print(10 / 2)     # 5.0   <- float, even though the answer is whole!
print(10 / 3)     # 3.3333333333333335
print(10 // 3)    # 3     <- floor division throws the fraction away
print(10 % 3)     # 1     <- remainder: 10 = 3*3 + 1
print(2 ** 10)    # 1024  <- 2 to the power of 10

print(-7 // 2)    # -4, NOT -3! floor division rounds toward NEGATIVE infinity
print(-7 % 2)     # 1     <- Python's modulo follows the divisor's sign
````

````python
# Use case 1: even or odd? A number is even when its remainder is 0
number = 42
print(number % 2 == 0)       # True -> even
print(7 % 2)                 # 1    -> odd numbers leave remainder 1

# Use case 2: wrap-around cycling (whose turn is it?)
players = ["Sarah", "Ali", "Rahim", "Nadia"]
turn = 7                     # the 8th move of the game
print(players[turn % len(players)])   # Rahim -> index wraps around safely

# Use case 3: converting units with // and % together
total_minutes = 135
print(total_minutes // 60, "hours and", total_minutes % 60, "minutes")
````

## 2. Assignment Operators

The plain `=` stores a value. Every other assignment here is an **augmented** shortcut that modifies a variable using its own current value: `score += 10` means exactly `score = score + 10`.

| Operator | Example | Same as |
| --- | --- | --- |
| `=` | `x = 5` | store `5` |
| `+=` | `x += 3` | `x = x + 3` |
| `-=` | `x -= 3` | `x = x - 3` |
| `*=` | `x *= 3` | `x = x * 3` |
| `/=` | `x /= 3` | `x = x / 3` (float!) |
| `//=` | `x //= 3` | `x = x // 3` |
| `%=` | `x %= 3` | `x = x % 3` |
| `**=` | `x **= 3` | `x = x ** 3` |


````python
score = 0
score += 25          # picked up a coin
score *= 2           # hit a double-points booster
score -= 10          # fell into a trap
print(score)         # 40

balance = 1000
balance /= 4         # NOTE: /= always produces a float
print(balance)       # 250.0

level = 2
level **= 3          # 2 cubed
print(level)         # 8

streak = 17
streak %= 10         # keep only the last digit
print(streak)        # 7
````

## 3. Comparison Operators

Comparisons ask questions and answer with a boolean: `== != > < >= <=`. They also **chain** beautifully — `min <= value <= max` reads like plain English and evaluates the middle value only once.

(Lesson 6 covered the boolean side in depth; here we focus on comparisons inside larger expressions.)


````python
temperature = 31

print(temperature >= 30)           # True -> officially a hot day
print(temperature != 25)           # True
print(20 <= temperature <= 35)     # True -> inside the comfortable band

password = "s3cret"
attempt = "hunter2"
print(password == attempt)         # False -> access denied
print(attempt != password)         # True  -> same fact, flipped
````

## 4. Logical Operators: `and`, `or`, `not`

Logical operators combine boolean expressions into bigger decisions:

| Python | C / Java / JavaScript | Meaning | Result is `True` when… |
| --- | --- | --- | --- |
| `and` | `&&` | both sides must hold | both operands are true |
| `or` | `\|\|` | either side is enough | at least one operand is true |
| `not` | `!` | flips the value | the operand is false |

If you come from another language, retrain your fingers: Python spells the logic out as words — `&&`, `\|\|` and `!` are simply invalid syntax here.

**Short-circuit evaluation:** `and` stops the moment its left side is `False` (the verdict is already decided), and `or` stops the moment its left side is `True`. The right-hand expression may never run at all — which turns short-circuiting into a safety tool for risky code.


````python
age = 20
has_ticket = True

print(age >= 18 and has_ticket)    # True  -> both conditions hold
print(age < 18 or has_ticket)      # True  -> one is enough
print(not has_ticket)              # False -> flipped

is_weekend = False
is_holiday = True
print(is_weekend or is_holiday)    # True  -> no office today
````

````python
x = 0

# Without the guard, 10 / x would crash with ZeroDivisionError.
# Because x != 0 is False, `and` SHORT-CIRCUITS: 10 / x never runs.
safe = (x != 0 and 10 / x > 2)
print(safe)                         # False -- no crash!

name = ""                           # imagine this came from input()
display_name = name or "Guest"      # or-shortcut: pick a fallback default
print(display_name)                 # Guest
````

## 5. Identity Operators: `is`, `is not`

`is` checks whether two names point at the **same object in memory** — not whether their values match (that is `==`'s job).

**Syntax:**
```python
x is None          # the idiomatic None test
x is not None      # its negation
```

Small integers are cached by CPython (remember interning from the Variables lesson?), which is why `a is b` *sometimes* works for numbers. Treat that as an accident — never as a technique.


````python
a = [1, 2, 3]
b = [1, 2, 3]
c = a

print(a == b)      # True  -> same contents
print(a is b)      # False -> two separate list objects
print(a is c)      # True  -> one object, two names

response = None                    # server hasn't answered yet
if response is None:               # the idiomatic None check
    print("Waiting for the server...")
````

## 6. Membership Operators: `in`, `not in`

Membership operators ask *"is this thing inside that collection?"* and answer with a boolean. They work on strings (substring search), lists, tuples, sets, ranges — and dictionaries, where they check **keys**, not values.

**Syntax:**
```python
"fun" in some_string
item in some_list
key in some_dict
```


````python
menu = ["biriyani", "khichuri", "fuchka"]

print("biriyani" in menu)          # True
print("pizza" not in menu)         # True

sentence = "Python is fun"
print("fun" in sentence)           # True -> substring search

prices = {"tea": 15, "coffee": 40}
print("tea" in prices)             # True  -> membership checks KEYS
print(15 in prices)                # False -> values are NOT checked
````

````python
lottery_numbers = {4, 8, 15, 16, 23}
print(23 in lottery_numbers)            # True -> set lookups are very fast
print(99 not in lottery_numbers)        # True

ticket = (7, 21)
print(ticket in [(7, 21), (1, 2)])      # tuples compare as wholes
print(5 in range(10))                   # True -> ranges support it too
````

## 7. Bitwise Operators: `& | ^ ~ << >>`

Bitwise operators treat integers as raw binary digits and work on them bit by bit. They look like their logical cousins but act on numbers, not booleans.

| Operator | Name | Example | Result |
| --- | --- | --- | --- |
| `&` | AND | `12 & 10` | `8` |
| `\|` | OR | `12 \| 10` | `14` |
| `^` | XOR | `12 ^ 10` | `6` |
| `~` | NOT (flip every bit) | `~5` | `-6` |
| `<<` | left shift (doubles per step) | `3 << 2` | `12` |
| `>>` | right shift (halves per step) | `12 >> 2` | `3` |

To see what actually happens, print numbers in binary with `bin()`: `12` is `1100` and `10` is `1010`. AND-ing the columns gives `1000`, which is `8`.


````python
a, b = 12, 10

print(bin(a), bin(b))     # 0b1100 0b1010  <- look at the bits directly

print(a & b)              # 8   (0b1000)  bits ON in BOTH numbers
print(a | b)              # 14  (0b1110)  bits ON in EITHER number
print(a ^ b)              # 6   (0b0110)  bits ON in EXACTLY ONE number
print(~5)                 # -6            flip every bit (= -n - 1)
print(3 << 2)             # 12            shift left = multiply by 2, twice
print(12 >> 2)            # 3             shift right = integer-divide by 4
````

````python
# Classic real-world use: pack several yes/no flags into ONE integer
READ, WRITE, EXECUTE = 4, 2, 1          # 0b100, 0b010, 0b001

my_perms = READ | WRITE                 # grant two permissions at once
print(bin(my_perms))                    # 0b110

print((my_perms & READ) == READ)        # True  -> can I read?
print((my_perms & EXECUTE) == EXECUTE)  # False -> can I execute?

my_perms |= EXECUTE                     # grant execute later on
print(bin(my_perms))                    # 0b111 -> all three now
````

## 8. Operator Precedence: Who Goes First?

When one expression mixes operators, Python follows a fixed priority order — highest first:

| Priority | Operators | Category |
| --- | --- | --- |
| 1 (highest) | `()` | parentheses |
| 2 | `**` | exponentiation |
| 3 | `+x`, `-x`, `~x` | unary plus / minus / bitwise NOT |
| 4 | `*`, `/`, `//`, `%` | multiplication family |
| 5 | `+`, `-` | addition / subtraction |
| 6 | `<<`, `>>` | bitwise shifts |
| 7 | `&` | bitwise AND |
| 8 | `^` | bitwise XOR |
| 9 | `\|` | bitwise OR |
| 10 | `==`, `!=`, `>`, `<`, `>=`, `<=`, `is`, `in` | comparisons, identity, membership |
| 11 | `not` | logical NOT |
| 12 | `and` | logical AND |
| 13 | `or` | logical OR |
| 14 (lowest) | `=`, `+=`, `-=`, … | assignment |

Note the logic reading order: `not` binds tighter than `and`, and `and` binds tighter than `or` — so `a or b and c` really means `a or (b and c)`.

**Best practice:** remember roughly where arithmetic sits, then use **parentheses for everything else**. Code is read far more often than it is written; `(a and b) or c` costs nothing and removes every doubt.


````python
print(2 + 3 * 4)            # 14  -> * runs before +
print((2 + 3) * 4)          # 20  -> parentheses win

print(2 ** 3 ** 2)          # 512 -> ** groups RIGHT to LEFT: 2**(3**2)
print(-2 ** 2)              # -4  -> unary minus binds LOOSER than **
print((-2) ** 2)            # 4   -> parentheses change everything

print(not 1 == 2)           # True -> == first, then not
print(True or False and False)   # True -> and before or
````

## ⚠️ Common Mistakes & Gotchas

| Mistake | Problem | Fix |
| --- | --- | --- |
| Expecting `10 / 2` to be `5` | `/` returns `5.0`, a float — which breaks indexing and `range()` | Use `//` when you need a whole number |
| Assuming `-7 // 2` is `-3` | Floor division rounds toward **negative infinity**: the answer is `-4` | Use `int(-7 / 2)` to truncate toward zero instead |
| Typing `&&`, `\|\|` or `!` | Invalid Python syntax | Use the words `and`, `or`, `not` |
| Using `&` where you mean `and` | Misleading, and precedence surprises (`&` binds tighter than comparisons) | Logic → `and`; bits → `&` |
| Testing values with `is` (`x is 5`) | Unreliable — depends on CPython's small-int cache | Compare with `==`; keep `is` for `None` |
| Trusting precedence in long lines | `not x == y` means `not (x == y)` — usually the opposite of what was intended | Add parentheses; give sub-expressions names |


## 💡 Best Practices & Pro Tips

- **Parentheses over memory.** Nobody should have to recall whether `%` beats `<<`; write `(a * b) % m`.
- **Use augmented assignments** (`total += price`) — they state intent ("update this variable") more clearly than `total = total + price`.
- **Break monster expressions into named steps:** `gross = price * quantity` then `net = gross - discount` beats one unreadable line.
- **Lean on `% n` for wrap-around and batching** — e.g. log progress only when `step % 100 == 0`, or cycle playlists with `i % len(songs)`.
- **Guard risky divisions with short-circuiting:** `denominator != 0 and total / denominator > target`.
- **🤖 AI-engineering relevance:** vectorised comparison operators build boolean masks over NumPy arrays (`arr[arr > 0]`), shifts and masks appear in hashing/quantisation code, and `and/or` guards keep expensive API calls from running on empty inputs.


## 📌 Summary

| Family | Operators | One-line memory hook |
| --- | --- | --- |
| Arithmetic | `+ - * / // % **` | `/` gives a float, `//` floors, `%` is the leftover |
| Assignment | `= += -= *= /= //= %= **=` | shortcut for `x = x <op> value` |
| Comparison | `== != > < >= <=` (chainable) | answers `True`/`False` |
| Logical | `and or not` | words, not `&& \|\| !`; they short-circuit |
| Identity | `is`, `is not` | same object? reserve for `None` |
| Membership | `in`, `not in` | substring, list item, dict keys |
| Bitwise | `& \| ^ ~ << >>` | binary digits — inspect them with `bin()` |
| Precedence | `()` → `**` → unary → `* / // %` → `+ -` → shifts → `&` → `^` → `\|` → comparisons → `not` → `and` → `or` → `=` | when unsure: add parentheses |

Key takeaways:

- `/` always returns a `float`; `//` rounds toward negative infinity; `%` gives the remainder and powers even/odd tests and wrap-around indexing.
- `and`/`or` short-circuit — the right side may never run, which is a feature you can exploit.
- `is` means "same object"; use it for `None` and nothing else.
- Precedence decides who goes first, but parentheses decide who is understood.
