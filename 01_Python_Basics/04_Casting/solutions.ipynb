## Part 1 — Warm-up

**1. Convert basics.** Four directions of casting.

```
print(int("25"), type(int("25")))     # 25 <class 'int'>
print(int(3.99), type(int(3.99)))     # 3 <class 'int'>
print(float(7), type(float(7)))       # 7.0 <class 'float'>
print(str(12), type(str(12)))         # 12 <class 'str'>
```

**2. Broken conversion.** `int()` refuses text that isn't a whole number.

```
# int("abc")
# ValueError: invalid literal for int() with base 10: 'abc'
#
# The message tells you exactly what failed: Python tried to read
# "abc" as a base-10 integer and found letters instead of digits.
```

**3. Chop vs round.**

```
print(int(3.99))    # 3 — int() TRUNCATES (chops the decimals)
print(round(3.99))  # 4 — round() rounds to nearest
```

## Part 2 — Practice

**4. Text math trap.** `+` means *concatenate* for strings, *addition* for numbers.

```
print("5" + "5")              # 55 — glued together as text
print(int("5") + int("5"))    # 10 — real math after casting
```

**5. Comma problem.** The comma isn't a valid character inside a number literal for `float()`, so we strip it first.

```
# float("1,000")
# ValueError: could not convert string to float: '1,000'

cleaned = "1,000".replace(",", "")   # "1000"
print(float(cleaned))                # 1000.0
```

**6. Round trip.** Each cast changes the type until we deliberately land on str again.

```
step1 = int("42")        # 42   <class 'int'>
step2 = step1 + 8        # 50
step3 = float(step2)     # 50.0 <class 'float'>
step4 = step3 / 2        # 25.0
final = str(step4)       # "25.0" <class 'str'>

print(final, type(final))
```

**7. Boolean casting.** Empty/zero values are falsy — but any non-empty string is truthy, even the word "False"!

```
print(bool(0))       # False — zero is falsy
print(bool(42))      # True  — non-zero number
print(bool(""))      # False — empty string
print(bool("False")) # True! — non-empty STRING, contents don't matter
print(bool(None))    # False — None is falsy
```

## Part 3 — Challenge

**8. Safe-ish casting.** `.isdigit()` predicts whether `int()` will succeed — once you learn `if`, you can gate the cast on it.

```
user_text = "123abc"
print(user_text.isdigit())   # False -> int() would raise ValueError

user_text = "123"
print(user_text.isdigit())   # True  -> safe to cast
print(int(user_text))        # 123
```

**9. Grade average.** `"90.5"` needs `float()`, not `int()` — pick the cast per value.

```
score_1 = int("85")      # whole number -> int
score_2 = float("90.5")  # has a decimal point -> float
score_3 = int("78")      # whole number -> int

average = (score_1 + score_2 + score_3) / 3
print(round(average, 2))     # 84.5
```
