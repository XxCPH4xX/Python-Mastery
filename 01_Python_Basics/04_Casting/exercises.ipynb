# Casting Exercises

Write your answer in the empty code cell under each question. Check yourself against `solutions.ipynb` when done.

**Part 1** = warm-up · **Part 2** = practice · **Part 3** = challenge.

## Part 1 — Warm-up

**1. Convert basics.** Cast `"25"` to int, `3.99` to int, `7` to float and `12` to str. Print each result **and** its type.

**2. Broken conversion.** Try `int("abc")`. Read the error carefully — what does the `ValueError` message tell you?

**3. Chop vs round.** Compare `int(3.99)` with `round(3.99)`. Which one truncates and which one rounds?

## Part 2 — Practice

**4. Text math trap.** Run `"5" + "5"`, then `int("5") + int("5")`. Explain the difference in a comment.

**5. Comma problem.** Try `float("1,000")` — why does it fail? Remove the comma with `.replace(",", "")` and cast successfully.

**6. Round trip.** Start with the string `"42"`: cast to int, add 8, cast to float, divide by 2, then cast back to str. Print the final value and type.

**7. Boolean casting.** Evaluate `bool(0)`, `bool(42)`, `bool("")`, `bool("False")` and `bool(None)`. Which results surprise you?

## Part 3 — Challenge

**8. Safe-ish casting.** For `user_text = "123abc"`, use `.isdigit()` to check whether casting to int would succeed. Print both the boolean and the string itself. Then try the same for `"123"`. (You'll automate this decision properly once you learn `if` statements.)

**9. Grade average.** Scores arrive as text: `"85"`, `"90.5"`, `"78"`. Cast each appropriately (careful — they're not all ints!) and print the average rounded to 2 decimals.
