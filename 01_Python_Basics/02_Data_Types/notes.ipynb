## 1. What Is a Data Type?

A **data type** defines the kind of value a variable can hold and what operations are allowed on it. Think of it as a label that tells Python: *"This is a number, so you can do math"* or *"This is a list, so you can add items to it."*

Python is **dynamically typed** — you never write `int x = 5` like in Java or C. Python figures out the type from the value you assign. Under the hood, every value is an **object**, and every object has a type.

**Why types matter:**
- They determine which operations work: `"hello" + "world"` concatenates strings, but `"hello" - "world"` crashes.
- They control memory layout: an `int` stores differently than a `list`.
- They enable Python to optimize: knowing the type lets the interpreter choose fast code paths.

**Syntax:**
```python
x = 42              # Python infers: x is an int
name = "Alice"      # Python infers: name is a str
type(x)             # inspect the type -> <class 'int'>
```

## 2. Python's Built-in Data Types

Python ships with **14 built-in types** organized into seven categories. You will use half of them daily; the rest appear in specialized scenarios like binary data handling or advanced collections.

| Category | Type | Description | Example |
| --- | --- | --- | --- |
| **Text** | `str` | Unicode character sequences (strings) | `"Hello, AI!"` |
| **Numeric** | `int` | Whole numbers, unlimited precision | `42`, `-1000` |
|  | `float` | Floating-point decimals (IEEE 754) | `3.14`, `-0.001` |
|  | `complex` | Complex numbers (real + imaginary) | `3+5j` |
| **Sequence** | `list` | Ordered, mutable collections | `[1, 2, 3]` |
|  | `tuple` | Ordered, **immutable** collections | `(1, 2, 3)` |
|  | `range` | Lazy integer sequences (for loops) | `range(10)` |
| **Mapping** | `dict` | Key-value pairs (hash table) | `{"name": "Alice"}` |
| **Set** | `set` | Unordered, unique items (mutable) | `{1, 2, 3}` |
|  | `frozenset` | Unordered, unique, **immutable** | `frozenset([1, 2])` |
| **Boolean** | `bool` | Truth values (`True`/`False`) | `True` |
| **Binary** | `bytes` | Immutable byte sequences | `b"Hello"` |
|  | `bytearray` | Mutable byte sequences | `bytearray([65, 66])` |
|  | `memoryview` | Zero-copy buffer access | `memoryview(b"x")` |
| **None** | `NoneType` | Absence of a value (singleton) | `None` |

> 🔍 **Under the Hood:** `NoneType` is a special singleton — there is exactly **one** `None` object in the entire Python runtime. Every `None` you see is the same object. That is why `x is None` works reliably (identity check), and it is the recommended way to test for `None`.

## 3. How Python Determines Types (Dynamic Typing)

Unlike statically typed languages (Java, C++, TypeScript), Python does not require you to declare types. The interpreter **infers** the type from the syntax you use:

| You write | Python sees | Resulting type |
| --- | --- | --- |
| `42` | bare number, no decimal | `int` |
| `3.14` | number with a decimal point | `float` |
| `"hello"` or `'hello'` | quotes | `str` |
| `[1, 2, 3]` | square brackets | `list` |
| `(1, 2, 3)` | parentheses | `tuple` |
| `{"a": 1}` | curly braces with `:` | `dict` |
| `{1, 2, 3}` | curly braces, no `:` | `set` |
| `True`, `False` | boolean literals | `bool` |
| `None` | the absence keyword | `NoneType` |

This is called **dynamic typing** — the same variable can hold different types at different times (though doing this often signals unclear design).

**Syntax:**
```python
x = 10          # x is int
x = "hello"     # now x is str (perfectly legal, but confusing!)
```

## 4. Using `type()` to Inspect Types

The built-in `type()` function returns the type of any object. This is your debugging superpower when a variable does not behave as expected — call `type()` to see what you are actually working with.

**Syntax:**
```python
type(object)     # returns the type object, e.g., <class 'int'>
```

You can also compare types directly:
```python
type(x) == int       # True if x is an integer
isinstance(x, int)   # preferred: also works with subclasses
```

> 💡 **Prefer `isinstance()`** when checking types in real code — it respects inheritance and works with multiple types: `isinstance(x, (int, float))` checks if `x` is numeric.

## 5. Mutable vs Immutable Types

One of Python's most important distinctions: some types can be **changed after creation** (mutable), while others are **frozen forever** (immutable). This affects how you pass data to functions, how memory works, and what bugs you will encounter.

| Mutable (can change) | Immutable (frozen) |
| --- | --- |
| `list` | `int`, `float`, `complex` |
| `dict` | `str` |
| `set` | `tuple` |
| `bytearray` | `frozenset` |
|  | `bytes` |
|  | `bool`, `NoneType` |

**What "mutable" means:**
```python
names = ["Alice", "Bob"]
names.append("Charlie")    # changes the SAME list object
```

**What "immutable" means:**
```python
x = 10
x = x + 5                  # creates a NEW int object, doesn't modify 10
```

> 🔍 **Under the Hood:** When you "change" an immutable value, Python creates a new object and updates the variable to point at it. The original object stays unchanged (and gets garbage-collected if nothing else references it). Mutables modify their contents **in place**, so every variable pointing to that object sees the change.

## 6. Examples: Exploring Each Type Category

Let's see each type category in action. Pay attention to how Python infers the type from the syntax.

````python
# 1. Numeric Types
age = 25  # int (integer)
price = 19.99  # float (floating point number)
z = 3 + 5j  # complex (complex number)

print(age, type(age))
print(price, type(price))
print(z, type(z))
````

````python
# 2. Text Type
greeting = "Hello, World!"  # str (string)

print(greeting, type(greeting))
````

````python
# 3. Sequence Types
fruits = ["apple", "banana"]  # list (mutable/changeable sequence)
coordinates = (10.0, 20.0)  # tuple (immutable sequence)
numbers = range(1, 5)  # range (sequence of numbers: 1, 2, 3, 4)

print(fruits, type(fruits))
print(coordinates, type(coordinates))
print(numbers, type(numbers))
print(list(numbers))
````

````python
# 4. Mapping Type
user = {"name": "Alice", "age": 30}  # dict (dictionary of key-value pairs)

print(user, type(user))
````

````python
# 5. Set Types
unique_numbers = {1, 2, 3, 3}  # set (unordered, unique elements -> {1, 2, 3})
immutable_set = frozenset([1, 2])  # frozenset (immutable version of set)

print(unique_numbers, type(unique_numbers))
print(immutable_set, type(immutable_set))
````

````python
# 6. Boolean Type
is_active = True  # bool (Boolean: True or False)

print(is_active, type(is_active))
````

````python
# 7. Binary Types
raw_data = b"Hello"  # bytes (immutable byte sequences)
mutable_bytes = bytearray(5)  # bytearray (mutable sequence of bytes)
mem_view = memoryview(b"Test")  # memoryview (allows buffer protocol access)

print(raw_data, type(raw_data))
print(mutable_bytes, type(mutable_bytes))
print(mem_view, type(mem_view))
````

````python
# 8. None Type
result = None  # NoneType (represents null / no value)

print(result, type(result))
````

## ⚠️ Common Mistakes & Gotchas

| Mistake | Problem | Fix |
| --- | --- | --- |
| Confusing `{}` for an empty set | `{}` creates an empty **dict**, not a set | Use `set()` for an empty set |
| Treating `(5)` as a tuple | Parentheses alone don't make a tuple — `(5)` is just `5` | Single-item tuple needs a comma: `(5,)` |
| Assuming `type(x) == float` catches all numbers | Misses `int` and `complex` | Use `isinstance(x, (int, float, complex))` |
| Modifying a "copy" that is really an alias | `b = a` does NOT copy a list; both point at the same object | Use `b = a.copy()` or `b = a[:]` |
| Changing types mid-program | `x = 10; x = "hello"` is legal but confusing | Keep variables to one type unless there is a clear reason |
| Using `==` to compare types with inheritance | `type(True) == int` is `False` (bool is a subclass) | Use `isinstance(True, int)` → `True` |

```python
# The empty-set trap:
x = {}
print(type(x))       # <class 'dict'> -- NOT a set!

y = set()
print(type(y))       # <class 'set'> -- correct

# The single-tuple trap:
a = (5)
print(type(a))       # <class 'int'> -- just a number in parens

b = (5,)             # trailing comma
print(type(b))       # <class 'tuple'> -- NOW it's a tuple
```

## 💡 Best Practices & Pro Tips

- **Use `type()` for debugging, `isinstance()` for checks.** When you need to branch on type, `isinstance(x, str)` is clearer and safer than `type(x) == str`.
- **Favor immutable types when data should not change.** Tuples as function return values, frozensets as dict keys — immutability prevents accidental modification bugs.
- **Match the type to the task.** Need unique items? Use `set`. Need order and duplicates? Use `list`. Need fast lookups? Use `dict`.
- **Understand mutability before passing to functions.** If you pass a list to a function, the function can modify it. Pass a tuple if you want protection.
- **Use type hints (annotations) for clarity.** Modern Python lets you write `def greet(name: str) -> str:` — it documents intent and enables editor autocomplete, though Python does not enforce it at runtime.
- **🤖 AI-engineering relevance:** Data pipelines juggle types constantly — NumPy arrays, pandas DataFrames, PyTorch tensors, JSON dicts. Knowing when Python will copy vs alias, and which operations create new objects, is the difference between fast code and memory leaks. Type annotations also help AI assistants (like GitHub Copilot) generate better suggestions.

## 📌 Summary

| Concept | What it means | Example |
| --- | --- | --- |
| **Data type** | The kind of value an object holds | `int`, `str`, `list` |
| **Dynamic typing** | Python infers type from syntax | `x = 42` → `int` |
| **`type(obj)`** | Returns the type of any object | `type("hi")` → `<class 'str'>` |
| **Mutable** | Can be changed after creation | `list`, `dict`, `set` |
| **Immutable** | Cannot be changed after creation | `int`, `str`, `tuple` |
| **`isinstance(obj, type)`** | Checks if obj is an instance of type | `isinstance(5, int)` → `True` |

**Key takeaways:**

- Python has 14 built-in types across seven categories: numeric, text, sequence, mapping, set, boolean, binary, and None.
- Python determines types automatically from syntax — quotes make strings, brackets make lists, curly braces with colons make dicts.
- Use `type()` to inspect types during debugging, and `isinstance()` for type checks in real code.
- Mutable types (list, dict, set) can be changed in place; immutable types (int, str, tuple) create new objects on modification.
- Understanding types prevents bugs and helps you choose the right data structure for each task.
