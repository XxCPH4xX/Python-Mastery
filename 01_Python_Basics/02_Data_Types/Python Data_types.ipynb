# Data Types

#### Data types - A data type in Python defines the kind of value a variable holds, telling the interpreter how to store the data in memory and what operations (like math or string concatenation) can be performed on it.

Python is dynamically typed, meaning you do not need to explicitly declare a variable's data type when creating it. Python determines the type automatically at runtime based on the assigned value.

You can check the data type of any object using the built-in type() function (e.g., type(x)).

Complete Python Data Structures & Types Cheat Sheet

| Category / Data Type | Delimiters / Enclosing Characters | Example Syntax |
| :--- | :--- | :--- |
| **String** (`str`) | Single `' '`, Double `" "`, Triple `""" """` | `"Hello"` |
| **List** (`list`) | Square brackets `[ ]` | `[1, 2, 3]` |
| **Tuple** (`tuple`) | Parentheses `( )` | `(1, 2, 3)` |
| **Set** (`set`) | Curly braces `{ }` | `{1, 2, 3}` |
| **Frozen Set** (`frozenset`) | `frozenset({ })` | `frozenset({1, 2, 3})` |
| **Dictionary** (`dict`) | Curly braces with key-value pairs `{ : }` | `{"a": 1, "b": 2}` |
| **Range** (`range`) | `range(start, stop)` | `range(1, 4)` |
| **Array** (`array` module) | `array('typecode', [ ])` | `array('i', [1, 2, 3])` |
| **NumPy Array** (`numpy`) | `np.array([ ])` | `np.array([1, 2, 3])` |
| **Bytes** (`bytes`) | `b' '` or `b" "` | `b'Hello'` |
| **Bytearray** (`bytearray`) | `bytearray([ ])` | `bytearray([65, 66, 67])` |

#### Data Types

| Category | Type Name | Description | Example |
| --- | --- | --- | --- |
| **Text** | `str` | Character sequences (strings) | `x = str("Hello World")` |
| **Numeric** | `int` | Whole integers | `x = int(20)` |
|  | `float` | Floating-point numbers (decimals) | `x = float(20.5)` |
|  | `complex` | Numbers with real and imaginary parts | `x = complex(1j)` |
| **Sequence** | `list` | Ordered, mutable (changeable) collections | `x = list(("apple", "banana", "cherry"))` |
|  | `tuple` | Ordered, immutable (unchangeable) collections | `x = tuple(("apple", "banana", "cherry"))` |
|  | `range` | Sequence of numbers (often used in loops) | `x = range(6)` |
| **Mapping** | `dict` | Key-value pairs | `x = dict(name="John", age=36)` |
| **Set** | `set` | Unordered collection of unique items | `x = set(("apple", "banana", "cherry"))` |
|  | `frozenset` | Immutable version of a set | `x = frozenset(("apple", "banana", "cherry"))` |
| **Boolean** | `bool` | Logical true or false values | `x = bool(5)` |
| **Binary** | `bytes` | Immutable sequence of bytes | `x = bytes(5)` |
|  | `bytearray` | Mutable sequence of bytes | `x = bytearray(5)` |
|  | `memoryview` | Memory buffer view of bytes object | `x = memoryview(bytes(5))` |

Code Example

```python
# 1. Numeric Types
age = 25  # int (integer)
price = 19.99  # float (floating point number)
z = 3 + 5j  # complex (complex number)

print(age, type(age))
print(price, type(price))
print(z, type(z))
```

```python
# 2. Text Type
greeting = "Hello, World!"  # str (string)

print(greeting, type(greeting))
```

```python
# 3. Sequence Types
fruits = ["apple", "banana"]  # list (mutable/changeable sequence)
coordinates = (10.0, 20.0)  # tuple (immutable sequence)
numbers = range(1, 5)  # range (sequence of numbers: 1, 2, 3, 4)

print(fruits, type(fruits))
print(coordinates, type(coordinates))
print(numbers, type(numbers))
print(list(numbers))
```

```python
# 4. Mapping Type
user = {"name": "Alice", "age": 30}  # dict (dictionary of key-value pairs)

print(user, type(user))
```

```python
# 5. Set Types
unique_numbers = {1, 2, 3, 3}  # set (unordered, unique elements -> {1, 2, 3})
immutable_set = frozenset([1, 2])  # frozenset (immutable version of set)

print(unique_numbers, type(unique_numbers))
print(immutable_set, type(immutable_set))
```

```python
# 6. Boolean Type
is_active = True  # bool (Boolean: True or False)

print(is_active, type(is_active))
```

```python
# 7. Binary Types
raw_data = b"Hello"  # bytes (immutable byte sequences)
mutable_bytes = bytearray(5)  # bytearray (mutable sequence of bytes)
mem_view = memoryview(b"Test")  # memoryview (allows buffer protocol access)

print(raw_data, type(raw_data))
print(mutable_bytes, type(mutable_bytes))
print(mem_view, type(mem_view))
```

```python
# 8. None Type
result = None  # NoneType (represents null / no value)

print(result, type(result))
```
