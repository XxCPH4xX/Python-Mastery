**5. Measuring length.** Apply `len()` to a string, a list, a tuple, a set and a dict. In one comment, describe what `len` counts for each.

```python
print(len("Hello"))
print(len([1, 2, 3]))
print(len((1, 2, 3, 4, 5)))
print(len({1, 2, 3, 4, 5, 6}))
print(len({"a": 1, "b": 2, "c": 3}))
```

**6. Key candidates.** Which of these work as a **dictionary key** and which raise an error? Test each one: `"name"`, `42`, `3.14`, `(1, 2)`, `[1, 2]`.

```python
test_dict = {}
test_dict["name"] = 1
test_dict["42"] = 2
test_dict["3.14"] = 3
test_dict[(1, 2)] = 4

try:
    test_dict[[1, 2]] = 5
except TypeError as err:
    print("list key failed:", err)
```

## Part 3 — Challenge

**8. Operation types.** Predict the resulting **value AND type** of each (write guesses as comments), then verify:

- `5 + 5.0`
- `"5" * 3`
- `True + True`
- `10 / 2`
- `[1] + [2]`

```python
print(5 + 5.0)  # 10.0
print("5" * 3)  # "555"
print(True + True)  # 2  bool act like int
print(10 / 2)  # 5
print([1] + [2])  # [1,2]
```
