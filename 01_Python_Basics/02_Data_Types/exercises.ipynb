# Data_Types Exercises

Write your answer in the empty code cell under each question. Check yourself against `solutions.ipynb` when done.

**Part 1** = warm-up · **Part 2** = practice · **Part 3** = challenge.

## Part 1 — Warm-up

**1. Identify them.** Print the type of each value: `42`, `3.14`, `"hello"`, `True`, `None`.

**2. Predict first.** Before running the code, write your guess for each type as a comment. Then verify: `[1, 2]`, `(1, 2)`, `{1, 2}`, `{"a": 1}`.

**3. One of each.** Create five variables holding: some text, a whole number, a decimal, a boolean, and "no value".

## Part 2 — Practice

**4. Sneaky booleans.** Check `isinstance(True, int)` and run `True + True`. Is `True` actually an integer in Python? What values does it take?

**5. Measuring length.** Apply `len()` to a string, a list, a tuple, a set and a dict. In one comment, describe what `len` counts for each.

**6. Key candidates.** Which of these work as a **dictionary key** and which raise an error? Test each one: `"name"`, `42`, `3.14`, `(1, 2)`, `[1, 2]`.

**7. Text vs bytes.** Create `s = "hi"` and `b = b"hi"`. Print the type and length of each — how do they differ?

## Part 3 — Challenge

**8. Operation types.** Predict the resulting **value AND type** of each (write guesses as comments), then verify:

- `5 + 5.0`
- `"5" * 3`
- `True + True`
- `10 / 2`
- `[1] + [2]`

**9. Profile card.** Build a dict called `profile` whose values cover different types (e.g. name, age, height, is_student, nickname=None). Then print each key together with `type()` of its value.
