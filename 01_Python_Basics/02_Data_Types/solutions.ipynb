## Part 1 — Warm-up

**1. Identify them.** Five core types straight from `type()`.

```
print(type(42))       # <class 'int'>
print(type(3.14))     # <class 'float'>
print(type("hello"))  # <class 'str'>
print(type(True))     # <class 'bool'>
print(type(None))     # <class 'NoneType'>
```

**2. Predict first.** Brackets tell the story: square = list, round = tuple, curly-only = set, curly-with-colon = dict.

```
print(type([1, 2]))    # list
print(type((1, 2)))    # tuple
print(type({1, 2}))    # set
print(type({"a": 1}))  # dict
```

## Part 2 — Practice

**4. Sneaky booleans.** `bool` is a *subclass* of `int` — `True` behaves like `1`, `False` like `0`.

```
print(isinstance(True, int))   # True!
print(isinstance(5.0, float))  # True
print(True + True)             # 2 — booleans ARE integers under the hood
```

**5. Measuring length.** Characters, items, items, unique items, keys.

```
print(len("hello"))         # 5    -> characters
print(len([1, 2, 3]))       # 3    -> elements
print(len((1, 2, 3, 4)))    # 4    -> elements
print(len({1, 1, 2, 3}))    # 3    -> UNIQUE elements
print(len({"a": 1, "b": 2}))# 2    -> keys
```

**6. Key candidates.** Only `[1, 2]` fails — lists are *mutable*, hence unhashable. Tuples work because they're immutable.

```
test_dict = {}
test_dict["name"] = 1   # ok
test_dict[42] = 2       # ok
test_dict[3.14] = 3     # ok
test_dict[(1, 2)] = 4   # ok — tuples are immutable/hashable

try:
    test_dict[[1, 2]] = 5          # TypeError!
except TypeError as err:
    print("list key failed:", err)
```

**7. Text vs bytes.** Same length (both hold two characters here), but `bytes` stores raw numbers 0–255 while `str` stores Unicode text.

```
s = "hi"
b = b"hi"

print(type(s), len(s))   # <class 'str'> 2
print(type(b), len(b))   # <class 'bytes'> 2
print(b)                 # b'hi' — note the b prefix
```

## Part 3 — Challenge

**8. Operation types.** Mixing types follows promotion rules: int+float → float, `/` always gives float, `bool` acts as int.

```
print(5 + 5.0)    # 10.0  -> float (int + float promotes)
print("5" * 3)    # 555   -> str repetition, NOT math
print(True + True)# 2     -> bool acts as int
print(10 / 2)     # 5.0   -> division ALWAYS returns float
print([1] + [2])  # [1, 2]-> list concatenation
```

**9. Profile card.** A dict whose values cover several types, inspected key by key.

```
profile = {
    "name": "Fahim",      # str
    "age": 21,            # int
    "height": 5.9,        # float
    "is_student": True,   # bool
    "nickname": None,     # NoneType
}

print(profile["name"], type(profile["name"]))
print(profile["age"], type(profile["age"]))
print(profile["height"], type(profile["height"]))
print(profile["is_student"], type(profile["is_student"]))
print(profile["nickname"], type(profile["nickname"]))
```
