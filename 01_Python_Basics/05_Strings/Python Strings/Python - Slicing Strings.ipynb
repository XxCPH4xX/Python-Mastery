# Slicing

You can return a range of characters by using the slice syntax.

Specify the start index and the end index, separated by a colon, to return a part of the string.

Get the characters from position 2 to position 5

```python
b = "Hello, World!"
print(
    b[2:5]
    # We use [ ] because it is a list and b is variable and we using [2:5] to get the characters from position of index 2 to position 5.
)
```

Slice From the Start

```python
b = "Hello, World!"
print(
    b[:5]
)  # By leaving out the start index, the range will start at the first character
```

Slice To the End

```python
b = "Hello, World!"
print(
    b[2:]
)  # By leaving out the stop index, the range will extend to the end of the string
```

Negative Indexing

```python
b = "Hello, World!"
print(b[-5:-2])  # Use negative indexes to start the slice from the end of the string
```
