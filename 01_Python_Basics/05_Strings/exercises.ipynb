# Strings Exercises

Write your answer in the empty code cell under each question. Check yourself against `solutions.ipynb` when done.

**Part 1** = warm-up · **Part 2** = practice · **Part 3** = challenge.

## Part 1 — Warm-up

**1. Character access.** For `word = "Python"`: print the first letter, the last letter using a **negative index**, and the length of the word.

**2. Slice it.** From `word = "Python"` get: `"Pyt"`, `"hon"`, every second character, and the whole string reversed.

**3. Change case.** Put `"hello world"` through `.upper()`, `.title()` and `.capitalize()` — print all three results.

## Part 2 — Practice

**4. Clean up.** Given `messy = "  Hello, World!  "`: strip the surrounding spaces, remove the comma with `.replace()`, and count how many times the letter `l` appears.

**5. Split & join.** Turn `csv_line = "red,green,blue"` into a list using `.split(",")`, then join the pieces back together with `" | "` as separator.

**6. F-string formatting.** With `product = "coffee"` and `price = 4.5`, print exactly:

```
Coffee costs $4.50 today.
```

Hint: an f-string plus `{price:.2f}`.

**7. Inside or out?** Check whether `"yth"` is inside `"Python"` using the `in` keyword, and find its starting position with `.find()`.

## Part 3 — Challenge

**8. Palindrome check.** For `word = "madam"`, write **one expression** that is `True` when the word reads the same backwards (combine `.lower()` with slice reversal). Test it on `"hello"` too.

**9. Username factory.** Start from `full_name = "  Ada Lovelace  "` and produce the username `ada.lovelace` using only string methods (strip, lower, split, join).
