## Part 1 — Warm-up

**1. Character access.** Index `0` is the first character; `-1` counts from the end.

```
word = "Python"

print(word[0])    # P
print(word[-1])   # n
print(len(word))  # 6
```

**2. Slice it.** Slices are `[start:stop:step]`, and `stop` is excluded.

```
word = "Python"

print(word[:3])    # Pyt    -> first three characters
print(word[-3:])   # hon    -> last three characters
print(word[::2])   # Pto    -> every second character
print(word[::-1])  # nohtyP -> negative step reverses
```

**3. Change case.** Note: none of these modify the original — they return a new string.

```
text = "hello world"

print(text.upper())       # HELLO WORLD
print(text.title())       # Hello World
print(text.capitalize())  # Hello world — only the first letter
```

## Part 2 — Practice

**4. Clean up.** Chain the methods; `count()` tallies non-overlapping occurrences.

```
messy = "  Hello, World!  "

cleaned = messy.strip().replace(",", "")
print(repr(cleaned))          # 'Hello World!'
print(cleaned.count("l"))     # 3
```

**5. Split & join.** `split` turns text into a list; `join` glues a list back into text.

```
csv_line = "red,green,blue"

parts = csv_line.split(",")
print(parts)                        # ['red', 'green', 'blue']

print(" | ".join(parts))            # red | green | blue
```

**6. F-string formatting.** `{price:.2f}` forces exactly two decimal places.

```
product = "coffee"
price = 4.5

print(f"{product.capitalize()} costs ${price:.2f} today.")
# Coffee costs $4.50 today.
```

**7. Inside or out?** `in` answers yes/no; `.find()` returns the index (or `-1` when missing).

```
print("yth" in "Python")     # True
print("Python".find("yth"))  # 2 -> starts at index 2
```

## Part 3 — Challenge

**8. Palindrome check.** Compare the word with its own reversal.

```
word = "madam"

is_palindrome = word.lower() == word.lower()[::-1]
print(is_palindrome)      # True

word = "hello"
print(word.lower() == word.lower()[::-1])   # False
```

**9. Username factory.** Clean → lowercase → split into pieces → join with dots.

```
full_name = "  Ada Lovelace  "

username = ".".join(full_name.strip().lower().split())
print(username)   # ada.lovelace
```
