## 1. What Is a String?

A **string** (`str`) is an ordered sequence of Unicode characters — letters, digits, spaces, punctuation, emojis, and more. It is Python's way of representing text. Strings are created by wrapping characters in quotes.

### 1.1 Single and Double Quotes

Python accepts both single quotes (`'...'`) and double quotes (`"..."`) to create a string. There is no functional difference — choose whichever makes your string more readable. Use double quotes when the text contains apostrophes, and single quotes when it contains double quotes.

````python
# Single and double quotes are interchangeable
poem_line = "Hello, World!"
prose_line = 'Hello, World!'

print(poem_line)
print(prose_line)

# Use the quote style that avoids escaping
with_apostrophe = "Python's strings are flexible"
with_quote = 'She said, "Python is amazing!"'

print(with_apostrophe)
print(with_quote)

# Both styles create the exact same str type
print(type(poem_line))
print(type(prose_line))
````

````python
# Strings are case-sensitive: "python" and "Python" are different
print("python" == "Python")    # False — different strings
print("python" == "python")    # True — same characters

# Use a consistent quote style in one project — most teams use double quotes
language = "Python"
version = "3.12"
print(language, version)
````

### 1.2 Triple-Quoted Strings

Triple-quoted strings (`'''...'''` or `"""..."""`) span multiple lines and preserve whitespace. They are ideal for multiline text, docstrings (explained later), and any text where line breaks matter.

````python
# Triple-quoted strings preserve newlines and indentation
haiku = """Line one
Line two
Line three"""

# Triple-quoted strings with single quotes too
sonnet = '''   indented
another line'''

print(haiku)
print(sonnet)

# Single-line triple-quoted strings are also valid
single = """Just one line"""
print(single)
````

````python
# Triple-quoted strings are commonly used for docstrings
def greet(name):
    """Return a friendly greeting for the given name."""
    return f"Hello, {name}!"

# The docstring is the string right after the def — accessible via help()
import inspect
print(inspect.getdoc(greet))
````

### 1.3 The `len()` Function with Strings

`len()` returns the number of characters in a string, including spaces and punctuation. Empty strings have length 0.

````python
word = "Hello"
sentence = "Hello, World!"
spacey = " "
empty = ""

print(len(word))       # 5 characters
print(len(sentence))   # 13 characters (including the comma and exclamation)
print(len(spacey))     # 1 — a single space is a valid character
print(len(empty))      # 0 — empty string
````

````python
# len() works on any string-like sequence
print(len("a"))          # 1
print(len("😀😁😂"))   # 3 — Unicode emojis count as single characters
print(len("café"))       # 4 — accented characters are single chars in Python 3
print(len("a\nb"))      # 3 — the \n is ONE character (the newline)
````

## 2. Strings Are Sequences — Indexing

In Python, a string is a **sequence** — an ordered collection where each element has a fixed position. You access individual characters using **square bracket notation** with a zero-based index.

| Index | 0 | 1 | 2 | 3 | 4 | 5 | ... | -3 | -2 | -1 |
|---|---|---|---|---|---|---|---|---|---|---|
| Character | H | e | l | l | o | , | ... | l | d | ! |

- **Positive indices** count from the start: index 0 is the first character.
- **Negative indices** count backwards from the end: index -1 is the last character.

````python
word = "Hello,"

print(word[0])   # First character: H
print(word[1])   # Second character: e
print(word[4])   # Fifth character: o
print(word[-1])  # Last character: ,
print(word[-2])  # Second-to-last: o
````

````python
# Negative indexing is especially useful when you want the last element
# without knowing the string's length
language = "Python"

print(language[0])    # First: P
print(language[-1])   # Last: n
print(language[-6])   # Same as [0]: P
print(language[-len(language)])  # First character, computed dynamically
````

````python
# Indexing returns a NEW string of length 1 — a character IS a string
word = "Python"
first_char = word[0]
print(type(first_char))  # <class 'str'>
print(first_char)        # 'P'

# A character is just a 1-character string — no separate 'char' type
print(len(word[0]))      # 1
print(word[0] + word[1]) # 'Py' — two characters concatenated
````

````python
# Accessing an index that doesn't exist raises an IndexError
word = "Python"

try:
    print(word[10])   # IndexError: string index out of range
except IndexError as e:
    print(e)

try:
    print(word[-10])  # Also IndexError for negative out of bounds
except IndexError as e:
    print(e)
````

## 3. Slicing Deep Dive

**Slicing** extracts a portion of a string using `[start:stop:step]`. The result is always a **new string** — the original is unchanged.

```
string[start:stop:step]
   │      │     │
   │      │     └── Step size (default 1). Use -1 to reverse.
   │      └──────── Stop index (excluded!)
   └───────────────── Start index (default 0)
```

| Parameter | Default | Behaviour |
|---|---|---|
| `start` | `0` | Index to begin slicing from |
| `stop` | len(string) | Index to stop BEFORE |
| `step` | `1` | Every Nth character; `-1` reverses |

````python
# Basic slicing: [start:stop] — stops BEFORE the stop index
word = "Python"

print(word[0:3])   # Characters at index 0, 1, 2 -> "Pyt"
print(word[3:6])   # Characters at index 3, 4, 5 -> "hon"
print(word[2:])    # From index 2 to end -> "thon"
````

````python
# Omitting start/stop uses sensible defaults
word = "Python"

print(word[:3])    # From start up to (but not including) index 3 -> "Pyt"
print(word[3:])    # From index 3 to the end -> "hon"
print(word[:])     # Entire string (a common way to copy a string) -> "Python"
print(word[:-1])   # All but the last character -> "Pytho"
````

````python
# Step of 2: take every 2nd character
word = "Python"

print(word[::2])   # P, t, o -> "Pto" (every 2nd from start)
print(word[1::2])  # y, h, n -> "yhn" (every 2nd from index 1)
print(word[4:1:-1])  # Reverse slice: index 4 down to (not including) 1
````

````python
# Negative step reverses the string
word = "Python"

print(word[::-1])   # Reversed: "nohtyP"
print(word[::-1][0])  # First character of reversed -> "n"
print(word[::-1][:3])  # First 3 of reversed -> "noh"

# Practical: get the last N characters
print(word[-3:])    # Last 3 characters: "hon"
print(word[-1])     # Just the last character: "n"
````

````python
# Slicing handles out-of-bounds gracefully (unlike indexing)
word = "Python"

print(word[:10])   # "Python" — no error, just returns what exists
print(word[3:100]) # "hon" — stops at end

# Classic slicing patterns
filename = "report.csv"
print(filename[-4:])   # ".csv" — file extension
print(filename[:-4])   # "report" — filename without extension
````

````python
# Practical slicing patterns you'll use all the time
message = "    Hello, World!    "

# Strip leading and trailing whitespace manually (use .strip() in practice)
clean = message[4:-4]
print(repr(message))   # Original unchanged
print(repr(clean))     # "Hello, World!"

# Extract first word using .find() and slicing
text = "Python is great"
space = text.find(" ")
first_word = text[:space]
print(first_word)  # "Python"
````

## 4. String Methods — Comprehensive Reference

Strings come with a rich set of built-in **methods** — functions that operate on the string object. Python's `str` type has over 40 methods. Here are the ones you will use every day, organised by purpose.

### 4.1 Case Methods

These methods return a **new string** with the case transformed. The original string is never modified.

````python
# .upper() — ALL CAPS
text = "python"
print(text.upper())

# .lower() — all lowercase
text = "PYTHON"
print(text.lower())

# .title() — First letter of each word capitalized
text = "hello world"
print(text.title())

# .capitalize() — First letter of the string capitalized, rest lower
text = "hello WORLD"
print(text.capitalize())

# .swapcase() — Swap uppercase <-> lowercase
text = "PyThOn"
print(text.swapcase())
````

````python
# Chaining methods
text = "  python IS amazing  "
print(text.strip().upper())   # Chain strip() then upper()

# Common pattern: normalize user input for comparison
user_input = "  PyThOn  "
print(user_input.strip().lower() == "python")  # True
````

````python
# Common real-world pattern: normalize names from a database
names = ["alice", "ALICE", "Alice", "aLiCe"]
normalized = [n.strip().lower().capitalize() for n in names]
print(normalized)

# Or with .title() for multi-word names
full_names = ["alice wonderland", "BOB BUILDER", "charlie brown"]
clean_names = [n.strip().title() for n in full_names]
for name in clean_names:
    print(name)
````

### 4.2 Whitespace & Cleaning

These methods handle spaces, tabs, and newlines at the edges of strings — essential when processing user input or reading files.

````python
# .strip() — remove whitespace from BOTH ends
text = "  hello  "
print(repr(text))         # repr() shows quotes + whitespace explicitly
print(repr(text.lstrip()))  # left strip
print(repr(text.rstrip()))  # right strip
print(repr(text.strip()))   # both sides
````

````python
# strip() removes any combination of the given characters
# No argument: removes ALL whitespace characters (space, tab, newline)
text = "\n\t  hello world  \n"
clean = text.strip()
print(repr(clean))

# Stripping specific characters
path = "/home/user/"
print(repr(path.strip("/")))   # Remove leading/trailing slashes

# Strip multiple characters
tagged = "---hello---"
print(repr(tagged.strip("-")))  # Remove leading/trailing dashes
````

````python
# .split() — split a string into a list of substrings
# Default separator is ANY whitespace (space, tab, newline)
sentence = "apple banana cherry"
print(sentence.split())   # Default: split on whitespace

# Split on a specific separator
csv = "apple,banana,cherry"
print(csv.split(","))     # Split on comma

# Split only N times (maxsplit)
data = "name,email,phone,address"
print(data.split(",", 2))  # Split on first 2 commas only
````

````python
# Splitting on newlines and other whitespace
multiline = "a\nb\nc\nd\ne"
lines = multiline.split("\n")  # Split on newline specifically
print(lines)

# Split multi-line text into lines
text = "line1\nline2\nline3\nline4"
print(text.splitlines())  # ["line1", "line2", "line3", "line4"]
````

````python
# str.join() — the opposite of split(): join a list into a string
# separator.join(list)
fruits = ["apple", "banana", "cherry"]
print(" | ".join(fruits))   # Join with pipe separator

# Join numbers (must convert to strings first)
numbers = [1, 2, 3, 4, 5]
print(", ".join(str(n) for n in numbers))

# Build a path safely
parts = ["2024", "08", "26"]
print("/".join(parts))
````

````python
# split() and join() together: clean and transform
# Remove extra spaces from a sentence
messy = "  apple   banana   cherry  "
clean = " ".join(messy.split())  # split on whitespace, join with single space
print(clean)

# Change the separator
joined = "apple and banana and cherry"
swapped = joined.replace(" and ", ", ")
print(swapped)

# Split into characters
letters = "ABCDE"
print(" ".join(letters))
````

````python
# Real-world: parsing comma-separated user input
user_input = "alice, bob , charlie"
names = [n.strip() for n in user_input.split(",")]
print(names)

# Joining back with a different separator
print(", ".join(names))
````

### 4.3 Search & Replace

These methods let you find substrings within strings, count occurrences, and replace text.

````python
# .find() — returns the STARTING index of the first occurrence
# Returns -1 if NOT found (no exception!)
text = "Hello, World!"

print(text.find("World"))    # 7 — "World" starts at index 7
print(text.find("Python"))   # -1 — not found

# .find() accepts optional start and stop
print(text.find("o", 8))    # Search from index 8 onwards: 8
print(text.find("o", 0, 8)) # Search in [0:8] range: 4
````

````python
# .index() — same as find() but raises ValueError if not found
text = "Hello, World!"

print(text.index("World"))   # 7 — same as find() when found
try:
    print(text.index("Python"))  # ValueError — unlike find(), this crashes!
except ValueError as e:
    print(e)
````

````python
# .replace() — replace ALL occurrences of a substring
# Returns a NEW string; original is unchanged
text = "Hello, World!"

print(text.replace("World", "Python"))

# Replace with count parameter (only first N occurrences)
text2 = "one one one"
print(text2.replace("one", "1", 1))   # Only first occurrence

# Chain operations
msg = "Hello, World! Welcome to World!"
print(msg.replace("World", "Python"))
````

````python
# .count() — count non-overlapping occurrences
text = "the cat sat on the mat"

print(text.count("at"))    # 3 — "cat", "sat", "mat"
print(text.count("the"))   # 2
print(text.count("z"))     # 0 — not found, returns 0 (not -1)
````

````python
# .startswith() and .endswith() — check prefix/suffix
# Returns True or False
filename = "report.csv"

print(filename.startswith("report"))   # True
print(filename.startswith("Report"))  # False — case-sensitive!

print(filename.endswith(".csv"))       # True
print(filename.endswith(".txt"))       # False

# Both accept a tuple of options
print(filename.endswith((".csv", ".xlsx", ".json")))  # True

url = "https://api.example.com/data"
print(url.startswith(("http://", "https://")))  # True
````

### 4.4 Boolean Check Methods

These methods return `True` or `False` — perfect for validation and conditional logic. They do **not** check the truthiness of the string itself.

````python
# .isalpha() — True if ALL characters are letters (A-Z, a-z)
print("abc".isalpha())    # True
print("abc123".isalpha()) # False — contains digits

# .isdigit() — True if ALL characters are digits (0-9)
print("123".isdigit())    # True
print("123.45".isdigit())  # False — contains period
print("42".isdigit())      # True

# .isalnum() — True if ALL characters are letters OR digits
print("abc123".isalnum())  # True
print("abc 123".isalnum()) # False — contains space
````

````python
# .isspace() — True if ALL characters are whitespace
# Whitespace includes: space, tab, newline, carriage return, etc.
print(" ".isspace())        # True
print("\t\n".isspace())     # True — tab and newline count
print("   a".isspace())     # False — contains letter

# .isupper() / .islower() — all uppercase or all lowercase
print("HELLO".isupper())     # True
print("Hello".isupper())    # False — mixed case
print("hello".islower())    # True
print("".islower())         # False — empty string is neither
````

````python
# Input validation is the most common use for boolean methods
age_str = "42"
if age_str.isdigit():
    age = int(age_str)
    print(f"Valid number: {age}")
else:
    print("Not a valid number")

age_str = "42 years"  # mixed input
if age_str.isdigit():
    print(f"Valid: {age_str}")
else:
    print("Not a valid number")
````

### 4.5 Padding & Alignment

These methods add whitespace or fill characters to align text — useful for formatted output, tables, and logging.

````python
# .ljust(width) — left-justify, pad on the right
# .rjust(width) — right-justify, pad on the left
# .center(width) — center, pad on both sides

print("hello".ljust(10))     # "hello     " — padded to width 10
print("hello".rjust(10))     # "     hello"
print("hello".center(10))    # "  hello   "  — equal pad on both sides

# With custom fill character
print("42".rjust(5, "0"))     # "00042"
print("hello".ljust(10, "-"))  # "hello-----"
````

````python
# Building a simple table with alignment
rows = [
    ("Alice", 25, 98.5),
    ("Bob", 30, 87.3),
    ("Charlie", 22, 92.1),
]

header = f"{'Name':<10} | {'Age':>4} | {'Score':>7}"
print(header)
print("-" * len(header))

for name, age, score in rows:
    print(f"{name:<10} | {age:>4} | {score:>7.1f}")
````

## 5. f-Strings — Modern String Formatting

**f-strings** (formatted string literals) were introduced in Python 3.6 and quickly became the preferred way to embed expressions inside strings. An f-string is a string with an `f` or `F` prefix; any expression inside curly braces `{}` is evaluated and inserted into the string at that position.

```python
f"Hello, {name}! You are {age} years old."
```

f-strings are:
- **Clearer** than `%`-formatting and `.format()`
- **Faster** — evaluated at runtime, no intermediate formatting objects
- **More flexible** — expressions work directly inside `{}`

````python
# Basic f-string usage
name = "Alice"
age = 30

print(f"Hello, {name}! You are {age} years old.")

# Expressions inside f-strings
x, y = 3, 4
print(f"The result is {x + y}")
print(f"{x} + {y} = {x + y}")
````

````python
# Format specifiers inside f-strings
price = 199.99
population = 1411778724
pi = 3.14159265
percentage = 0.123

print(f"Price: ${price:.2f}")        # 2 decimal places: 199.99

print(f"Population: {population:,}") # Thousands separator: 1,411,778,724

# Format as currency (manual)
amount = 1234.567
print(f"${amount:,.2f}")

# Format as percentage
print(f"{percentage:.1%}")   # 12.3%
````

````python
# More numeric format specifiers
import math

print(f"{math.pi}")        # Full precision: 3.141592653589793
print(f"{math.pi:.2f}")    # 2 decimals: 3.14
print(f"{math.pi:.4f}")    # 4 decimals: 3.1416
print(f"{math.pi:.2e}")    # Scientific: 3.14e+00
print(f"{42:05d}")          # Zero-padded int: 00042
print(f"{42:d}")           # Plain int: 42
````

````python
# Alignment in f-strings: :< (left), :> (right), :^ (center)
# Width specifier pads to a minimum width

print(f"{42:>5}")     # Right-align in 5 chars: "   42"
print(f"{42:<5}")     # Left-align in 5 chars: "42   "
print(f"{42:^5}")     # Center in 5 chars: " 42  "

# Fill character with alignment
print(f"{42:0>5}")    # Pad with zeros: "00042"
print(f"{42:=^10}")   # Fill with =: "====42===="
````

````python
# Combining alignment, width, and numeric formatting
# Invoice table example

items = [
    ("Apples", 10, 12.50),
    ("Bread", 2, 5.99),
    ("Milk", 1, 3.49),
]

print(f"{'Item':<10} | {'Qty':>5} | {'Price':>7}")
print("-" * 34)

total = 0
for name, qty, price in items:
    print(f"{name:<10} | {qty:>5} | ${price:>6.2f}")
    total += qty * price

print("-" * 34)
print(f"{'Total':<10} | {'':>5} | ${total:>6.2f}")
````

````python
# f-strings with function calls and method chaining
name = "Alice"
words = "hello world"

print(f"{name.lower()} is 30 years old")  # Method inside f-string
print(f"Hello, {name.strip().title()}!")   # Chain inside f-string

# f-strings with !r (repr) and !s (str) conversions
word = "Python"
print(f"Python has {len(word)} characters")
````

````python
# f-strings with method calls and complex expressions
x, y = 3, 4
print(f"The sum is {x + y}")
print(f"3 raised to 4 is {3 ** 4}")

# Conditional expressions in f-strings
numbers = [1, 5, 9, 2, 7]
print(f"Max: {max(numbers)}, Min: {min(numbers)}")
````

### 5.1 Why f-strings over `.format()`?

Before f-strings, `.format()` was the modern choice. Compare the two:

```python
name = "Alice"
# .format() style:
"Hello, {}! You are {} years old.".format(name, 30)
"Hello, {0}! You are {1} years old.".format(name, 30)    # positional
"Hello, {n}! You are {a} years old.".format(n=name, a=30)  # named
# f-string style:
f"Hello, {name}! You are {30} years old."
```

f-strings are shorter, read more naturally, and are now the standard. You will still see `.format()` in older codebases.

````python
# .format() style — still works, but f-strings are preferred
name = "Alice"

print("Hello, {}! You are {} years old.".format(name, 30))
print("Hello, {0}! You are {1} years old.".format(name, 30))   # positional
print("Hello, {n}! You are {a} years old.".format(n=name, a=30))  # named
````

````python
# Debugging trick: f-strings support = for self-documenting output
name = "Alice"
age = 30
print(f"{name=}, {age=}")  # Python 3.8+: name='Alice', age=30

# Useful for quick debugging
x = 42
print(f"{x=}")  # x=42
````

````python
# Multiline f-strings (Python 3.12+ improved this, earlier versions need parentheses)
name = "Alice"
age = 30

message = (
    f"Name: {name}\n"
    f"Age: {age}"
)
print(message)

# Triple-quoted f-strings
bio = f"""
Name: {name}
Age: {age}
"""
print(bio.strip())
````

## 6. String Concatenation & Repetition

Strings can be combined (`+`) and repeated (`*`) to build new strings. These are the most basic string operations.

````python
# Concatenation: + joins two strings
first = "Hello"
second = "World"

print(first + second)        # "HelloWorld"
print(first + ", " + second + "!")  # "Hello, World!"

# Build a full name
first_name = "Three"
last_name = "Musketeers"
print(first_name + " " + last_name)
````

````python
# Repetition: * repeats a string N times
print("ha-" * 3)       # "ha-ha-ha-"
print("=" * 12)         # "============"
print("wow" * 3)        # "wowwowwow"

# Practical: decorative separators
title = " Python Strings "
print("=" * 50)
print(f"{title:^50}")
print("=" * 50)
````

````python
# f-string interpolation: the cleanest way to mix variables and text
greeting = "Hello"
target = "World"

print(f"{greeting} {target}")    # f-string: most readable

# Comparison: all three approaches
name, age = "Alice", 30

print(name + ", Age: " + str(age))           # Concatenation — messy
print("{}, Age: {}".format(name, age))        # .format() — verbose
print(f"Name: {name}, Age: {age}")            # f-string — cleanest
````

````python
# Gotcha: concatenation only works with strings — not with other types
try:
    print("Age: " + 30)  # TypeError
except TypeError as e:
    print(e)

# Convert first, or use f-string
print("Age: " + str(30))   # "Age: 30"
print(f"Age: {30}")       # f-string handles conversion automatically
````

## 7. Escape Sequences

Some characters have special meaning inside a string literal. To include them literally, Python uses **escape sequences** — a backslash (`\`) followed by a character.

| Sequence | Meaning | Example |
|---|---|---|
| `\\` | Backslash | `"\\"` → `\` |
| `\'` | Single quote | `'\'A\'B'` → `'A'B` |
| `\"` | Double quote | `"A\"B"` → `A"B` |
| `\n` | Newline | `"A\nB"` → A on one line, B on next |
| `\t` | Tab | `"A\tB"` → A, then tab, then B |
| `\r` | Carriage return | Moves cursor to start of line |
| `\b` | Backspace | Deletes previous character |

````python
# Newline: \n — move to a new line
print("Line 1\nLine 2")

# Tab: \t — insert a tab character
print("Column A\tColumn B")

# Backslash: \\ — include a literal backslash
print("Backslash: \\")

# Quotes — use the opposite quote type to avoid escaping
print('Quote: \'')      # Single quote inside single-quoted string
print("Double quote: \"")  # Double quote inside double-quoted string
````

````python
# Including quotes in strings
quote = 'She said "Python is fun"'
print(quote)

# Windows file paths need double backslashes (or raw strings — see below)
windows_path = "C:\\Users\\Fahim"
print("Path:", windows_path)

# \n vs actual newlines in source code
with_newline = "First line\nSecond line"
print(repr(with_newline))   # Shows the \n escape
print(with_newline)          # Shows the actual newlines
````

````python
# Less common escapes: \r (carriage return), \b (backspace)
print("Line 1\rLine 2")   # \r returns cursor to start, overwrites
print("Hello\bWorld")      # \b moves back one char, so 'o' is overwritten

# \x and \u for hex/unicode escapes
print("\x41")    # Hex escape: \x41 = 'A'
print("\u0041")   # Unicode escape: \u0041 = 'A'
print("\U0001F600")  # 32-bit Unicode: grinning face emoji
````

### 7.1 Raw Strings

A **raw string** is prefixed with `r` and treats backslashes as literal characters — no escape sequences are interpreted. Raw strings are essential for **file paths on Windows** and for **regular expressions** (covered later).

````python
# Raw string: r"..." — backslashes are literal
windows_path = r"C:\Users\Fahim\Documents"
print(windows_path)    # Exactly: C:\Users\Fahim\Documents

# Compare with a regular string
regular_path = "C:\\Users\\Fahim\\Documents"
print(regular_path)    # Same output, but requires double backslashes

# Raw strings with \n
raw = r"Line 1\nLine 2"   # The \n is LITERAL: two chars: backslash + n
print(raw)

# Regular string with \n
regular = "Line 1\nLine 2"  # The \n creates an actual newline
print(regular)
````

````python
# Real-world: regex patterns need raw strings
import re

# Match a Windows path — raw string keeps the backslashes literal
path_pattern = r"C:\Users\Fahim"
match = re.search(path_pattern, r"C:\Users\Fahim\file.txt")
print("Match found:", match is not None)

# Without raw string, you'd need to double each backslash
path_pattern2 = "C:\\\\Users\\\\Fahim"  # Confusing!
print(repr(path_pattern))
print(repr(path_pattern2))
````

## 8. String Immutability

**Strings in Python are immutable** — once created, a string can never be changed. Every string method that seems to "modify" the string actually returns a **new string**. The original string is always preserved.

This is a fundamental design decision that:
- Makes strings safe to share (no accidental modifications)
- Allows Python to intern/cache short strings for performance
- Requires you to reassign rather than modify in place

````python
# Trying to change a character raises TypeError
word = "hello"
try:
    word[0] = "H"   # TypeError: 'str' object does not support item assignment
except TypeError as e:
    print(e)
````

````python
# The correct approach: create a NEW string and reassign
word = "hello"
print(word)           # Original unchanged

word = word.capitalize()  # Creates a NEW string
print(word)           # "Hello" — new value

# Confirm the original was not mutated
original = "hello"
modified = original.upper()

print(original)   # Still "hello"
print(modified)   # "HELLO"
````

````python
# Immutability means operations always return NEW strings
# Multiple references can share the same string safely

a = "hello"
b = a           # b points to the SAME string object
c = a.upper()   # c points to a NEW string object

print(a)        # "hello" — a is unchanged
print(c)        # "HELLO" — c is the new string
# Both a and b still point to "hello" — safe sharing!
````

````python
# Demonstration: NO string method modifies in place
s = "hello"
s.upper()  # Returns new string, but we didn't capture it
print(s)   # Still "hello" — original unchanged!

# To actually change s, you must reassign
s = s.upper()
print(s)   # Now "HELLO"

# This is a common beginner trap:
name = "alice"
name.capitalize()   # Result discarded
print(f"Original '{name}' was NOT modified")
````

````python
# Building strings in a loop — a common anti-pattern
# AVOID THIS: string concatenation in a loop is O(n^2)
result = ""
for i in range(5):
    result = result + str(i)  # Creates a new string each iteration

print(result)

# BETTER: use a list and join() — O(n)
parts = []
for i in range(5):
    parts.append(str(i))

print("".join(parts))
````

## 10. String Encoding & Unicode

Every string in Python 3 is stored as a sequence of **Unicode code points** — a universal character set that covers characters from every written language on Earth. Python 3 uses UTF-8 by default internally, so strings can contain any character: Latin letters, Chinese characters, emoji, mathematical symbols, and more.

````python
# Unicode in action — Python 3 handles all languages
chinese = "你好"       # Chinese: hello
arabic = "مرحبا"     # Arabic: hello
emoji = "😀"          # Emoji

print(chinese)
print(len(chinese))  # 2 — length counts characters
print(emoji)
print(len(emoji))    # 1
````

````python
# UTF-8 encoding: characters are stored as bytes
text = "hello"
encoded = text.encode("utf-8")
print(f"Total bytes (UTF-8): {len(encoded)}")  # 5 bytes for "hello"

text2 = "你好"
encoded2 = text2.encode("utf-8")
print(f"Total bytes (UTF-8): {len(encoded2)}")  # 6 bytes for 2 Chinese chars

text3 = "😀"
encoded3 = text3.encode("utf-8")
print(f"Total bytes (UTF-8): {len(encoded3)}")  # 4 bytes for one emoji
````

````python
# Encoding and decoding between str and bytes
# .encode() converts str -> bytes
# .decode() converts bytes -> str

text = "Hello, World!"
encoded = text.encode("utf-8")
print("Encoded:", encoded)

decoded = encoded.decode("utf-8")
print("Decoded:", decoded)

# Different encodings use different byte sizes
text = "你好世界"   # Chinese: Hello World
print(f"UTF-16 bytes: {len(text.encode('utf-16'))}")
print(f"UTF-32 bytes: {len(text.encode('utf-32'))}")
````

## 11. Working with Text: A Practical Pipeline

Here is how the string methods combine in a real-world text preprocessing pipeline — a task you will do constantly in AI engineering before sending text to an LLM.

````python
# A realistic text preprocessing pipeline
raw_input = "  The Quick Brown FOX!  "

# Step 1: Strip whitespace from both ends
cleaned = raw_input.strip()
print(f"Original: {repr(raw_input)}")
print(f"Cleaned: {repr(cleaned)}")

# Step 2: Normalize case
normalized = cleaned.lower()

# Step 3: Split into words
words = normalized.split()
print(f"Words: {words}")
print(f"Word count: {len(words)}")

# Step 4: Search operations
print(f"Has 'fox': {'fox' in normalized}")
print(f"Replaced: {normalized.replace('fox', 'dog')}")

# Step 5: Check prefix/suffix
print(f"Starts with 'the': {normalized.startswith('the')}")
print(f"Ends with '!': {normalized.endswith('!')}")
````

## 15. String Comparison

Strings support the standard comparison operators: `==`, `!=`, `<`, `<=`, `>`, `>=`. String ordering is **lexicographic** (alphabetical, using the underlying Unicode code points).

````python
# Equality and inequality
print("apple" == "apple")    # True
print("apple" != "banana")  # True
print("Apple" == "apple")    # False — case-sensitive

# Lexicographic ordering
print("apple" < "banana")    # True — 'a' < 'b'
print("z" > "a")             # True — 'z' (122) > 'a' (97)
print("Z" < "a")             # True — uppercase letters come first in ASCII
````

````python
# Sorting strings
words = ["banana", "apple", "cherry", "date"]
sorted_words = sorted(words)
print(sorted_words)  # ['apple', 'banana', 'cherry', 'date']

# Sort by length instead
by_length = sorted(words, key=len)
print(by_length)

# Sort case-insensitively
mixed = ["Alpha", "alpha", "Bravo", "bravo"]
case_insensitive = sorted(mixed, key=str.lower)
print(case_insensitive)  # ['Alpha', 'alpha', 'Bravo', 'bravo']

# Reverse sort
reverse_sorted = sorted(words, reverse=True)
print(reverse_sorted)
````

## 17. Advanced String Patterns

A few more patterns that appear in real code — useful for interviews, data processing, and AI text work.

````python
# Check if a string is a palindrome
def is_palindrome(s):
    # Normalize: keep only letters, lowercase
    cleaned = "".join(c.lower() for c in s if c.isalnum())
    return cleaned == cleaned[::-1]

print(is_palindrome("racecar"))                    # True
print(is_palindrome("hello"))                      # False
print(is_palindrome("Madam, I'm Adam"))            # True
print(is_palindrome("A man, a plan, a canal: Panama"))  # True
````

````python
# Reverse words in a sentence
sentence = "Hello World"
reversed_words = " ".join(sentence.split()[::-1])
print(reversed_words)  # "World Hello"

# Capitalize alternating letters
text = "hello world"
alternating = "".join(c.upper() if i % 2 == 0 else c.lower()
                       for i, c in enumerate(text))
print(alternating)  # "HeLlO WoRlD"

# Swap case (an in-place equivalent already exists as .swapcase())
swapped = "".join(c.lower() if c.isupper() else c.upper() for c in text)
print(swapped)  # "HELLO WORLD"
````

````python
# Common interview tasks: string transformations
# 1. Capitalize the first letter of each word without using .title()
def capitalize_words(s):
    words = s.split()
    return " ".join(w[0].upper() + w[1:] for w in words)

print(capitalize_words("hello world"))  # "Hello World"

# 2. Compress consecutive duplicate characters
def compress(s):
    if not s:
        return ""
    result = []
    current = s[0]
    count = 1
    for c in s[1:]:
        if c == current:
            count += 1
        else:
            result.append(current + (str(count) if count > 1 else ""))
            current = c
            count = 1
    result.append(current + (str(count) if count > 1 else ""))
    return "".join(result)

print(compress("aabcccdd"))  # "a2bc3d2"
````

## 18. String Interning (Behind the Scenes)

Python has a small optimization called **interning**: short strings and string-like objects that look like identifiers are cached in memory, so creating two equal short strings may return the **same object**. This is invisible in most code but explains why `is` and `==` behave differently for strings.

````python
# Short strings (and identifier-like strings) are interned automatically
a = "hello"
b = "hello"
print(a is b)    # True — same cached object
print(a == b)    # True — same value

# Longer strings may or may not be interned
c = "a very long string that is unlikely to be interned"
d = "a very long string that is unlikely to be interned"
print(c == d)    # True — same value
print(c is d)    # False — two different objects (usually)

# Never rely on `is` for string comparison — use `==`
print(a is b and c == d)  # True in both cases
````

````python
# sys.intern() lets you explicitly intern a string
# Useful in performance-critical code that compares the same string many times
import sys

a = sys.intern("a long string that we will compare many times")
b = sys.intern("a long string that we will compare many times")
print(a is b)  # True — both are now the same interned object

# Plain "hello" comparison
x = "hello"
y = "hello"
print(x is y)  # True — already interned
````

## ⚠️ Common Mistakes & Gotchas

| Mistake | Problem | Fix |
|---|---|---|
| Strings are 0-indexed | First char is index 0, not 1 | `s[0]` gets the first character |
| Slice `[:stop]` includes stop | Slice stops *before* stop index | `s[:5]` gets indices 0-4, not 0-5 |
| `"hello"[0] = "H"` | TypeError — strings are immutable | Reassign: `s = "H" + s[1:]` |
| `f"{x, y}"` creates a tuple | Creates a tuple inside the string | `f"{x}, {y}"` — no comma inside `{}` |
| `" "` (space) is truthy | `" ".isalpha()` is False, but `" "` is not empty | Use `if s.strip():` to catch whitespace-only |
| `.strip()` removes all whitespace | Removes ALL whitespace from both ends | Not just one space |
| `.find()` returns -1 | Returns -1 for "not found" (not an exception) | Use `if "X" in msg:` for boolean check |
| Case sensitivity | `"Python" != "python"` | Normalize: `.lower()` before comparing |
| Empty `""` vs space `" "` | Both look similar but different | `""` is falsy, `" "` is truthy |
| `"False"` being truthy | Any non-empty string is truthy | Compare: `s == "False"` or `s.lower() == "false"` |

````python
# DEMONSTRATION: The most common string traps

print("TRAP 1: Trying to modify a character in place")
print("  word = 'hello'")
print("  word[0] = 'H'  # TypeError!")
print()
print("Solution: reassign to a new string")
print("  word = 'hello'")
print("  word = 'H' + word[1:]  # 'Hello'")
print()
print("TRAP 2: f-string with comma inside braces creates a tuple")
x, y = 1, 2
print(f"  f'{{x, y}}'  # '(1, 2)' — not what you want!")
print()
print("Solution: separate the variables in the braces")
print(f"  f'{{x}} and {{y}}'  # '{x} and {y}'")
print()
print("TRAP 3: find() returns -1, not None or raises")
print(f"  'hello'.find('z')  # { -1}")
print("  if 'hello'.find('z'): print('found')  # WRONG: -1 is truthy!")
print()
print("Solution: use 'in' for boolean checks")
print(f"  if 'z' in 'hello': print('found')  # {'z' in 'hello'}")
print()
print("TRAP 4: Strings are case-sensitive")
print(f"  'Python' == 'python'  # {'Python' == 'python'}")
print()
print("Solution: normalize case before comparing")
print(f"  'Python'.lower() == 'python'.lower()  # {'Python'.lower() == 'python'.lower()}")
````

## 💡 Best Practices & Pro Tips

- **Use f-strings for all formatting** — they are the clearest, most readable, and fastest option in Python 3.6+.
- **Normalize before comparing** — always `.strip().lower()` user input before comparing to known values.
- **Use `.split()` and `.join()`** instead of string concatenation in loops. Joining is O(n); concatenation is O(n^2).
- **Use `.isdigit()` / `.isalpha()` / `.isalnum()`** for input validation before converting strings to numbers.
- **Use raw strings `r"..."`** for file paths on Windows and regex patterns to avoid backslash hell.
- **Strings are immutable** — every method returns a new string. Never write `word[0] = "H"`; instead do `word = "H" + word[1:]`.
- **`len()` counts characters, not bytes** in Python 3. Use `len(s.encode("utf-8"))` if you need byte count.
- **Test substring membership with `in`** (`"cat" in text`) rather than `find() != -1` when you only need a boolean.
- **🤖 AI-engineering relevance:** Every LLM prompt and response is a string. Text preprocessing — `.strip()`, `.lower()`, `.split()`, `.replace()` — cleans user input before tokenization. When building RAG pipelines, you split documents into chunks and cast embeddings to strings for storage. Most AI API errors come from incorrect string formatting — extra spaces, wrong quotes, missing `f` prefix — so mastering strings early pays off constantly in AI engineering.
