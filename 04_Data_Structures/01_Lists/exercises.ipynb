# Lists Exercises

Write your answer in the empty code cell under each question. Check yourself against `solutions.ipynb` when done.

**Part 1** = warm-up · **Part 2** = practice · **Part 3** = challenge.

## Part 1 — Warm-up

**1. Your first list.** Create a list called `fruits` holding `"apple"`, `"banana"` and `"cherry"`. Print the list, its type, and how many items it has.

**2. First and last.** Given `scores = [72, 85, 90, 66]`, print the first item, the last item (use a negative index), and the second-to-last item.

**3. Is it in the cart.** Given `cart = ["laptop", "mouse", "keyboard"]`, print whether `"mouse"` is in the cart and whether `"monitor"` is NOT in it.

## Part 2 — Practice

**4. Slice of the week.** Given `days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]`, print four slices: the first three days, the days from `"Wed"` onward, every second day, and the whole week reversed.

**5. Grow the inventory.** Start with `inventory = ["sword", "shield"]`. Add `"potion"` to the end, squeeze `"helmet"` in at index `1`, then extend with the loot list `["gem", "ring"]`. Print the list after every step.

**6. Trim the to-do list.** Start with `todo = ["email boss", "buy milk", "gym", "read"]`. Remove `"buy milk"` by value, then pop the last task and print what it was, then delete the first task with `del`. Print the list after each step.

**7. Sort it twice.** Given `numbers = [42, 7, 19, 3, 88]`, sort the list in place with `.sort()` and print what `.sort()` itself returned (surprise!). Then build a separate descending list with `sorted(..., reverse=True)` and confirm `numbers` did not change this time.

## Part 3 — Challenge

**8. The aliasing trap.** Run `a = [1, 2, 3]` and `b = a`. Append `4` through `b`, then print `a` - what happened? Prove it with `id(a) == id(b)`. Now make three *real* copies (`a.copy()`, `list(a)`, `a[:]`), show their ids differ from `a`, and demonstrate that appending to a copy leaves `a` alone.

**9. Grade book.** Given `grades = [91, 45, 78, 62, 55, 100]`, report how many grades there are, the highest, the lowest, and the average rounded to 2 decimals - using `len`, `max`, `min` and `sum`. Then snapshot the list with a slice copy, set `grades[0] = 0`, and print both lists to prove the backup survived.
