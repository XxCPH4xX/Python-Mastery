# Python Lists

> 📘 **Python Mastery** · Module 04 — Data Structures · Lesson 1/5

Lists are the most used container in Python: one name that holds many values, in order, and editable at any time. Almost every program you will ever write — from a shopping cart to a neural network's training data — stores things in lists.

## 🎯 Learning Objectives

By the end of this lesson you will be able to:

- **Create** lists with literals, the `list()` constructor, and from other iterables.
- **Read** single items with positive and negative indexing, and cut out sub-lists with slicing `[start:stop:step]`.
- **Test membership** with `in` and `not in`.
- **Modify** lists: change items, and add/remove with `append`, `insert`, `extend`, `remove`, `pop`, `del`, and `clear`.
- **Sort** lists correctly with `sort()` vs `sorted()`, including `reverse=True` and `key=len`.
- **Explain** the shallow-copy trap: why `b = a` is not a copy and how nested lists stay connected through copies.

## 1. Creating a List

A list is written as comma-separated values inside square brackets `[]`. Think of it as a numbered shopping list: the paper (the list object) holds many entries, and each entry keeps its position. A list can mix types freely — numbers, strings, even other lists.

**Syntax:**
```python
my_list = [item1, item2, item3]      # literal (most common)
empty = []                           # empty list
also = list(iterable)                # build from a string, range, tuple...
```

````python
shopping_list = ["rice", "eggs", "milk", "bananas"]
mixed = [42, "hello", 3.14, True]     # any types can live together
empty = []                            # valid: a list with nothing in it
from_range = list(range(1, 6))        # [1, 2, 3, 4, 5]
from_string = list("abc")             # ['a', 'b', 'c']

print(shopping_list)
print(mixed)
print(empty)
print(from_range)
print(from_string)
print(type(shopping_list), "holding", len(shopping_list), "items")
````

## 2. Indexing — Reading One Item

Every item has a position called an **index**, starting at `0`, not `1`. Negative indexes count from the end: `-1` is always the last item. This zero-based system is used everywhere in programming, so burn it in now.

**Syntax:**
```python
my_list[index]     # index 0 = first item, -1 = last item
```

````python
scores = [72, 85, 90, 66, 78]

first = scores[0]        # 0-based indexing: first item is [0]
second = scores[1]
last = scores[-1]        # -1 means "the last item"
second_last = scores[-2]

print(first, second, last, second_last)

# Asking for a position that does not exist raises IndexError
try:
    print(scores[10])
except IndexError as e:
    print("Error:", e)
````

## 3. Slicing — Cutting Out a Range `[start:stop:step]`

Slicing takes a **slice of the list** and hands you back a brand-new list. The `stop` index is *excluded* — think of it as a fence post you stop *before*. Slices never raise errors for out-of-range values; they just return what exists.

**Syntax:**
```python
my_list[start:stop:step]   # defaults: start=0, stop=end, step=1
my_list[::-1]              # classic trick: the whole list reversed
my_list[:]                 # a shallow copy of the entire list
```

````python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

print(days[0:3])      # ['Mon', 'Tue', 'Wed']  <- stop index NOT included
print(days[2:])       # from index 2 to the end
print(days[:4])       # from the start up to (not including) index 4
print(days[::2])      # every 2nd item
print(days[-3:-1])    # negative indexes work here too
print(days[::-1])     # step -1 reverses the whole list

weekend = days[5:]        # slicing always builds a NEW list
weekend.append("Nap day")
print(days[5:], "<- original untouched")
````

## 4. Checking Membership — `in` and `not in`

The `in` operator asks a yes/no question: *is this value somewhere in the list?* It returns a boolean and is far cleaner than writing your own loop.

**Syntax:**
```python
value in my_list         # True if present
value not in my_list     # True if absent
```

````python
cart = ["laptop", "mouse", "keyboard"]

print("mouse" in cart)          # True
print("monitor" in cart)        # False
print("monitor" not in cart)    # True

search = "webcam"               # Simulated input: what the user searched for
if search in cart:
    print(search, "is already in your cart")
else:
    print(search, "is NOT in your cart yet")
````

## 5. Changing Items — Lists Are Mutable

Lists are **mutable**: you can overwrite any position after creation. Contrast this with strings, which can never be edited in place — you always build a new string instead. Slice assignment even lets you replace several items at once.

**Syntax:**
```python
my_list[i] = new_value           # replace one item
my_list[start:stop] = [v1, v2]   # replace a whole range
```

````python
prices = [199.99, 49.50, 12.25]

prices[0] = 179.99              # replace one item
prices[-1] = 15.00              # works from the end too
prices[0:2] = [185.00, 45.00]   # replace a slice (can even change length)
print(prices)

word = list("cat")              # ['c', 'a', 't']
word[0] = "b"                   # strings can't do this, lists can
print(word)
````

## 6. Adding Items — `append()`, `insert()`, `extend()`

Three ways to grow a list, each with a different job: `append()` adds **one** item at the end (fast), `insert()` squeezes an item in **at a position** (everything after shifts right), and `extend()` adds **all items** from another iterable.

**Syntax:**
```python
my_list.append(x)        # one item, at the end
my_list.insert(i, x)     # one item, at position i
my_list.extend(other)    # every item of other, appended one by one
```

````python
inventory = ["sword", "shield"]

inventory.append("potion")          # add ONE item at the end
print(inventory)

inventory.insert(1, "helmet")       # slot it in at index 1, others shift right
print(inventory)

loot = ["gem", "ring"]
inventory.extend(loot)              # add MANY items from another list
print(inventory)

trap = ["a", "b"]
trap.append(["c", "d"])             # append adds the LIST ITSELF as one item!
print(trap, "<- nested, probably not what you wanted")
````

## 7. Removing Items — `remove()`, `pop()`, `del`, `clear()`

Four removal tools: `remove(value)` deletes the **first match by value** (error if absent), `pop(index)` deletes **by position and returns the item**, `del` deletes by position or slice without returning anything, and `clear()` empties the whole list.

**Syntax:**
```python
my_list.remove(value)     # first occurrence by value (ValueError if missing)
x = my_list.pop()         # remove + return last item (or pop(index))
del my_list[i]            # delete by position, no return value
my_list.clear()           # remove everything
```

````python
todo = ["email boss", "buy milk", "gym", "read", "buy milk"]

todo.remove("buy milk")        # kills only the FIRST match
print(todo)

finished = todo.pop()          # removes AND returns the last item
print("finished:", finished)
started = todo.pop(0)          # or pop any index
print("started:", started)

del todo[1]                    # delete by position, returns nothing
print(todo)

todo.clear()                   # empty the list completely
print(todo)

try:
    todo.remove("ghost task")  # value not in list -> ValueError
except ValueError as e:
    print("Error:", e)
````

## 8. Sorting — `sort()` vs `sorted()`

Two sorting tools that behave differently. `list.sort()` rearranges the list **in place** and returns `None`. `sorted(any_iterable)` leaves the original alone and returns a **new** sorted list. Both accept `reverse=True` and a `key=` function that decides what to compare (e.g. `key=len` compares word lengths).

**Syntax:**
```python
my_list.sort(reverse=False, key=None)   # in place, returns None
new_list = sorted(my_list, reverse=False, key=None)
```

````python
numbers = [42, 7, 19, 3, 88]

result = numbers.sort()      # sorts IN PLACE, and returns None!
print(numbers)               # the list itself is now sorted
print(result)                # None -- classic beginner surprise

descending = sorted(numbers, reverse=True)   # NEW list, original safe
print(descending)

names = ["Sarah", "bob", "Alice"]
print(sorted(names))                  # ALL capitals sort before lowercase!
print(sorted(names, key=str.lower))   # true alphabetical order

animals = ["elephant", "cat", "hippo"]
print(sorted(animals, key=len))       # sort by word length
````

## 9. Copying Lists — the Shallow-Copy Trap ⚠️

This is the single most important idea in this lesson. Assigning `b = a` does **not** copy anything — it just gives the same list a second name. Use `.copy()`, `list(a)` or `a[:]` to get an independent top-level copy. But beware: those are *shallow* copies, so **nested lists inside are still shared**.

> 🔍 **Under the Hood:** a Python variable never "contains" a list — it stores a *reference* (a pointer) to a list object living elsewhere in memory. `b = a` copies the pointer, so both names aim at the exact same object (`id(a) == id(b)` proves it). A shallow copy allocates a fresh outer list and fills it with *copies of the pointers*, which is why the top level becomes independent while any inner list objects remain shared. Only `copy.deepcopy()` walks the whole structure and clones every level.

**Syntax:**
```python
b = a              # NOT a copy: another name for the same list
c = a.copy()       # real (shallow) copy  - or list(a), or a[:]
deep = copy.deepcopy(a)   # copy including nested lists
```

````python
a = [1, 2, 3]
b = a                     # NOT a copy: both names -> the SAME list
print(id(a) == id(b))     # True: identical object in memory

b.append(4)               # modify through b...
print(a, "<- a changed too!")

c = a.copy()              # three idiomatic ways to get a REAL copy
d = list(a)
e = a[:]
print(id(a) == id(c), id(a) == id(d), id(a) == id(e))   # all False now

c.append(99)
print(a, "<- safe: c is a different object")
````

````python
import copy

grid_a = [[1, 2], [3, 4]]
grid_b = grid_a.copy()      # SHALLOW copy: new outer list...

grid_b.append([5, 6])       # ...so adding at the TOP level is safe
print(grid_a)

grid_b[0][0] = 999          # but the INNER lists are still shared!
print(grid_a, "<- oops, original changed")

grid_c = copy.deepcopy(grid_a)   # deep copy clones every level
grid_c[0][0] = 0
print(grid_a, "<- untouched now")
````

## 10. List Operators — `+` and `*`

Lists support two arithmetic-style operators: `+` concatenates two lists into a new one, and `*` repeats a list several times. Handy for building quick patterns and combined results.

**Syntax:**
```python
combined = list1 + list2      # new list, both joined
repeated = [x] * n            # the list repeated n times
my_list += other              # shorthand for extend()
```

````python
fruits = ["apple", "banana"] + ["cherry"]   # + joins two lists
print(fruits)

pattern = ["hi"] * 4                        # * repeats the contents
print(pattern)

scores = [10]
scores += [20, 30]                          # += grows the list in place
print(scores)

x = [1, 2]
y = [1, 2]
print(x == y, x is y)    # == compares VALUES; is compares IDENTITY
````

## 11. Nested Lists — Grids and Matrices

A list can contain other lists, which turns it into a 2-D structure — a table of rows. You reach an inner item with two brackets: `m[row][col]`. This is the mental warm-up for NumPy arrays later in the course.

**Syntax:**
```python
matrix = [[r1c1, r1c2], [r2c1, r2c2]]
matrix[row][column]
```

````python
matrix = [
    [1, 2, 3],      # row 0
    [4, 5, 6],      # row 1
    [7, 8, 9],      # row 2
]

print(matrix[1])        # the whole second row
print(matrix[1][2])     # row 1, column 2 -> 6

matrix[2][0] = 70       # editing deep inside works fine
for row in matrix:
    print(row)
````

## 12. Built-in Helpers — `len()`, `min()`, `max()`, `sum()`

Python ships four workhorse functions that accept any list of numbers (and `len` works on everything). They are the fastest way to answer "how many / how small / how big / how much in total".

**Syntax:**
```python
len(my_list)      # item count
min/max(my_list)  # extremes (numbers or comparable strings)
sum(my_list)      # total of numbers
```

````python
temps = [21.5, 19.8, 24.1, 22.6]

print(len(temps))                  # how many readings
print(min(temps), max(temps))      # coldest, hottest
print(sum(temps))                  # total
print(round(sum(temps) / len(temps), 2))   # average

students = ["Rafi", "Mina", "John"]
print(min(students), max(students))   # strings too: alphabetical ends
````

## ⚠️ Common Mistakes & Gotchas

| Mistake | Problem | Fix |
| :--- | :--- | :--- |
| `top = nums.sort()` | `sort()` returns `None`, so `top` becomes `None` and the "sorted result" vanishes | Use `top = sorted(nums)`, or call `nums.sort()` alone |
| `copy = original` | Copies the *reference*, not the list — both names mutate together | Use `original.copy()`, `list(original)` or `original[:]` |
| `lst.append([4, 5, 6])` | Adds the whole list as ONE nested element | Use `lst.extend([4, 5, 6])` or `lst += [4, 5, 6]` |
| Removing items while looping `for x in lst:` | The list shrinks underneath the loop, so items get skipped | Loop over `lst.copy()`, or build a comprehension: `[x for x in lst if keep(x)]` |
| `lst[len(lst)]` | Valid indexes are `0` to `len(lst)-1`; this is one past the end → `IndexError` | Use `lst[-1]` for the last item |
| Editing a list inside a function and expecting the caller's list unchanged | Lists are passed by reference — the caller sees every mutation | Pass `data.copy()` or build and return a new list |

## 💡 Best Practices & Pro Tips

- Prefer `append()` (amortized O(1)). `insert(0, x)` shifts every element and is O(n) — if you push/pop at the front constantly, reach for `collections.deque` instead.
- Membership tests (`x in big_list`) scan item by item — O(n). If you test repeatedly against thousands of items, convert to a `set` (next lessons).
- Keep lists **homogeneous** (all scores, all names). Mixed-type lists work but make every later line harder to reason about.
- Functions that receive a list can permanently change the caller's data. Document mutating functions clearly, or operate on a copy.
- 🔬 **AI-engineering relevance:** lists are the raw material of ML code — a batch is a list of samples, a document is a list of tokens, a dataset is a list of rows. Before feeding them to NumPy/PyTorch, you will slice, filter, and reshape exactly as practiced here. And the aliasing trap from Section 9 is the #1 source of "why did my training data change?" bugs in real projects.

## 📌 Summary

| Method / Syntax | What it does | Example |
| :--- | :--- | :--- |
| `[a, b, c]` | create a list | `nums = [1, 2, 3]` |
| `lst[i]` / `lst[-1]` | read one item (0-based, negative from end) | `lst[0]`, `lst[-1]` |
| `lst[a:b:c]` | slice (stop excluded) | `lst[::2]`, `lst[::-1]` |
| `x in lst` | membership test | `"milk" in cart` |
| `lst[i] = v` | change an item | `prices[0] = 9.99` |
| `lst.append(x)` | add one item at the end | `cart.append("tea")` |
| `lst.insert(i, x)` | add at a position | `lst.insert(0, "first")` |
| `lst.extend(other)` | add many items | `lst.extend([4, 5])` |
| `lst.remove(v)` | delete first match by value | `lst.remove("spam")` |
| `lst.pop(i)` | delete by index, returns item | `last = lst.pop()` |
| `del lst[i]` | delete by index, returns nothing | `del lst[0]` |
| `lst.clear()` | remove everything | `lst.clear()` |
| `lst.sort(key=..., reverse=...)` | sort in place, returns `None` | `names.sort(key=len)` |
| `sorted(lst)` | new sorted list | `sorted(nums, reverse=True)` |
| `lst.copy()` / `list(lst)` / `lst[:]` | shallow copy | `backup = scores.copy()` |
| `len/min/max/sum` | built-in helpers | `sum(scores) / len(scores)` |

**Key takeaways**

- Lists are **ordered** and **mutable** — index, slice, and edit them freely.
- `b = a` aliases; `.copy()` protects only the top level; `copy.deepcopy()` protects nested data.
- `sort()` mutates and returns `None`; `sorted()` returns a new list.
- Zero-based indexing means the last item is `lst[-1]`, never `lst[len(lst)]`.
