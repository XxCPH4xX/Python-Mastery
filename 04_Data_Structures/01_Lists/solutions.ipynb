## Part 1 — Warm-up

**1. Your first list.** A list literal holds all three fruits; `len()` counts them.

```
fruits = ["apple", "banana", "cherry"]

print(fruits)          # ['apple', 'banana', 'cherry']
print(type(fruits))    # <class 'list'>
print(len(fruits))     # 3
```

**2. First and last.** Index `0` reads from the front; `-1` and `-2` count back from the end.

```
scores = [72, 85, 90, 66]

print(scores[0])    # 72  - first item
print(scores[-1])   # 66  - last item
print(scores[-2])   # 90  - second to last
```

**3. Is it in the cart.** `in` and `not in` answer membership directly with a boolean.

```
cart = ["laptop", "mouse", "keyboard"]

print("mouse" in cart)          # True
print("monitor" not in cart)    # True
```

## Part 2 — Practice

**4. Slice of the week.** Stop index is excluded, missing values fall back to the edges, and `[::-1]` walks backwards.

```
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

print(days[0:3])     # ['Mon', 'Tue', 'Wed']  - stop NOT included
print(days[2:])      # ['Wed', 'Thu', 'Fri', 'Sat', 'Sun']
print(days[::2])     # ['Mon', 'Wed', 'Fri', 'Sun']
print(days[::-1])    # the whole week, reversed
```

**5. Grow the inventory.** Three growth tools, three jobs: append one, insert at a spot, extend by many.

```
inventory = ["sword", "shield"]

inventory.append("potion")         # ONE item, at the end
print(inventory)

inventory.insert(1, "helmet")      # at index 1, others shift right
print(inventory)

inventory.extend(["gem", "ring"])  # MANY items, one by one
print(inventory)
```

**6. Trim the to-do list.** `remove()` matches by value, `pop()` hands back what it took, `del` works silently by position.

```
todo = ["email boss", "buy milk", "gym", "read"]

todo.remove("buy milk")     # kills only the FIRST match
print(todo)

finished = todo.pop()       # removes AND returns the last item
print("finished:", finished)

del todo[0]                 # delete by position, returns nothing
print(todo)
```

**7. Sort it twice.** `.sort()` mutates and returns `None`; `sorted()` returns a brand-new list.

```
numbers = [42, 7, 19, 3, 88]

result = numbers.sort()      # sorts IN PLACE...
print(numbers)               # [3, 7, 19, 42, 88]
print(result)                # None - classic beginner surprise

descending = sorted(numbers, reverse=True)   # brand-new list
print(descending)            # [88, 42, 19, 7, 3]
print(numbers)               # untouched: still ascending
```

## Part 3 — Challenge

**8. The aliasing trap.** Assignment shares one object; the three copy idioms each build a fresh list.

```
a = [1, 2, 3]
b = a                     # NOT a copy: a second name, same object
print(id(a) == id(b))     # True

b.append(4)
print(a, "<- changed through b!")

c = a.copy()              # three real-copy idioms
d = list(a)
e = a[:]
print(id(a) == id(c), id(a) == id(d), id(a) == id(e))  # False x3

c.append(99)
print(a, "<- safe: c is a different object")
```

**9. Grade book.** Built-ins answer count/extremes/total instantly, and `grades[:]` snapshots the top level.

```
grades = [91, 45, 78, 62, 55, 100]

print(len(grades))                # 6
print(max(grades), min(grades))   # 100 45
average = round(sum(grades) / len(grades), 2)
print(average)                    # 71.83

backup = grades[:]      # slice copy: independent at the top level
grades[0] = 0
print(grades)           # [0, 45, 78, 62, 55, 100]
print(backup)           # [91, 45, 78, 62, 55, 100] - survived
```
