## Part 1 — Warm-up

**1. Duplicates vanish.** Repeats disappear the moment the set is built.

```
unique_ids = {101, 102, 103, 101, 102}
print(unique_ids)      # {101, 102, 103} (display order may vary)
print(len(unique_ids)) # 3

print(set([3, 1, 3, 2, 1]))   # {1, 2, 3}
```

**2. The empty-set trap.** Curly braces were claimed by dictionaries first - `{}` is an empty DICT.

```
print(type({}))      # <class 'dict'>  - not a set!
print(type(set()))   # <class 'set'>   - the ONLY way to write an empty set
```

**3. No positions here.** Sets are unordered, so indexing fails; membership and looping are the way in.

```
tags = {"python", "coding", "ai"}

try:
    tags[0]
except TypeError as e:
    print("Error:", e)

print("ai" in tags)     # True
for tag in sorted(tags):
    print(tag)
```

## Part 2 — Practice

**4. Growing a set.** `add()` takes one element, `update()` pulls in a whole iterable, and duplicate adds are silent no-ops.

```
skills = {"python"}

skills.add("sql")
print(sorted(skills))     # ['python', 'sql']

skills.add("python")      # already present: nothing changes
print(sorted(skills))

skills.update(["git", "docker"])
print(sorted(skills))     # ['docker', 'git', 'python', 'sql']
```

**5. remove vs discard.** `discard()` forgives missing values; `remove()` does not.

```
nums = {10, 20, 30, 40}

nums.remove(20)
nums.discard(99)          # absent -> no complaint
print(sorted(nums))       # [10, 30, 40]

try:
    nums.remove(20)       # already gone -> KeyError
except KeyError as e:
    print("KeyError:", e)
```

**6. Skill overlap.** Four operators straight from mathematics: `|`, `&`, `-`, `^`.

```
frontend = {"html", "css", "js"}
backend = {"python", "js", "sql"}

print(sorted(frontend | backend))   # union: everyone
print(sorted(frontend & backend))   # intersection: shared
print(sorted(frontend - backend))   # frontend only
print(sorted(frontend ^ backend))   # in exactly one of them
```

**7. Subset checks.** `<=` tests subset, `<` proper subset, `>=` superset; equality ignores order.

```
basics = {"python", "sql"}
full_stack = {"python", "sql", "git"}

print(basics <= full_stack)          # True
print(basics < full_stack)           # True - subset AND smaller
print(full_stack >= basics)          # True - superset
print(basics == {"sql", "python"})   # True - order never matters
```

## Part 3 — Challenge

**8. Dedupe but keep the order.** Dict keys are unique AND remember insertion order - perfect for ordered dedup.

```
clicks = ["home", "shop", "home", "cart", "shop", "home"]

print(sorted(set(clicks)))          # unique but reshuffled
print(list(dict.fromkeys(clicks)))  # ['home', 'shop', 'cart'] - order kept
```

**9. The frozen set.** A frozenset can never change, which makes it hashable enough to be a dict key.

```
locked = frozenset([1, 2, 3])

try:
    locked.add(4)
except AttributeError as e:
    print("Error:", e)

flags = frozenset({"debug", "verbose"})
settings = {flags: "dev-profile"}
print(settings[frozenset({"verbose", "debug"})])   # dev-profile
```
