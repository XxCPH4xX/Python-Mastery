## Part 1 — Warm-up

**1. Duplicates vanish.** Create a set literal containing `101`, `102`, `103`, `101`, `102` and print it plus its length. Then pass `[3, 1, 3, 2, 1]` to `set()` and print the deduplicated result.

**2. The empty-set trap.** Print `type({})` and `type(set())`. Which one is really a set? Say why in a comment.

**3. No positions here.** With `tags = {"python", "coding", "ai"}`, try `tags[0]` inside `try/except` and print the error. Then query the set the right way: test whether `"ai"` is in it and loop over `sorted(tags)` to print each tag in stable order.

## Part 2 — Practice

**4. Growing a set.** Start with `skills = {"python"}`. Add `"sql"`, then try adding `"python"` again (what happens?), then `update` with the list ["git", "docker"]. Print `sorted(skills)` after every step.

**5. remove vs discard.** With `nums = {10, 20, 30, 40}`: remove `20`, then `discard` the missing value `99` (notice: silent), then try `remove(20)` a second time inside `try/except` and catch the `KeyError`.

**6. Skill overlap.** Given `frontend = {"html", "css", "js"}` and `backend = {"python", "js", "sql"}`, print (each wrapped in `sorted()` for stable display): the union, the intersection, the frontend-only skills, and the skills in exactly one of the two.

**7. Subset checks.** With `basics = {"python", "sql"}` and `full_stack = {"python", "sql", "git"}`: is `basics` a subset of `full_stack`? A *proper* subset? Does `full_stack >= basics` hold? And are `basics` and `{"sql", "python"}` equal?

## Part 3 — Challenge

**8. Dedupe but keep the order.** Given `clicks = ["home", "shop", "home", "cart", "shop", "home"]`: show that `sorted(set(clicks))` loses the visit order, then deduplicate while keeping first-seen order with `dict.fromkeys`.

**9. The frozen set.** Create `locked = frozenset([1, 2, 3])` and try `locked.add(4)` inside `try/except`. Finally, use a frozenset of flags as a DICTIONARY KEY and look up its value.
