# Python Sets

> 📘 **Python Mastery** · Module 04 — Data Structures · Lesson 3/5

A set stores each value **at most once**, in no particular order — exactly like the mathematical set you met at school. In exchange for giving up order and duplicates, Python gives you near-instant membership tests and elegant operations like union and intersection.

## 🎯 Learning Objectives

By the end of this lesson you will be able to:

- **Create** sets and explain how duplicates vanish automatically.
- **Avoid** the classic trap of using `{}` for an empty set (that builds a dict!).
- **Explain** why sets have no indexing or slicing.
- **Add and remove** elements with `add`, `update`, `remove`, `discard`, and `pop`.
- **Combine** sets with union, intersection, difference, and symmetric difference (`|`, `&`, `-`, `^`).
- **Test relationships** with subset/superset operators and **justify** why membership testing is O(1).

## 1. Creating a Set — Duplicates Vanish

Curly braces around comma-separated values build a set. If you repeat a value, Python silently keeps one copy. Passing any iterable to `set()` does the same deduplication on the fly.

**Syntax:**
```python
my_set = {v1, v2, v3}        # literal
from_iterable = set(x)       # from a list / string / tuple
```

````python
unique_ids = {101, 102, 103, 101, 102}   # duplicates removed instantly
print(unique_ids)
print(len({1, 1, 1}), "element survived out of three")

print(set([3, 1, 3, 2, 1]))              # dedupe any list
print(sorted(set("banana")))             # unique characters (sorted for stable display)
````

## 2. The Empty Set Trap — `set()`, NOT `{}`

Here is a genuine rite of passage: `{}` creates an empty **dictionary**, not an empty set — curly braces were claimed by dicts first. The empty set must be written `set()`.

**Syntax:**
```python
empty_dict = {}      # dict!
empty_set = set()    # the ONLY way to write an empty set
```

````python
empty_dict = {}
empty_set = set()

print(type(empty_dict))
print(type(empty_set))
````

## 3. Unordered — No Indexing, No Slicing

Sets have no positions, so `[0]` is meaningless and raises `TypeError`. You can still loop over a set, ask its length, and test membership — you just cannot ask for "the third element", because there is no such thing.

> 🔍 **Under the Hood:** a set is implemented as a *hash table*. Each value is run through `hash()` and that number picks its slot in an underlying array. Slot placement has nothing to do with insertion order — which is why there are no indexes, why display order can look scrambled, and (next section) why lookups skip the search entirely.

**Syntax:**
```python
my_set[0]          # TypeError: 'set' object is not subscriptable
for x in my_set:   # looping: fine
x in my_set        # membership: fine (and fast!)
```

````python
tags = {"python", "coding", "ai"}

try:
    tags[0]                       # no positions -> error
except TypeError as e:
    print("Error:", e)

print("ai" in tags)               # membership is the way to query a set
for tag in sorted(tags):          # sorted() only for stable DISPLAY order
    print(tag)
print(len(tags), "tags total")
````

## 4. Adding Elements — `add()` and `update()`

`add()` inserts a single element; `update()` merges in every element of another iterable. Adding a value the set already has changes nothing and raises no error.

**Syntax:**
```python
my_set.add(x)          # one element
my_set.update(other)   # all elements from any iterable
```

````python
skills = {"python"}

skills.add("sql")
print(sorted(skills))

skills.add("python")            # duplicate add: silent no-op
print(sorted(skills))

skills.update(["git", "docker"])     # many at once from a list
skills.update("ml")                  # careful: iterates the STRING -> 'm', 'l'
print(sorted(skills))
````

## 5. Removing Elements — `remove()` vs `discard()` vs `pop()`

The pair to remember: `remove(x)` blows up with `KeyError` if `x` is missing, while `discard(x)` stays silent either way. `pop()` removes and returns an *arbitrary* element — remember, there is no "first".

**Syntax:**
```python
my_set.remove(x)     # KeyError if x is absent
my_set.discard(x)    # no error if x is absent
x = my_set.pop()     # remove + return an arbitrary element
my_set.clear()       # empty the set
```

````python
nums = {10, 20, 30, 40}

nums.remove(20)               # exact value required
nums.discard(99)              # missing? discard shrugs it off
print(sorted(nums))

try:
    nums.remove(20)           # already gone -> KeyError
except KeyError as e:
    print("KeyError:", e)

taken = nums.pop()            # arbitrary element out
print("popped:", taken, "| left:", sorted(nums))
````

## 6. Set Algebra — Union, Intersection, Difference

This is where sets shine. Two sets combine with single-character operators straight from mathematics: `|` everything from either side, `&` only what both share, `-` what the left has and the right lacks, `^` what exactly one side has. Method forms do the same job and accept *any* iterable as the argument.

| Operator | Method | Meaning |
| :--- | :--- | :--- |
| `a \| b` | `a.union(b)` | in a OR b |
| `a & b` | `a.intersection(b)` | in a AND b |
| `a - b` | `a.difference(b)` | in a but not b |
| `a ^ b` | `a.symmetric_difference(b)` | in exactly one of them |

**Syntax:**
```python
a | b    # union
a & b    # intersection
a - b    # difference
a ^ b    # symmetric difference
```

````python
frontend = {"html", "css", "js"}
backend = {"python", "js", "sql"}

print(sorted(frontend | backend))    # union: everyone
print(sorted(frontend & backend))    # intersection: shared skills ("js")
print(sorted(frontend - backend))    # frontend-only
print(sorted(backend - frontend))    # backend-only
print(sorted(frontend ^ backend))    # in exactly ONE of the two

print(frontend.union({"js", "rust"}))       # methods accept ANY iterable
print(frontend.intersection({"js"}))
print(frontend.difference(["html"]))        # even a list works here
````

## 7. Relationships — Subsets and Supersets

Ask "is every element of A also in B?" with `<=` (or the wordier `issubset`). `>=` asks the reverse question. A strict `<` means subset *and* fewer elements.

**Syntax:**
```python
a <= b    # a is a subset of b        (a.issubset(b))
a >= b    # a is a superset of b      (a.issuperset(b))
a < b     # proper subset (strictly smaller)
a.isdisjoint(b)   # True if they share NOTHING
```

````python
basics = {"python", "sql"}
full_stack = {"python", "sql", "git"}

print(basics <= full_stack)     # every basic skill is in full_stack
print(full_stack >= basics)     # ...same fact from the other side
print(basics < full_stack)      # proper subset: yes, and strictly smaller
print(basics == {"sql", "python"})   # equality ignores order AND duplicates

print(basics.isdisjoint({"java", "go"}))   # nothing in common -> True
````

## 8. 🔍 Membership Is O(1) — Benchmark vs List

The killer feature. Asking `"x" in some_list` walks the list item by item — O(n), slow when huge. Asking `"x" in some_set` hashes `x` and jumps straight to its slot — O(1) on average, whether the set holds ten items or ten million.

> 🔍 **Under the Hood:** `hash(value)` produces a big integer; masking it down picks an index into the set's internal array of slots. Lookup therefore costs one hash + one jump, regardless of size. Collisions (two values landing in one slot) are handled by probing, which is why the guarantee is *average* O(1), not absolute. Lists store items by position only, so they must be scanned linearly.

**Example:**

````python
import time, random

random.seed(42)
big_list = list(range(200_000))
big_set = set(big_list)
target = 190_001                    # present in both

start = time.perf_counter()
for _ in range(1000):
    target in big_list
list_ms = (time.perf_counter() - start) * 1000

start = time.perf_counter()
for _ in range(1000):
    target in big_set
set_ms = (time.perf_counter() - start) * 1000

print(f"1000 lookups in LIST: {list_ms:8.1f} ms")
print(f"1000 lookups in SET : {set_ms:8.3f} ms")
print(f"the set was roughly {list_ms / max(set_ms, 0.001):,.0f}x faster")
````

## 9. Idiom Corner — Deduplicate a List While Keeping Order

`set()` deduplicates fast but scatters the order. When you need uniqueness **and** first-seen order, the idiomatic tool is `dict.fromkeys()` — since Python 3.7 dicts remember insertion order, and their keys are by definition unique.

**Syntax:**
```python
unique_ordered = list(dict.fromkeys(my_list))
```

````python
clicks = ["home", "shop", "home", "cart", "shop", "home"]

print(sorted(set(clicks)))                       # unique, but order lost
print(list(dict.fromkeys(clicks)), "<- order kept")

# same idea, spelled out with a manual loop (what fromkeys does internally)
seen = []
for c in clicks:
    if c not in seen:
        seen.append(c)
print(seen)
````

## 10. `frozenset` — an Immutable Set

A `frozenset` behaves like a set but rejects all modifying calls. Because it can never change, it is hashable — so unlike a plain set, it may serve as a dictionary key or live inside another set.

**Syntax:**
```python
locked = frozenset(iterable)   # no .add/.remove/.update allowed
{locked: value}                # legal: frozensets are hashable
```

````python
locked = frozenset([1, 2, 3])
print(locked)

try:
    locked.add(4)
except AttributeError as e:
    print("Error:", e)

config_flags = frozenset({"debug", "verbose"})
settings = {config_flags: "dev-profile"}          # frozenset as a dict key!
print(settings[frozenset({"verbose", "debug"})])  # equality + hash still match
````

## ⚠️ Common Mistakes & Gotchas

| Mistake | Problem | Fix |
| :--- | :--- | :--- |
| `empty = {}` | Creates an empty **dict**, not a set | Use `set()` |
| `tags[0]` | Sets are unordered → `TypeError: 'set' object is not subscriptable` | Loop, or test `x in tags`; convert with `sorted(tags)` or `list(tags)` |
| `my_set.add([1, 2])` | Lists are unhashable → `TypeError: unhashable type: 'list'` | Add immutable types only, or use a tuple / `frozenset` |
| Expecting `remove()` to fail quietly | `remove(missing)` raises `KeyError` | Use `discard()` when absence is normal |
| Relying on set print order | Order depends on hashing, not insertion | Sort for display: `sorted(s)`; never index by position |
| `s.update("abc")` intending one item | Strings iterate char by char → adds `'a'`, `'b'`, `'c'` | Use `s.add("abc")` for a single element |

## 💡 Best Practices & Pro Tips

- Convert once, test often: building a set from a large list pays for itself after just a handful of `in` checks.
- Reach for sets whenever the words "unique", "common", or "overlap" appear in the problem — they map directly onto `set()`, `&`, and `|`.
- Prefer `discard()` over `remove()` unless a missing element genuinely indicates a bug worth crashing on.
- Keep a canonical order for output by wrapping in `sorted(...)` at print time.
- 🔬 **AI-engineering relevance:** NLP vocabulary building is literally `set(tokens)`; measuring precision/recall overlap between predicted and gold label sets uses intersection/difference; deduplicating prompts before paying for LLM API calls is a set (or `dict.fromkeys`) away. And the O(1) lookup is why tokenizers store vocab as a dict/set, not a list.

| :--- | :--- | :--- |

**Key takeaways**

- Sets = **unordered + unique**: no duplicates ever, no indexing ever.
- Empty set is `set()`; `{}` belongs to dictionaries.
- Membership, unions, intersections are near-instant thanks to the hash table.
- Need uniqueness *and* order? `list(dict.fromkeys(items))`.
