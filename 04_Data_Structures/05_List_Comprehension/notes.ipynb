# Python List Comprehension

> 📘 **Python Mastery** · Module 04 — Data Structures · Lesson 5/5

List comprehension is Python's one-line recipe for building a list: describe *what you want*, not *how to loop*. It replaces the five-line append-loop with a single readable expression, and the same idea extends to dicts, sets, and memory-friendly generators.

## 🎯 Learning Objectives

By the end of this lesson you will be able to:

- **Write** the comprehension pattern `[expression for item in iterable]` and translate any simple for-loop into it.
- **Distinguish** filtering (`if` at the END) from transforming (`if-else` BEFORE the `for`).
- **Call functions** and methods inside comprehensions for real data-cleaning pipelines.
- **Flatten and transpose** nested matrices with nested comprehensions.
- **Build** dict and set comprehensions with `{k: v ...}` and `{x ...}`.
- **Choose** generator expressions when memory matters, guided by the readability rule.

## 1. The Syntax — `[expression for item in iterable]`

Read it out loud: *"give me `n ** 2`, for each `n` in `numbers`."* It is exactly the loop below it, compressed: the `for n in numbers` part is identical, and what used to be `squares.append(____)` moves to the front as the expression.

> 🔍 **Under the Hood:** a comprehension is not just shorter — it is faster. The compiler runs a dedicated `LIST_APPEND` bytecode inside the loop, skipping the repeated attribute lookup of `.append` on every pass. Typical speedup is 20–50% over the equivalent `for`/`append` loop.

**Syntax:**
```python
[expression for item in iterable]

# equivalent to:
result = []
for item in iterable:
    result.append(expression)
```

````python
numbers = [1, 2, 3, 4, 5]

squares = []                  # the classic loop version
for n in numbers:
    squares.append(n ** 2)
print(squares)

squares_lc = [n ** 2 for n in numbers]    # same result, one line
print(squares_lc)

names = ["sarah", "rafi", "mina"]
upper = [name.upper() for name in names]
print(upper)
````

## 2. Filtering — Condition AFTER the `for`

When the `if` comes at the **end**, it acts as a doorkeeper: items that fail the test never make it into the list. The expression stays unchanged; you are simply choosing *which* items to keep.

**Syntax:**
```python
[expression for item in iterable if condition]     # FILTER: keep some items
```

````python
scores = [45, 82, 91, 67, 38, 95]

passing = [s for s in scores if s >= 60]        # keep scores >= 60 only
print(passing)

evens = [n for n in range(1, 11) if n % 2 == 0]
print(evens)

words = ["ai", "model", "gpu", "tensor"]
long_words = [w for w in words if len(w) > 2]
print(long_words)
````

## 3. Transforming — Condition BEFORE the `for`

Put an `if-else` **before** the `for` and every item passes through, but gets transformed on the way: one value for truthy items, another for falsy ones. Nothing is filtered — everything appears, possibly changed. (This position is required because `if-else` is an *expression* here; a plain `if` is not.)

| Position | Meaning | Keeps how many items? |
| :--- | :--- | :--- |
| `[x if cond else y for x in it]` | transform | all of them |
| `[x for x in it if cond]` | filter | only those passing |

**Syntax:**
```python
[value_if_true if condition else value_if_false for item in iterable]
```

````python
scores = [45, 82, 91, 67, 38, 95]

labels = ["PASS" if s >= 60 else "FAIL" for s in scores]   # transform each item
print(labels)

parity = ["even" if n % 2 == 0 else "odd" for n in range(1, 6)]
print(parity)

raw_readings = [3, -1, 7, -8, 2]
clipped = [max(x, 0) for x in raw_readings]      # no if-else needed here!
print(clipped)
````

## 4. Calling Functions Inside Comprehensions

The expression slot accepts anything that returns a value: method chains, built-ins, your own functions. This is where data cleaning becomes a tidy pipeline.

**Syntax:**
```python
[clean(item) for item in messy_items]
```

````python
messy = ["  AI ", "ML\n", " DATA ", "llm"]
clean = [token.strip().lower() for token in messy]   # method chain per item
print(clean)

names = ["sarah akter", "rafi hasan"]
initials = [name.title() for name in names]
print(initials)

def normalize(x, lo=0, hi=1):     # custom function: scale lo..hi -> 0..1
    return round((x - lo) / (hi - lo), 2)

temps_c = [10, 20, 30, 40]
temps_f = [(c * 9 / 5) + 32 for c in temps_c]        # call math per item
print(temps_f)
print([normalize(c, 0, 100) for c in temps_c])       # custom function too
````

## 5. Nested Comprehensions — Flatten & Transpose Matrices

Two `for` clauses read left-to-right like nested loops — outer first, inner second. That ordering lets you flatten a matrix into one list, or build a new matrix cell by cell (the transpose).

**Syntax:**
```python
[f(x) for row in matrix for x in row]         # flatten (loop order preserved)
[[row[i] for row in matrix] for i in cols]    # nested -> builds rows
```

````python
matrix = [[1, 2, 3],
          [4, 5, 6]]

flat = [num for row in matrix for num in row]   # reads: for row... for num...
print(flat)

transposed = [[row[i] for row in matrix] for i in range(3)]
print(transposed)          # columns become rows

pairs = [(x, y) for x in range(1, 3) for y in range(1, 3)]   # mini grid
print(pairs)
````

## 6. Dictionary Comprehensions — `{key: value for ...}`

Swap the brackets for braces and add a colon: now each iteration produces a key-value pair. Perfect for lookup tables, inverted mappings, and filtered dictionaries.

**Syntax:**
```python
{key_expr: value_expr for item in iterable}            # build a dict
{k: v for k, v in pairs if condition}                  # with a filter
```

````python
words = ["apple", "banana", "cherry"]

length_map = {w: len(w) for w in words}
print(length_map)

cities = ["dhaka", "sydney", "london"]
temps = [31, 18, 12]
weather = {city: temp for city, temp in zip(cities, temps)}   # zip pairs them up
print(weather)

hot_only = {city: t for city, t in weather.items() if t > 15}
print(hot_only)

inverted = {temp: city for city, temp in weather.items()}     # flip direction
print(inverted)
````

## 7. Set Comprehensions — `{x for ...}`

Braces *without* a colon give you a set: same comprehension machinery, but duplicates collapse and order disappears — exactly the set rules from Lesson 3.

**Syntax:**
```python
{expression for item in iterable}
```

````python
sentences = ["AI is fun", "AI is powerful", "fun times"]

unique_words = {word.lower() for s in sentences for word in s.split()}
print(sorted(unique_words))                 # sorted() just for stable display

word_lengths = {len(w) for w in unique_words}   # which lengths occur?
print(sorted(word_lengths))

digits_used = {ch for ch in "call 999 or 112"}
print(sorted(digits_used))
````

## 8. Generator Expressions — Same Syntax, Lazy Results

Replace the brackets with parentheses and you get a **generator**: nothing is computed until you ask, and it produces values one at a time instead of storing them all. Feed it straight into `sum()`, `max()`, `any()`, `all()` or `join()` — even the extra parentheses may be dropped.

> 🔍 **Under the Hood:** a list comprehension allocates memory for every result up front — a million items means a million slots reserved immediately. A generator expression stores only a tiny iterator object (~200 bytes regardless of size) plus its current position, yielding each value on demand. For one-shot aggregations, that difference decides whether your script fits in RAM.

**Syntax:**
```python
(expr for item in iterable)          # generator object, computed lazily
sum(expr for item in iterable)       # common: feed directly to a function
```

````python
import sys

as_list = [n ** 2 for n in range(1_000_000)]   # all million squares NOW
as_gen = (n ** 2 for n in range(1_000_000))    # computes NOTHING yet

print(sys.getsizeof(as_list), "bytes for the finished list")
print(sys.getsizeof(as_gen), "bytes for the lazy generator")

total = sum(n ** 2 for n in range(1_000_001))   # genexp feeds sum() directly
print("sum of squares up to 1,000,000:", total)

print(any(n > 5 for n in range(10)))            # short-circuits early
print(as_gen)                                   # the object itself, not values
````

## 9. The Readability Rule

Comprehensions exist for clarity, not cleverness. The community bar: **if you can't read it aloud in one breath, use a loop.** No `else` without an `if` in the filter slot, avoid side effects inside, and flatten at most two levels.

**Example:**

````python
weird = [x * 2 if x % 2 == 0 else x
         for x in [y + 1 for y in range(10)]
         if x != 5]                    # legal... but who can read this?

readable = []
for x in range(1, 11):                 # same computation, obvious at a glance
    if x == 5:
        continue
    readable.append(x * 2 if x % 2 == 0 else x)

print(readable == weird)               # identical results
print(readable)

# good default: simple filter + simple transform
data = [12, 45, 7, 33, 21]
big_doubled = [n * 2 for n in data if n > 10]
print(big_doubled)
````

## ⚠️ Common Mistakes & Gotchas

| Mistake | Problem | Fix |
| :--- | :--- | :--- |
| Putting `if-else` after the `for`: `[x for x in it if x>1 else 0]` | SyntaxError — filters can't have an `else` | Transform goes BEFORE the `for`: `[x if x>1 else 0 for x in it]`; filter stays after |
| Expecting `[x for x in it if cond else ...]` to filter *and* replace | One `if` cannot do both jobs | Combine: `[x if cond else fallback for x in it]` (transform) or nest conditions carefully |
| Building a million-item list just to run `sum(...)` | Wastes memory and time | Use a generator expression: `sum(x for x in big)` |
| Side effects inside a comprehension (`[print(x) for x in xs]`) | Works, but returns junk (`[None,...]`) and confuses readers | Use a plain `for` loop for actions; comprehensions produce values |
| Four-deep nested comprehensions | Unreadable and bug-prone | Flatten with loops, or extract helper functions |
| Assuming the loop variable leaks out afterwards | In Python 3 it doesn't — `n` after `[n for n in r]` is gone/unbound | Don't rely on it; assign explicitly if you need the last value |

## 💡 Best Practices & Pro Tips

- Default ladder: simple `for`+`append` → comprehension → generator expression → `map`/`filter`. Move right only with a reason.
- Name things well: `[score for score in scores if score >= 60]` beats `[s for s in x if s > 60]` in shared code.
- Reach for `sum()/any()/all()/"".join(genexp)` before building throwaway lists.
- Keep transformations pure: same input, same output, no printing or mutating — then comprehensions stay safe to reorder and parallelize later.
- 🔬 **AI-engineering relevance:** preprocessing pipelines are comprehension country — normalizing text tokens, clipping pixel values, building `{word: index}` vocabularies, batching samples. And the genexp lesson scales up: PyTorch DataLoader and pandas iterate lazily for exactly the memory reasons shown in Section 8.

## 📌 Summary

| Pattern | Syntax | Example |
| :--- | :--- | :--- |
| Basic build | `[expr for x in it]` | `[n**2 for n in nums]` |
| Filter (if AFTER for) | `[expr for x in it if cond]` | `[s for s in scores if s >= 60]` |
| Transform (if-else BEFORE for) | `[a if cond else b for x in it]` | `["PASS" if s>=60 else "FAIL" for s in scores]` |
| Nested loops | `[expr for row in m for x in row]` | flatten a matrix |
| Nested lists | `[[...] for i in range(k)]` | transpose a matrix |
| Function calls | `[clean(x) for x in raw]` | `[t.strip().lower() for t in messy]` |
| Dict comp | `{k: v for x in it}` | `{w: len(w) for w in words}` |
| Set comp | `{x for x in it}` | `{w.lower() for w in words}` |
| Generator exp | `(x for x in it)` | `sum(x*x for x in range(10**6))` |

**Key takeaways**

- `[expression for item in iterable]` — read it aloud; it *is* the loop.
- `if` after the `for` filters; `if-else` before the `for` transforms.
- Braces with a colon → dict comp; braces without → set comp.
- Parentheses → lazy generator: constant memory for huge one-pass jobs.
- Unreadable? Use a loop. Clarity always wins.
