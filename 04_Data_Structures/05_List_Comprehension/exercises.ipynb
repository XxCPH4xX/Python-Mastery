# List Comprehension Exercises

Write your answer in the empty code cell under each question. Check yourself against `solutions.ipynb` when done.

**Part 1** = warm-up · **Part 2** = practice · **Part 3** = challenge.

## Part 1 — Warm-up

**1. Squares, twice.** Build the squares of `[1, 2, 3, 4, 5]` the classic way (empty list + `for` loop + `.append()`), then again as a one-line list comprehension. Print both and confirm they match.

**2. Filter the evens.** Using ONE comprehension, collect the even numbers from `range(1, 11)`. Remember: the filtering `if` goes AFTER the `for`.

**3. Pass or fail.** For `scores = [45, 82, 91, 67, 38]`, build a list of the SAME length where each score becomes `"PASS"` (60 or more) or `"FAIL"`. The `if-else` goes BEFORE the `for`.

## Part 2 — Practice

**4. Clean the tokens.** Given `messy = ["  AI ", "ML\n", " DATA ", "llm"]`, use one comprehension to produce `["ai", "ml", "data", "llm"]` - chain `.strip()` and `.lower()` on each token.

**5. Flatten the matrix.** Given `matrix = [[1, 2, 3], [4, 5, 6]]`, use a nested comprehension (two `for` clauses) to flatten it into `[1, 2, 3, 4, 5, 6]`.

**6. Weather lookup table.** Given `cities = ["dhaka", "sydney", "london"]` and `temps = [31, 18, 12]`, build a `{city: temp}` dictionary with a dict comprehension fed by `zip`. Then build a SECOND dict containing only the cities warmer than 15 degrees.

**7. Unique words.** Given `sentences = ["AI is fun", "AI is powerful", "fun times"]`, use a set comprehension to collect every lowercase word across all sentences. Print it wrapped in `sorted()` for stable display.

## Part 3 — Challenge

**8. Transpose.** The matrix `[[1, 2, 3], [4, 5, 6]]` has 2 rows and 3 columns. Use a nested comprehension to transpose it into `[[1, 4], [2, 5], [3, 6]]` - columns become rows.

**9. Lazy squares.** Compare memory: build `[n ** 2 for n in range(1_000_000)]` and its generator-expression twin, then print `sys.getsizeof` for both. Finish by feeding a generator expression straight into `sum()` to total the squares from 1 to 1,000,000.
