## Part 1 — Warm-up

**1. Squares, twice.** Same loop, compressed - the appended expression moves to the front.

```
numbers = [1, 2, 3, 4, 5]

squares = []
for n in numbers:                 # classic version
    squares.append(n ** 2)

squares_lc = [n ** 2 for n in numbers]   # one line

print(squares)
print(squares_lc)
print(squares == squares_lc)      # True
```

**2. Filter the evens.** The trailing `if` acts as a doorkeeper: failing items never enter.

```
evens = [n for n in range(1, 11) if n % 2 == 0]
print(evens)    # [2, 4, 6, 8, 10]
```

**3. Pass or fail.** `if-else` before the `for` transforms every item instead of filtering.

```
scores = [45, 82, 91, 67, 38]

labels = ["PASS" if s >= 60 else "FAIL" for s in scores]
print(labels)   # ['FAIL', 'PASS', 'PASS', 'PASS', 'FAIL']
```

## Part 2 — Practice

**4. Clean the tokens.** The expression slot accepts whole method chains - a pipeline in one line.

```
messy = ["  AI ", "ML\n", " DATA ", "llm"]

clean = [token.strip().lower() for token in messy]
print(clean)    # ['ai', 'ml', 'data', 'llm']
```

**5. Flatten the matrix.** Two `for` clauses read left to right, exactly like the nested loops they replace.

```
matrix = [[1, 2, 3], [4, 5, 6]]

flat = [num for row in matrix for num in row]
print(flat)    # [1, 2, 3, 4, 5, 6]
```

**6. Weather lookup table.** `zip` pairs the lists up; a filter at the end keeps only the hot cities.

```
cities = ["dhaka", "sydney", "london"]
temps = [31, 18, 12]

weather = {city: temp for city, temp in zip(cities, temps)}
print(weather)    # {'dhaka': 31, 'sydney': 18, 'london': 12}

hot_only = {city: t for city, t in weather.items() if t > 15}
print(hot_only)   # {'dhaka': 31, 'sydney': 18}
```

**7. Unique words.** Braces WITHOUT a colon make a set comprehension - duplicates collapse.

```
sentences = ["AI is fun", "AI is powerful", "fun times"]

unique_words = {word.lower() for s in sentences for word in s.split()}
print(sorted(unique_words))
# ['ai', 'fun', 'is', 'powerful', 'times']
```

## Part 3 — Challenge

**8. Transpose.** The OUTER clause picks the column index; the INNER clause gathers that column from every row.

```
matrix = [[1, 2, 3],
          [4, 5, 6]]

transposed = [[row[i] for row in matrix] for i in range(3)]
print(transposed)    # [[1, 4], [2, 5], [3, 6]]
```

**9. Lazy squares.** The finished list reserves a million slots; the generator is a tiny lazy object.

```
import sys

as_list = [n ** 2 for n in range(1_000_000)]   # computed NOW
as_gen = (n ** 2 for n in range(1_000_000))    # computed on demand

print(sys.getsizeof(as_list), "bytes for the list")
print(sys.getsizeof(as_gen), "bytes for the generator")

total = sum(n ** 2 for n in range(1_000_001))  # genexp feeds sum()
print(total)                                   # 333333833333500000
```
