# Python Data Structures Cheat Sheet

#### Complete Python Data Structures & Types Cheat Sheet

| Category / Data Type | Delimiters / Enclosing Characters | Example Syntax |
| :--- | :--- | :--- |
| **String** (`str`) | Single `' '`, Double `" "`, Triple `""" """` | `"Hello"` |
| **List** (`list`) | Square brackets `[ ]` | `[1, 2, 3]` |
| **Tuple** (`tuple`) | Parentheses `( )` | `(1, 2, 3)` |
| **Set** (`set`) | Curly braces `{ }` | `{1, 2, 3}` |
| **Frozen Set** (`frozenset`) | `frozenset({ })` | `frozenset({1, 2, 3})` |
| **Dictionary** (`dict`) | Curly braces with key-value pairs `{ : }` | `{"a": 1, "b": 2}` |
| **Range** (`range`) | `range(start, stop)` | `range(1, 4)` |
| **Array** (`array` module) | `array('typecode', [ ])` | `array('i', [1, 2, 3])` |
| **NumPy Array** (`numpy`) | `np.array([ ])` | `np.array([1, 2, 3])` |
| **Bytes** (`bytes`) | `b' '` or `b" "` | `b'Hello'` |
| **Bytearray** (`bytearray`) | `bytearray([ ])` | `bytearray([65, 66, 67])` |

Code Example

```python
import array
import numpy as np

# 1. Tuple (uses parentheses directly)
my_tuple = (1, 2, 3)
print(my_tuple)  # Output: (1, 2, 3)

# 2. List (created with parentheses via constructor)
my_list = list((1, 2, 3))
print(my_list)  # Output: [1, 2, 3]

# 3. Set (created with parentheses via constructor)
my_set = set((1, 2, 3))
print(my_set)  # Output: {1, 2, 3}

# 4. Standard Python Array (requires typecode 'i' for integers)
my_array = array.array("i", (1, 2, 3))
print(my_array)  # Output: array('i', [1, 2, 3])

# 5. NumPy Array
my_numpy_array = np.array((1, 2, 3))
print(my_numpy_array)  # Output: [1 2 3] (printed without commas)

# 6. String Type
greeting = "Hello, World!"  # str (string)

print(greeting, type(greeting))
```
