## 1. The Big Comparison Table

| Structure | Syntax | Ordered? | Mutable? | Duplicates? | Indexing? | Typical use |
| :--- | :--- | :---: | :---: | :---: | :---: | :--- |
| **List** `list` | `[1, 2, 3]` | ✅ insertion order | ✅ yes | ✅ allowed | ✅ `lst[0]`, slices | shopping lists, batches, anything that changes |
| **Tuple** `tuple` | `(1, 2, 3)` · one item `(5,)` | ✅ insertion order | ❌ frozen | ✅ allowed | ✅ `t[0]`, slices | coordinates, fixed records, dict keys |
| **Set** `set` | `{1, 2, 3}` · empty is `set()` | ❌ unordered | ✅ yes | ❌ auto-removed | ❌ no positions | dedupe, membership tests, set algebra |
| **Dict** `dict` | `{"k": v}` · empty is `{}` or `dict()` | ✅ since Python 3.7 | ✅ yes | keys unique; values may repeat | ❌ by position, ✅ by key (`d["k"]`) | lookups, JSON, configs, records |

> 🔍 **Under the Hood:** list/tuple keep items in a contiguous array of references (position = index). Set/dict are hash tables: every key/element is hashed to find its slot, which is why they trade order for O(1) average lookups.

## 2. Lists — Quick Reference

| Method / Syntax | Returns / Does | Mini-example |
| :--- | :--- | :--- |
| `[a, b]` / `list(x)` | create | `nums = [3, 1, 2]` |
| `lst[i]` / `lst[-1]` | item at position (0-based / from end) | `nums[0]` → `3` |
| `lst[a:b:c]` | slice, stop excluded | `nums[::-1]` → `[2, 1, 3]` |
| `x in lst` | membership scan, O(n) | `1 in nums` → `True` |
| `.append(x)` | add one item at end | `nums.append(4)` |
| `.insert(i, x)` | add at position, shifts rest | `nums.insert(0, 0)` |
| `.extend(it)` / `lst += it` | append each element of it | `nums.extend([5, 6])` |
| `.remove(v)` | delete first occurrence by value | `nums.remove(2)` |
| `.pop(i)` | remove + return item i (default last) | `nums.pop()` |
| `del lst[i]` | delete by position, no return | `del nums[0]` |
| `.clear()` | remove all items | `nums.clear()` |
| `.sort(key=, reverse=)` | sort IN PLACE, returns `None` | `nums.sort(reverse=True)` |
| `sorted(lst)` | NEW sorted list | `sorted(nums, key=len)` |
| `.count(v)` / `.index(v)` | occurrences / first position | `nums.count(1)` |
| `.copy()` / `list(lst)` / `lst[:]` | shallow copy | `backup = nums.copy()` |
| `len/min/max/sum` | built-in helpers | `sum(nums)` |

⚠️ Remember: `b = a` aliases (no copy); nested inner lists survive even `.copy()` — use `copy.deepcopy()` for full isolation.

## 3. Tuples — Quick Reference

| Method / Syntax | Returns / Does | Mini-example |
| :--- | :--- | :--- |
| `(1, 2)` / `1, 2` | create (comma makes it, parens optional) | `point = 3, 7` |
| `(5,)` | ONE-item tuple — trailing comma required | `(5,)` not `(5)` |
| `()` | empty tuple | `empty = ()` |
| `t[i]` / slicing | read only, same as lists | `point[-1]` |
| `v in t` | membership | `3 in point` |
| `.count(v)` | how many times v appears | `t.count("win")` |
| `.index(v)` | first position of v | `t.index("draw")` |
| `t[0] = x` | ❌ TypeError — immutable | rebuild instead: `t + (x,)` |
| `list(t)` ↔ `tuple(lst)` | unfreeze / freeze | `tuple([1, 2])` |
| Packing / unpacking | assign & split by position | `x, y = point`; `a, b = b, a` |
| Extended unpacking | star collects the rest | `first, *rest = t` |

💡 Hashable when all elements are immutable → works as dict key / set member.

## 4. Sets — Quick Reference

| Method / Syntax | Returns / Does | Mini-example |
| :--- | :--- | :--- |
| `{1, 2}` / `set(x)` | create, duplicates auto-dropped; empty is `set()` NOT `{}` | `set([1, 1, 2])` → `{1, 2}` |
| `.add(x)` | insert one element | `tags.add("ai")` |
| `.update(it)` | insert many (iterates!) | `tags.update(["ml", "llm"])` |
| `.remove(x)` | delete x — KeyError if absent | `nums.remove(20)` |
| `.discard(x)` | delete x silently if absent | `nums.discard(99)` |
| `.pop()` | remove + return arbitrary element | `nums.pop()` |
| `.clear()` | remove everything | `nums.clear()` |
| `a \| b` / `.union(it)` | union: in either | `fe \| be` |
| `a & b` / `.intersection(it)` | shared elements | `fe & be` |
| `a - b` / `.difference(it)` | in a but not b | `fe - be` |
| `a ^ b` / `.symmetric_difference(it)` | in exactly one | `fe ^ be` |
| `a <= b` / `a >= b` | subset / superset | `basics <= stack` |
| `.isdisjoint(b)` | True if nothing shared | `basics.isdisjoint({"go"})` |
| `frozenset(x)` | immutable + hashable set | `frozenset([1, 2])` |

⚠️ No indexing/slicing (unordered). Elements must be hashable — no lists/dicts inside.

💡 Order-preserving dedupe idiom: `list(dict.fromkeys(items))`

## 5. Dictionaries — Quick Reference

| Method / Syntax | Returns / Does | Mini-example |
| :--- | :--- | :--- |
| `{"k": v}` / `dict(pairs)` | create; keys must be hashable & unique | `user = {"name": "Sarah"}` |
| `d[k]` | value for k — KeyError if missing | `user["name"]` |
| `.get(k, default)` | value or default, never crashes | `stock.get("mango", 0)` |
| `d[k] = v` | add new key or overwrite | `user["age"] = 22` |
| `.update(other)` | merge many pairs, overwriting | `prefs.update({"theme": "dark"})` |
| `.setdefault(k, dv)` | fetch-or-create with default | `log.setdefault(day, []).append(x)` |
| `.pop(k)` | remove k, return its value | `cache.pop("tmp")` |
| `.popitem()` | remove + return last-inserted pair | `cache.popitem()` |
| `del d[k]` | delete entry, no return | `del cache["img1"]` |
| `.clear()` | remove all entries | `cache.clear()` |
| `.keys()` / `.values()` / `.items()` | LIVE views over the dict | `for k, v in d.items():` |
| `k in d` | membership checks KEYS only, O(1) avg | `"sarah" in users` |

⚠️ Views are live windows, not copies — snapshot with `list(d.keys())`. Keys must be immutable (str/int/tuple OK; list raises TypeError).

💡 Dict comprehension: `{k: v for x in it}` — e.g. `{w: len(w) for w in words}`

## 6. Converting Between Structures

| Conversion | Call | Notes |
| :--- | :--- | :--- |
| list → tuple | `tuple(lst)` | freeze it |
| tuple → list | `list(t)` | unfreeze it |
| any iterable → set | `set(x)` | dedupes; order lost |
| set/list/tuple → sorted list | `sorted(x)` | dedupe AND order back |
| pairs → dict | `dict(pairs)` | needs (key, value) tuples |
| two lists → dict | `dict(zip(keys, values))` | zip pairs them up |
| string → list of chars/words | `list(s)` / `s.split()` | |
| list of strings → string | `"-".join(lst)` | any separator |
| dict → pairs | `d.items()` | loop or `list(...)` |

**Example:**

```python
data = [1, 2, 2, 3]
pairs = [("a", 1), ("b", 2)]

print(tuple(data))                       # (1, 2, 2, 3)
print(set(data))                         # {1, 2, 3}
print(sorted(set(data)))                 # [1, 2, 3]  dedupe + order
print(dict(pairs))                       # {'a': 1, 'b': 2}
print(dict(zip(["x", "y"], [10, 20])))   # {'x': 10, 'y': 20}
print("-".join(["learn", "python"]))     # learn-python
print(list(dict.fromkeys(data)))         # [1, 2, 3]  dedupe keeping order
```

## 7. Choosing the Right Structure — Decision Guide

Answer in order:

1. **Need label→value lookup?** → **dict** (configs, records, JSON, caches).
2. **Need uniqueness / fast membership / overlap math?** → **set**.
3. **Ordered collection that must change?** → **list**.
4. **Ordered, fixed forever (or needed as a dict key)?** → **tuple**.

| You have... | Reach for |
| :--- | :--- |
| A shopping cart that grows and shrinks | list |
| (lat, lon) GPS points used as map keys | tuple |
| "Which of these emails are already registered?" | set |
| User profiles keyed by user ID | dict |
| Unique vocabulary words from a corpus | set |
| A function returning several related values | tuple |
| Counting word frequencies | dict |
| Deduplicating a playlist but keeping play order | `list(dict.fromkeys(...))` |

## 8. Big-O Speed Table

Average cost per operation (`n` = number of items):

| Operation | List | Tuple | Set | Dict |
| :--- | :---: | :---: | :---: | :---: |
| Access by index `x[i]` | O(1) | O(1) | — | — |
| Access by key | — | — | — | O(1) avg |
| Membership `in` | O(n) 😞 | O(n) | O(1) avg ⚡ | O(1) avg ⚡ (keys) |
| Append / add at end | O(1)* amortized | — | O(1) avg | O(1) avg |
| Insert / delete in middle | O(n) | — | O(1) avg | O(1) avg |
| Pop from end | O(1) | — | — | `popitem()` O(1) |
| Pop from front | O(n) 😞 (use `deque`) | — | — | — |
| Sort | O(n log n) | — | — | — |
| Copy | O(n) | — | O(n) | O(n) |

\* amortized: occasional resizes, cheap on average.

**The two rules worth memorizing:** searching a *list* scans everything; searching a *set/dict* jumps straight there. If you test membership more than once against a big collection, convert to a set first.

## ⚠️ Cross-Type Gotchas Recap

| Mistake | Problem | Fix |
| :--- | :--- | :--- |
| `empty_set = {}` | That's an empty **dict** | `set()` |
| `single = (5)` | Plain int — comma makes tuples | `(5,)` |
| `b = a` on a list | Alias, not a copy | `a.copy()` (+ `deepcopy` if nested) |
| `nums.sort()` captured as result | Returns `None` | `sorted(nums)` |
| `value in my_dict` expecting values | Checks keys only | `value in my_dict.values()` |
| `my_list` as a dict/set member | Unhashable → TypeError | Use a tuple instead |
| Looping a list/dict while mutating it | Skipped items / RuntimeError | Iterate over a snapshot or comprehension |

## 💡 Best Practices & Pro Tips

- Default to **list** for sequences, upgrade to **set/dict** the moment lookup speed or uniqueness matters.
- Keep collections homogeneous (all numbers, all strings) — future-you reads them faster.
- Prefer built-ins over hand-rolled loops: `sum`, `min`, `max`, `len`, `sorted`, `zip`, `enumerate`.
- Comprehensions for building, generator expressions for consuming huge streams.
- 🔬 **AI-engineering relevance:** this sheet is the daily vocabulary of data work — rows as list-of-dicts before pandas, vocab as token→id dict, label sets for metrics, tensor shapes as tuples, config dicts passed to models. Master these four containers and every library in modules 10–18 reads easier.

## 📌 Summary

| Structure | One-line identity | Signature move |
| :--- | :--- | :--- |
| **List** | ordered, mutable, duplicates OK | `lst.append(x)` · slice `lst[::-1]` |
| **Tuple** | ordered, frozen, hashable | `x, y = point` · dict key |
| **Set** | unordered, unique, O(1) `in` | `a & b` · dedupe `set(x)` |
| **Dict** | key→value, ordered since 3.7 | `d.get(k, default)` · `for k, v in d.items():` |
| **Comprehensions** | one-line builders | `[f(x) for x in xs if p(x)]` · `{k: v for ...}` |

**Key takeaways**

- Pick structure by the questions you'll ask of the data: by position → list/tuple; by identity/uniqueness → set; by name → dict.
- Mutability decides what can be a key and whether data stays put: mutable list/set/dict vs immutable tuple/frozenset.
- Speed lives in hash tables: sets and dicts give average O(1) membership; lists cost O(n).
- Everything above converts cleanly with `list()`, `tuple()`, `set()`, `dict()`, `sorted()` — so choosing wrong early is cheap to fix.
