Write your answer in the empty code cell under each question. Check yourself against `solutions.ipynb` when done.

## Part 1 — Warm-up

**1. Your first dict.** Create `student` with keys `"name"`, `"age"` and `"city"` (values up to you). Print the name and age using square brackets, and the number of entries.

**2. Safe reading.** Given `stock = {"apple": 12, "banana": 7}`: print `stock["apple"]`, then fetch `"mango"` with square brackets inside `try/except` to see the `KeyError`, and finally fetch `"mango"` safely twice - once with plain `.get()` and once with a default of `0`.

**3. Add and overwrite.** Start with `profile = {"user": "sarah_dev"}`. Add an `"email"` entry, then overwrite `"user"` with a new handle. Print the dict and its length after each step.

## Part 2 — Practice

**4. Emptying the cache.** Given `cache = {"img1": "blob-a", "img2": "blob-b", "tmp": "junk-data"}`: pop `"tmp"` while keeping its value, then `popitem()` the newest pair, then `del` `"img1"`. Print the dict after each removal.

**5. Loop the menu.** Given `menu = {"coffee": 3.5, "latte": 4.25, "tea": 2.75}`, loop with `.items()` and print one line per drink like `coffee   $3.50`. Then find the most expensive drink with `max(menu, key=menu.get)`.

**6. Keys only.** Given `users = {"sarah": "online", "rafi": "offline"}`, predict and print the results of `"sarah" in users`, `"online" in users` and `"online" in users.values()`. Add a comment explaining what `in` searches by default.

**7. Hashable keys only.** Build one dictionary with a string key, an int key and a tuple key, then look up the tuple entry. Next, try `{[1, 2]: "x"}` inside `try/except` - why does Python refuse?

## Part 3 — Challenge

**8. Mini user database.** Build `users_db` where `"u001"` maps to `{"name": "Sarah", "premium": True, "orders": ["book", "pen"]}` and `"u002"` maps to `{"name": "Rafi", "premium": False, "orders": []}`. Print Sarah's name with chained brackets, append `"lamp"` to Rafi's orders deep inside the dict, then loop over `.items()` printing uid, name and order count.

**9. Merge and group.** Start `prefs = {"theme": "dark"}` and `update` it with `{"theme": "light", "lang": "bn"}` - what happened to theme? Then build a log dictionary: use `setdefault(day, [])` to group the events ("monday", 9), ("monday", 10) and ("tuesday", 8) into per-day lists.
