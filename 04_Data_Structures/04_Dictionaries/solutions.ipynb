## Part 1 — Warm-up

**1. Your first dict.** Square brackets read a value by its key; `len()` counts entries.

```
student = {"name": "Sarah", "age": 22, "city": "Dhaka"}

print(student["name"])    # Sarah
print(student["age"])     # 22
print(len(student))       # 3 entries
```

**2. Safe reading.** `[]` crashes on missing keys; `.get()` degrades gracefully.

```
stock = {"apple": 12, "banana": 7}

print(stock["apple"])          # 12

try:
    stock["mango"]
except KeyError as e:
    print("KeyError:", e)

print(stock.get("mango"))      # None
print(stock.get("mango", 0))   # 0 - our chosen default
```

**3. Add and overwrite.** Assigning to a new key adds an entry; assigning to an existing key replaces it.

```
profile = {"user": "sarah_dev"}

profile["email"] = "sarah@mail.com"   # new key -> added
print(profile, len(profile))

profile["user"] = "sarah2024"         # existing key -> replaced
print(profile, len(profile))
```

## Part 2 — Practice

**4. Emptying the cache.** `pop()` returns the removed value, `popitem()` removes the last-inserted pair, `del` deletes silently.

```
cache = {"img1": "blob-a", "img2": "blob-b", "tmp": "junk-data"}

evicted = cache.pop("tmp")    # remove AND take the value out
print(evicted, "|", cache)

last_in = cache.popitem()     # most recently inserted pair leaves
print(last_in, "|", cache)

del cache["img1"]             # delete by key, returns nothing
print(cache)
```

**5. Loop the menu.** `.items()` hands you `(key, value)` pairs ready to unpack in the loop header.

```
menu = {"coffee": 3.5, "latte": 4.25, "tea": 2.75}

for drink, price in menu.items():
    print(f"{drink:<8} ${price:.2f}")

best = max(menu, key=menu.get)   # rank KEYS by their VALUES
print("most expensive:", best)   # latte
```

**6. Keys only.** `in` scans KEYS; values need the explicit `.values()` view.

```
users = {"sarah": "online", "rafi": "offline"}

print("sarah" in users)             # True  - keys by default
print("online" in users)            # False - values are invisible
print("online" in users.values())   # True  - ask explicitly
```

**7. Hashable keys only.** Keys are hashed, so they must be immutable; lists can never qualify.

```
ok = {"name": "str key", 42: "int key", (1, 2): "tuple key"}
print(ok[(1, 2)])     # tuple key

try:
    {[1, 2]: "x"}     # a list could mutate after insertion
except TypeError as e:
    print("Error:", e)
```

## Part 3 — Challenge

**8. Mini user database.** Chain brackets to descend into nested records; `.items()` drives the report.

```
users_db = {
    "u001": {"name": "Sarah", "premium": True, "orders": ["book", "pen"]},
    "u002": {"name": "Rafi", "premium": False, "orders": []},
}

print(users_db["u001"]["name"])          # descend two levels

users_db["u002"]["orders"].append("lamp")   # mutate deep inside

for uid, info in users_db.items():
    print(uid, info["name"], len(info["orders"]), "order(s)")
```

**9. Merge and group.** `update()` merges (overwriting collisions); `setdefault()` fetch-or-creates the per-day list.

```
prefs = {"theme": "dark"}
prefs.update({"theme": "light", "lang": "bn"})
print(prefs)    # theme got OVERWRITTEN to light, lang added

log = {}
log.setdefault("monday", []).append(9)
log.setdefault("monday", []).append(10)   # same list reused
log.setdefault("tuesday", []).append(8)
print(log)      # {'monday': [9, 10], 'tuesday': [8]}
```
