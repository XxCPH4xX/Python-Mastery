## Part 1 — Warm-up

**1. Make some tuples.** The comma creates the tuple; parentheses are optional decoration.

```
point = (3, 7)
rgb = 255, 144, 0        # no parentheses needed

print(point, type(point))
print(rgb, type(rgb))
```

**2. The one-item trap.** Bare parentheses are grouping; only the comma signals "tuple".

```
print(type((5)))      # <class 'int'>  - surprise!
print(type((5,)))     # <class 'tuple'>
print(len((5,)))      # 1
```

**3. Read-only days.** Reading a tuple works exactly like reading a list.

```
days = ("Sat", "Sun", "Mon")

print(days[0])        # Sat
print(days[-1])       # Mon
print(days[1:])       # ('Sun', 'Mon')
print(len(days))      # 3
print("Mon" in days)  # True
```

## Part 2 — Practice

**4. Immutability, demonstrated.** Item assignment raises `TypeError`; the list round-trip is the escape hatch.

```
days = ("Sat", "Sun", "Mon")

try:
    days[0] = "Tuesday"
except TypeError as e:
    print("Error:", e)

flexible = list(days)      # unfreeze
flexible[0] = "Tuesday"
days = tuple(flexible)     # refreeze
print(days)
```

**5. Swap and star.** Packing sends values into a tuple; unpacking (with `*`) spreads them out.

```
a, b = 1, 2
a, b = b, a            # the famous Pythonic swap
print(a, b)            # 2 1

first, *rest = (10, 20, 30, 40)
print(first)           # 10
print(rest)            # [20, 30, 40]
```

**6. Only two methods.** Immutability leaves exactly `count()` and `index()` behind.

```
results = ("win", "loss", "win", "draw", "win")

print(results.count("win"))    # 3
print(results.index("draw"))   # 3
```

**7. Freeze and unfreeze.** `tuple()` freezes, `list()` unfreezes - conversions go both ways.

```
my_list = [1, 2, 3]

frozen = tuple(my_list)
print(frozen, type(frozen))      # (1, 2, 3) <class 'tuple'>

unfrozen = list(frozen)
print(unfrozen, type(unfrozen))  # [1, 2, 3] <class 'list'>

letters = tuple("abc")
print(letters)                   # ('a', 'b', 'c')
```

## Part 3 — Challenge

**8. Coordinates as dictionary keys.** Tuples are hashable so they work as keys; lists never are.

```
locations = {
    (23.81, 90.41): "Dhaka",
    (40.71, -74.01): "New York",
}
print(locations[(23.81, 90.41)])    # Dhaka

try:
    {[1, 2]: "list key"}
except TypeError as e:
    print("Error:", e)              # unhashable type
```

**9. Return two values at once.** Returning `min(numbers), max(numbers)` builds one tuple that unpacks neatly.

```
def min_max(numbers):
    return min(numbers), max(numbers)

low, high = min_max([14, 3, 27, 9])
print(low, high)    # 3 27
```
