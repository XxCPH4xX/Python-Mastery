# Tuples Exercises

Write your answer in the empty code cell under each question. Check yourself against `solutions.ipynb` when done.

**Part 1** = warm-up · **Part 2** = practice · **Part 3** = challenge.

## Part 1 — Warm-up

**1. Make some tuples.** Create `point = (3, 7)` and an RGB colour holding `255`, `144`, `0` written WITHOUT parentheses (commas only). Print both values and their types.

**2. The one-item trap.** Predict first, then run: what are `type((5))` and `type((5,))`? Print both, plus the length of `(5,)`.

**3. Read-only days.** Given `days = ("Sat", "Sun", "Mon")`, print the first item, the last item, the slice `days[1:]`, the length, and whether `"Mon"` is in it.

## Part 2 — Practice

**4. Immutability, demonstrated.** With `days = ("Sat", "Sun", "Mon")`, try `days[0] = "Tuesday"` inside `try/except` and print the error. Then still get the job done: convert to a list, edit it, and convert back to a tuple.

**5. Swap and star.** Set `a = 1` and `b = 2`, then swap them in ONE line using tuple unpacking. After that, unpack `(10, 20, 30, 40)` into `first` and `rest`, where `rest` collects everything after the first item.

**6. Only two methods.** Given `results = ("win", "loss", "win", "draw", "win")`, use the only two methods a tuple has: count how many wins there are and find the index of `"draw"`.

**7. Freeze and unfreeze.** Turn the list `[1, 2, 3]` into a tuple and back into a list again, printing the type after each conversion. Then build a tuple straight from the string `"abc"`.

## Part 3 — Challenge

**8. Coordinates as dictionary keys.** Build `locations` mapping `(23.81, 90.41)` to `"Dhaka"` and `(40.71, -74.01)` to `"New York"`, then look up Dhaka. Next, try using a LIST `[1, 2]` as a key inside `try/except` - why does Python refuse?

**9. Return two values at once.** Write a function `min_max(numbers)` that returns the smallest and largest value as a tuple. Call it on `[14, 3, 27, 9]` and unpack the result into `low` and `high`.
