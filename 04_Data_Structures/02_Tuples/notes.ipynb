# Python Tuples

> 📘 **Python Mastery** · Module 04 — Data Structures · Lesson 2/5

A tuple is a list that has been sealed shut: ordered and indexable like a list, but impossible to change after creation. That restriction is a feature — it protects data that must not drift (coordinates, RGB colors, database records) and lets Python treat tuples as fixed, hashable values.

## 🎯 Learning Objectives

By the end of this lesson you will be able to:

- **Create** tuples with literals, without parentheses, and via `tuple()`.
- **Explain** immutability and predict the exact `TypeError` you get when trying to modify one.
- **Avoid** the classic `(5)` vs `(5,)` single-item trap.
- **Use packing and unpacking**, including the elegant `a, b = b, a` swap and extended unpacking with `*rest`.
- **Compare** tuples vs lists on memory, speed, data integrity, and hashability.
- **Decide** when a tuple is the right container for a job.

## 1. Creating a Tuple

Tuples are written as comma-separated values — usually wrapped in parentheses for readability. Surprisingly, it is the **comma** that creates the tuple, not the parentheses; they are optional.

**Syntax:**
```python
my_tuple = (v1, v2, v3)     # literal, with parentheses
shorter = v1, v2            # parentheses optional: the comma does the work
empty = ()                  # empty tuple
from_iterable = tuple(x)    # from a list, string, range...
```

````python
point = (3, 7)
rgb = 255, 144, 0             # no parentheses needed - commas make the tuple
singleton = (5,)              # one-item tuple NEEDS the trailing comma
not_a_tuple = (5)             # just the number 5 inside grouping parens

coords = tuple([10.5, 20.5])  # build from a list
letters = tuple("abc")        # build from a string -> ('a', 'b', 'c')

print(point, rgb, singleton)
print(coords, letters)
print(type(not_a_tuple), "<- surprise!")
````

## 2. Immutability — Tuples Never Change

Reading from a tuple works exactly like reading from a list. Writing does not exist: any attempt to assign to an element raises `TypeError`. If you truly need to change the contents, convert to a list, edit, and convert back.

> 🔍 **Under the Hood:** because a tuple can never change, CPython can optimize it hard. The empty tuple is a singleton created once at startup and reused forever, and constant tuples written directly in your source are pre-built by the compiler and stored in the code object — so `(1, 2, 3)` costs nothing to construct at runtime. Lists get none of these benefits precisely because they may change.

**Syntax:**
```python
t[0]                 # reading: fine
t[0] = x             # TypeError: 'tuple' object does not support item assignment
list(t) -> edit -> tuple(...)   # the escape hatch
```

````python
days = ("Sat", "Sun", "Mon")

print(days[0], days[-1], days[1:])   # indexing & slicing behave like lists
print(len(days), "Mon" in days)      # so do len() and membership

try:
    days[0] = "Tuesday"              # forbidden!
except TypeError as e:
    print("Error:", e)

flexible = list(days)      # workaround: list -> edit -> back to tuple
flexible[0] = "Tuesday"
days = tuple(flexible)
print(days)
````

## 3. The One-Item Tuple Trap — `(5,)` Not `(5)`

Beginners write `(5)` expecting a tuple and receive the plain integer `5`. Parentheses here are ordinary math grouping; only the **comma** signals "tuple". This bug hides quietly until something later demands a tuple and gets an `int`.

**Syntax:**
```python
(5)     # int 5          -- parentheses are just grouping
(5,)    # tuple (5,)     -- the comma is what matters
5,      # also a tuple   -- parentheses optional, even for one item
```

````python
tricky = (5)           # an int wearing parentheses
real = (5,)            # a real one-item tuple
also_ok = 5,           # comma alone is enough

print(type(tricky), tricky)
print(type(real), real)
print("length of real:", len(real))

empty_t = ()
print(empty_t, "length:", len(empty_t))
````

## 4. Packing and Unpacking

Writing `x = 1, 2` *packs* values into a tuple; writing `a, b = t` *unpacks* a tuple into separate variables — Python matches them up by position. This powers two idioms you will use constantly: the one-line swap, and returning several values from a function. A starred name collects "everything left over".

**Syntax:**
```python
t = a, b               # packing
a, b = t               # unpacking
first, *rest = t       # extended unpacking (* collects the remainder)
a, b = b, a            # swap with no temp variable
```

````python
position = 12.5, 8.0          # packing into a tuple
x, y = position               # unpacking into two variables
print(x, y)

a, b = 1, 2
a, b = b, a                   # the famous Pythonic swap!
print(a, b)

first, *rest = (10, 20, 30, 40)     # star grabs all remaining items
print(first, rest)

name, *middles, country = "Sarah", "Akter", "Dhaka", "Bangladesh"
print(name, middles, country)

t1, t2 = (1, 2), (3, 4)
merged = (*t1, *t2)                 # merge tuples by unpacking
print(merged)
````

## 5. Tuple Methods — Only `count()` and `index()`

Run `dir(())` and you will find exactly two methods beyond the built-ins: `count(value)` tells you how often a value appears, and `index(value)` gives the position of its first occurrence. No `append`, no `sort`, no `remove` — immutability means there is simply nothing to add or rearrange.

**Syntax:**
```python
t.count(value)   # how many times value appears
t.index(value)   # position of first occurrence (ValueError if missing)
```

````python
results = ("win", "loss", "win", "draw", "win")

print(results.count("win"))       # 3 wins
print(results.index("draw"))      # first position of "draw"
print(len(results), "games total")

print("win" in results)           # membership works normally
print(results[1:4])               # slicing returns a new tuple
print(sorted(results))            # sorted() still works -> returns a LIST
````

## 6. Tuple vs List — Which One and Why

Same data, different guarantees. Tuples use less memory, build faster, and — crucially — are **hashable**, which means a tuple can be used as a dictionary key or inside a set, while a list cannot.

| Aspect | List `[ ]` | Tuple `( )` |
| :--- | :--- | :--- |
| Mutable | yes | no |
| Memory | larger (spare capacity for growth) | smaller |
| Creation speed | slower | faster |
| Dict key / set member | ❌ unhashable | ✅ hashable |
| Best for | collections that grow/change | fixed records, coordinates, keys |

**Syntax:**
```python
import sys
sys.getsizeof([1, 2, 3])   # compare sizes yourself
{(lat, lon): city}         # tuple as dict key
```

````python
import sys, timeit

data_list = [1, 2, 3, 4, 5]
data_tuple = (1, 2, 3, 4, 5)

print(sys.getsizeof(data_list), "bytes for the list")
print(sys.getsizeof(data_tuple), "bytes for the tuple")

t_list = timeit.timeit(stmt="[1, 2, 3, 4, 5]", number=1_000_000)
t_tuple = timeit.timeit(stmt="(1, 2, 3, 4, 5)", number=1_000_000)
print(f"create 1,000,000 lists: {t_list:.3f}s | tuples: {t_tuple:.3f}s")
print("(numbers vary per machine - tuples consistently win)")
````

````python
# Hashability: tuples can be dict keys, lists cannot
locations = {
    (23.81, 90.41): "Dhaka",
    (40.71, -74.01): "New York",
}
print(locations[(23.81, 90.41)])

bad_key = ([1, 2], "oops")        # contains a mutable list
try:
    {bad_key: "nope"}
except TypeError as e:
    print("Error:", e)
````

## 7. Nested Tuples

Tuples nest like lists do — and unpacking makes nested structures pleasant to read. The pattern below (name + grade per student) is a classic fixed-shape record.

**Syntax:**
```python
records = ((a1, b1), (a2, b2))
records[i][j]                  # access
for x, y in records: ...       # unpack straight in the loop
```

````python
students = (
    ("Sarah", 92),
    ("Rafi", 85),
    ("Mina", 78),
)

print(students[0])          # one whole record
print(students[0][1])       # Sarah's grade

for name, grade in students:      # unpack directly in the loop header
    print(name, "->", grade)
````

## 8. Converting Between Lists and Tuples

`tuple()` freezes a list; `list()` unfreezes a tuple. Conversions are shallow — inner containers keep their type — but for flat data this round-trip is the standard "lock / unlock" move.

**Syntax:**
```python
tuple(my_list)   # list -> tuple (freeze)
list(my_tuple)   # tuple -> list (unfreeze)
```

````python
my_list = [1, 2, 3]

frozen = tuple(my_list)        # freeze it
back_to_list = list(frozen)    # unfreeze it
print(frozen, back_to_list)

sentence = "learn python daily"
words = tuple(sentence.split())     # str -> list of words -> tuple
joined = "-".join(words)            # join accepts any iterable of strings
print(words)
print(joined)
````

## 9. When to Choose Which

Rule of thumb: **does this collection need to change?** If yes, use a list. If it represents one fixed thing (a coordinate, a color, a config row), use a tuple and enjoy the safety.

| Situation | Choose |
| :--- | :--- |
| Shopping cart, growing scores, to-do items | **List** |
| GPS coordinates, RGB color, (width, height) | **Tuple** |
| Dictionary key / set member | **Tuple** |
| Function returning multiple values | **Tuple** |
| Data shared across many parts of a program that must stay consistent | **Tuple** |

**Example:**

````python
def min_max(numbers):
    return min(numbers), max(numbers)     # returns ONE tuple of two results

low, high = min_max([14, 3, 27, 9])
print(low, high)

screen = (1920, 1080)      # fixed fact about the world -> tuple
resolution_map = {screen: "Full HD"}
print(resolution_map[(1920, 1080)])
````

## ⚠️ Common Mistakes & Gotchas

| Mistake | Problem | Fix |
| :--- | :--- | :--- |
| `single = (5)` | Grouping parentheses, not a tuple — you silently get an `int` | Write `(5,)` — the comma creates the tuple |
| `t[0] = new_value` | `TypeError`: tuples forbid item assignment | Convert: `t = tuple(list(t)[:1] + [new_value] + ...)` or rebuild |
| Expecting `t.append(x)` to exist | Tuples have only `count()` and `index()` → `AttributeError` | Build a new tuple: `t = t + (x,)`, or reconsider using a list |
| Putting a list inside a tuple used as a dict key | The tuple stays unhashable while it holds a list → `TypeError` | Keep every element immutable (numbers, strings, tuples) |
| Thinking immutability protects *objects inside* the tuple | `t = ([1, 2],)`; then `t[0].append(3)` succeeds! The tuple's references are frozen, not the objects | Treat mutable contents as a design smell; use fully immutable elements |
| `for x, y in points:` when some rows have 3 values | Unpacking mismatch → `ValueError: too many values to unpack` | Match shapes exactly, or use `*rest` for flexible tails |

## 💡 Best Practices & Pro Tips

- Default to a tuple whenever the data has a **fixed shape**: coordinates, dates as `(year, month, day)`, hyperparameter combos. The interpreter enforces your intent for free.
- Return tuples from functions instead of printing results — callers can unpack or ignore parts (`low, _ = min_max(data)`).
- Upgrade path: when a positional tuple becomes hard to read (`row[2]`? what is index 2?), graduate to `collections.namedtuple` or a `dataclass`.
- Use tuples as composite dictionary keys to build lookup tables (e.g. `{(from_lang, to_lang): translator}`).
- 🔬 **AI-engineering relevance:** tensor shapes are always tuples — `x.shape == (32, 128, 768)` — precisely because shapes must be hashable and stable. Model configs, random seeds bundles, and cache keys (e.g. hashing a prompt tuple before calling an LLM API) all rely on tuple immutability.

## 📌 Summary

| Operation | Syntax | Works on tuples? |
| :--- | :--- | :--- |
| Create | `(1, 2)` or `1, 2`; one item needs `(1,)` | ✅ |
| Index / slice | `t[0]`, `t[-1]`, `t[1:3]` | ✅ (slice returns a tuple) |
| Count occurrences | `t.count(v)` | ✅ |
| Find position | `t.index(v)` | ✅ |
| Length / membership | `len(t)`, `v in t` | ✅ |
| Modify an item | `t[0] = 9` | ❌ `TypeError` — immutable |
| Append / remove / sort methods | `append`, `extend`, ... | ❌ don't exist |
| Convert | `list(t)` ↔ `tuple(lst)` | ✅ |
| Pack / unpack | `a, b = b, a`, `first, *rest = t` | ✅ |
| Dict key / set member | `{t: value}` | ✅ if all elements immutable |

**Key takeaways**

- Tuples = ordered + **immutable**; the comma, not the parentheses, makes a tuple.
- Only two methods exist: `count()` and `index()` — immutability leaves nothing else to do.
- Hashable → usable as dict keys and set members; also lighter and faster than lists.
- Use lists for changing collections, tuples for fixed records.
