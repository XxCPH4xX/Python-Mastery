# Read_File Exercises

Write your answer in the empty code cell under each question. Check yourself against `solutions.ipynb` when done.

**Part 1** = warm-up · **Part 2** = practice · **Part 3** = challenge.

## Part 1 — Warm-up

**1. Create, then read it all.** Files can't be read until they exist, so make one first: build a string holding three lines (end each with `\n`), save it with `Path("sample_data", "poem.txt").write_text(text, encoding="utf-8")`, then open it with `with open(...) as f:` and print everything using `.read()`.

**2. Line by line by hand.** Using the same `sample_data/poem.txt` (recreate it from Question 1 if you skipped that), call `.readline()` three times on one open file. Print the raw first line with `repr()` to see its hidden `\n`, the second line tidied with `.strip()`, and the third line shouted with `.upper()`.

**3. All lines at once.** Read `sample_data/poem.txt` with `.readlines()`. Print how many lines came back, then loop over the list with `enumerate(lines, start=1)` printing lines like `1 -> Roses are red` (strip each line).

## Part 2 — Practice

**4. Character budget.** Open `poem.txt` once and make two reads from the same handle: first `.read(11)`, then `.read()` with no argument. Print `repr()` of the short chunk and the length of the remainder. Add a comment naming the hidden thing both reads share (the notes call it the **cursor**).

**5. The missing file.** Wrap an attempt to read `sample_data/unicorn.txt` inside `try`/`except FileNotFoundError` so the program prints a friendly message instead of crashing. Afterwards, ask `Path("sample_data", "unicorn.txt").exists()` what it thinks and print the answer.

**6. Close encounters.** Open `poem.txt` the manual way (no `with`), read it, and print `f.closed` — then close it and print `f.closed` again. Underneath, redo the same read using `with` and print `g.closed` after the block. Finish with a one-line comment on why `with` is the safer habit.

**7. One-line pathlib read.** Skip `open()` entirely: fetch the whole poem with `Path("sample_data", "poem.txt").read_text(encoding="utf-8")`. Print the word count using `len(text.split())` and the last line using index `-1` on `text.splitlines()`.

## Part 3 — Challenge

**8. Stream the sensor log.** First create `sample_data/sensor.log` with these five lines: `08:00 21.4`, `10:00 24.0`, `12:00 27.3`, `14:00 29.9`, `16:00 25.6`. Then read it back using ONLY `for line in f:` (no `.read()`, no `.readlines()`) to find the highest temperature and the total number of readings. Add a comment saying why this style is kinder to huge files.

**9. Encoding round-trip.** Write the note `"Cafe bill: 250 taka ✓"` to `sample_data/note_unicode.txt` twice: once letting the machine guess the encoding (wrap that attempt in `try`/`except UnicodeEncodeError`), and once with `encoding="utf-8"`. Read the UTF-8 version back and print it. Add a comment explaining why spelling out the encoding is the professional habit.
