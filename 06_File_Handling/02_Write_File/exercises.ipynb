## Part 1 — Warm-up

**1. Hello, disk.** Open `sample_data/note.txt` with `with open(..., "w", encoding="utf-8") as f:` and make two separate `.write()` calls to land two lines (you supply the `\n`). Prove it worked by printing `Path("sample_data/note.txt").read_text(encoding="utf-8")`.

**2. Counting characters.** `.write()` hands back the number of characters it wrote. Capture the return value of three writes into `sample_data/counted_note.txt` — the words `"Files "` and `"remember "` plus one `"\\n"` — print the total, and check it against `len()` of the file's contents with `==`.

**3. Numbers are not welcome.** Text mode stores characters only. Try `f.write(95)` inside `try`/`except TypeError` and print the error message. Then write a score properly to `sample_data/score.txt` two ways in one block: `str(score)` glued with `+`, and an f-string like `f"Bonus rate: {rate:.0%}"` for a rate of `0.256`.

## Part 2 — Practice

**4. The silent overwrite trap.** Write the line `Version 1: precious results` (ending with `\n`) to `sample_data/version.txt` using `"w"` mode and print the file. Then open the very same path with `"w"` again and write `Version 2` (also ending with `\n`). Print the file once more and add a comment saying exactly WHEN Version 1 disappeared (it was not during the second `.write()`).

**5. Glued shopping list.** Loop over `items = ["rice", "eggs", "milk"]` writing each one to `sample_data/glued.txt` with NO newline, then print `repr()` of the file to see the damage. Rewrite the list properly (one item per line) into `sample_data/tidy.txt` and print that file too.

**6. `writelines()` does not mean lines.** Call `f.writelines(players)` with `players = ["Sarah", "Tanvir", "Amina"]` on `sample_data/glued_players.txt` and inspect the result with `repr()`. Then fix it with a single `writelines` call that attaches `"\\n"` to every name (the notes show a generator expression), writing to `sample_data/tidy_players.txt`.

**7. Born blank.** Print `Path("sample_data", "auto_created.txt").exists()` BEFORE touching it. Then simply open that path with `"w"` mode and write one line, and check `.exists()` again. Add a comment stating what `"w"` does when the file is missing — and what it does when the file is NOT missing.
