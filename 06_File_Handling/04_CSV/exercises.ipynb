## Part 1 — Warm-up

**1. Build a gradebook.** A CSV is just text: store this in one triple-quoted string — header `Name,Math,Science` followed by rows `Sarah,88,92`, `Tanvir,75,81` and `Amina,95,89` — and save it to `sample_data/class.csv` with `Path(...).write_text(..., encoding="utf-8")`. Print the raw text to see the commas doing their job.

**2. Rows come back as lists.** Open `sample_data/class.csv` with `open(..., newline="", encoding="utf-8")` and loop over `csv.reader(f)`, printing each row and `type(row).__name__`. Notice two things: rows are lists, and the marks are strings, not numbers.

**3. Peel off the header.** Re-read `sample_data/class.csv` with `csv.reader`, but this time call `next(reader)` to capture the header, then `next(reader)` again for the first data row. Print the header, the number of columns (`len(header)`) and the first student's row.

## Part 2 — Practice

**4. Write it properly.** Use `csv.writer` to save this table to `sample_data/subjects.csv`: header `Subject,Credits,Grade`, then rows `Math,4,A`, `Science,3,A-` and `History,2,B+`. Emit the header with `writerow()` and the rest with a single `writerows()`. Remember `newline=""` — then print the file's text.

**5. Commas inside values.** Write `[["City", "Country"], ["Chattogram", "Bangladesh"], ["Kolkata", "India, West Bengal"]]` to `sample_data/cities.csv` with `csv.writer` and print the raw text. Spot the quotes the writer added around one value, and add a comment on why building that row with `",".join(...)` would have corrupted the file.

**6. DictReader by name.** Loop over `sample_data/class.csv` with `csv.DictReader` and print each student's name and Math mark — accessing them by column NAME, not index. Convert marks with `int()` and, after the loop, print the class's average Math mark rounded to 1 decimal place.

**7. Why `newline=""` matters.** Write the two-row table `[["City", "Population"], ["Dhaka", 22479000]]` to `sample_data/no_flag.csv` WITHOUT `newline=""` and to `sample_data/with_flag.csv` WITH it. Print both files with `repr(Path(...).read_bytes())` and compare. Add a comment on what Windows inserts without the flag and what that looks like in Excel.

## Part 3 — Challenge

**8. Export averages with `DictWriter`.** Starting from `sample_data/class.csv`, compute each student's average across Math and Science (convert every mark with `int()` and round the average to 1 decimal place). Collect the results as dicts with keys `Student` and `Average`, then export to `sample_data/student_averages.csv` with `csv.DictWriter` — `fieldnames=["Student", "Average"]`, `writeheader()` first, then `writerows()`. Print the finished file.

**9. Delimiter detective.** Write the table `[["name", "score"], ["Sarah", 88], ["Tanvir", 75]]` twice: as tab-separated values to `sample_data/scores.tsv` (`delimiter="\\t"`) and as semicolon-separated values to `sample_data/scores_semicolon.csv`. Print both files with `repr()` to see the different separators, then read the semicolon file back with the MATCHING delimiter and print its rows. Add a comment naming the delimiter many European spreadsheet exports use instead of a comma.
