## Part 1 — Warm-up

**1. Gentle append.** Create `sample_data/diary.txt` with one line (`"Day 1: started file handling\\n"`) using `"w"` mode. Then append `"Day 2: learned append mode\\n"` with `"a"` mode in a SEPARATE `with` block. Print the file — Day 1 must still be there.

**2. Append creates missing files too.** Append a line to `sample_data/fresh_log.txt` — a file that does not exist yet. Print whether it exists afterwards and print its contents. Add a comment naming the OTHER mode you know that also creates missing files (and what that mode does to existing ones).

**3. The welded note.** Append `"buy milk"` and then `"call the plumber"` to `sample_data/sticky.txt` — deliberately forgetting `"\\n"`. Print `repr()` of the file to witness the weld. Repair it by appending a lone `"\\n"` plus one clean entry that DOES end with `"\\n"`, then print each line via `.splitlines()`.

## Part 2 — Practice

**4. Three runs, one log.** Simulate three program runs: loop over the three strings `"2026-08-24 | studied loops"`, `"2026-08-25 | studied functions"` and `"2026-08-26 | studied file handling"`, appending EACH one in its own `with` block to `sample_data/study_log.txt`. Print the finished log and add a comment on why the earlier entries survived.

**5. The invisible entry.** Open `sample_data/cursor.txt` with mode `"a+"`. Write one line, then immediately print `repr(f.read())` (spoiler: it is empty!) and `f.tell()`. Now rewind with `f.seek(0)` and print the whole file through the same handle. Add a comment explaining what the cursor was doing.

**6. `x` marks the exclusive.** Try to create `sample_data/exam_submission.txt` with mode `"x"` and write `"final answers, locked forever\\n"` — catching `FileExistsError` so a second attempt cannot destroy the first. Make the attempt twice (a loop over `(1, 2)` works nicely) and finish by printing the file's content to prove the original survived.

## Part 3 — Challenge

**7. A tiny event logger.** Define `log_event(message)` that appends a line shaped `[2026-08-26] message` to `sample_data/events.log` (end every line with `"\\n"`). Call it three times with different messages, then print the log by STREAMING it with `for line in f:` and `enumerate(..., start=1)` so entries come out numbered.

**8. High-score table, three modes.** Work on `sample_data/scores.txt` in stages: (1) seed it with `"w"` so reruns start clean, writing `Sarah 120` and `Tanvir 95` on separate lines; (2) append `Amina 150` and `Rafi 60` with `"a"`; (3) read all lines, sort them by score descending (highest first), and REWRITE the file with `"w"` so it becomes a ranked leaderboard. Print the final file. Add a comment: which mode would you use to protect a snapshot from EVER being overwritten?
