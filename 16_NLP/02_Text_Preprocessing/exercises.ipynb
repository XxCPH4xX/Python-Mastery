## Part 1 — Warm-up

**1. Peel the packaging.** Clean this scraped review in four stages, printing after each: strip HTML tags with `re.sub(r"<[^>]+>", " ", s)`, strip URLs with `re.sub(r"https?://\S+|www\.\S+", " ", s)`, decode entities with `html.unescape`, then collapse whitespace runs and strip.
Start from: `scraped = "<p>Screen cracked within <b>two days</b>!</p> Details: http://shop.example.com/r42 &amp; photos."`

**2. Three tokenizers.** Tokenize `sentence = "Don't panic: the refund arrived, and it's correct!"` three ways: plain `.split()`, `re.findall(r"\w+", ...)`, and `re.findall(r"\w+(?:'\w+)?|[^\w\s]", ...)`. Print each list, then comment on what happened to `don't` and `it's` in each.

**3. Expand contractions.** Build a small `CONTRACTIONS` dict mapping tokens to LISTS of words (include `don't`, `won't`, `isn't`, `wouldn't`, `it's`, `i'm`). Write `expand(tokens)` that replaces each token with its expansion (or itself), and run it on `"don't worry: it's working, she wouldn't lie".split()`. Why did `worry:` survive untouched? Comment.

## Part 2 — Practice

**4. Case decisions.** Lowercase `pair_a = "I work at Apple."` and `pair_b = "I eat an apple daily."` and note in a comment what lowercasing destroys in the first. Then test `"NOT HAPPY WITH THIS".isupper()`. Finally prove `"Straße".casefold() == "STRASSE".casefold()` while plain `.lower()` fails to match them.

**5. The stopword trap.** With `STOPWORDS` containing `"this", "is", "phone"`'s neighbours (the, is, this, was, ...) INCLUDING `not`: filter `positive = "this phone is good"` and `negative = "this phone is not good"` and print both results — they come out identical! Then repair it: build `safe_stopwords = STOPWORDS - NEGATION_KEEP` with `NEGATION_KEEP = {"not", "no", "never"}` and re-filter the negative sentence.

**6. Build a mini stemmer.** Define the ordered rule list `STEM_RULES` (from the notes: `("ies","y")` must fire BEFORE `("es","")`). Implement `mini_stem(word)` that applies the FIRST matching suffix rule but only if the remaining stem keeps at least 3 letters. Run it on `["playing", "studies", "scratches", "studied", "university", "universes"]` and comment on the `university`/`universes` collision.

**7. Lemmatize by lookup.** Keep `mini_stem` from the previous question. Build `LEMMA_MAP` with irregular pairs (`ran`→`run`, `better`→`good`, `mice`→`mouse`, `arrived`→`arrive`, `happily`→`happy`, `went`→`go`) and write `lemmatize(word)` returning the map hit or falling back to `mini_stem`. Print original/stemmed/lemmatized rows for `["ran", "better", "mice", "studied", "arrived", "happily"]`.
