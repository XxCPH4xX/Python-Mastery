## Part 1 — Warm-up

**1. Your first array.** Convert the list `[10, 20, 30, 40]` into a NumPy array called `prices`, then double EVERY element with a single expression (no loop!). Print the array before and after.

**2. Inspect the array.** Given `grid = np.array([[78, 85, 90], [62, 71, 88]])`, print its `ndim`, `shape`, `size`, `dtype` and `itemsize`. Predict each value before printing.

**3. Blank canvases.** Create: a 2x3 array of zeros, a length-4 array of ones, a 2x2 array filled entirely with `7`, and the 3x3 identity matrix. Print all four.

## Part 2 — Practice

**4. arange workout.** Using only `np.arange`, build four arrays: `0` to `4`, `2` to `7`, every 5th number from 0 up to 20, and a countdown `10, 8, 6, 4, 2`. Print each.

**5. linspace workout.** Build exactly 5 evenly spaced points from 0 to 10 with the endpoint included, then 5 points from 0 to 1 with the endpoint EXCLUDED (`endpoint=False`). Print both.

**6. arange vs linspace showdown.** Run `np.arange(0, 1, 0.1)` and `np.linspace(0, 1, 11)`. Print each with its `.size`. In a comment: which one guarantees that `1.0` appears, and why?

**7. Vectorize the VAT.** Given `prices_list = [120, 250, 80, 430, 95]`: add 15% VAT twice - once with a list comprehension, once by converting to an array and writing ONE expression. Then add the fees `np.array([5, 10, 5, 10, 5])` to the array element-wise. Round the printed arrays to 2 decimals with `np.round(..., 2)`.

## Part 3 — Challenge

Time each with `time.perf_counter()` and print all three durations in milliseconds plus how many times faster NumPy was than the loop.

**9. Image batch inspector.** Create `batch = np.zeros((4, 28, 28))` - four grayscale 28x28 images. Print its `ndim`, `shape`, `size` and `nbytes`. Then create a length-3 array of ones with `dtype=bool` and a 2x2 array of zeros with `dtype=np.float32`, printing both arrays and their dtypes.
