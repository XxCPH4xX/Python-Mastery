## Part 1 — Warm-up

**1. Your first array.** One operator applied to the whole array replaces the whole loop.

```
import numpy as np

prices = np.array([10, 20, 30, 40])
print(prices)
print(prices * 2)      # every element doubled at once
```

**2. Inspect the array.** These five attributes are the first thing to check whenever array code misbehaves.

```
import numpy as np

grid = np.array([[78, 85, 90],
                 [62, 71, 88]])

print("ndim    :", grid.ndim)      # 2
print("shape   :", grid.shape)    # (2, 3)
print("size    :", grid.size)     # 6
print("dtype   :", grid.dtype)     # int64
print("itemsize:", grid.itemsize)  # 8 bytes per element
```

```
import numpy as np

print(np.zeros((2, 3)))
print()
print(np.ones(4))
print()
print(np.full((2, 2), 7))
print()
print(np.eye(3))
```

## Part 2 — Practice

**4. arange workout.** Start included, stop EXCLUDED, negative steps count down.

```
import numpy as np

print(np.arange(5))           # [0 1 2 3 4]
print(np.arange(2, 8))        # [2 3 4 5 6 7]
print(np.arange(0, 21, 5))    # [ 0  5 10 15 20]
print(np.arange(10, 0, -2))   # [10  8  6  4  2]
```

**5. linspace workout.** `linspace` thinks in COUNTS, not steps - and includes the stop by default.

```
import numpy as np

print(np.linspace(0, 10, 5))                  # 10 IS included
print(np.linspace(0, 1, 5, endpoint=False))   # stop excluded
```

**6. arange vs linspace showdown.** Fractional steps make arange's endpoint depend on rounding; linspace promises the endpoints.

```
import numpy as np

risky = np.arange(0, 1, 0.1)
safe = np.linspace(0, 1, 11)

print(risky, "->", risky.size, "values")
print(safe)
# linspace wins for fractional spacing: 1.0 is GUARANTEED included,
# while arange's floating-point step may stop early.
```

**7. Vectorize the VAT.** Whole-array expressions replace comprehensions - and array + array is pairwise math, not concatenation.

```
import numpy as np

prices_list = [120, 250, 80, 430, 95]
prices = np.array(prices_list)

with_vat_list = [p * 1.15 for p in prices_list]   # the old habit
with_vat_np = prices * 1.15                       # one expression

print(np.round(with_vat_list, 2))
print(np.round(with_vat_np, 2))

fees = np.array([5, 10, 5, 10, 5])
print(prices + fees)      # element-wise, position by position
```

## Part 3 — Challenge

**8. Speed race: loop vs NumPy.** Compiled C loops over contiguous memory beat Python objects by orders of magnitude.

```
import time
import numpy as np

N = 1_000_000
py_list = list(range(N))
np_array = np.arange(N)

start = time.perf_counter()
total = 0
for x in py_list:
    total += x
loop_ms = (time.perf_counter() - start) * 1000

start = time.perf_counter()
total = sum(py_list)
builtin_ms = (time.perf_counter() - start) * 1000

start = time.perf_counter()
total = np.sum(np_array)
numpy_ms = (time.perf_counter() - start) * 1000

print(f"for-loop : {loop_ms:8.2f} ms")
print(f"sum()    : {builtin_ms:8.2f} ms")
print(f"np.sum() : {numpy_ms:8.2f} ms")
print(f"NumPy was roughly {loop_ms / max(numpy_ms, 1e-9):.0f}x faster than the loop")
```

**9. Image batch inspector.** Creation helpers take any shape tuple, and `dtype=` controls memory and precision.

```
import numpy as np

batch = np.zeros((4, 28, 28))
print("ndim  :", batch.ndim)     # 3
print("shape :", batch.shape)   # (4, 28, 28)
print("size  :", batch.size)    # 3136
print("nbytes:", batch.nbytes)  # 3136 * 8 bytes

flags = np.ones(3, dtype=bool)
light = np.zeros((2, 2), dtype=np.float32)
print(flags, flags.dtype)
print(light.dtype)
```
