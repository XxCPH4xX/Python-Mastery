## Part 1 — Warm-up

**1. Element-wise math.** Every operator applies position by position; scalars hit every element.

```
import numpy as np

a = np.array([10, 20, 30])
b = np.array([1, 2, 3])

print(a + b)
print(a * b)      # pairwise product, NOT matrix multiplication
print(a ** 2)
print(a % 7)
print(a + 100)
```

**2. Universal functions.** Ufuncs run compiled C loops per element; `round` uses banker's rounding (ties go to the EVEN side).

```
import numpy as np

squares = np.array([0, 4, 9, 16, 25])
print(np.sqrt(squares))
print(np.abs(np.array([-5, 3, -0.5])))

values = np.array([2.1, 2.5, 2.7, 3.5, -2.5])
print("round :", np.round(values))   # .5 ties -> even neighbor
print("floor :", np.floor(values))
print("ceil  :", np.ceil(values))
```

**3. One number per array.** Aggregations collapse the whole array into one summary number.

```
import numpy as np

salaries = np.array([35, 48, 52, 61, 90])

print("sum :", salaries.sum())
print("min :", salaries.min(), "| max :", salaries.max())
print("mean:", salaries.mean())
print("var :", salaries.var())
print("std :", round(salaries.std(), 2))
```

## Part 2 — Practice

**4. Masks are numbers too.** `True == 1`, so summing a mask counts and averaging gives a rate.

```
import numpy as np

scores = np.array([45, 82, 67, 91, 58])

passed = scores >= 60
print(passed)
print("passed count:", passed.sum())
print("pass rate   :", passed.mean())
```

**5. Axis demystified.** The axis argument is the dimension that DISAPPEARS: axis=0 gives one value per column, axis=1 one per row.

```
import numpy as np

m = np.arange(12).reshape(3, 4)
print(m)
print()

print(m.sum(), "-> shape", m.sum().shape)          # one number
print(m.sum(axis=0), "-> shape", m.sum(axis=0).shape)  # per column
print(m.sum(axis=1), "-> shape", m.sum(axis=1).shape)  # per row
# axis=0 collapsed the rows away; axis=1 collapsed the columns away.
```

**6. Outlier-proof statistics.** Means get dragged by extremes; medians and percentiles stay put.

```
import numpy as np

skewed = np.array([35, 48, 52, 61, 90, 95, 300])

print("mean  :", skewed.mean())          # dragged way up
print("median:", np.median(skewed))     # robust
print(np.percentile(skewed, [25, 50, 90]))
```

**7. Where is the extreme.** `argmax` returns the POSITION of the maximum - perfect for looking up a parallel labels array.

```
import numpy as np

votes = np.array([120, 340, 275, 90])
candidates = np.array(["Amina", "Rafi", "Karim", "Nadia"])

winner_idx = votes.argmax()
print("winner:", candidates[winner_idx], "with", votes.max(), "votes")
print("loser :", candidates[votes.argmin()])

daily = np.array([12, 9, 15, 7, 11])
print(daily.cumsum())
```

## Part 3 — Challenge

**8. Yes-or-no sanity checks.** `.any()` and `.all()` collapse an array of verdicts into one decision.

```
import numpy as np

temps = np.array([31.5, 33.8, 30.2, 34.1])

print("any above 33?    ", (temps > 33).any())
print("all above 30?    ", (temps > 30).all())
print("any below 30?    ", (temps < 30).any())

readings = np.array([3.2, 0.5, 1.1])
print("all non-negative?", (readings >= 0).all())
```

**9. NaN hygiene.** NaN spreads through every calculation; nan-aware functions skip it automatically.

```
import numpy as np

readings = np.array([36.5, np.nan, 37.1, 36.9])

print(readings.mean())     # nan - contagious!
print(np.isnan(readings))
clean_mean = readings[~np.isnan(readings)].mean()
print("DIY clean mean:", clean_mean)
print("np.nanmean:", np.nanmean(readings))
print("np.nansum :", np.nansum(readings))
```
