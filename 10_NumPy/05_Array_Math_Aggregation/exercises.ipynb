## Part 1 — Warm-up

**1. Element-wise math.** Given `a = np.array([10, 20, 30])` and `b = np.array([1, 2, 3])`, print `a + b`, `a * b`, `a ** 2`, `a % 7` and `a + 100`. Is `a * b` matrix multiplication?

**2. Universal functions.** Given `squares = np.array([0, 4, 9, 16, 25])`, print `np.sqrt` of it and `np.abs` of `[-5, 3, -0.5]`. Then round `values = np.array([2.1, 2.5, 2.7, 3.5, -2.5])` with `np.round` - why do the `.5` ties go where they go? Compare with `np.floor` and `np.ceil`.

**3. One number per array.** Given `salaries = np.array([35, 48, 52, 61, 90])` (thousands), print the sum, min, max, mean, variance and standard deviation (std rounded to 2 decimals).

## Part 2 — Practice

**4. Masks are numbers too.** Given `scores = np.array([45, 82, 67, 91, 58])`: print the boolean array `scores >= 60`, then use `.sum()` on the mask to count passes and `.mean()` on it to get the pass rate. How many students passed?

**5. Axis demystified.** Build `m = np.arange(12).reshape(3, 4)`. Print the matrix, then `m.sum()` with no axis, `m.sum(axis=0)` and `m.sum(axis=1)` - each with its result's `.shape`. In comments: which direction did each axis collapse?

**6. Outlier-proof statistics.** Given `skewed = np.array([35, 48, 52, 61, 90, 95, 300])` (one CEO salary ruins everything): print the mean, the median (`np.median`), and the 25th/50th/90th percentiles via `np.percentile(skewed, [25, 50, 90])`. Which summary survives the outlier better?

**7. Where is the extreme.** Given `votes = np.array([120, 340, 275, 90])` and `candidates = np.array(["Amina", "Rafi", "Karim", "Nadia"])`: find the winning INDEX with `.argmax()`, then the winner's name and vote count. Also find the loser with `.argmin()`. Finally print the running total of `daily = np.array([12, 9, 15, 7, 11])` sales with `.cumsum()`.

## Part 3 — Challenge

**8. Yes-or-no sanity checks.** Given `temps = np.array([31.5, 33.8, 30.2, 34.1])`, answer three questions with `.any()`/`.all()`: is ANY temp above 33? Are ALL above 30? Is any below 30? Then check the safety rule `readings = np.array([3.2, 0.5, 1.1])`: are all readings non-negative?

**9. NaN hygiene.** Given `readings = np.array([36.5, np.nan, 37.1, 36.9])`: print `readings.mean()` (poisoned!), the `np.isnan` map of missing positions, the DIY clean mean via `readings[~np.isnan(readings)]`, and finally `np.nanmean` and `np.nansum`. What do the nan-family functions buy you?
