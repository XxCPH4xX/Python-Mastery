## Part 1 — Warm-up

**1. What type did NumPy pick.** Create `a = np.array([1, 2, 3])`, `b = np.array([1.0, 2.5])` and `c = np.array([True, False, True])`. Print each array with its `dtype` and notice what NumPy guessed for each.

**2. Force the dtype.** Build a float32 array from `[1, 2, 3]` by passing `dtype=` at creation, and a length-3 array of ones with `dtype=bool`. Print both arrays and their dtypes.

**3. astype: convert (but carefully).** Given `prices = np.array([19.99, 5.49, 8.75])`, convert to int two ways: directly with `.astype(int)` (watch it truncate toward zero!), and properly with `np.round(prices).astype(int)`. Then convert the strings `raw = np.array(["42", "17", "93"])` to ints and add 8 to every one.

## Part 2 — Practice

**4. Overflow watch.** Create `tiny = np.array([100, 126, 127], dtype=np.int8)` and print `tiny + 1` - some values wrapped silently! Then brighten `pixels = np.array([200, 240, 250], dtype=np.uint8)` by adding 60 two ways: directly (broken) and after widening with `.astype(np.int16)` (correct). Print all results.

**5. Assignment is not copying.** Set `a = np.array([1, 2, 3, 4])` and `b = a`. Check whether they are the same object with `a is b`. Then set `b[0] = 99` and print `a` - explain in a comment what happened.

**6. view(): new wrapper, same bytes.** Take `a = np.array([1, 2, 3, 4])` and create `v = a.view()`. Print `v is not a`, `np.shares_memory(a, v)` and `v.base is a`. Then set `v[1] = 500` and print `a`. What is shared, and what is not?

**7. copy(): fully independent.** Take `a = np.array([1, 2, 3, 4])` and create `c = a.copy()`. Print `np.shares_memory(a, c)` and whether `c.base is None` (a true copy owns its data). Set `c[0] = 777` and prove `a` survived untouched.
