## Part 1 — Warm-up

**1. What type did NumPy pick.** Whole numbers become int64, decimals become float64 - the defaults you will see constantly.

```
import numpy as np

a = np.array([1, 2, 3])
b = np.array([1.0, 2.5])
c = np.array([True, False, True])

print(a, a.dtype)    # int64
print(b, b.dtype)    # float64
print(c, c.dtype)    # bool
```

**2. Force the dtype.** `dtype=` works in `np.array`, `zeros`, `ones` - everywhere.

```
import numpy as np

light = np.array([1, 2, 3], dtype=np.float32)
flags = np.ones(3, dtype=bool)

print(light, light.dtype)
print(flags, flags.dtype)
```

**3. astype: convert (but carefully).** `astype` truncates floats; round explicitly when that matters.

```
import numpy as np

prices = np.array([19.99, 5.49, 8.75])
print(prices.astype(int))             # truncates!
print(np.round(prices).astype(int))   # round first, then cast

raw = np.array(["42", "17", "93"])
nums = raw.astype(int)
print(nums + 8)
```

## Part 2 — Practice

**4. Overflow watch.** Fixed-size types WRAP instead of growing - widen before arithmetic near the limits.

```
import numpy as np

tiny = np.array([100, 126, 127], dtype=np.int8)   # int8 max is 127
print(tiny + 1)        # wrapped around, no warning!

pixels = np.array([200, 240, 250], dtype=np.uint8)   # max 255
print(pixels + 60)                    # naive: wraps past 255
print(pixels.astype(np.int16) + 60)   # widened first: correct
```

**5. Assignment is not copying.** `b = a` creates a second NAME, not a second array.

```
import numpy as np

a = np.array([1, 2, 3, 4])
b = a                  # NO copy: another name for the same object

print("same object?", a is b)   # True
b[0] = 99
print(a)    # changed through b - there was only ever one array
```

**6. view(): new wrapper, same bytes.** A view is a NEW array object aimed at the SAME data buffer.

```
import numpy as np

a = np.array([1, 2, 3, 4])
v = a.view()

print("different objects? ", v is not a)
print("shares memory?    ", np.shares_memory(a, v))
print("v.base is a?      ", v.base is a)

v[1] = 500
print(a)    # data is shared, so a changed
```

**7. copy(): fully independent.** Only `.copy()` allocates a second buffer - nothing shared afterwards.

```
import numpy as np

a = np.array([1, 2, 3, 4])
c = a.copy()

print("shares memory?", np.shares_memory(a, c))   # False
print("c owns its data?", c.base is None)         # True

c[0] = 777
print("a unchanged :", a)
print("c modified  :", c)
```

## Part 3 — Challenge

**8. View-or-copy detective.** Basic slicing -> view; masks, fancy indexing and arithmetic -> copies.

```
import numpy as np

a = np.arange(6)

candidates = [("slice  a[1:5]", a[1:5]),
              ("fancy  a[[1,2,3]]", a[[1, 2, 3]]),
              ("mask   a[a>2]", a[a > 2]),
              ("arith  a+100", a + 100)]

for label, candidate in candidates:
    print(label, "-> shares memory:", np.shares_memory(a, candidate))

# Rule: a regular patch of existing memory is a VIEW;
# anything gathered arbitrarily (masks, fancy, math) is a COPY.
```

**9. Backup done right.** Widen BEFORE arithmetic near a type limit, and keep backups with `.copy()`.

```
import numpy as np

data = np.array([200, 210, 220], dtype=np.uint8)
backup = data.copy()                 # real copy this time

naive = data + 60                    # uint8 math: wraps around 255
print("naive :", naive)

safe = data.astype(np.int16) + 60    # widen first, then add
print("safe  :", safe)

print("backup:", backup, "<- untouched")
```
