## Part 1 — Warm-up

**1. Reshape basics.** Reshape reinterprets the SAME buffer - usually returning a view.

```
import numpy as np

flat = np.arange(12)
grid = flat.reshape(3, 4)

print(grid)
print("shape:", grid.shape)
print("shares memory?", np.shares_memory(flat, grid), "- it is a view!")
```

**2. The wildcard -1.** `-1` means "you compute this dimension" - allowed exactly once.

```
import numpy as np

readings = np.arange(24)

print(readings.reshape(4, -1).shape)   # (4, 6)
print(readings.reshape(-1, 8).shape)   # (3, 8)

images = readings.reshape(2, 3, 4)
back = images.reshape(-1)
print(images.shape, "->", back.shape)
```

**3. Transpose it.** `.T` flips the axes as a view; a 1-D vector's transpose is itself.

```
import numpy as np

grades = np.array([[78, 85, 90],
                   [62, 71, 88]])

print(grades.T)
print("shape:", grades.T.shape)
print("still a view?", np.shares_memory(grades, grades.T))

v = np.array([1, 2, 3])
print(v.T.shape, "<- unchanged: 1-D transposes to itself")
```

## Part 2 — Practice

**4. ravel vs flatten.** `ravel` returns a view when possible; `flatten` always copies.

```
import numpy as np

grid = np.arange(6).reshape(2, 3)
flat_view = grid.ravel()      # window onto grid's memory
flat_view[0] = 99
print(grid)                   # grid changed!

grid = np.arange(6).reshape(2, 3)
flat_copy = grid.flatten()    # fresh independent buffer
flat_copy[0] = 99
print(grid)                   # untouched
```

**5. Add an axis.** `newaxis` changes rank, not data - crucial for broadcasting and matmul.

```
import numpy as np

v = np.array([1.0, 2.0, 3.0])
print("vector   :", v.shape)

col = v[:, np.newaxis]
row = v[np.newaxis, :]
print("column   :", col.shape)
print("row      :", row.shape)
print("squeezed :", np.squeeze(row).shape)
```

**6. Stack them together.** Reshaping changes one array; stacking JOINS arrays - `stack` adds a brand-new axis.

```
import numpy as np

week1 = np.array([31.5, 32.0, 33.8])
week2 = np.array([30.2, 29.9, 34.1])

print(np.vstack([week1, week2]))
print(np.hstack([week1, week2]))

a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])

print(np.concatenate([a, b], axis=0))
print()
print(np.concatenate([a, b], axis=1))
layers = np.stack([a, b])
print("stacked shape:", layers.shape)
```

**7. Broadcast or break.** Right-align shapes: every dimension must match or be 1.

```
import numpy as np

M = np.arange(12).reshape(3, 4)

bonus = np.array([10, 20, 30, 40])       # (4,) fits the last axis
col = np.array([[100], [200], [300]])    # (3, 1) stretches down rows
print((M + bonus).shape, (M + col).shape)

vec = np.array([1, 2, 3])                # (3,)
try:
    M + vec                              # last axis: 4 vs 3 clash
except ValueError as err:
    print("ValueError:", err)

print(M + vec[:, np.newaxis])            # (3, 1) broadcasts fine
```

## Part 3 — Challenge

**8. Normalize the columns.** Column stats come out shape `(n_features,)` and broadcast down every row automatically.

```
import numpy as np

X = np.array([[1200, 3, 85],
              [1450, 4, 110],
              [900, 2, 60],
              [1600, 3, 130]], dtype=float)

means = X.mean(axis=0)
stds = X.std(axis=0)
X_norm = (X - means) / stds     # broadcasting does the rest

print(np.round(X_norm, 2))
print("column means:", np.round(X_norm.mean(axis=0), 6))
print("column stds :", np.round(X_norm.std(axis=0), 6))
```

**9. From vector to batch.** Adding a batch axis turns one-off math into batched matrix multiplication - the shape ML models actually use.

```
import numpy as np

W = np.arange(6, dtype=float).reshape(2, 3)
x = np.array([1.0, 2.0, 3.0])

print("W @ x       =", W @ x)               # (2,3) @ (3,) -> (2,)

X_batch = x[np.newaxis, :]                  # (1, 3)
out = X_batch @ W.T                         # (1,3) @ (3,2) -> (1, 2)
print("batch out   =", out, "shape:", out.shape)
```
