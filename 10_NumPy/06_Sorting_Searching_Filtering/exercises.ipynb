## Part 1 — Warm-up

**1. Copy vs in-place sort.** Given `times = np.array([14.2, 11.9, 13.5, 10.8])`: sort with `np.sort(times)` and print both the result AND `times` afterwards. Then call `times.sort()` and print `times` again. What is the difference between the two spellings?

**2. Descending order.** With `scores = np.array([78, 92, 61, 88, 95, 73])`, produce a DESCENDING sorted array WITHOUT destroying `scores` (hint: sort a copy, then reverse with `[::-1]`). Print both the descending result and the original.

**3. Sort rows and columns.** Given `m = np.array([[9, 3, 7], [4, 8, 1], [6, 5, 2]])`, print `np.sort(m, axis=1)` (each row sorted) and `np.sort(m, axis=0)` (each column sorted). Remember which axis collapses!

## Part 2 — Practice

**4. argsort: rank the runners.** Given `names = np.array(["Rafi", "Nadia", "Omar", "Priya"])` and `points = np.array([72, 91, 65, 88])`: compute `order = np.argsort(points)`, then use it to print the points AND the names in ascending-score order. What does `argsort` actually return - values or positions?

**5. Top-3 leaderboard.** Given `scores = np.array([78, 92, 61, 88, 95, 73])`, find the positions of the three biggest scores with `np.argsort(scores)[-3:][::-1]` and print their positions and their scores, best first.

**6. The three faces of where.** With `temps = np.array([31.5, 33.8, 30.2, 34.1])`, use `np.where(temps > 32)` to get hot-day POSITIONS (remember to unwrap with `[0]`) and print the corresponding temperatures. Then with `scores = np.array([45, 82, 67, 91, 58])`, build PASS/FAIL verdicts with the two-value form, and clamp `signal = np.array([3, -1, 7, -5, 2])` with `np.where(signal < 0, 0, signal)`.

**7. Instant frequency table.** Given `votes = np.array([2, 3, 2, 1, 3, 2, 4, 1, 2])` (votes for candidates 1-4), use `np.unique(votes, return_counts=True)` and print one line per candidate. Then announce the winner via `counts.argmax()` indexing back into the candidates array.

## Part 3 — Challenge

With the sorted `queue = np.array([10, 20, 30, 40, 50])`, use `np.searchsorted(queue, 35)` to find where 35 belongs, insert it with `np.insert`, and compare `side="left"` vs `side="right"` for the value 20 in `np.array([10, 20, 20, 30])`.
