## Part 1 — Warm-up

**1. Copy vs in-place sort.** `np.sort(arr)` returns a new sorted COPY; `arr.sort()` reorders in place and destroys the original order.

```
import numpy as np

times = np.array([14.2, 11.9, 13.5, 10.8])

ordered = np.sort(times)      # sorted COPY
print(ordered)
print(times, "<- untouched")

times.sort()                  # IN PLACE, returns None
print(times)                  # order permanently changed
```

**2. Descending order.** NumPy sorts ascending; reversing the sorted copy gives descending.

```
import numpy as np

scores = np.array([78, 92, 61, 88, 95, 73])

descending = np.sort(scores)[::-1]
print(descending)
print(scores, "<- original intact")
```

**3. Sort rows and columns.** `axis=1` travels along each row; `axis=0` travels down each column.

```
import numpy as np

m = np.array([[9, 3, 7],
              [4, 8, 1],
              [6, 5, 2]])

print(np.sort(m, axis=1))
print()
print(np.sort(m, axis=0))
```

## Part 2 — Practice

**4. argsort: rank the runners.** `argsort` hands back INDICES; index ANY parallel array with them.

```
import numpy as np

names = np.array(["Rafi", "Nadia", "Omar", "Priya"])
points = np.array([72, 91, 65, 88])

order = np.argsort(points)
print("ranking order :", order)
print("sorted points :", points[order])
print("sorted names  :", names[order])
```

**5. Top-3 leaderboard.** argsort ascending, take the tail, reverse - the classic top-k pattern.

```
import numpy as np

scores = np.array([78, 92, 61, 88, 95, 73])

idx_top = np.argsort(scores)[-3:][::-1]
print("top-3 positions:", idx_top)
print("their scores    :", scores[idx_top])
```

**6. The three faces of where.** Form 1 finds indices (a tuple!), form 2 is a vectorized if/else.

```
import numpy as np

temps = np.array([31.5, 33.8, 30.2, 34.1])
hot_idx = np.where(temps > 32)[0]     # unwrap the tuple
print("hot day indices  :", hot_idx)
print("hot temperatures :", temps[hot_idx])

scores = np.array([45, 82, 67, 91, 58])
print(np.where(scores >= 60, "PASS", "FAIL"))

signal = np.array([3, -1, 7, -5, 2])
print(np.where(signal < 0, 0, signal))
```

**7. Instant frequency table.** `unique + return_counts` is a value_counts with zero dependencies.

```
import numpy as np

votes = np.array([2, 3, 2, 1, 3, 2, 4, 1, 2])

candidates, counts = np.unique(votes, return_counts=True)
for c, n in zip(candidates, counts):
    print(f"candidate {c}: {n} votes")

winner = candidates[counts.argmax()]
print("winner: candidate", winner)
```

## Part 3 — Challenge

**8. Clamp and insert.** `clip` replaces cleaning if/else chains; `searchsorted` needs SORTED input.

```
import numpy as np

readings = np.array([-12, 5, 42, 87, 103, 60])
print(np.clip(readings, 0, 100))
print(np.clip(readings, 0, None))

queue = np.array([10, 20, 30, 40, 50])
pos = np.searchsorted(queue, 35)
print("35 belongs at index", pos)
print(np.insert(queue, pos, 35))

dups = np.array([10, 20, 20, 30])
print("left :", np.searchsorted(dups, 20))
print("right:", np.searchsorted(dups, 20, side="right"))
```

**9. Grade pipeline.** Pipeline hygiene: clip inputs, branch with nested where, summarize with unique counts.

```
import numpy as np

scores = np.array([45, 82, 67, 91, 58, 73])
cleaned = np.clip(scores, 0, 100)

grades = np.where(cleaned >= 80, "A",
                  np.where(cleaned >= 60, "B", "C"))
for score, grade in zip(scores.tolist(), grades.tolist()):
    print(score, "->", grade)

print("number of A grades:", np.count_nonzero(cleaned >= 80))
letters, counts = np.unique(grades, return_counts=True)
for letter, n in zip(letters, counts):
    print(letter, n)
```
