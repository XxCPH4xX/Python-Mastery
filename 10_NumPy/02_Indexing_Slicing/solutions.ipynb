## Part 1 — Warm-up

**1. Pick single elements.** Positions start at 0; negatives count back from the end.

```
import numpy as np

scores = np.array([78, 92, 61, 88, 95, 73])

print(scores[0])     # 78
print(scores[-1])    # 73
print(scores[-2])    # 95
```

**2. Slice it five ways.** Start included, stop excluded, optional step; `[::-1]` walks backwards.

```
import numpy as np

scores = np.array([78, 92, 61, 88, 95, 73])

print(scores[1:4])    # [92 61 88]
print(scores[:3])     # [78 92 61]
print(scores[-2:])    # [95 73]
print(scores[::2])    # [78 61 95]
print(scores[::-1])   # reversed
```

**3. Matrix coordinates.** One bracket pair with `m[row, col]` beats chained brackets; `:` means "everything on this axis".

```
import numpy as np

grades = np.array([[78, 85, 90],
                   [62, 71, 88],
                   [90, 93, 79]])

print(grades[1, 2])    # 88
print(grades[-1, 0])   # 90
print(grades[1, :])    # whole second row
print(grades[:, 0])    # whole first column
```

## Part 2 — Practice

**4. Cut out sub-blocks.** A range on each axis carves out a rectangular patch.

```
import numpy as np

grades = np.array([[78, 85, 90],
                   [62, 71, 88],
                   [90, 93, 79]])

print(grades[0:2, 0:2])
print()
print(grades[0:2, 1:])
```

**5. The view trap.** Basic slices are VIEWS onto the same memory; `.copy()` detaches them.

```
import numpy as np

scores = np.array([78, 92, 61, 88, 95])
top = scores[2:5]          # a VIEW, not a copy
top[0] = 999
print(scores)              # 999 leaked into the original!
print(np.shares_memory(scores, top))   # True

scores = np.array([78, 92, 61, 88, 95])
safe = scores[2:5].copy()  # independent buffer
safe[0] = 999
print(safe)
print(scores)              # untouched this time
```

**6. Filter with a mask.** Indexing with a boolean array keeps only the True positions.

```
import numpy as np

temps = np.array([31.5, 32.0, 33.8, 30.2, 29.9, 34.1])

hot_days = temps > 32
print(hot_days)
print(temps[hot_days])       # named-mask form
print(temps[temps > 32])     # idiomatic one-liner
print("hot days:", np.count_nonzero(hot_days))
```

**7. Combine conditions.** Arrays demand `&`, `|`, `~` - never `and`/`or`/`not`.

```
import numpy as np

scores = np.array([45, 82, 67, 91, 58, 73, 39, 88])

print(scores[(scores >= 50) & (scores < 80)])
print(scores[(scores >= 80) | (scores < 40)])
print(scores[~(scores < 40)])
```

## Part 3 — Challenge

**8. Fancy indexing.** Integer-array indexing selects arbitrary positions, in your order, with repeats.

```
import numpy as np

menu_prices = np.array([150, 220, 90, 310, 175])
print(menu_prices[[3, 0, 3]])     # repeats allowed

grid = np.array([[10, 20, 30],
                 [40, 50, 60],
                 [70, 80, 90]])
print(grid[[0, 2], [0, 2]])       # pairs: (0,0) and (2,2)

grades = np.array([[78, 85, 90],
                   [62, 71, 88],
                   [90, 93, 79]])
print(grades[:, [0, 2]])          # all rows, chosen columns
```

**9. Masked assignment cleanup.** Anything you can READ with a mask, you can ASSIGN through - whole if/else loops gone.

```
import numpy as np

sensor = np.array([12, 98, 15, 240, 18, 5, 190])
clean = sensor.copy()
clean[clean > 100] = 100     # clip every spike in one statement
print("raw  :", sensor)
print("fixed:", clean)

board = np.zeros((3, 3), dtype=int)
board[board == 0] = 7
print(board)
```
