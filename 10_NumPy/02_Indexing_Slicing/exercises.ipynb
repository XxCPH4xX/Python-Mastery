# NumPy Indexing & Slicing Exercises

Write your answer in the empty code cell under each question. Check yourself against `solutions.ipynb` when done.

**Part 1** = warm-up · **Part 2** = practice · **Part 3** = challenge.

## Part 1 — Warm-up

**1. Pick single elements.** Given `scores = np.array([78, 92, 61, 88, 95, 73])`, print the first element, the last element, and the second-to-last element (use negative indexes for the last two).

**2. Slice it five ways.** With the same `scores = np.array([78, 92, 61, 88, 95, 73])`, print: positions 1 to 3, the first three, the last two, every other element, and the whole array reversed.

**3. Matrix coordinates.** Given `grades = np.array([[78, 85, 90], [62, 71, 88], [90, 93, 79]])`, print the element at row 1, column 2 (tuple form!), the first element of the last row, ALL of row 1, and ALL of column 0.

## Part 2 — Practice

**4. Cut out sub-blocks.** From the same `grades` array (3 students x 3 subjects), extract: the top-left 2x2 sub-block, and the first two students' last two subject columns. Print both rectangles.

**5. The view trap.** Given `scores = np.array([78, 92, 61, 88, 95])`, take the slice `top = scores[2:5]`, set `top[0] = 999`, then print `scores` - what leaked through? Prove the connection with `np.shares_memory(scores, top)`. Finally fix it: rebuild `scores` fresh and repeat using `.copy()` on the slice so the original survives.

**6. Filter with a mask.** Given `temps = np.array([31.5, 32.0, 33.8, 30.2, 29.9, 34.1])`, build the boolean mask `temps > 32`, print it, then use it to print only the hot temperatures (both via the named mask and the one-line form). Count the hot days with `np.count_nonzero`.

**7. Combine conditions.** Given `scores = np.array([45, 82, 67, 91, 58, 73, 39, 88])`, select: scores from 50 to 79 (`&`), scores that are 80+ OR below 40 (`|`), and everything that is NOT below 40 (`~`). Wrap every comparison in parentheses.

## Part 3 — Challenge

**8. Fancy indexing.** Given `menu_prices = np.array([150, 220, 90, 310, 175])`, pick the prices at positions 3, 0 and 3 AGAIN (duplicates allowed) in one expression. Then with `grid = np.array([[10, 20, 30], [40, 50, 60], [70, 80, 90]])`, grab both diagonal corners with paired index arrays, and pull columns 0 and 2 of `grades = np.array([[78, 85, 90], [62, 71, 88], [90, 93, 79]])` for every student.

**9. Masked assignment cleanup.** Given `sensor = np.array([12, 98, 15, 240, 18, 5, 190])`, copy it, then cap every reading above 100 down to 100 with a single masked assignment. Print raw and fixed side by side. Then fill a 3x3 int array of zeros with 7s everywhere using masked assignment (`board[board == 0] = 7`) and print it.
