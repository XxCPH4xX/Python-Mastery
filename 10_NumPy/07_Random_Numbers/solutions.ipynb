## Part 1 — Warm-up

**1. Meet the Generator.** Each draw ADVANCES the generator's private stream - the seed pins the sequence, not every individual value.

```
import numpy as np

rng = np.random.default_rng(seed=42)

print(type(rng))
print(rng.random())    # first draw
print(rng.random())    # second draw - the stream moved on
```

**2. Roll the dice.** `high` is exclusive by default; `endpoint=True` makes it inclusive.

```
import numpy as np

rng = np.random.default_rng(seed=7)

print(rng.integers(1, 7))                      # one roll: 1..6
print(rng.integers(1, 7, size=5))              # five rolls
print(rng.integers(1, 6, size=5, endpoint=True))  # same range, inclusive form
```

**3. Uniform floats.** `random` covers the unit interval; `uniform` stretches it anywhere.

```
import numpy as np

rng = np.random.default_rng(seed=7)

print(np.round(rng.random(size=3), 3))
print(np.round(rng.uniform(10, 20, size=3), 3))
```

## Part 2 — Practice

**4. Bell curve samples.** `normal` takes any mean/std; `standard_normal` is the mean-0, std-1 shortcut.

```
import numpy as np

rng = np.random.default_rng(seed=42)

heights = rng.normal(loc=170, scale=8, size=5)
errors = rng.standard_normal(size=4)
print(np.round(heights, 1))
print(np.round(errors, 3))
```

**5. Pick from a population.** `choice` samples any array; `p=` biases the odds and must sum to 1.

```
import numpy as np

rng = np.random.default_rng(seed=3)
students = np.array(["Rafi", "Nadia", "Omar", "Priya"])

print(rng.choice(students, size=2, replace=False))
print(rng.choice(students, size=5))

picks = rng.choice(students, size=20, p=[0.2, 0.5, 0.2, 0.1])
names, counts = np.unique(picks, return_counts=True)
for name, n in zip(names, counts):
    print(name, "was picked", int(n), "times")
```

**6. shuffle vs permutation.** `shuffle` mutates in place and returns None; `permutation` returns a shuffled COPY.

```
import numpy as np

rng = np.random.default_rng(seed=11)

deck = np.arange(1, 11)
result = rng.shuffle(deck)
print("shuffle returned:", result)    # None!
print("deck now        :", deck)

fresh = np.arange(1, 11)
mixed = rng.permutation(fresh)
print("permutation     :", mixed)
print("original intact :", fresh)
```

**7. Prove reproducibility.** Same seed -> identical stream. One generator per experiment, created ONCE.

```
import numpy as np

first = np.random.default_rng(seed=123).integers(0, 100, size=5)
second = np.random.default_rng(seed=123).integers(0, 100, size=5)

print(first)
print(second)
print("identical?", np.array_equal(first, second))
# A fresh default_rng(123) inside a loop restarts the SAME stream
# every iteration - fake variety. Create the generator once, outside.
```

## Part 3 — Challenge

**8. Sanity-check the bell curve.** Simulations earn trust when their aggregate statistics land on theory.

```
import numpy as np

rng = np.random.default_rng(seed=0)
sample = rng.normal(loc=100, scale=15, size=100_000)

print("samples      :", sample.size)
print("empirical mean:", round(sample.mean(), 2), "(target 100)")
print("empirical std :", round(sample.std(), 2), "(target 15)")
within_1sd = np.mean(np.abs(sample - 100) < 15) * 100
within_2sd = np.mean(np.abs(sample - 100) < 30) * 100
print(round(within_1sd, 1), "% within 1 sd (theory ~68%)")
print(round(within_2sd, 1), "% within 2 sd (theory ~95%)")
```

**9. Two-dice casino audit.** Many outcomes summed approach a bell shape - the law of large numbers made visible with unique + counts.

```
import numpy as np

rng = np.random.default_rng(seed=1)
die1 = rng.integers(1, 7, size=10_000)
die2 = rng.integers(1, 7, size=10_000)
totals = die1 + die2

faces, counts = np.unique(totals, return_counts=True)
for face, count in zip(faces, counts):
    print(f"total {face:2d}: {count / totals.size:.1%}")

best = faces[counts.argmax()]
print("most likely total:", best)
```
