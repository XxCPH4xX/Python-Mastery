## Part 1 — Warm-up

**1. Meet the Generator.** Create `rng = np.random.default_rng(seed=42)`. Print `type(rng)`, then call `rng.random()` TWICE and print both floats. Why are the two values different even though the seed is fixed?

**2. Roll the dice.** With `rng = np.random.default_rng(seed=7)`: roll one six-sided die with `rng.integers(1, 7)` (why is 7 excluded?), roll five dice with `size=5`, and roll five dice using `endpoint=True` with the arguments `(1, 6)` so 6 can appear. Print all three results.

**3. Uniform floats.** With `rng = np.random.default_rng(seed=7)`: draw three floats in `[0, 1)` with `rng.random(size=3)` and three floats in `[10, 20)` with `rng.uniform(10, 20, size=3)`. Print both rounded to 3 decimals.

## Part 2 — Practice

**4. Bell curve samples.** With `rng = np.random.default_rng(seed=42)`: draw five heights from `rng.normal(loc=170, scale=8, size=5)` (cm) and four zero-centered noise values with `rng.standard_normal(size=4)`. Print the heights rounded to 1 decimal and the noise to 3.

**5. Pick from a population.** With `students = np.array(["Rafi", "Nadia", "Omar", "Priya"])` and `rng = np.random.default_rng(seed=3)`: pick 2 students WITHOUT repeats (`replace=False`), pick 5 WITH repeats, then make 20 weighted picks with `p=[0.2, 0.5, 0.2, 0.1]` and count them with `np.unique(picks, return_counts=True)`. Who gets picked most?

**6. shuffle vs permutation.** With `rng = np.random.default_rng(seed=11)` and `deck = np.arange(1, 11)`: call `rng.shuffle(deck)`, print its RETURN VALUE (surprise?) and the deck afterwards. Then take `fresh = np.arange(1, 11)`, shuffle a copy with `rng.permutation(fresh)`, and print both the mixed copy and the untouched original.

**7. Prove reproducibility.** Draw five integers in `[0, 100)` from `np.random.default_rng(seed=123)`, then draw five MORE from a SECOND, freshly created generator with the same seed. Compare with `np.array_equal`. Now explain in a comment what would happen if you created a NEW `default_rng(123)` inside a loop that should produce different samples each iteration.

## Part 3 — Challenge

**8. Sanity-check the bell curve.** With `rng = np.random.default_rng(seed=0)`, draw 100,000 samples from `rng.normal(loc=100, scale=15, size=100_000)`. Print the empirical mean and std (rounded to 2 decimals, targets 100 and 15), then the PERCENTAGE of samples within one std of the mean and within two (theory: about 68% and 95%) - use `np.mean(np.abs(sample - 100) < 15) * 100`.

Print each total with its percentage (`count / totals.size` formatted with `:.1%`), then announce the most likely total via `counts.argmax()`. Which total wins, and why?
