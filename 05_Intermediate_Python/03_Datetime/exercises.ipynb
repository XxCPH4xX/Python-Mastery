# Datetime Exercises

Write your answer in the empty code cell under each question. Check yourself against `solutions.ipynb` when done.

**Part 1** = warm-up · **Part 2** = practice · **Part 3** = challenge.

## Part 1 — Warm-up

**1. Four building blocks.** Create all four objects the module offers: `independence_day = date(1971, 3, 26)`, `class_start = time(9, 30)`, `order_placed = datetime(2026, 8, 26, 18, 45)` and `shipping_window = timedelta(days=5)`. Print all four.

**2. Read it back.** For `d = date(2026, 8, 26)` print `year`, `month`, `day` and `weekday()` (Monday = 0). Then build the same date next year with `.replace(year=2027)` and prove in a printout that the original object is unchanged.

**3. Right now.** Store `datetime.now()` in `right_now` and `date.today()` in `today`. Print `today` along with its weekday name via `.strftime("%A")`.

## Part 2 — Practice

**4. One timestamp, four outfits.** For `d = datetime(2026, 8, 26, 18, 45, 7)`, print three formatted versions: ISO style (`%Y-%m-%d`), everyday style (`%d/%m/%Y %I:%M %p`) and formal-invitation style (`%A, %d %B %Y`).

**5. Parse it back.** Turn the text `"26/08/2026 18:45"` into a `datetime` with `strptime`, then re-format it as ISO (`%Y-%m-%dT%H:%M:%S`). Afterwards try parsing `"2026-08-26"` with the template `"%d/%m/%Y"` inside `try/except ValueError` and print the error message.

**6. Deadlines and ages.** An order is placed at `datetime(2026, 8, 26, 18, 45)` and arrives after `5` days and `6` hours. Compute the arrival and format it with `"%A %d %b, %H:%M"`. Then, for someone born on `date(2002, 3, 14)`, print their age in days as of `date(2026, 8, 26)`.

## Part 3 — Challenge

**7. Deadline triage.** Given `deadlines = [date(2026, 9, 15), date(2026, 8, 30), date(2026, 11, 2)]` and `today = date(2026, 8, 26)`: print the earliest, the latest and the sorted list. Then build a list of deadlines due within 30 days using a comprehension comparing against `today + timedelta(days=30)`.

**8. Timezone-aware timestamps.** A flight takes off at `datetime(2026, 8, 26, 14, 0, tzinfo=timezone.utc)`. Print its `isoformat()`, convert it to Dhaka time with `timezone(timedelta(hours=6))` and print that too. Then show that comparing it with a *naive* `datetime(2026, 8, 26, 14, 0)` raises `TypeError`. Finally round-trip the takeoff through `.timestamp()` and `datetime.fromtimestamp(..., tz=timezone.utc)` and prove you got the same moment back.
