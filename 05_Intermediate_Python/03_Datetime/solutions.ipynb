## Part 1 — Warm-up

**1. Four building blocks.** Each type models one idea: a day, a clock reading, day+clock together, a duration.

```
from datetime import date, time, datetime, timedelta

independence_day = date(1971, 3, 26)
class_start = time(9, 30)
order_placed = datetime(2026, 8, 26, 18, 45)
shipping_window = timedelta(days=5)

print(independence_day)    # 1971-03-26
print(class_start)         # 09:30:00
print(order_placed)        # 2026-08-26 18:45:00
print(shipping_window)     # 5 days, 0:00:00
```

**2. Read it back.** Attributes read a date out; `.replace()` returns a NEW object because dates are immutable.

```
from datetime import date

d = date(2026, 8, 26)
print(d.year, d.month, d.day)           # 2026 8 26
print("weekday number:", d.weekday())   # 2 (Wednesday; Mon=0)

next_year = d.replace(year=2027)
print(next_year)                        # 2027-08-26
print("original untouched:", d)         # 2026-08-26
```

**3. Right now.** `now()` / `today()` are the only non-deterministic calls in the module — great live, avoid them in tests.

```
from datetime import date, datetime

right_now = datetime.now()
today = date.today()

print("this moment:", right_now)
print("just the day:", today, "->", today.strftime("%A"))
```

## Part 2 — Practice

**4. One timestamp, four outfits.** `strftime` fills a template of `%codes` — object becomes human-readable string.

```
from datetime import datetime

d = datetime(2026, 8, 26, 18, 45, 7)

print(d.strftime("%Y-%m-%d"))            # 2026-08-26
print(d.strftime("%d/%m/%Y %I:%M %p"))   # 26/08/2026 06:45 PM
print(d.strftime("%A, %d %B %Y"))        # Wednesday, 26 August 2026
```

**5. Parse it back.** `strptime` needs the template to match the text exactly, or it raises `ValueError`.

```
from datetime import datetime

stamp = datetime.strptime("26/08/2026 18:45", "%d/%m/%Y %H:%M")
print(stamp)                                   # 2026-08-26 18:45:00
print(stamp.strftime("%Y-%m-%dT%H:%M:%S"))     # 2026-08-26T18:45:00

try:
    datetime.strptime("2026-08-26", "%d/%m/%Y")   # template disagrees
except ValueError as e:
    print("ValueError:", e)
```

**6. Deadlines and ages.** Subtracting dates yields a `timedelta`; adding one shifts a date forward. There is no `months=` — month lengths vary.

```
from datetime import date, datetime, timedelta

order = datetime(2026, 8, 26, 18, 45)
delivery = order + timedelta(days=5, hours=6)
print("arrives:", delivery.strftime("%A %d %b, %H:%M"))   # Tuesday 01 Sep, 00:45

born = date(2002, 3, 14)
today = date(2026, 8, 26)
print(f"{(today - born).days} days old")                  # 8931 days old
```

## Part 3 — Challenge

**7. Deadline triage.** Date objects compare and sort naturally — no string gymnastics needed.

```
from datetime import date, timedelta

deadlines = [date(2026, 9, 15), date(2026, 8, 30), date(2026, 11, 2)]
today = date(2026, 8, 26)

print("earliest:", min(deadlines))    # 2026-08-30
print("latest  :", max(deadlines))    # 2026-11-02
print("sorted  :", sorted(deadlines))

horizon = today + timedelta(days=30)             # 2026-09-25
urgent = [d for d in deadlines if d <= horizon]
print("due within a month:", urgent)             # [2026-08-30, 2026-09-15]
```

**8. Timezone-aware timestamps.** Aware datetimes pin down one instant worldwide; mixing naive with aware is refused rather than guessed.

```
from datetime import datetime, timedelta, timezone

takeoff = datetime(2026, 8, 26, 14, 0, tzinfo=timezone.utc)
dhaka = timezone(timedelta(hours=6))             # UTC+6

print("UTC  :", takeoff.isoformat())             # 2026-08-26T14:00:00+00:00
print("Dhaka:", takeoff.astimezone(dhaka).isoformat())   # 2026-08-26T20:00:00+06:00

naive = datetime(2026, 8, 26, 14, 0)
try:
    naive < takeoff                              # guessing is not allowed
except TypeError as e:
    print("TypeError:", e)

back = datetime.fromtimestamp(takeoff.timestamp(), tz=timezone.utc)
print("round-trip equal?", back == takeoff)      # True
```
