# Python Datetime

> 📘 **Python Mastery** · Module 05 — Intermediate Python · Lesson 3/7

Deadlines, birthdays, delivery estimates, log stamps -- every real program touches time. The `datetime` module turns calendar maths from string hacking into ordinary arithmetic.

## 🎯 Learning Objectives

- Create and inspect `date`, `time` and `datetime` objects
- Get the current moment with `today()` / `now()`
- Format dates to text with `strftime` and parse text back with `strptime`
- Add and subtract durations using `timedelta`
- Compare, sort and convert dates; produce clean ISO 8601 strings
- Explain naive vs aware datetimes and Unix timestamps

## 1. Four Objects, One Module

The module gives you four building blocks -- learn which one a problem needs and half the work is done:

| Type | Represents | Example |
|---|---|---|
| `date` | a calendar day (no clock) | your birthday |
| `time` | a clock reading (no day) | 09:30 AM |
| `datetime` | day + clock together | an order timestamp |
| `timedelta` | a *duration* | "30 days" |

**Syntax:**
```python
from datetime import date, time, datetime, timedelta

date(2026, 8, 26)          # year, month, day
time(14, 30)               # hour, minute
datetime(2026, 8, 26, 14, 30)
timedelta(days=30)
```

**Example:** one of each.

````python
from datetime import date, time, datetime, timedelta

independence_day = date(1971, 3, 26)
class_start = time(9, 30)
order_placed = datetime(2026, 8, 26, 18, 45)
shipping_window = timedelta(days=5)

print(independence_day, "|", class_start, "|", order_placed, "|", shipping_window)
print(type(order_placed).__name__)
````

## 2. Building Specific Dates

Construct with integers, read back with attributes (`year`, `month`, `day`, ...), and derive variants with `.replace(...)` -- which returns a **new** object because dates are immutable.

**Syntax:**
```python
d = date(2026, 8, 26)
d.year, d.month, d.day
d.replace(year=2027)     # new object; original untouched
```

**Example:**

````python
from datetime import date

d = date(2026, 8, 26)
print(d.year, d.month, d.day)
print("weekday number:", d.weekday(), "(Mon=0 ... Sun=6)")
next_year = d.replace(year=2027)
print(next_year, "- same date, next year")
print("original unchanged:", d)
````

## 3. Right Now: `today()` and `now()`

`datetime.now()` returns the machine's current local time; `date.today()` just the day. (`datetime.today()` exists too and behaves like `now()`.) These are the only non-deterministic calls in this notebook -- perfect for live dashboards, but in examples and tests prefer fixed dates so results can be checked.

**Syntax:**
```python
from datetime import datetime, date
datetime.now()    # full local timestamp
date.today()      # today's date only
```

**Example:**

````python
from datetime import datetime, date

right_now = datetime.now()
print("full :", right_now)
print("parts: Y={} M={} D={} H={}".format(
    right_now.year, right_now.month, right_now.day, right_now.hour))

today = date.today()
print("just the day:", today, "->", today.strftime("%A"))
````

## 4. Formatting with `strftime` (object -> string)

`strftime` = "**s**tring **f**rom **t**ime". You hand it a template of `%codes`, it fills them in. This is how you produce "26 Aug 2026" or "06:45 PM" for humans.

| Code | Meaning | Example |
|---|---|---|
| `%Y` | 4-digit year | 2026 |
| `%y` | 2-digit year | 26 |
| `%m` | month, zero-padded | 08 |
| `%d` | day, zero-padded | 05 |
| `%H` | hour 00-23 | 18 |
| `%I` | hour 01-12 | 06 |
| `%p` | AM / PM | PM |
| `%M` | minute | 45 |
| `%S` | second | 07 |
| `%A` | weekday, full | Wednesday |
| `%a` | weekday, short | Wed |
| `%B` | month, full | August |
| `%b` | month, short | Aug |
| `%j` | day of year | 238 |
| `%%` | a literal percent sign | % |

**Syntax:**
```python
some_datetime.strftime("%d/%m/%Y %H:%M")
```

**Example:** one timestamp, four outfits.

````python
from datetime import datetime

d = datetime(2026, 8, 26, 18, 45, 7)

print(d.strftime("%Y-%m-%d"))                 # ISO style
print(d.strftime("%d/%m/%Y %I:%M %p"))        # everyday style
print(d.strftime("%A, %d %B %Y"))             # formal invitation
print(d.strftime("%H:%M:%S (day %j of %Y)"))  # nerdy log line
````

## 5. Parsing with `strptime` (string -> object)

`strptime` is the mirror image: given a string *and* the template describing it, you get back a real `datetime` you can compute with. The template must match exactly -- wrong separators or field order raise `ValueError`.

**Syntax:**
```python
datetime.strptime("26/08/2026 18:45", "%d/%m/%Y %H:%M")
```

**Example:**

````python
from datetime import datetime

stamp = datetime.strptime("26/08/2026 18:45", "%d/%m/%Y %H:%M")
print(stamp, type(stamp).__name__)

print(stamp.strftime("%Y-%m-%dT%H:%M:%S"))    # round-trip into another format

bad = "2026-08-26"
try:
    datetime.strptime(bad, "%d/%m/%Y")        # template disagrees with text
except ValueError as e:
    print("ValueError:", e)
````

## 6. Durations with `timedelta`

Subtracting two datetimes yields a `timedelta`; adding one shifts a date forward. That single idea covers ages, deadlines, trials and ETAs. Note what's missing: there is no `months=` -- months have unequal lengths, so use `timedelta(days=...)` or the third-party `dateutil.relativedelta`.

**Syntax:**
```python
delta = end - start                 # timedelta
later = start + timedelta(days=30, hours=6)
delta.days, delta.total_seconds()
```

**Example:** age in days and an order deadline.

````python
from datetime import date, datetime, timedelta

# age, done properly
born = date(2002, 3, 14)
today = date(2026, 8, 26)
age_years = today.year - born.year - ((today.month, today.day) < (born.month, born.day))
print(f"Sarah is {age_years} ({(today - born).days} days old)")

# a deadline
order = datetime(2026, 8, 26, 18, 45)
delivery = order + timedelta(days=5, hours=6)
print("ordered :", order)
print("arrives :", delivery.strftime("%A %d %b, %H:%M"))

fortnight = timedelta(days=14)
print("two weeks =", fortnight.total_seconds() / 86400, "days")
````

## 7. Comparing and Sorting Dates

Because they are proper objects, dates compare with `< <= > >= == !=` and sort naturally -- no string gymnastics needed. Chained comparisons give you elegant range checks.

**Syntax:**
```python
if deadline < today: ...
sorted(deadlines)
min(deadlines), max(deadlines)
start <= check_date <= end
```

**Example:**

````python
from datetime import date, timedelta

deadlines = [
    date(2026, 9, 15),
    date(2026, 8, 30),
    date(2026, 11, 2),
]
print("earliest:", min(deadlines))
print("latest  :", max(deadlines))
print("sorted  :", sorted(deadlines))

today = date(2026, 8, 26)
urgent = [d for d in deadlines if d <= today + timedelta(days=30)]
print("due within a month:", urgent)

print("15 Sep inside September?", date(2026, 9, 1) <= date(2026, 9, 15) <= date(2026, 9, 30))
````

## 8. `.date()`, `.time()` and `isoformat()`

A `datetime` carries both halves; sometimes you want only one. `.date()` and `.time()` slice them apart. And when sending dates to APIs, databases or config files, `isoformat()` produces the universal ISO 8601 form -- sortable, unambiguous, language-neutral (`2026-08-26T18:45:00`). Its inverse is `datetime.fromisoformat(...)`.

**Syntax:**
```python
dt.date(), dt.time(), dt.isoformat()
datetime.fromisoformat("2026-08-26T18:45:00")
```

**Example:**

````python
from datetime import datetime

order = datetime(2026, 8, 26, 18, 45, 7)

day_only = order.date()
clock_only = order.time()
print(day_only, "-", type(day_only).__name__)
print(clock_only, "-", type(clock_only).__name__)

as_text = order.isoformat()
print(as_text)
print("parsed back equal?", datetime.fromisoformat(as_text) == order)
````

## 9. Naive vs Aware Datetimes

A **naive** datetime has no timezone attached -- it's just numbers, and every machine assumes its own local meaning. An **aware** datetime carries a UTC offset, so it pins down one exact instant worldwide. For anything shared across machines or countries, store aware values (usually UTC) and convert for display.

> 🔍 **Under the Hood:** internally a datetime stores three plain integers -- days, seconds within the day, microseconds -- so arithmetic is exact. "Awareness" adds only a stored offset that participates in comparisons and conversions. Mixing the two kinds in one comparison raises `TypeError`, because Python refuses to guess the missing context.

**Syntax:**
```python
from datetime import datetime, timezone, timedelta
aware = datetime(2026, 8, 26, 14, 0, tzinfo=timezone.utc)
local = aware.astimezone(timezone(timedelta(hours=6)))   # Dhaka (UTC+6)
```

**Example:**

````python
from datetime import datetime, timedelta, timezone

utc_flight = datetime(2026, 8, 26, 14, 0, tzinfo=timezone.utc)
dhaka = timezone(timedelta(hours=6))          # UTC+6
print("takeoff UTC  :", utc_flight.isoformat())
print("takeoff Dhaka:", utc_flight.astimezone(dhaka).isoformat())

naive = datetime(2026, 8, 26, 14, 0)
try:
    naive < utc_flight                        # guessing is not allowed
except TypeError as e:
    print("TypeError:", e)

now_utc = datetime.now(timezone.utc)           # the modern way to say utcnow()
print("this very instant (UTC):", now_utc.strftime("%Y-%m-%d %H:%M"))
````

## 10. Timestamps and the Epoch

Computers love counting. A **Unix timestamp** is simply the number of seconds since the epoch -- 1 January 1970, 00:00 UTC. Logs, databases and APIs store this integer everywhere. `.timestamp()` converts a (preferably aware) datetime into that integer; `fromtimestamp()` converts back.

**Syntax:**
```python
seconds = some_datetime.timestamp()
datetime.fromtimestamp(seconds, tz=timezone.utc)
```

**Example:**

````python
from datetime import datetime, timezone

epoch = datetime(1970, 1, 1, tzinfo=timezone.utc)
launch = datetime(2026, 8, 26, 14, 30, tzinfo=timezone.utc)

ts = launch.timestamp()
print("epoch was       :", epoch.timestamp(), "seconds")
print("launch happens  :", ts, "seconds after epoch")

back = datetime.fromtimestamp(ts, tz=timezone.utc)
print("round-trip works:", back == launch)

days = (launch - epoch).total_seconds() / 86400
print(f"that is about {days:,.0f} days since 1970")
````

## ⚠️ Common Mistakes & Gotchas

| Mistake | Problem | Fix |
|---|---|---|
| Comparing a naive with an aware datetime | instant `TypeError` at runtime | pick one convention; make everything aware (UTC) |
| `datetime.utcnow()` | deprecated, and returns a *naive* value pretending to be UTC | `datetime.now(timezone.utc)` |
| `timedelta(months=30)` | doesn't exist -- month lengths vary | `timedelta(days=...)` or `dateutil.relativedelta(months=1)` |
| Assuming `weekday()` counts from Sunday | Python says Monday = 0 ... Sunday = 6 | use `.isoweekday()` for Mon = 1 ... Sun = 7 |
| Storing local wall-clock strings in databases | ambiguous across timezones and DST shifts | store UTC timestamps or ISO strings; convert at display time |
| Formatting with `%m` expecting "8" | codes are zero-padded ("08"); Linux-only `%-m` breaks on Windows | accept padding, or strip manually |

## 💡 Best Practices & Pro Tips

- Store UTC everywhere; convert to local only at the edge where humans look. Real zones come from `zoneinfo.ZoneInfo("Asia/Dhaka")`.
- Speak ISO 8601 (`isoformat()`) between systems -- it even sorts correctly as plain text.
- Use fixed dates in tests and teaching examples; reserve `now()` for genuinely live features.
- Parse user-typed dates defensively -- wrap `strptime` in `try/except ValueError` with a friendly message.
- **AI-engineering relevance:** datasets are split by date (train on the past, test on the future), logs are mined by timestamp, and feature engineering extracts hour/day-of-week columns -- all of it runs through this module (wrapped by pandas' `.dt` accessor).

## 📌 Summary

| Tool | What it does | Example |
|---|---|---|
| `date / time / datetime` | build calendar & clock objects | `date(2026, 8, 26)` |
| `datetime.now()` | current local moment | `datetime.now(timezone.utc)` |
| `.strftime(fmt)` | object -> formatted string | `d.strftime("%d %b %Y")` |
| `datetime.strptime(s, fmt)` | string -> object | strict template match |
| `timedelta(days=30)` | duration; supports arithmetic | `now + timedelta(days=30)` |
| comparisons & `sorted` | natural ordering of dates | `min(deadlines)` |
| `.date() / .time()` | split a datetime | `order.time()` |
| `.isoformat()` | standard interchange string | `2026-08-26T18:45:00` |
| `.timestamp()` | seconds since the 1970 epoch | `datetime.fromtimestamp(ts, tz=timezone.utc)` |

Key takeaways:

- Compute with objects, never with date *strings* -- parse once, then let `timedelta` do the work.
- `strftime` formats out, `strptime` parses in; the template must match the data exactly.
- Aware beats naive: attach `timezone.utc` whenever the value leaves your machine.
