## Part 1 — Warm-up

**1. First round trip.** Given `student = {"name": "Sarah", "age": 24, "courses": ["math", "cs"], "active": True}`, convert it to JSON text with `json.dumps` and print the text and its type. Convert back with `json.loads` and print whether the result equals the original.

**2. Lost in translation.** Dump this dictionary: `{"id": 7, "tags": ("ml", "llm"), "score": 91.5, "passed": True, "note": None}`. Print the JSON text, load it back, and print the type of the loaded `tags` value. Comment on what happened to the tuple, `True` and `None`.

**3. Readable Unicode.** Dump `customer = {"name": "আরিফ হাসান", "city": "Dhaka"}` twice — once with defaults and once with `ensure_ascii=False` — and print both results. Which one can a human actually read?
