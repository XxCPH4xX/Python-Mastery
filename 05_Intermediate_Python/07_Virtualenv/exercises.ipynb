# Virtual Environments Exercises

Write your answer in the empty code cell under each question. Check yourself against `solutions.ipynb` when done.

**Part 1** = warm-up · **Part 2** = practice · **Part 3** = challenge.

## Part 1 — Warm-up

**1. Why isolate.** Project A needs `numpy 1.x`; Project B needs `numpy 2.x`. Explain in comments what happens if both install into the global Python, and write the commands that create Project B's own private environment named `.venv` inside its project folder.

**2. Flip the switch.** Write the activation command for each shell as comments: Windows PowerShell, Windows CMD, Windows Git Bash, macOS/Linux. Then the universal command to LEAVE the environment, and what the successful prompt looks like.

**3. Ask Python, not the prompt.** From inside Python, print `sys.executable`, `sys.prefix`, `sys.base_prefix` and whether the two prefixes differ. Also report whether the `VIRTUAL_ENV` environment variable is set. Add a comment interpreting what each value means inside an activated venv.

## Part 2 — Practice

**4. The standard workflow.** Put the five workflow steps in execution order as commented commands: install `requests` and `pandas`, snapshot with freeze, ignore `.venv` in git, create `.venv`, activate it (Git Bash). Number them 1-5.

**5. Rebuild it anywhere.** A teammate cloned your repo and it contains only `requirements.txt`. Give the two commands that hand them a working environment. Bonus: the no-activation shortcut for running `train.py` with the venv's own Python.

**6. Anatomy of .venv.** Name the four things you find inside a fresh `.venv` folder and each one's job (comments). Confirm from Python that the `venv` builder module is available. Is a venv a full copy of Python? Answer in a comment.

## Part 3 — Challenge

**7. The honest detector.** Write a function `in_venv()` that returns `True` only when running inside a virtual environment, plus a `report()` that prints the executable, both prefixes and a verdict. Call `report()` and explain in a comment why this check beats glancing at the terminal prompt.

**8. Three broken environments.** Diagnose and fix each scenario in comments: (a) running `.venv\Scripts\Activate.ps1` inside cmd throws odd errors; (b) after renaming the project folder the venv stopped working; (c) a teammate complains the repo is huge because someone committed `.venv` to git. Bonus: a colleague runs `pip install` while forgetting to activate — where do those packages land?
