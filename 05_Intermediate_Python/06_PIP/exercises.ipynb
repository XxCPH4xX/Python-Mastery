## Part 1 — Warm-up

**1. Who is your pip.** In your TERMINAL (not inside Python), check the pip version two ways: bare `pip --version` and the preferred `python -m pip --version`. Paste both outputs into the cell as comments. Why is the second form safer when several Pythons share the machine?

**2. Fetch a package.** Write the terminal commands to: install `requests` (latest), install the exact version `requests==2.32.3`, and install `numpy`, `pandas` and `matplotlib` in ONE command. Keep each as a comment in the cell.

**3. Upgrade, then remove.** Write the commands to upgrade `requests` to the newest allowed version, and to uninstall it WITHOUT being asked for confirmation. Which flag skips the y/n prompt? Any command to update everything at once?

## Part 2 — Practice

**4. Take inventory.** Which three terminal commands list every installed package, show which have newer releases, and inspect ONE package deeply (`numpy`)? Then, from inside Python, print the folder where third-party packages land and the path of the current interpreter.

**5. Inventory from inside Python.** Without leaving the notebook, print the installed version of `numpy`, `pandas` and `matplotlib` from Python, handling the case where one is missing gracefully. Bonus: fetch pip's own metadata and print its Name and Version.

**6. The environment recipe.** Create a `requirements.txt` in the notebook containing exactly these lines, each annotated with what its specifier means: `numpy==2.3.1`, `pandas>=2.2,<3`, `requests~=2.32`, `matplotlib`. Then add comments with the two terminal commands: one that GENERATES the file from the current environment, one that REBUILDS an environment from it.
