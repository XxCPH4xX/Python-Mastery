## Part 1 — Warm-up

**1. Raw dogma.** Show that `"\\d+"` (escaped backslash) and `r"\d+"` (raw string) are the SAME string. Then use the raw version with `re.findall` to pull every run of digits out of `"Table 4 needs 3 chairs, room 42 needs 21."`

**2. Words and digits.** On `log = "ERROR 404 at 12:01 | ok | ERROR 500 at 13:15"`: extract every `ERROR <number>` pair with the pattern `r"ERROR \d+"`, then every `minute:second` timestamp with `r"\d{2}:\d{2}"`.

**3. search vs match vs fullmatch.** On the same log, use `re.search(r"\d{2}:\d{2}", log)` and print `.group()` and `.start()`. Then report whether `re.match` with the same pattern finds anything (and say why in a comment). Finally confirm `re.fullmatch(r"\d{2}", "40")` succeeds while `re.fullmatch(r"\d{2}", "404")` fails.

## Part 2 — Practice

**4. Character menu.** Using character classes only: count the vowels in `"beautiful day"` with `[aeiou]`; find the hex-ish characters in `"fa3 G7 #zz"` with `[0-9a-f]`; and list everything that is NOT a letter or space in `"Room 42, floor #7!"` with `[^a-zA-Z ]`.

**5. Fixed shapes.** Extract product codes shaped like two capital letters followed by three digits (`r"[A-Z]{2}\d{3}"`) and ISO dates (`r"\d{4}-\d{2}-\d{2}"`) from `code = "ID-AB123 2026-08-26 x9 zz"`.

**6. Flags change the rules.** From `reply = "YES please.\nno thanks\nMAYBE later"`, find every yes/no/maybe word in ONE `findall` using `re.IGNORECASE` with the pattern `r"\byes\b|\bno\b|\bmaybe\b"`. Then, from `build_log = "INFO boot\nERROR disk full\nINFO login"`, compare what `^ERROR.*` finds with and without `re.MULTILINE`.

**7. Greedy vs lazy.** On `row = "<td>Sarah</td><td>Dhaka</td>"`, extract cell contents two ways: greedy `r"<td>(.*)</td>"` and lazy `r"<td>(.*?)</td>"`. Comment on the difference between the two result lists.

## Part 3 — Challenge

**8. Groups extract fields.** For `entry = "Parcel delivered on 2026-08-26 to Dhaka."`: capture year, month and day with `r"(\d{4})-(\d{2})-(\d{2})"` and print the whole match plus `.groups()`. Repeat with named groups and print `group("month")`. Finally use `re.sub` with back-references to rewrite the date inside the sentence as day/month/year (`26/08/2026`).

**9. Compile once, validate everywhere.** Compile two patterns up top with clear names: `USERNAME = re.compile(r"[A-Za-z0-9_]{3,16}")` and `EMAIL = re.compile(r"[\w.+-]+@[\w-]+\.[A-Za-z]{2,}")`. Validate the candidates `["sarah_k", "ab", "has space", "way_too_long_username_x", "Rafi2026"]` with `.fullmatch` (print accepted/REJECTED for each). Then harvest every email address from: `message = "Write to sarah.rahman@gmail.com or arif+bhuyian@yahoo.co.uk today. Fragments like @nope.com or name@ are ignored."`
