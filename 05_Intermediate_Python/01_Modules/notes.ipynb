## 1. What Is a Module?

A **module** is simply a `.py` file containing Python code — variables, functions, classes. The file name (without `.py`) becomes the module name: `math.py` is the module `math`.

Think of it as a labelled toolbox: instead of keeping every tool on one giant bench, you store wrenches in one box, hammers in another, and grab exactly the box you need.

**Syntax:**
```python
import module_name

module_name.function()      # use its tools through the dot
module_name.CONSTANT
```

Python ships with hundreds of modules ready-made (**the standard library**), and millions more can be downloaded (**third-party packages** — Lesson 6).

**Example:** the `math` module gives us square roots, pi, and friends.

````python
import math  # load the toolbox called "math"

print(math.sqrt(625))   # square root of 625
print(math.pi)          # a constant stored in the module
print(math.floor(7.9))  # round down to 7
````

## 2. Importing a Whole Module

`import math` loads everything in the module under the name `math`. You then reach each tool through the dot: `math.sqrt(...)`. This style is slightly longer to type but crystal clear — every reader can see *where* `sqrt` came from.

One import gives you access to the entire namespace of the module.

> 🔍 **Under the Hood:** when Python imports a module it checks `sys.modules` — a dictionary of already-loaded modules — *before* touching the disk. If it's there, the import returns instantly. That's why a module's top-level code runs only **once per session**, and why editing a file mid-session appears to do nothing until you restart the kernel or call `importlib.reload(module)`.

**Syntax:**
```python
import math
import random, sys        # several modules in one line (less common)

math.sqrt(x)
```

**Example:** dice rolls with `random` and circles with `math`.

````python
import math
import random

random.seed(42)  # reproducible "randomness" (Lesson 2 covers this in depth)

radius = 5.5
area = math.pi * radius ** 2
print(f"A pizza of radius {radius} cm has area {area:.1f} cm^2")

dice = random.randint(1, 6)
print(f"You rolled a {dice}")
````

## 3. Importing Specific Names

If you only need one or two tools, bring them straight into your namespace with `from ... import ...`. After that you call them **without the prefix**.

The trade-off: readers must remember where `sqrt` came from, and if *you* also define your own `sqrt`, yours silently wins. Use it when the source is obvious and the names are distinctive.

**Syntax:**
```python
from math import sqrt
from datetime import datetime, date   # several names, comma-separated

sqrt(x)
```

**Example:**

````python
from math import sqrt, factorial

print(sqrt(225))         # no "math." needed anymore
print(factorial(6))      # 6! = 6*5*4*3*2*1

from collections import Counter
votes = ["rap", "rock", "rap", "lofi", "rap", "rock"]
print(Counter(votes).most_common(1))  # the winner
````

## 4. Aliases with `as`

Long module names get tiring. `import X as Y` loads module `X` but binds it to the shorter name `Y`. The whole Python community agrees on a few canonical aliases — using them makes your code instantly readable to everyone else:

| Convention | Why |
|---|---|
| `import numpy as np` | NumPy arrays everywhere in data science |
| `import pandas as pd` | DataFrames |
| `import matplotlib.pyplot as plt` | plotting |
| `import tensorflow as tf` | deep learning |
| `from datetime import datetime as dt` | saves typing twice |

Invent your own aliases sparingly — `import numpy as n` will annoy every reviewer.

**Syntax:**
```python
import module_name as short_name
from module import long_function_name as fn
```

**Example:**

````python
import statistics as stats          # our own friendly alias

scores = [72, 88, 95, 61, 88]
print(stats.mean(scores))
print(stats.median(scores))
print(stats.multimode(scores))

from math import factorial as fact  # alias an individual function too
print(fact(10))
````

## 5. Creating Your Own Module

Any `.py` file you write **is already a module**. To prove it, we'll build `util.py` — a small grab-bag of helpers — right here in `sample_data/`, then import it like any professional library.

Two things make this work:
1. `Path.write_text(...)` creates the file.
2. `sys.path.insert(0, "sample_data")` tells Python where to look. Python searches the folders listed in `sys.path`; our notebook's folder doesn't contain `util.py`, so we add the subfolder ourselves.

**Syntax:**
```python
import sys
sys.path.insert(0, "path/to/my/folder")  # teach Python where to search

import my_module                          # now it behaves like stdlib
```

This is the content we are about to write to `sample_data/util.py`:
```python
"""util.py - our first home-made module."""

TAX_RATE = 0.15  # VAT in Bangladesh

def greet(name):
    """Return a friendly welcome."""
    return f"Hello, {name}! Welcome aboard."

def celsius_to_fahrenheit(celsius):
    """Convert Celsius to Fahrenheit."""
    return celsius * 9 / 5 + 32

def with_tax(amount):
    """Add 15% VAT to a price."""
    return amount * (1 + TAX_RATE)
```

**Example:** create the file, point `sys.path` at the folder, import it, use it.

````python
from pathlib import Path

UTIL_SOURCE = '''
"""util.py - our first home-made module."""

TAX_RATE = 0.15  # VAT in Bangladesh

def greet(name):
    """Return a friendly welcome."""
    return f"Hello, {name}! Welcome aboard."

def celsius_to_fahrenheit(celsius):
    """Convert Celsius to Fahrenheit."""
    return celsius * 9 / 5 + 32

def with_tax(amount):
    """Add 15% VAT to a price."""
    return amount * (1 + TAX_RATE)
'''

Path("sample_data").mkdir(exist_ok=True)
path = Path("sample_data/util.py")
path.write_text(UTIL_SOURCE, encoding="utf-8")
print("Wrote", path.resolve())
````

````python
import sys
sys.path.insert(0, "sample_data")  # add our folder to the search path

import util                        # Python finds sample_data/util.py

print(util.greet("Sarah"))
print(util.celsius_to_fahrenheit(37))

dinner = 850.0                     # BDT
print(f"Dinner {dinner:.0f} BDT -> with VAT: {util.with_tax(dinner):.2f} BDT")

print("Module file lives at:", util.__file__)
````

````python
from util import greet, with_tax   # selective import works too

print(greet("Arif"))
print(f"{with_tax(1200):.2f}")

names = [n for n in dir(util) if not n.startswith("_")]
print("Public names in util:", names)   # section 7 explains dir()
````

## 6. `if __name__ == "__main__":` Explained

Every Python file has a built-in string variable called `__name__`:

- When you **run the file directly** (`python greet_demo.py`), Python sets `__name__ = "__main__"` — the file is the star of the show.
- When the file is **imported**, `__name__` becomes the module's actual name (`"greet_demo"`).

The famous guard block uses this so that demo/self-test code runs only when the file is executed directly, never when someone imports it. It lets one file be both a reusable library *and* a runnable script.

**Syntax:**
```python
def main():
    ...                       # things that should happen when RUN

if __name__ == "__main__":
    main()                    # skipped entirely on import
```

**Example:** write `greet_demo.py`, then execute it "as a script" using `runpy.run_path(..., run_name="__main__")` — the notebook equivalent of typing `python sample_data/greet_demo.py` in a terminal.

````python
from pathlib import Path
import runpy

DEMO_SOURCE = '''
"""greet_demo.py - one file, two personalities."""

def hello(name):
    return f"Hello, {name}!"

def main():
    print(hello("Sarah"))
    print("(banner) Running DIRECTLY - __name__ is '__main__'")

if __name__ == "__main__":
    main()
'''

Path("sample_data/greet_demo.py").write_text(DEMO_SOURCE, encoding="utf-8")

runpy.run_path("sample_data/greet_demo.py", run_name="__main__")  # act as a script
````

````python
import sys
sys.path.insert(0, "sample_data")  # (needed again if cells run out of order)

import greet_demo                  # importing does NOT trigger the guard

print("__name__ of imported module:", greet_demo.__name__)
print(greet_demo.hello("Rafi"))
print("No banner above -> the guard kept main() quiet on import.")
````

## 7. Exploring a Module with `dir()`

Given any module, `dir(module)` lists every name inside it — functions, constants, classes. That's how professionals poke around unfamiliar libraries without opening documentation. Filter out the underscore-prefixed internals to see the public API, and call `help(module.function)` for a built-in manual page.

**Syntax:**
```python
dir(module)                 # list of all names
help(module.func)           # mini-manual for one function
module.__doc__              # the module's own description
```

**Example:**

````python
import random

public = [n for n in dir(random) if not n.startswith("_")]
print(public[:12], "...")            # peek at the first dozen

help(random.shuffle)                 # read the built-in docs

print(random.__doc__.strip().splitlines()[0])  # module's one-liner
````

## 8. A Tour of the Standard Library

These ten modules cover most everyday needs. You'll meet several properly in later lessons — this table is your map.

| Module | Superpower | Typical use |
|---|---|---|
| `os` | talks to the operating system | folders, environment variables |

````python
from collections import Counter
from pathlib import Path
import json, re, random, sys

random.seed(42)
grades = [random.choice(["A", "B", "C"]) for _ in range(20)]
print("Counter :", Counter(grades))

record = {"course": "python", "week": 5}
print("JSON    :", json.dumps(record))

print("digits  :", re.findall(r"\d+", "classroom 42, floor 7"))
print("cwd     :", Path.cwd().name)
print("python  :", sys.version.split()[0])
````

## 9. Standard Library vs Third-Party Packages

Both arrive through `import`, but they come from different places:

| | Standard library | Third-party packages |
|---|---|---|
| Source | ships **with** Python | downloaded from **PyPI** |
| Install step | none | `pip install package` (Lesson 6) |
| Maintained by | Python core team | thousands of volunteers/companies |
| Examples | `math`, `json`, `random`, `pathlib` | `numpy`, `pandas`, `requests` |
| Trust & review | heavily vetted | check popularity & maintenance yourself |

Rule of thumb: try the standard library first — it solves a surprising amount — and reach for PyPI when you need specialised power (`numpy` for matrices, `requests` for HTTP).

**Pro tip:** run `help("modules")` in a Python shell to list every module available in *your* installation.

## ⚠️ Common Mistakes & Gotchas

| Mistake | Problem | Fix |
|---|---|---|
| Naming your file `math.py` (or `random.py`) | Python imports **your** file instead of the real module, then fails with baffling errors like `AttributeError: module 'math' has no attribute 'sqrt'` | Never name scripts after standard-library modules |
| `from module import *` | Floods your namespace; you can't tell where any name came from; silent collisions | Import the module, or named functions explicitly |
| Editing a module while the session is running | Imports are cached in `sys.modules`, so changes look ignored | Restart the kernel, or `importlib.reload(my_module)` |
| Circular imports (`a.py` imports `b`, `b.py` imports `a`) | One module gets half-initialised → `ImportError` | Move shared code into a third module both can import |
| `ModuleNotFoundError` although the file exists | The folder isn't in `sys.path` (notebooks often start elsewhere) | `sys.path.insert(0, "folder")` or run from the right directory |

## 💡 Best Practices & Pro Tips

- One job per module, with a lowercase descriptive name: `utils.py`, `prices.py`, `text_clean.py`.
- Prefer `import module` for clarity; alias only with community conventions (`np`, `pd`, `plt`).
- Put anything meant to be *run* behind `if __name__ == "__main__":` so the file stays safely importable.
- Keep `sys.path` hacks out of real projects — proper projects install their own package (or use a `src/` layout) instead.
- **AI-engineering relevance:** ML repos live or die by clean modules (`data.py`, `model.py`, `train.py`). Remember also that heavy top-level setup in a module runs **once** thanks to import caching — great for loading a model once, painful if it was meant to reload.

## 📌 Summary

| Statement | What it does | Example |
|---|---|---|
| `import math` | load a whole module | `math.sqrt(16)` |
| `from math import sqrt` | load selected names | `sqrt(16)` |
| `import numpy as np` | load with an alias | `np.array([1, 2])` |
| `sys.path.insert(0, p)` | add a folder to the search path | `import my_util` |
| `dir(module)` | list everything inside | `[n for n in dir(math) if not n.startswith('_')]` |
| `help(obj)` | built-in documentation | `help(random.choice)` |
| `if __name__ == "__main__":` | run only when executed directly | calls `main()` |

Key takeaways:

- A module is just a `.py` file; `import` makes its contents available through a dot.
- Import caching means top-level code runs once per session — restart or `reload()` to see edits.
- `__name__` tells a file whether it's being run (`"__main__"`) or imported.
- Standard library = free and installed; third-party = powerful but fetched via pip.
