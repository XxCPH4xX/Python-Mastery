## Part 1 — Warm-up

**1. Your first import.** Dot access: everything inside the module lives behind the name you imported.

```
import math

print(math.sqrt(144))       # 12.0
print(round(math.pi, 2))    # 3.14
```

**2. Pick just one tool.** `from ... import ...` brings the name straight into your namespace — no prefix needed.

```
from random import randint

roll = randint(1, 6)        # inclusive on both ends
print(f"You rolled a {roll}")
```

**3. Shorter names.** `import X as Y` binds the module to a shorter name; the dot still works as usual.

```
import statistics as stats

scores = [72, 88, 95, 61]
print(stats.mean(scores))    # 79.0
print(stats.median(scores))  # 80.0
```

## Part 2 — Practice

**4. Count the votes.** Selective imports work for classes exactly like they do for functions.

```
from collections import Counter

votes = ["rap", "rock", "rap", "lofi", "rap", "rock"]
winner = Counter(votes).most_common(1)
print(winner)                # [('rap', 3)]
```

**5. Peek inside a module.** Filtering out the underscore-prefixed internals reveals the public API — no documentation required.

```
import random

public = [n for n in dir(random) if not n.startswith("_")]
print(len(public), "public names")   # count varies by Python version
print(public[:10])
# e.g. ['Random', 'SystemRandom', 'betavariate', 'choice', ...]
```

**6. Build your own module.** Any `.py` file is a module once Python can find it — `sys.path.insert` tells it where to look.

```
from pathlib import Path
import sys

Path("sample_data").mkdir(exist_ok=True)
Path("sample_data/helpers.py").write_text("\n".join([
    "TAX_RATE = 0.15",
    "",
    "def with_tax(amount):",
    "    # Adds 15% VAT to a price.",
    "    return amount * (1 + TAX_RATE)",
]) + "\n", encoding="utf-8")

sys.path.insert(0, "sample_data")   # teach Python where to search
import helpers                      # finds sample_data/helpers.py

print(helpers.with_tax(850))        # 977.5
print(helpers.TAX_RATE)             # 0.15
```

**7. Imports are cached.** The second import hands back the cached object from `sys.modules` — top-level code runs once per session.

```
import math      # first import loads it
import math      # second import is (almost) free
import sys

print(math is sys.modules["math"])   # True - same cached object reused

# Editing a module mid-session looks ignored because of this cache:
# restart the kernel, or call importlib.reload(module), to see edits.
```

## Part 3 — Challenge

**8. One file, two personalities.** Run directly, `__name__` is `"__main__"` and the guard fires; imported, `__name__` becomes `"demo"` and the guard is skipped.

```
from pathlib import Path
import runpy
import sys

demo_lines = [
    'def hello(name):',
    '    return f"Hello, {name}!"',
    '',
    'def main():',
    "    print('(banner) running directly - __name__ is __main__')",
    '    print(hello("Sarah"))',
    '',
    "if __name__ == '__main__':",
    '    main()',
]
Path("sample_data").mkdir(exist_ok=True)
demo_path = Path("sample_data/demo.py")
demo_path.write_text("\n".join(demo_lines) + "\n", encoding="utf-8")

runpy.run_path(str(demo_path), run_name="__main__")   # acts like: python demo.py

sys.path.insert(0, "sample_data")
import demo                    # importing does NOT trigger the guard

print("__name__ of the imported module:", demo.__name__)
print(demo.hello("Rafi"))      # library functions still available
```

**9. Standard-library sampler.** The batteries are included — one import each and you are productive.

```
import json
import random
import re

random.seed(42)                       # reproducible pick below
record = {"course": "python"}

print(json.dumps(record))                     # {"course": "python"}
print(re.findall(r"\d+", "room 42, floor 7"))  # ['42', '7']
print(random.choice(["A", "B", "C"]))
```
