# Modules Exercises

Write your answer in the empty code cell under each question. Check yourself against `solutions.ipynb` when done.

**Part 1** = warm-up · **Part 2** = practice · **Part 3** = challenge.

## Part 1 — Warm-up

**1. Your first import.** Import the `math` module and print the square root of `144`, then `math.pi` rounded to 2 decimal places.

**2. Pick just one tool.** Instead of the whole module, import only `randint` from `random` and use it to roll a six-sided dice. Print the result.

**3. Shorter names.** `statistics` is a mouthful. Import it with the alias `stats` and print the mean and median of `scores = [72, 88, 95, 61]`.

## Part 2 — Practice

**4. Count the votes.** Import only `Counter` from `collections` and count `votes = ["rap", "rock", "rap", "lofi", "rap", "rock"]`. Print the winner with `.most_common(1)`.

**5. Peek inside a module.** `dir(module)` lists every name a module contains. Print how many *public* names (no leading underscore) the `random` module has, plus its first ten public names in alphabetical order.

**6. Build your own module.** Create a file `helpers.py` containing `TAX_RATE = 0.15` and a function `with_tax(amount)` returning `amount * (1 + TAX_RATE)`. Add its folder to `sys.path`, then `import helpers` and print `helpers.with_tax(850)` and `helpers.TAX_RATE`.

**7. Imports are cached.** Write `import math` twice in a row, then check whether `math is sys.modules["math"]`. What does the result tell you about how often a module's top-level code runs? Answer in a comment.

## Part 3 — Challenge

Execute the file *as a script* with `runpy.run_path(..., run_name="__main__")`, then `import demo` normally and show that the banner stays quiet.
