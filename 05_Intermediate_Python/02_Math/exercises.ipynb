# Math Exercises

Write your answer in the empty code cell under each question. Check yourself against `solutions.ipynb` when done.

**Part 1** = warm-up · **Part 2** = practice · **Part 3** = challenge.

## Part 1 — Warm-up

**1. Absolute power.** Use the built-ins to print `abs(-27)`, `pow(2, 16)` and `divmod(47, 6)`. Which two numbers does `divmod` return? Say it in a comment.

**2. Cart totals.** Given `prices = [420, 155, 95]`, print the cheapest price, the dearest price and the total.

**3. Round things.** Print `round(92.4567, 2)` and then `round(15842, -3)`. A negative second argument rounds to tens (-1), hundreds (-2), thousands (-3).

## Part 2 — Practice

**4. Banker's rounding.** Predict first, then run: `round(2.5)`, `round(0.5)`, `round(1.5)` and `round(3.5)`. Add a comment explaining the rule you observed.

**5. The bus problem.** `people = 53` need transport and each bus has `20` seats. Use `math.ceil` to compute the number of buses. Then print `ceil`, `floor`, `trunc` and `round` of `-2.7` and comment on how they disagree on negatives.

**6. Float betrayal.** Print `0.1 + 0.2`, test it against `== 0.3`, then repair the comparison with `math.isclose`. Finish by comparing `sum([0.1] * 10)` and `math.fsum([0.1] * 10)` against `1.0`.

**7. Sharing fairly.** Use `math.gcd(48, 180)` to simplify the fraction 48/180, printing the simplified form. Then use `math.lcm(4, 6)` to find when two alarms that ring every 4 and 6 minutes ring together.

## Part 3 — Challenge

**8. The invoice floats fail.** A T-shirt costs `19.99` and a customer buys `3`. Show that float multiplication drifts (`19.99 * 3`), then redo it exactly with `Decimal`. Compute 15% VAT on `Decimal("4500.00")`, quantized to 2 decimal places. Finally round `Decimal("2.5")` twice: once with plain `round()` and once with `ROUND_HALF_UP`.

**9. Deterministic game night.** Seed the random generator with `42`, then roll five dice with a list comprehension over `randint(1, 6)`, pick one move from `["rock", "paper", "scissors"]` with `choice`, and draw 2 raffle winners from `["Sara", "Arif", "Nadia", "Rafi", "Mina"]` with `sample`. Every rerun must produce identical results.
