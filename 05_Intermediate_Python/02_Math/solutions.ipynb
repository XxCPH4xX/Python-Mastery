## Part 1 — Warm-up

**1. Absolute power.** `abs` strips the sign, `pow` raises, and `divmod` answers quotient and remainder together.

```
print(abs(-27))        # 27
print(pow(2, 16))      # 65536
print(divmod(47, 6))   # (7, 5) -> quotient and remainder
```

**2. Cart totals.** The three great reducers — `min`, `max`, `sum` — each swallow an iterable.

```
prices = [420, 155, 95]

print(min(prices))   # 95
print(max(prices))   # 420
print(sum(prices))   # 670
```

**3. Round things.** `round(x, n)` handles decimal places; negative `n` rounds left of the decimal point.

```
print(round(92.4567, 2))   # 92.46
print(round(15842, -3))    # 16000 - nearest thousand
```

## Part 2 — Practice

**4. Banker's rounding.** Exact halfway cases go to the nearest EVEN neighbour — that's banker's rounding, not a bug.

```
print(round(2.5))   # 2 - tie goes DOWN to the even neighbour
print(round(0.5))   # 0
print(round(1.5))   # 2 - tie goes UP to the even neighbour
print(round(3.5))   # 4

# Banker's rounding: halfway cases always land on the EVEN number,
# so tiny round-up biases cancel out across many transactions.
```

**5. The bus problem.** You cannot order 2.65 buses — ceiling rounds up. On negatives the four functions split into three different answers.

```
import math

people, seats = 53, 20
buses = math.ceil(people / seats)      # 2.65 -> must round UP
print(f"{people} people need {buses} buses")

x = -2.7
print("ceil :", math.ceil(x))    # -2 (towards +infinity)
print("floor:", math.floor(x))   # -3 (towards -infinity)
print("trunc:", math.trunc(x))   # -2 (chopped towards zero)
print("round:", round(x))        # -3 (nearest)
```

**6. Float betrayal.** Floats are binary fractions, so storage noise accumulates; `isclose` compares safely and `fsum` adds precisely.

```
import math

print(0.1 + 0.2)                      # 0.30000000000000004
print(0.1 + 0.2 == 0.3)               # False - binary storage noise
print(math.isclose(0.1 + 0.2, 0.3))   # True  - the safe comparison

print(sum([0.1] * 10) == 1.0)         # False - error accumulates
print(math.fsum([0.1] * 10) == 1.0)   # True  - fsum tracks precision
```

**7. Sharing fairly.** `gcd` shrinks fractions to lowest terms; `lcm` finds the first shared multiple.

```
import math

g = math.gcd(48, 180)                 # 12
print(g, "-> 48/180 simplifies to", 48 // g, "/", 180 // g)
print("alarms ring together every", math.lcm(4, 6), "minutes")   # 12
```

## Part 3 — Challenge

**8. The invoice floats fail.** `Decimal` built from strings keeps base-10 money exact, and `.quantize()` applies an explicit rounding rule.

```
from decimal import Decimal, ROUND_HALF_UP

print(19.99 * 3)              # 59.97000000000001 - drift!
print(Decimal("19.99") * 3)   # 59.97 - exact

price = Decimal("4500.00")
vat = (price * Decimal("0.15")).quantize(Decimal("0.01"))
print("VAT:", vat)            # VAT: 675.00 - clean

print(round(Decimal("2.5")))  # 2 - banker's again
print(Decimal("2.5").quantize(Decimal("1"), rounding=ROUND_HALF_UP))   # 3
```

**9. Deterministic game night.** `seed` fixes the Mersenne Twister sequence, so the same calls replay identically forever.

```
import random

random.seed(42)               # reproducible from here on

rolls = [random.randint(1, 6) for _ in range(5)]
print("dice:", rolls)         # same five numbers on every run
print("move:", random.choice(["rock", "paper", "scissors"]))
winners = random.sample(["Sara", "Arif", "Nadia", "Rafi", "Mina"], 2)
print("winners:", winners)    # unique names, original list untouched
```
