## Part 1 — Warm-up

**1. Grab a seat.** `break` jumps PAST the loop the moment the goal is reached.

```python
seats = ["taken", "empty", "taken", "empty"]

for i, seat in enumerate(seats, start=1):
    if seat == "empty":
        print(f"Sitting in seat {i}")
        break                    # stop scanning, we have a seat
    print(f"Seat {i}: occupied, moving on...")

print("Loop is over.")           # break lands HERE
```

**2. Refund desk.** `continue` cancels only this pass; the loop itself carries on.

```python
orders = [
    {"id": 101, "amount": 500},
    {"id": 102, "amount": 0},
    {"id": 103, "amount": 1250},
]

total = 0
for order in orders:
    if order["amount"] <= 0:
        continue  # cancelled: skip the rest of this pass
    total += order["amount"]

print("Revenue:", total, "BDT")  # Revenue: 1750 BDT
```

**3. Deliberate no-op.** `pass` satisfies the 'block must contain something' rule while doing nothing.

```python
error_code = 404

if error_code == 404:
    pass                         # missing pages are expected; ignore them
else:
    print("Unexpected error:", error_code)

print("Server keeps running.")

class FutureFeature:              # stub class - pass's most common job
    pass


# pass is NOT continue: execution falls through to the very next
# statement in the same block. It exists purely as a placeholder.
```

## Part 2 — Practice

**4. Same loop, three ways.** break quits the loop, continue quits the pass, pass quits nothing.

```python
numbers = [1, 3, 4, 5, 7]

# break: abandons the WHOLE loop at the first even
for n in numbers:
    if n % 2 == 0:
        print("even found:", n)
        break
    print("odd:", n)
# odd: 1 / odd: 3 / even found: 4

# continue: cancels only THIS pass; the loop keeps walking
for n in numbers:
    if n % 2 == 0:
        continue
    print("odd:", n)
# odd: 1 / odd: 3 / odd: 5 / odd: 7

# pass: placeholder - the print below STILL runs for evens
for n in numbers:
    if n % 2 == 0:
        pass
    print("saw:", n)
# saw: 1 ... saw: 7 (all five numbers)
```

**5. Innermost only.** break binds to the innermost enclosing loop - Row 1 dies early, Row 0 never knew.

```python
grid = [[2, 4, 6], [8, 9, 10]]

for row_i, row in enumerate(grid):
    print("Row", row_i)
    for value in row:
        if value % 2 != 0:
            print("  odd value", value, "-> abandoning THIS row only")
            break                # exits the INNER loop; outer marches on
        print("  even:", value)

# Row 0 is all even and prints completely; Row 1 stops at 9 -
# yet the outer loop still moved on to print Row 1's header first.
# There is no 'break outer': the keyword only ever leaves ONE loop.
```

**6. First affordable pick.** Keep the winner in a variable, break immediately - don't keep scanning.

```python
products = [
    ("keyboard", 1500),
    ("mouse", 700),
    ("monitor", 9800),
    ("webcam", 2400),
]
budget = 1000

found = None
for name, price in products:
    if price <= budget:
        found = (name, price)
        break                   # first affordable product wins

print(f"Picked {found[0]} at {found[1]} BDT")   # Picked mouse at 700 BDT
```

**7. Stock check.** No break means the loop exhausted everything: the else says 'wasn't there'.

```python
products = [
    ("keyboard", 1500),
    ("mouse", 700),
    ("monitor", 9800),
    ("webcam", 2400),
]
target = "monitor"

for name, price in products:
    if name == target:
        print(target, "is in stock at", price, "BDT")
        break
else:
    print(target, "is out of stock")
# monitor is in stock at 9800 BDT
# For target = "printer": no break fires, so the else runs ->
# printer is out of stock.
```

## Part 3 — Challenge

**8. Escape both loops.** `return` exits the whole function - the cleanest way out of nested loops.

```python
def first_odd(grid):
    for row_i, row in enumerate(grid):
        for col_i, value in enumerate(row):
            if value % 2 != 0:
                return (row_i, col_i)   # escapes BOTH loops at once
    return None


print(first_odd([[2, 4], [6, 9]]))      # (1, 1)
```

**9. Rescue the retry loop.** Update the counter before any continue; else fires only when no break happened.

```python
MAX_RETRIES = 3
attempts = ["timeout", "timeout", "success"]

attempt = 0
while attempt < MAX_RETRIES:
    attempt += 1                # update FIRST, before any continue
    result = attempts[attempt - 1]
    if result == "timeout":
        continue                # safe now: the counter already moved
    print("Attempt", attempt, "->", result)
    break
else:
    print(f"Gave up after {MAX_RETRIES} attempts")

# Attempt 3 -> success
```
