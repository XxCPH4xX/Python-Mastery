# Break & Continue Exercises

Write your answer in the empty code cell under each question. Check yourself against `solutions.ipynb` when done.

**Part 1** = warm-up · **Part 2** = practice · **Part 3** = challenge.

## Part 1 — Warm-up

**1. Grab a seat.** For `seats = ["taken", "empty", "taken", "empty"]`, scan with `enumerate(seats, start=1)`, print `Seat N: occupied, moving on...` for taken ones, and `break` at the first free seat with `Sitting in seat 2`.

**2. Refund desk.** Orders carry `[{"id": 101, "amount": 500}, {"id": 102, "amount": 0}, {"id": 103, "amount": 1250}]`. Skip cancelled orders (`amount <= 0`) with `continue`, charge the rest, and print `Revenue: 1750 BDT`.

**3. Deliberate no-op.** With `error_code = 404`, write an `if...else` where a 404 deserves to do nothing at all (use `pass` plus a comment explaining why) and anything else prints `Unexpected error:` plus the code. Then declare a stub class `FutureFeature` whose whole body is `pass`.

## Part 2 — Practice

**4. Same loop, three ways.** For `numbers = [1, 3, 4, 5, 7]`, write three loops over the same list: one that `break`s at the first even number, one that `continue`s past evens, and one whose even branch holds only `pass`. Predict all three outputs in comments before running.

**5. Innermost only.** For `grid = [[2, 4, 6], [8, 9, 10]]`, loop over rows and then values, printing each even value, and `break` out of a row the moment you hit an odd one. In comments: which rows survive fully, and which loop does the `break` actually leave?

**6. First affordable pick.** Products cost keyboard 1500, mouse 700, monitor 9800, webcam 2400 (build the list of tuples yourself) and `budget = 1000`. Use the result-variable-plus-`break` idiom to stop at the first product you can afford and print `Picked mouse at 700 BDT`.

**7. Stock check.** Using the same four-product list, search for `target = "monitor"` with a `for...else` loop that reports it in stock (with its price), or - via the `else` - prints `<target> is out of stock`. Add a comment about what happens for a `target = "printer"`.

## Part 3 — Challenge

**8. Escape both loops.** Write `first_odd(grid)` returning the `(row, col)` position of the first odd value anywhere. Remember `break` cannot leave two loops, so wrap the nested search in a function and `return` from it. Test on `[[2, 4], [6, 9]]` and expect `(1, 1)`.

**9. Rescue the retry loop.** A teammate wrote a retry loop whose `continue` sits ABOVE `attempt += 1`, so one failure makes it spin forever. Rebuild it correctly with `MAX_RETRIES = 3` and simulated attempts `["timeout", "timeout", "success"]`: skip timeouts with `continue`, `break` on success, and use `while...else` to print `Gave up after 3 attempts` if retries run out.
