## Part 2 — Practice

**4. Same loop, three ways.** For `numbers = [1, 3, 4, 5, 7]`, write three loops over the same list: one that `break`s at the first even number, one that `continue`s past evens, and one whose even branch holds only `pass`. Predict all three outputs in comments before running.

```python
numbers = [1, 3, 4, 5, 7]

# break: abandons the WHOLE loop at the first even
for n in numbers:
    if n % 2 == 0:
        print("even found:", n)
        break
    print("odd:", n)
# odd: 1 / odd: 3 / even found: 4

# continue: cancels only THIS pass; the loop keeps walking
for n in numbers:
    if n % 2 == 0:
        continue
    print("odd:", n)
# odd: 1 / odd: 3 / odd: 5 / odd: 7

# pass: placeholder - the print below STILL runs for evens
for n in numbers:
    if n % 2 == 0:
        pass
    print("saw:", n)
# saw: 1 ... saw: 7 (all five numbers)
```
