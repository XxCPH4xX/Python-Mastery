# For Loop Exercises

Write your answer in the empty code cell under each question. Check yourself against `solutions.ipynb` when done.

**Part 1** = warm-up · **Part 2** = practice · **Part 3** = challenge.

## Part 1 — Warm-up

**1. Letter by letter.** Loop over the string `"Dhaka"` and print its characters on ONE line separated by spaces (hint: `end=" "`, then a bare `print()` afterwards).

**2. Fruit stand.** For `fruits = ["apple", "banana", "mango"]`, print one line per item like `Today's fruit: apple`.

**3. Range predictions.** Before running anything, write a comment predicting the numbers produced by `range(5)`, `range(25, 29)`, `range(500, 1001, 250)` and `range(10, 0, -2)`. Then print each sequence in its own small loop and check yourself.

## Part 2 — Practice

**4. Basket total.** For `prices = {"rice": 70, "eggs": 140, "milk": 95}`, walk `.items()` and print each line as `rice    70 BDT` (pad with format specs), then print `Basket total: 305 BDT`.

**5. Podium numbering.** For `runners = ["Tina", "Rahim", "Sarah"]`, use `enumerate()` with `start=1` to print `#1 Tina`, `#2 Rahim`, `#3 Sarah`.

**6. Study partners.** Pair `students = ["Sarah", "Rahim", "Tina"]` with `scores = [88, 92, 79]` using `zip()` and print `Sarah scored 88`-style lines. Then build a `marks` dictionary from the two lists in one line with `dict(zip(...))` and print it.

**7. At the enrollment desk.** With `enrolled = ["sarah", "rahim", "tina"]` and `target = "john"`, search with a `for` loop and use the loop `else` to print `john not found - sending offer email`. Add a comment describing what happens when `target = "tina"` instead.

## Part 3 — Challenge

**8. Times table.** Use two nested loops over `range(1, 6)` to print the 1-5 multiplication table: five rows, each product padded via `f"{row * col:>4}"`, staying on one line within a row and starting a new line after each row.

**9. Cleanup crew.** From `nums = [1, 2, 4, 5, 6]`: first demonstrate the bug - removing evens with `.remove()` *while* looping leaves a survivor; run it and print what's left. Then do it safely: build a brand-new `odds` list, skipping evens with `continue`, and print `[1, 5]`.
