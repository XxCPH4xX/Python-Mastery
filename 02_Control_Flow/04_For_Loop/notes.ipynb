# Python For Loops

> **Python Mastery** · Module 02 — Control Flow · Lesson 4/5

A `for` loop visits every item of a collection - every character of a name, every product in a cart, every row of a dataset. Unlike C or Java, Python's `for` is not a counting loop; it's a **"for each item in..."** loop, and that one idea works on strings, lists, tuples, sets, dicts, files, and more.

## 🎯 Learning Objectives

By the end of this lesson you will be able to:

- **Loop** over strings, lists, tuples, sets, and dictionaries (including `.items()`)
- **Generate** number sequences with `range(start, stop, step)`
- **Steer** loops with `break`, `continue`, and the loop-`else` clause
- **Combine** `enumerate()` and `zip()` instead of manual index juggling
- **Nest** loops to build grids like a times table
- **Modify** collections safely by building new lists instead of editing mid-loop

## 1. The `for...in` Statement

The header names a temporary variable and the collection to walk through; the indented body runs once per item. The loop variable (`fruit`, `ch`) is created fresh on each pass and still exists after the loop ends.

**Syntax:**
```python
for variable in iterable:
    # runs once per item
    ...
```

It works on any **iterable** - here are the everyday ones:

````python
fruits = ["apple", "banana", "mango"]

for fruit in fruits:
    print("Today's fruit:", fruit)

print("---")

for ch in "Dhaka":            # a string iterates character by character
    print(ch, end=" ")
print()

print("---")

dimensions = (1920, 1080)      # tuples work too
for value in dimensions:
    print(value, "px")
# Sets are also iterable - but remember they have NO guaranteed order.
````

## 2. Looping Over Dictionaries

A plain `for key in d:` walks the **keys**. For both halves at once, use `.items()`, which yields `(key, value)` pairs you can unpack into two variables.

**Syntax:**
```python
for key in my_dict:               # keys only
for value in my_dict.values():    # values only
for key, value in my_dict.items():  # both, unpacked
```

````python
prices = {"rice": 70, "eggs": 140, "milk": 95}   # BDT per pack

for item in prices:                 # keys by default
    print(item)

print("---")

total = 0
for item, price in prices.items():  # unpack each pair
    total += price
    print(f"{item:<6} {price:>3} BDT")

print("Basket total:", total, "BDT")
````

## 3. Counting with `range()`

When you *do* want numbers, `range()` builds them lazily. It follows the same stop-rule as slicing: the stop value is **never included**.

| Call | Produces |
|---|---|
| `range(5)` | `0 1 2 3 4` |
| `range(2, 6)` | `2 3 4 5` |
| `range(0, 20, 5)` | `0 5 10 15` |
| `range(10, 0, -2)` | `10 8 6 4 2` |

> 🔍 **Under the Hood:** In Python 2, `range()` built the whole list in memory. Python 3's `range` is a lazy sequence object - it computes each number on demand with simple arithmetic, so `range(1_000_000_000)` costs about as much memory as `range(10)`. That laziness is why `sum(range(...))` over huge spans stays fast. Underneath *every* `for` loop, Python calls `iter()` on the collection and keeps calling `next()` until a `StopIteration` signal ends the loop.

````python
# One-arg: start at 0, stop excluded
for n in range(5):
    print(n, end=" ")
print()

# Two-arg: custom start
for day in range(25, 29):
    print(f"Aug {day}")

# Three-arg: step
for taka in range(500, 1001, 250):
    print("Installment:", taka, "BDT")

# Negative step counts DOWN
for n in range(10, 0, -2):
    print(n, end=" ")
print()

# range + len trick for indices (prefer enumerate, shown later)
names = ["Sarah", "Rahim", "Tina"]
for i in range(len(names)):
    print(i, "->", names[i])
````

## 4. Steering Loops: `break` and `continue`

Both keywords from the while-loop lesson work identically here:

- `break` - leave the loop entirely.
- `continue` - abandon this item, jump to the next.

**Syntax:**
```python
for item in items:
    if skip_it:
        continue
    if done_searching:
        break
    ...work...
```

````python
queue = ["ann", "spam-bot", "raj", "ad-bot", "mia"]

processed = []
for user in queue:
    if user.endswith("bot"):
        continue          # bots: skip silently
    processed.append(user)
    if user == "raj":
        print("VIP found - priority handling, stopping queue")
        break             # nobody after raj gets processed now

print("Processed:", processed)
````

## 5. The Loop `else`

Same rule as while: the `else` block runs only when the loop **wasn't interrupted by `break`** - i.e., it ran through *every* item. The classic use is search: *"if we got through everything without finding it, say so."*

**Syntax:**
```python
for item in items:
    if matches(item):
        break
else:
    # no break -> nothing matched
    ...
```

````python
enrolled = ["sarah", "rahim", "tina"]
target = "john"

for student in enrolled:
    if student == target:
        print(target, "is already enrolled")
        break
else:
    print(target, "not found - sending offer email")
# Swap target to 'tina' and re-run: break fires, else is skipped.
````

## 6. Nested Loops

A loop inside a loop runs its inner body **completely** for every single outer pass - which is why nested loops multiply their work (outer × inner). Great for grids and pairings; watch out on giant data.

**Syntax:**
```python
for outer in outers:
    for inner in inners:
        ...   # runs len(outers) * len(inners) times
```

````python
print("Times table 1-5:")
for row in range(1, 6):
    for col in range(1, 6):
        print(f"{row * col:>4}", end="")   # stay on this line
    print()                                  # newline after each row

print()
# Pairing example: every shirt with every trouser
shirts = ["white", "black"]
trousers = ["jeans", "khaki", "slacks"]
count = 0
for shirt in shirts:
    for trousers_ in trousers:
        count += 1
        print(f"Outfit {count}: {shirt} shirt + {trousers_}")
````

## 7. `enumerate()`: Index and Value Together

Need the position *and* the item? Skip `range(len(...))` bookkeeping - `enumerate()` hands you both, neatly unpacked. Pass `start=1` to number things like a human.

**Syntax:**
```python
for index, value in enumerate(iterable):
    ...

for rank, value in enumerate(iterable, start=1):
    ...
```

````python
runners = ["Tina", "Rahim", "Sarah"]

for i, name in enumerate(runners):
    print(i, name)

print("--- podium style ---")

for rank, name in enumerate(runners, start=1):
    medal = "🥇" if rank == 1 else ("🥈" if rank == 2 else "🥉")
    print(f"{medal} #{rank}: {name}")
````

## 8. `zip()`: Walking Two Lists in Step

Two parallel lists - names here, scores there - become paired up by `zip()`. It stops at whichever list runs out first (pass `strict=True` on Python 3.10+ to raise an error instead, catching mismatched data early).

**Syntax:**
```python
for a, b in zip(list_a, list_b):
    ...
```

````python
students = ["Sarah", "Rahim", "Tina"]
scores = [88, 92, 79]

for name, score in zip(students, scores):
    status = "PASS" if score >= 50 else "FAIL"
    print(f"{name:<7} {score:>3} {status}")

print("---")

# Three lists at once? zip takes any number
roll = [1, 2, 3]
grades = ["A", "A+", "B"]
for r, n, g in zip(roll, students, grades):
    print(f"Roll {r}: {n} ({g})")

# zip also builds dicts in one line
marks = dict(zip(students, scores))
print(marks)
````

> 🔍 **Under the Hood:** A `for` loop tracks its position inside the collection by an internal counter handed to `next()`. If you `remove()` an item mid-loop, everything after shifts left by one slot - but the counter doesn't know - so the element that moved into the just-visited slot is **skipped silently**, no error raised. Adding items during iteration can even change how many passes happen. The fix in both cases: never mutate the thing you're iterating; iterate over a copy, or build a brand-new list.

````python
# BUGGY: removing while looping - an even number sneaks through
buggy = [1, 2, 4, 5, 6]
for n in buggy:
    if n % 2 == 0:
        buggy.remove(n)
print("After buggy removal:", buggy)   # 4 survived - it was SKIPPED

# SAFE approach 1: build a NEW list (usually clearest)
nums = [1, 2, 4, 5, 6]
odds = []
for n in nums:
    if n % 2 != 0:
        odds.append(n)
nums = odds
print("Rebuilt list:", nums)

# SAFE approach 2: iterate over a copy, mutate the original
cart = ["milk", "candy", "eggs"]
for item in cart[:]:              # cart[:] is a shallow copy
    if item == "candy":
        cart.remove(item)
print("Cart now:", cart)
````

## ⚠️ Common Mistakes & Gotchas

| Mistake | Problem | Fix |
|---|---|---|
| Expecting `range(1, 5)` to include 5 | Stop value is always excluded | Use `range(1, 6)` |
| Removing/appending to a list while looping over it | Items skipped or extra passes - silent bugs | Build a new list, or loop over `list[:]` |
| Adding keys to a dict while looping it | `RuntimeError: dictionary changed size during iteration` | Collect changes first, apply after the loop |
| Trusting set order in output | Sets are unordered - results may print differently | Sort first (`sorted(my_set)`) if order matters |
| `for i in range(len(data)):` then using only `data[i]` | Works, but noisy and un-Pythonic | `for item in data:` or `enumerate()` |
| Reusing the loop variable afterwards expecting clean state | It still holds the LAST item | Don't rely on it; pick a clearly-named result variable |

## 💡 Best Practices & Pro Tips

- **Name the loop variable after one item**, not the whole collection: `for student in students:` reads like English.
- **Prefer `enumerate()` / `zip()`** over index arithmetic - fewer moving parts, fewer off-by-one bugs.
- **Keep loop bodies short.** If a body grows past ~10 lines, move it into a well-named function (Module 03).
- **Filter-and-build beats filter-in-place**: constructing a new list is safe, obvious, and previews list comprehensions (`[n for n in nums if n % 2]`) coming in Module 04.
- **AI-engineering relevance:** nearly all data prep is a for loop - iterating dataset rows, zipping features with labels (`for x, y in zip(X, y):`), batching samples into chunks before feeding a model. Getting comfortable with `zip`/`enumerate` now makes NumPy/Pandas vectorization feel natural later, because you'll recognize exactly *which* loop is being replaced.

## 📌 Summary

| Tool | What it does | Example |
|---|---|---|
| `for x in seq:` | run once per item | `for fruit in fruits:` |
| `dict.items()` | yield (key, value) pairs | `for k, v in prices.items():` |
| `range(a, b, step)` | numbers a..b-1 by step | `range(0, 10, 2)` |
| `enumerate(seq)` | index + value together | `for i, v in enumerate(xs, 1):` |
| `zip(a, b)` | walk lists in parallel | `for n, s in zip(names, scores):` |
| `break` / `continue` | exit early / skip one item | `if hit: break` |
| loop `else` | runs only if no break | search-not-found pattern |

Key takeaways:

- Python's `for` means **for-each**: hand it any iterable, no counters required.
- `range()` excludes its stop value - same mental model as slices.
- Never mutate a collection while looping it; build a new one or iterate a copy.
- `enumerate` replaces index juggling; `zip` replaces parallel-list gymnastics.
