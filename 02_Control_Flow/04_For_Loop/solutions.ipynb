## Part 1 — Warm-up

**1. Letter by letter.** Strings iterate character by character; `end=` keeps prints on the same line.

```python
for ch in "Dhaka":
    print(ch, end=" ")
print()
```

**2. Fruit stand.** The body runs once per item; the loop variable holds the current one.

```python
fruits = ["apple", "banana", "mango"]

for fruit in fruits:
    print("Today's fruit:", fruit)
```

**3. Range predictions.** Stop is always excluded; a negative step counts down.

```python
# Predictions:
#   range(5)              -> 0 1 2 3 4
#   range(25, 29)         -> 25 26 27 28
#   range(500, 1001, 250) -> 500 750 1000
#   range(10, 0, -2)      -> 10 8 6 4 2

for n in range(5):
    print(n, end=" ")
print()

for n in range(25, 29):
    print(n, end=" ")
print()

for n in range(500, 1001, 250):
    print(n, end=" ")
print()

for n in range(10, 0, -2):
    print(n, end=" ")
print()
```

## Part 2 — Practice

**4. Basket total.** .items() yields (key, value) pairs you can unpack into two variables.

```python
prices = {"rice": 70, "eggs": 140, "milk": 95}

total = 0
for item, price in prices.items():
    total += price
    print(f"{item:<6}{price:>4} BDT")

print("Basket total:", total, "BDT")
```

**5. Podium numbering.** enumerate hands you index and value together - no range(len(...)) juggling.

```python
runners = ["Tina", "Rahim", "Sarah"]

for rank, name in enumerate(runners, start=1):
    print(f"#{rank} {name}")
```

**6. Study partners.** zip walks two lists in step; dict(zip(keys, values)) builds a mapping instantly.

```python
students = ["Sarah", "Rahim", "Tina"]
scores = [88, 92, 79]

for name, score in zip(students, scores):
    print(f"{name} scored {score}")

marks = dict(zip(students, scores))
print(marks)
```

**7. At the enrollment desk.** else runs only when the loop finished without breaking - the not-found case.

```python
enrolled = ["sarah", "rahim", "tina"]
target = "john"

for student in enrolled:
    if student == target:
        print(target, "is already enrolled")
        break
else:
    print(target, "not found - sending offer email")

# With target = "tina": the break fires on the third pass, so the
# else block is skipped and only 'tina is already enrolled' prints.
```

## Part 3 — Challenge

**8. Times table.** The inner loop completes fully for every outer pass - rows here, columns there.

```python
for row in range(1, 6):
    for col in range(1, 6):
        print(f"{row * col:>4}", end="")
    print()
```

**9. Cleanup crew.** Removing mid-loop shifts items under your feet; building a new list never skips.

```python
nums = [1, 2, 4, 5, 6]

# BUGGY: removals shift everything after the removed item one slot
# left, and the element sliding into the visited slot gets skipped.
for n in nums:
    if n % 2 == 0:
        nums.remove(n)
print("After buggy removal:", nums)  # [1, 4, 5] - the 4 sneaked through!

# SAFE: never mutate what you iterate; build a new list instead.
nums = [1, 2, 4, 5, 6]
odds = []
for n in nums:
    if n % 2 == 0:
        continue  # abandon this pass, move to the next item
    odds.append(n)
print("Rebuilt list:", odds)  # [1, 5]
```
