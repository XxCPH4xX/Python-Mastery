## 1. Indentation Defines the Block

Most languages wrap a block of code in curly braces `{ }`. Python uses **whitespace** instead: every line indented under a decision *belongs* to that decision. The standard indent is **4 spaces** (PEP 8), and every IDE inserts them for you with Tab.

Think of indentation as the code's family tree - the indented lines are the *children* of the `if` line above them.

**Syntax:**
```python
if condition:
    # these lines run ONLY when condition is True
    do_this()
    also_this()
# this line is NOT indented -> it always runs
```

````python
temperature = 34  # a hot summer day in Dhaka

if temperature > 30:
    print("It's hot today!")
    print("Drink plenty of water.")

print("This line always runs, rain or shine.")
# Both indented lines belong to the `if`.
# Change temperature to 22 and re-run: the two messages disappear,
# but the last print still executes.
````

> 🔍 **Under the Hood:** Indentation is not cosmetic in Python - it *is* the grammar. When CPython reads your file, its tokenizer emits special `INDENT` and `DEDENT` markers every time the indent level changes, and the parser builds the program tree from those markers. This is why mixing tabs and spaces raises `TabError`: the tokenizer cannot agree on how wide one "step" is. Pick 4 spaces, configure your editor to turn Tab into spaces, and never think about it again.

## 2. The `if` Statement

An `if` statement runs its block **only when** the condition is `True`. If the condition is `False`, Python simply skips the block and moves on - no error, no message.

Conditions are built with **comparison operators**, which always produce `True` or `False`:

| Operator | Meaning | Example |
|---|---|---|
| `==` | equal to | `score == 100` |
| `!=` | not equal to | `password != guess` |
| `>` `<` | greater / less than | `marks > 49` |
| `>=` `<=` | greater/less or equal | `age >= 18` |

**Syntax:**
```python
if condition:
    statement(s)   # runs only when condition is True
```

````python
score = 850

if score >= 500:
    print("Level up unlocked!")

if score < 500:
    print("Keep grinding...")

print("Game over. Final score:", score)
# Only ONE of the two ifs fires - they are checked independently.
````

## 3. `if...else`

When you want exactly one of **two** paths to run, attach an `else`. The `else` takes no condition - it is simply *"everything the `if` didn't catch"*.

**Syntax:**
```python
if condition:
    # runs when condition is True
else:
    # runs when condition is False
```

````python
marks = 48

if marks >= 50:
    print("Result: PASS 🎉")
else:
    print("Result: FAIL - see you at the retake")

print(f"Marks obtained: {marks}/100")
````

## 4. `if...elif...else`

For **more than two** outcomes, chain conditions with `elif` (short for *"else if"*). Python checks them **top to bottom** and runs **only the first** whose condition is `True` - everything after is skipped, even if it would also be true. A final `else` catches whatever fell through.

**Syntax:**
```python
if condition_1:
    ...
elif condition_2:      # checked only if condition_1 was False
    ...
elif condition_3:
    ...
else:
    ...                # none of the above were True
```

````python
score = 78

if score >= 90:
    grade = "A+"
elif score >= 80:
    grade = "A"
elif score >= 70:
    grade = "B"
elif score >= 60:
    grade = "C"
else:
    grade = "F"

print(f"Score {score} -> grade {grade}")
# Trace: 78 fails >= 90, fails >= 80, passes >= 70 -> 'B'.
# The lower branches are never even looked at.
````

## 5. `elif` Chains vs Many Separate `if`s

These look similar but behave differently:

- **Separate `if`s** are independent questions. Every one gets asked, and *several can answer yes*.
- **An `elif` chain** is one question with many options. The first match wins; the rest are skipped.

Use separate `if`s when the situations genuinely overlap. Use `elif` when the options are mutually exclusive.

````python
age = 65

# --- Separate ifs: BOTH messages can fire ---
if age >= 18:
    print("Charges as adult")
if age >= 60:
    print("Gets senior discount")

# --- elif chain: exactly ONE label sticks ---
if age >= 60:
    category = "senior"
elif age >= 18:
    category = "adult"
else:
    category = "child"

print("Category:", category)
````

## 6. Nested `if`

An `if` inside another `if` checks a second thing **only after** the first passes. Great for step-by-step gates (ticket first, then age), but keep nesting to 2 levels deep - beyond that, code becomes hard to read.

**Syntax:**
```python
if outer_condition:
    if inner_condition:
        ...   # both True
    else:
        ...   # outer True, inner False
else:
    ...       # outer False
```

````python
has_ticket = True
age = 16

if has_ticket:
    if age >= 18:
        print("Enjoy the movie!")
    else:
        print("Sorry, this film is adults-only.")
else:
    print("Please buy a ticket first.")

# Same logic, flatter with `and`:
if has_ticket and age >= 18:
    print("(Flat version) Enjoy the movie!")
````

## 7. Short-Hand `if` and the Ternary Expression

Two compact forms:

1. **Short-hand `if`** - the whole statement on one line. Handy for tiny actions.
2. **Ternary (conditional expression)** - an `if...else` that produces a **value**, so it fits on the right side of an `=`.

Read the ternary out loud as: *"`x` if the condition holds, otherwise `y`."*

**Syntax:**
```python
# short-hand if
if condition: do_something()

# ternary expression
value = x if condition else y
```

````python
a, b = 12, 30

# Short-hand if
if a < b: print("a is smaller")

# Ternary: pick a value in one line
age = 20
ticket_type = "full fare" if age >= 18 else "half fare"
print(ticket_type)

# Realistic use: member pricing
price = 1200          # BDT
is_member = True
payable = price * 0.9 if is_member else price
print(f"Payable: {payable:.0f} BDT")

# Ternaries can chain, but readability drops fast -
# use a normal if...elif...else once it gets this dense
size_label = "S" if age < 13 else ("M" if age < 20 else "L")
print(size_label)
````

## 8. Combining Conditions: `and`, `or`, `not`

Real decisions usually depend on **several facts at once**:

| Operator | Runs when... | Example |
|---|---|---|
| `and` | **both** sides are True | `has_ticket and age >= 18` |
| `or` | **at least one** side is True | `day == "fri" or day == "sat"` |
| `not` | flips True <-> False | `not is_banned` |

> 🔍 **Under the Hood:** Python **short-circuits** these operators. In `A and B`, if `A` is falsy, `B` is never even evaluated - Python already knows the answer. Likewise `A or B` skips `B` when `A` is truthy. That makes cheap-first ordering a real optimization: write the inexpensive check first (`if user and user.is_active:`), and the second check protects itself from running on missing data.

````python
balance = 900        # BDT
item_price = 850
kyc_verified = True

if balance >= item_price and kyc_verified:
    print("Purchase approved")

day = "Friday"
if day == "Friday" or day == "Saturday":
    print("Weekend schedule applies")

maintenance_mode = False
if not maintenance_mode:
    print("Service is up - processing request")
````

## 9. Truthiness: Values That Act Like True/False

Any Python value can stand alone as a condition - Python quietly converts it to `True` or `False`. These are the **falsy** values (everything else is truthy):

`False`, `None`, `0`, `0.0`, `""` (empty string), `[]` (empty list), `{}` (empty dict), `()` , `set()`

So `if cart:` naturally means *"if the cart has anything in it"* - no `len(cart) > 0` needed.

> 🔍 **Under the Hood:** Truthiness comes from each type's `__bool__` method (falling back to `__len__` for containers). `bool([])` calls `list.__len__`, gets `0`, and reports falsy. Custom classes can join in: define `__bool__` and your own objects decide whether `if obj:` is true. Careful trap: `"False"` and `[0]` are **non-empty**, hence truthy.

````python
cart = []               # empty list -> falsy

if cart:
    print("Proceeding to checkout...")
else:
    print("Your cart is empty")

attempts_left = 0       # zero -> falsy
if not attempts_left:
    print("Account locked. Try again later.")
else:
    print(f"{attempts_left} attempts left")

# See the conversion yourself with bool():
print(bool("hello"), bool(""), bool(42), bool(0.0), bool([0]), bool(None))
````

## 10. `pass` - Doing Nothing on Purpose

Python refuses to run a block with nothing in it. When you want to sketch a branch now and fill it in later, put `pass` there as a placeholder. It executes successfully and changes nothing.

`pass` is a real statement; a `# comment` is not. A block containing only a comment still crashes with `SyntaxError: expected an indented block`, because the parser sees no statement there at all.

**Syntax:**
```python
if condition:
    pass   # TODO: implement later
```

````python
subscription = "free"

if subscription == "pro":
    pass   # TODO: unlock analytics dashboard (coming in sprint 4)
else:
    print("Upgrade to Pro to unlock analytics.")

class PaymentGateway:      # you will meet classes in Module 08
    pass                   # 'stub' classes are pass's most common job
````

## ⚠️ Common Mistakes & Gotchas

| Mistake | Problem | Fix |
|---|---|---|
| `if score = 100:` | `=` assigns, so Python raises `SyntaxError` | Compare with `==` |
| `if score > 80` (no colon) | `SyntaxError: expected ':'` | Every `if`/`elif`/`else` header ends with `:` |
| Random indent widths (3 spaces, then 5) | `IndentationError: unexpected indent` | Consistent 4 spaces throughout |
| `if x == 1 or 2:` | Always `True` - `2` by itself is truthy, so the whole `or` wins | `if x == 1 or x == 2:` |
| Building a tower of nested `if`s 5 levels deep | Unreadable, easy to lose track of which block you're in | Combine with `and`, or return early ("guard clauses") |

````python
x = 5

# WRONG - this prints for EVERY value of x:
if x == 1 or 2:
    print("bug: entered the branch anyway")

# RIGHT - repeat the full comparison on each side:
if x == 1 or x == 2:
    print("x really is 1 or 2")
else:
    print("correct: x is neither 1 nor 2")

# Even tidier with `in`:
if x in (1, 2):
    print("membership check - cleanest of all")
````

## 💡 Best Practices & Pro Tips

- **Test the positive, invert with `not`**: `if items:` reads better than `if len(items) > 0:`; `if not error:` beats `if error == False:` (never write `== True`/`== False` at all).
- **Order `elif` branches from most specific to most general** - otherwise a broad early condition swallows the narrow ones behind it.
- **Prefer guard clauses**: handle the bad case first and `return`/`continue`, keeping the happy path un-indented at the top level.
- **Reserve ternaries for picking between two obvious values**; the moment a branch needs side effects, go back to a full `if...else`.
- **AI-engineering relevance:** control flow like this is everywhere in ML glue code - feature flags (`if ENABLE_CACHE:`), quality gates (`if val_accuracy >= 0.90: deploy()`), and cleaning pipelines that branch on missing fields (`if not row.get("email"): skip`). Mastering readable branching now pays off in every pipeline you write later.

## 📌 Summary

| Construct | What it does | Example |
|---|---|---|
| `if cond:` | runs block only when cond is True | `if score > 80:` |
| `else:` | fallback when no branch matched | `else:` |
| `elif cond:` | extra alternative, checked top-down | `elif score > 70:` |
| `and` / `or` / `not` | combine or flip conditions | `if age >= 18 and has_id:` |
| `x if cond else y` | ternary - picks a value inline | `"pass" if m >= 50 else "fail"` |
| `pass` | legal no-op placeholder | `if todo: pass` |

Key takeaways:

- **Indentation is the syntax** - 4 spaces define what belongs to each branch.
- An `elif` chain stops at the **first** True; separate `if`s all get their chance.
- Empty things are falsy: write `if cart:` instead of counting lengths.
- `=` stores, `==` asks. Forgetting the colon is the #1 first-week error.
