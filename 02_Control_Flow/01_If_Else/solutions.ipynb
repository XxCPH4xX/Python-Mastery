## Part 1 — Warm-up

**1. Hot-day alert.** Indentation binds both prints to the `if`; the last line sits outside the block.

```
temperature = 34

if temperature > 30:
    print("It's hot today!")
    print("Drink plenty of water.")

print("This line always runs, rain or shine.")
# Set temperature = 22 and only the last line prints.
```

**2. Pass or fail.** `else` catches everything the `if` didn't - exactly one of the two lines runs.

```
marks = 48

if marks >= 50:
    print("Result: PASS")
else:
    print("Result: FAIL - see you at the retake")
# Result: FAIL - see you at the retake
```

**3. Fare in one line.** A ternary is an `if...else` that *produces a value*, so it fits on the right of `=`.

```
age = 20
ticket_type = "full fare" if age >= 18 else "half fare"
print(ticket_type)   # full fare
```

## Part 2 — Practice

**4. Grade ladder.** 78 fails >= 90 and >= 80, passes >= 70 - the chain stops at the first True.

```
score = 78

if score >= 90:
    grade = "A+"
elif score >= 80:
    grade = "A"
elif score >= 70:
    grade = "B"
elif score >= 60:
    grade = "C"
else:
    grade = "F"

print(f"Score {score} -> grade {grade}")   # Score 78 -> grade B
```

**5. Three little gates.** One gate per fact: `and` demands both, `or` accepts either, `not` flips the flag.

```
balance = 900
item_price = 850
kyc_verified = True
day = "Friday"
maintenance_mode = False

if balance >= item_price and kyc_verified:
    print("Purchase approved")

if day == "Friday" or day == "Saturday":
    print("Weekend schedule applies")

if not maintenance_mode:
    print("Service is up")
```

**6. Empty means falsy.** `[]` and `0` are falsy, so `if cart:` alone asks 'does it have anything?'

```
cart = []
attempts_left = 0

if cart:
    print("Proceeding to checkout...")
else:
    print("Your cart is empty")

if attempts_left:
    print(f"You still have {attempts_left} attempts left")
else:
    print("Account locked")

print(bool("hello"), bool(""), bool(0.0))   # True False False
```

**7. One question or many.** Separate `if`s are independent questions; an `elif` chain stops at the first match.

```
age = 65

# Separate ifs: both messages get their chance
if age >= 18:
    print("Charges as adult")
if age >= 60:
    print("Gets senior discount")

# elif chain: first True wins, the rest are never checked
if age >= 60:
    category = "senior"
elif age >= 18:
    category = "adult"
else:
    category = "child"

print("Category:", category)   # Category: senior
```

## Part 3 — Challenge

**8. Ticket, then age.** The inner check only ever runs once the outer one passed - then `and` flattens it.

```
has_ticket = True
age = 16

if has_ticket:
    if age >= 18:
        print("Enjoy the movie!")
    else:
        print("Sorry, this film is adults-only.")
else:
    print("Please buy a ticket first.")

if has_ticket and age >= 18:
    print("(Flat version) Enjoy the movie!")
```

**9. Syntax clinic.** `=` stores while `==` asks, headers need `:`, and a bare truthy `2` hijacks `or`.

```
# (a) '=' assigns instead of comparing -> SyntaxError. Equality is '=='.
# (b) Missing ':' after the condition -> SyntaxError.
# (c) '2' by itself is truthy, so 'x == 1 or 2' is ALWAYS True.

score = 100
if score == 100:
    print("Perfect score!")

score = 85
if score > 80:
    print("Well above 80")

x = 5
if x == 1 or x == 2:          # repeat the full comparison on each side
    print("x is 1 or 2")
else:
    print("x is neither 1 nor 2")

if x in (1, 2):               # membership test - cleanest fix of all
    print("x is 1 or 2")
else:
    print("still neither")
```
