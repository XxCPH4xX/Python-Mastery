## 🔴 Exercise 9 — Nested `if` with Logical Operators (Hard)

Simulate an **ATM withdrawal**:

- If `has_card` is False → print **"Insert card first"**
- Else if `pin_correct` is False → print **"Incorrect PIN"**
- Else if `balance >= amount` → print **"Withdrawal successful"**
- Else → print **"Insufficient funds"**

**Use a flat structure with `elif` (not deeply nested `if`s).**

**Starter code:**

```python
has_card = True
pin_correct = True
balance = 5000
amount = 6000

# Your code here
if has_card:
    if pin_correct:
        if balance >= amount:
            print("Transaction successful")
        else:
            print("Insufficient balance")
    else:
        print("Incorrect PIN")
else:
    print("Insert card first")
```

```python
has_card = True
pin_correct = True
balance = 5000
amount = 6000

# Your code here
if not has_card:
    print("Insert card first")
elif not pin_correct:
    print("Incorrect PIN")
elif balance < amount:
    print("Insufficient balance")
else:
    print("Transaction successful")
```

## 🔴 Exercise 10 — Real-World Scenario: Shipping Calculator (Hard)

Write a program that calculates shipping cost based on:

- `weight` (kg): base cost is `weight * 50` BDT
- `is_member`: members get **10% discount** on base cost
- `is_express`: express shipping adds **flat 200 BDT** (after discount)
- `is_fragile`: fragile items add **15% surcharge** on the total so far

Print the final cost formatted to 2 decimal places.

**Use ternary expressions for the conditional adjustments.**

**Starter code:**

```python
weight = 2.5
is_member = True
is_express = True
is_fragile = False

# Your code here
primary_cost = float(weight * 50)

final_cost = primary_cost if not is_member else primary_cost * 0.9
final_cost = final_cost + 200 if is_express else final_cost
final_cost = final_cost * 1.15 if is_fragile else final_cost

print(f"Final shipping cost: {final_cost:.2f} BDT")
```
