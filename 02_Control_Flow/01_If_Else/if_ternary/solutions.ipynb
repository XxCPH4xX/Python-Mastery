## 🟢 Exercise 3 — `if...elif...else` Chain (Easy)

**Solution:**

````python
score = 82

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
elif score >= 60:
    grade = "D"
else:
    grade = "F"

print(grade)
````

## 🔴 Exercise 9 — Nested `if` with Logical Operators (Hard)

**Solution:**

Note: This uses a flat `elif` chain as requested, avoiding deep nesting.

````python
has_card = True
pin_correct = True
balance = 5000
amount = 6000

if not has_card:
    print("Insert card first")
elif not pin_correct:
    print("Incorrect PIN")
elif balance >= amount:
    print("Withdrawal successful")
else:
    print("Insufficient funds")
````

## 🔴 Exercise 10 — Real-World Scenario: Shipping Calculator (Hard)

**Solution:**

Applies adjustments in order: base → member discount → express fee → fragile surcharge

````python
weight = 2.5
is_member = True
is_express = True
is_fragile = False

# Base cost
cost = weight * 50

# Member discount (10% off base)
cost = cost * 0.9 if is_member else cost

# Express shipping (flat 200 after discount)
cost = cost + 200 if is_express else cost

# Fragile surcharge (15% on total so far)
cost = cost * 1.15 if is_fragile else cost

print(f"Final shipping cost: {cost:.2f} BDT")
````

## 💡 Explanation Notes

### Exercise 8 — Chained Ternary
The parentheses are important for readability and correct evaluation order:
```python
category = "Child" if age < 13 else ("Teen" if age < 20 else ("Adult" if age < 60 else "Senior"))
```
Read right-to-left: each `else` clause contains the next ternary.

### Exercise 9 — Flat `elif` vs Nested `if`
The flat version is preferred:
- Easier to read
- Each condition checked in order
- No deep indentation

### Exercise 10 — Order of Operations
The order matters:
1. Base: `weight * 50`
2. Member discount: 10% off **base**
3. Express: +200 **after** discount
4. Fragile: 15% on **total so far**

With the given inputs:
- Base: 2.5 × 50 = 125
- Member: 125 × 0.9 = 112.5
- Express: 112.5 + 200 = 312.5
- Fragile: False → no change
- Final: 312.50 BDT
