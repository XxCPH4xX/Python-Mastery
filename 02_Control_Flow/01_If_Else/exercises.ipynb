# If/Else Exercises

Write your answer in the empty code cell under each question. Check yourself against `solutions.ipynb` when done.

**Part 1** = warm-up · **Part 2** = practice · **Part 3** = challenge.

## Part 1 — Warm-up

**1. Hot-day alert.** Let `temperature = 34`. Write an `if` that prints two reminders *only* when `temperature > 30`, then one unindented line that prints no matter what. Change the value to `22` and re-run to watch the indented lines vanish.

**2. Pass or fail.** With `marks = 48`, use an `if...else` to print `Result: PASS` when the marks are 50 or more and `Result: FAIL - see you at the retake` otherwise.

**3. Fare in one line.** Let `age = 20`. Using a ternary expression, set `ticket_type` to `"full fare"` when `age >= 18`, else `"half fare"`, and print it.

## Part 2 — Practice

**4. Grade ladder.** With `score = 78`, build an `if...elif...else` chain that assigns `A+` from 90, `A` from 80, `B` from 70, `C` from 60, and `F` below that. Print a line like `Score 78 -> grade B`.

**5. Three little gates.** Given `balance = 900`, `item_price = 850`, `kyc_verified = True`, `day = "Friday"` and `maintenance_mode = False`: print `Purchase approved` when the balance covers the price **and** KYC is verified; print `Weekend schedule applies` when the day is Friday **or** Saturday; print `Service is up` when the system is **not** in maintenance mode. Use each of `and`, `or`, `not` once.

**6. Empty means falsy.** With `cart = []` and `attempts_left = 0`, branch on the values directly - no `len()` and no `== 0` - to print `Your cart is empty` vs `Proceeding to checkout...`, and `Account locked` vs an attempts-left message. Finish by printing `bool("hello")`, `bool("")` and `bool(0.0)` to watch the conversion happen.

**7. One question or many.** With `age = 65`: first ask two **separate** `if`s (`age >= 18` charges as adult, `age >= 60` gets senior discount) and notice both can fire. Then build a single `if...elif...else` chain that sticks exactly one label on the person - `senior`, `adult` or `child` - and print it.

## Part 3 — Challenge

**8. Ticket, then age.** A cinema checks two gates in a row: `has_ticket = True` and `age = 16`. Using a **nested** `if`, print `Enjoy the movie!`, `Sorry, this film is adults-only.` or `Please buy a ticket first.` as appropriate. Then rewrite the happy path flat with one condition using `and`.

**9. Syntax clinic.** Three patients await diagnosis. In comments, name each disease, then write the cured runnable version: (a) `if score = 100:` which should test equality; (b) `if score > 80` which refuses to run at all; (c) `if x == 1 or 2:` with `x = 5`, which wrongly enters the branch for *every* x. For (c), show both the repeated-comparison fix and the `x in (1, 2)` version.
