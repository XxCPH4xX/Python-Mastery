## 1. The `while` Statement

A `while` loop keeps running its block while the condition stays `True`. Each round works like this:

1. Test the condition. False? Skip past the whole loop - done.
2. True? Run the indented block.
3. Go back to step 1.

Because the test happens **before** each pass, a condition that starts `False` means the body runs **zero times**.

**Syntax:**
```python
while condition:
    # body: repeats until condition becomes False
    ...
    # something must eventually make the condition False!
```

````python
countdown = 5

while countdown > 0:
    print(countdown)
    countdown -= 1        # shrink the counter each pass

print("Lift off!")
# Comment out the countdown -= 1 line and this loop would never end.
````

## 2. Leaving Early with `break`

`break` aborts the loop **immediately** and jumps to the first line after it. Classic use: stop as soon as you've found what you were looking for.

Since real `input()` would block these notebooks, we simulate typed guesses with a list - same logic, runnable output.

**Syntax:**
```python
while True:          # 'forever' on purpose...
    ...
    if found_what_we_want:
        break        # ...but we always have an exit
```

````python
secret = 7
guesses = [3, 9, 7, 1]     # Simulated input: player typed these

i = 0
while i < len(guesses):
    guess = guesses[i]
    print("Guess:", guess)
    if guess == secret:
        print("Correct! Unlocked in", i + 1, "tries.")
        break               # stop searching - no point continuing
    i += 1

print("Game finished.")
````

## 3. Skipping a Round with `continue`

`continue` abandons the **current** pass and jumps straight back to the condition test. Everything below it in the block is skipped for that round only; the loop itself carries on.

> ⚠️ In a `while` loop, `continue` jumps to the **condition**, not to "the next statement". If your counter update sits below the `continue`, it never runs - and you've built an infinite loop. Update **first**, then decide whether to continue.

**Syntax:**
```python
while condition:
    update_counter()      # do this BEFORE any continue
    if should_skip:
        continue
    ...rest of the work...
```

````python
readings = [22, -1, 25, -7, 30]   # negative = faulty sensor

i = 0
while i < len(readings):
    value = readings[i]
    i += 1                 # update FIRST so continue can't trap us
    if value < 0:
        continue           # discard bad reading, move to the next
    print(f"Valid reading: {value}C")
````

## 4. The `while...else` Clause

Python lets a loop own an `else` block. It runs **only when the loop finishes normally** - meaning the condition became `False` without a single `break`. If `break` fired, the `else` is skipped.

The usual reading is: *"loop until you find it... otherwise (`else`) report that it wasn't there."*

**Syntax:**
```python
while condition:
    ...
    if done_early:
        break
else:
    # reached ONLY if no break happened
    ...
```

````python
attempts = ["1234", "1111", "0000"]   # Simulated input: PIN tries
correct_pin = "4321"

i = 0
while i < len(attempts):
    if attempts[i] == correct_pin:
        print("Phone unlocked")
        break
    print("Wrong PIN:", attempts[i])
    i += 1
else:
    print("Too many failures - phone locked")   # ran because no break

# Try correct_pin = "1111" and re-run: now the else is skipped.
````

## 5. Pattern: The Counter Loop

The oldest trick in programming: start a counter, run while it's inside range, bump it each pass. Use it when you *could* use a number of repetitions but want full control of the steps.

**Syntax:**
```python
i = 1
while i <= n:
    ...work with i...
    i += 1
```

````python
total = 0
number = 1

while number <= 100:
    total += number
    number += 1

print("Sum of 1..100 =", total)
````

## 6. Pattern: The Accumulator

An **accumulator** is a variable that collects a running result across passes - a sum, a product, a string being built, a list being filled. Initialize it *before* the loop, update it *inside*, read it *after*.

Factorial is the textbook case: multiply instead of add, starting from 1 (never 0!).

**Syntax:**
```python
result = <neutral value>     # 0 for sums, 1 for products, [] for lists
while ...:
    result = result <combine> item
```

````python
n = 6
factorial = 1            # accumulator starts at the multiplicative identity
i = 2

while i <= n:
    factorial *= i
    i += 1

print(f"{n}! = {factorial}")
# 6! = 1*2*3*4*5*6 = 720
````

````python
# Accumulating into a LIST: keep only passing students
scores = [45, 82, 91, 58, 77]
passing = []
i = 0

while i < len(scores):
    if scores[i] >= 60:
        passing.append(scores[i])
    i += 1

print("Passing scores:", passing)
print("Pass rate:", f"{len(passing) / len(scores):.0%}")
````

## 7. Sentinel-Controlled Loops

A **sentinel** is a special value that means *"stop here"* - like typing `done` at a checkout prompt or an empty line ending a chat message. The loop shape is usually `while True:` plus `break` when the sentinel shows up.

**Syntax:**
```python
while True:
    data = get_next_input()
    if data == SENTINEL:
        break
    process(data)
```

````python
SENTINEL = "done"
incoming = ["rice", "eggs", "milk", "done"]   # Simulated input: cashier scans

cart = []
while True:
    if not incoming:                 # nothing left to scan
        item = SENTINEL
    else:
        item = incoming.pop(0)
    if item == SENTINEL:
        break                        # sentinel ends the shopping trip
    cart.append(item)

print("Cart:", cart)
print(f"Items: {len(cart)} | Total lines processed: yes")
````

> 🔍 **Under the Hood:** An infinite loop is just a condition that never turns `False`. Two everyday causes: (1) the body forgot to move toward the exit - e.g. `i = 1` initialized but never incremented; (2) floating-point drift - `x += 0.1` never lands *exactly* on `1.0`, because 0.1 has no exact binary representation, so `while x != 1.0:` spins forever. Compare floats with `<`/`>=` against a tolerance, never with `!=`. When a runaway loop hits your terminal, press **Ctrl+C** to raise `KeyboardInterrupt` and stop it.

````python
# ACCIDENTAL infinite loop - the fix is one line:
i = 1
limit = 3
steps_taken = 0

while i <= limit:
    # BUG (commented out): print(i) with no i += 1 would spin forever
    print(i)
    i += 1                # <- progress toward exit: i grows, limit won't
    steps_taken += 1

print("Exited safely after", steps_taken, "passes")

# Intentional 'infinite' loop: while True + break is idiomatic Python
# for servers and game loops - 'run forever' really means 'until told to stop'.
````

````python
# Float drift demo: why == fails on decimal steps
x = 0.0
hops = 0
while x < 1.0 and hops < 20:
    x += 0.1
    hops += 1

print("hops needed:", hops)      # more than 10!
print("final x:", repr(x))       # not exactly 1.0
````

## ⚠️ Common Mistakes & Gotchas

| Mistake | Problem | Fix |
|---|---|---|
| Forgetting to update the counter (`i = 1` but no `i += 1`) | Infinite loop - the condition can never change | Make sure every loop body makes progress toward exit |
| `while i < 5:` vs `while i <= 5:` confusion | Runs one pass too few or too many (off-by-one) | Decide the boundary, then test with tiny inputs |
| `continue` placed above the counter update | Update skipped -> infinite loop | Update before any `continue` |
| Comparing floats with `!=` / `==` | Rounding drift makes the condition never true | Use `<` / `>=`, or `abs(a - b) < tolerance` |
| Reaching for `while` when the count is known | More moving parts than needed | Known count → `for` (next lesson); unknown → `while` |

## 💡 Best Practices & Pro Tips

- **Ask on every loop: *what makes this condition false?*** If you can't answer instantly, you may have written an infinite loop.
- **Prefer `while True:` + `break` over a convoluted compound condition** when the natural exit is in the middle of the body - readers see the exit at the right moment.
- **Initialize accumulators immediately above their loop** and name them after what they hold (`total_price`, not `t`).
- **Cap retry loops**: `while attempt < MAX_RETRIES:` beats blind repetition whenever the loop talks to hardware, files, or networks.
- **AI-engineering relevance:** training and tuning jobs are while-loops at heart - iterate epochs *until loss < tolerance*, poll an API *until status == "ready"* (with retries capped!), batch-consume a queue *until empty*. Same patterns, higher stakes: an uncapped retry loop against a paid LLM endpoint costs real money per minute.

## 📌 Summary

| Construct | What it does | Example |
|---|---|---|
| `while cond:` | repeat while cond is True (tested first) | `while lives > 0:` |
| `break` | exit the loop immediately | `if guess == secret: break` |
| `continue` | skip to the next pass | `if invalid: continue` |
| `else:` on a loop | runs only if no `break` happened | `while ...: ... else: ...` |
| `+=` / `-=` | update a counter in place | `i += 1` |
| `while True:` + `break` | deliberate open-ended loop | servers, games, sentinels |

Key takeaways:

- The condition is checked **before** every pass, so a False start means zero passes.
- Every loop must make **progress toward its exit** - or it runs forever (Ctrl+C rescues you).
- Loop-`else` means *"no break occurred"*, which makes search loops read beautifully.
- Counters/accumulators follow one rhythm: initialize outside, update inside, use after.
