## Part 1 — Warm-up

**1. Countdown.** Each pass prints, then decrements; when the counter hits 0 the condition fails.

```
countdown = 5

while countdown > 0:
    print(countdown)
    countdown -= 1        # progress toward exit

print("Lift off!")
```

**2. Gauss, manually.** Initialize outside, update inside, read after - the classic accumulator rhythm.

```
total = 0
number = 1

while number <= 100:
    total += number
    number += 1

print("Sum of 1..100 =", total)   # Sum of 1..100 = 5050
```

**3. Never ran.** The condition is tested BEFORE the first pass, so a False start means zero passes.

```
battery = 100

while battery < 20:
    print("Low battery")
    battery -= 10

print("Loop finished; battery still", battery)
# The body ran 0 times: 100 < 20 was False from the very first test,
# so Python skipped straight past the whole loop.
```

## Part 2 — Practice

**4. Six factorial.** Same rhythm as a sum, but the neutral value for products is 1.

```
n = 6
factorial = 1            # multiplicative identity
i = 2

while i <= n:
    factorial *= i
    i += 1

print(f"{n}! = {factorial}")   # 6! = 720
```

**5. Faulty sensor filter.** Update the index BEFORE any continue, or the skipped round traps the loop forever.

```
readings = [22, -1, 25, -7, 30]

i = 0
while i < len(readings):
    value = readings[i]
    i += 1                 # update FIRST so continue cannot trap us
    if value < 0:
        continue           # discard the bad reading
    print(f"Valid reading: {value}C")
```

**6. Stop at the right guess.** `break` aborts the loop immediately; no point looking at later guesses.

```
secret = 7
guesses = [3, 9, 7, 1]

i = 0
while i < len(guesses):
    guess = guesses[i]
    print("Guess:", guess)
    if guess == secret:
        print("Correct! Unlocked in", i + 1, "tries.")
        break
    i += 1

print("Game finished.")
```

**7. Lockout.** Loop-`else` runs only when NO break fired - perfect for found/not-found.

```
attempts = ["1234", "1111", "0000"]
correct_pin = "4321"

i = 0
while i < len(attempts):
    if attempts[i] == correct_pin:
        print("Phone unlocked")
        break
    print("Wrong PIN:", attempts[i])
    i += 1
else:
    print("Too many failures - phone locked")   # ran because no break

# With correct_pin = "1111": the break fires on the second try,
# so 'Phone unlocked' prints and the else block is skipped entirely.
```

## Part 3 — Challenge

**8. Checkout sentinel.** `while True:` + `break` is the idiomatic shape when the exit is mid-body.

```
SENTINEL = "done"
incoming = ["rice", "eggs", "milk", "done"]

cart = []
while True:
    item = incoming.pop(0)      # next scanned item
    if item == SENTINEL:
        break                   # sentinel ends the shopping trip
    cart.append(item)

print("Cart:", cart)
```

**9. Runaway rescue.** Every loop needs progress toward exit; floats drift, so compare with `<`, never `!=`.

```
# (a) Nothing moves i, so 'i <= 3' stays True forever.
i = 1
while i <= 3:
    print(i, end=" ")
    i += 1              # <- the missing line: progress toward exit
print()

# (b) 0.1 has no exact binary form, so x lands on 0.9999999999999999,
#     1.0999..., etc. and NEVER equals exactly 1.0 - '!=' spins forever.
#     Test the direction of travel instead:
x = 0.0
while x < 1.0:
    x += 0.1
print("final x:", repr(x))   # 1.0999999999999999 - close, but not exact
```
