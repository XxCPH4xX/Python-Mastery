# While Loop Exercises

Write your answer in the empty code cell under each question. Check yourself against `solutions.ipynb` when done.

**Part 1** = warm-up · **Part 2** = practice · **Part 3** = challenge.

## Part 1 — Warm-up

**1. Countdown.** Starting from `countdown = 5`, print `5` down to `1`, one per line, then print `Lift off!` after the loop ends. Shrink the counter inside the body.

**2. Gauss, manually.** Use a counter loop to add every whole number from `1` to `100` into a `total` accumulator, then print `Sum of 1..100 = 5050`.

**3. Never ran.** Let `battery = 100` and write `while battery < 20:` whose body would print `Low battery` and drain the battery by 10. Before running, predict how many times the body executes; after running, explain in a comment why the answer is what it is.

## Part 2 — Practice

**4. Six factorial.** Compute `6!` with a `while` loop and a product accumulator starting at `1` (never 0!). Print `6! = 720`.

**5. Faulty sensor filter.** The readings `[22, -1, 25, -7, 30]` include faulty negatives. Walk the list with a `while` loop, use `continue` to discard bad values, and print `Valid reading: 22C`-style lines for the good ones. Mind where you bump the index!

**6. Stop at the right guess.** With `secret = 7` and guesses `[3, 9, 7, 1]`, loop through them printing each guess, and `break` the moment one is correct - reporting `Correct! Unlocked in 3 tries.`

**7. Lockout.** PIN tries are `["1234", "1111", "0000"]` and `correct_pin = "4321"`. Loop through them, print `Wrong PIN:` plus each failed try, and use a `while...else` clause to print `Too many failures - phone locked` because nothing succeeded. Add a comment saying what changes when `correct_pin` is `"1111"` instead.

## Part 3 — Challenge

**8. Checkout sentinel.** Scanned items arrive as `["rice", "eggs", "milk", "done"]` where `"done"` is the sentinel meaning stop. Fill a `cart` list using a `while True:` loop that `break`s when the sentinel appears, then print `Cart: ['rice', 'eggs', 'milk']`.

**9. Runaway rescue.** (a) A loop starts `i = 1`, tests `while i <= 3:` and prints `i` - but never ends. Name the missing line in a comment, then write the fixed version that prints `1 2 3` and exits. (b) A friend waits with `while x != 1.0:` while growing `x += 0.1` from `0.0`. Explain in a comment why that spins forever, then rewrite the test so it terminates.
