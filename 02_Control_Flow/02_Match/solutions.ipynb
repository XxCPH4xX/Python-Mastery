## Part 1 — Warm-up

**1. Status desk.** Literal cases compared top-down; the `_` wildcard plays the role of `else`.

```
def status_message(code):
    match code:
        case 200:
            return "OK"
        case 404:
            return "Not Found"
        case 500:
            return "Server Error - our fault"
        case _:
            return f"Unknown status {code}"

for code in (200, 404, 418):
    print(code, "->", status_message(code))
# 200 -> OK
# 404 -> Not Found
# 418 -> Unknown status 418
```

**2. Robot remote.** `|` lets several literals share one case body - read it as 'or'.

```
command = "go"

match command:
    case "start" | "go" | "run":
        print("Robot moving forward")
    case "stop" | "halt" | "quit":
        print("Robot stopping")
    case _:
        print(f"Unrecognized command: {command}")
```

**3. Method spotlight.** DELETE fits none of the listed groups, so the wildcard safety net catches it.

```
http_method = "DELETE"

match http_method:
    case "GET" | "HEAD":
        print("Read-only request")
    case "POST" | "PUT" | "PATCH":
        print("Writes data")
    case _:
        print("Method not handled:", http_method)
```

## Part 2 — Practice

**4. Capture the leftovers.** A bare name in a pattern CAPTURES whatever reached it instead of comparing.

```
def describe(x):
    match x:
        case 0:
            return "zero"
        case 1 | -1:
            return "plus or minus one"
        case other:               # captures anything left over
            return f"something else: {other!r}"

for sample in (0, -1, "banana", 3.14):
    print(describe(sample))
```

**5. Priced by the kilo.** A guard adds `if <condition>` after the pattern; failing guards let matching continue.

```
weight_kg = 12

shipping = None
match weight_kg:
    case n if n <= 0:
        shipping = "invalid weight"
    case n if n <= 1:
        shipping = 60       # BDT
    case n if n <= 5:
        shipping = 120
    case n if n <= 10:
        shipping = 200
    case _:
        shipping = "quote required"

print(f"Shipping for {weight_kg} kg: {shipping}")
# Shipping for 12 kg: quote required
```

**6. Where on the grid?.** Tuple patterns check the shape AND the slots, top-down: exact shapes go first.

```
points = [(0, 0), (0, 7), (3, 0), (2, 9)]

for point in points:
    match point:
        case (0, 0):
            print(point, "-> the origin")
        case (0, y):
            print(point, f"-> on the y-axis, height {y}")
        case (x, 0):
            print(point, f"-> on the x-axis, position {x}")
        case (x, y):
            print(point, f"-> general point ({x}, {y})")
```

**7. The capture trap.** `case LIMIT:` is a bare name, so it captures - silently rebinding your constant.

```
LIMIT = 100
value = 42

match value:
    case LIMIT:          # bare name -> captures; LIMIT is rebound to 42!
        pass
print("After the buggy match, LIMIT holds:", LIMIT)

LIMIT = 100             # restore the constant
match value:
    case v if v == LIMIT:   # a guard compares instead of capturing
        print("equals the limit")
    case _:
        print("below the limit")   # <- this one prints
```

## Part 3 — Challenge

**8. Leaderboard shapes.** Sequence patterns match length and slots; `*rest` soaks up any extras into a list.

```
scores = [95, 88, 72, 61]

match scores:
    case []:
        print("Empty leaderboard")
    case [top]:
        print(f"Solo leader with {top}")
    case [top, second, *others]:
        print(f"Gold {top}, silver {second}, plus {len(others)} more players")
# Gold 95, silver 88, plus 2 more players
```

**9. JSON dispatcher.** Mapping patterns check only the keys you list and ignore everything else.

```
records = [
    {"name": "Sarah", "role": "admin", "city": "Dhaka"},
    {"name": "Rahim", "role": "member"},
    {"city": "Chattogram"},
]

for record in records:
    match record:
        case {"role": "admin", "name": who}:
            print(f"Admin detected: {who} - full access")
        case {"role": role, "name": who}:
            print(f"{who} is a {role}")
        case _:
            print(f"Incomplete profile: {record}")

# Sarah's record matched the admin pattern even though it also had
# a 'city' key - dict patterns simply ignore extra keys.
```
