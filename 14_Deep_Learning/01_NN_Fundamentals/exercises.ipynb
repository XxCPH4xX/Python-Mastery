## Part 1 — Warm-up

**1. Your first neuron.** A single neuron decides whether Sarah goes running. With `w = np.array([0.6, 0.9, 0.4])`, bias `b = -3.0` and inputs `x = np.array([7.0, 8.0, 6.0])`, compute the weighted sum `z = np.dot(w, x) + b`. Define a step activation returning 1 when `z > 0` else 0, and print whether the neuron fires. Then check the gloomy input `np.array([3.0, 2.0, 5.0])` too.

**2. Two activations.** Implement `sigmoid(z) = 1 / (1 + np.exp(-z))` and `relu(z) = np.maximum(0, z)` as functions. Apply BOTH to `z = np.array([-2.0, -0.5, 0.0, 1.5])` and print the results. In a comment: which of the two can output negative numbers?

**3. Score your guesses with MSE.** The true rents are `rent = np.array([20.0, 35.0])` (thousands). One model predicts `pred_good = np.array([21.0, 34.0])`, another `pred_bad = np.array([30.0, 45.0])`. Write `mse(y_true, y_pred) = np.mean((y_true - y_pred) ** 2)` and print both models' scores.

## Part 2 — Practice

**4. Forward pass of a 2-4-1 network.** Seed `rng = np.random.default_rng(42)` and create `W1 = rng.normal(0, 0.5, size=(2, 4))`, `b1 = np.zeros((1, 4))`, `W2 = rng.normal(0, 0.5, size=(4, 1))`, `b2 = np.zeros((1, 1))`. For `X = np.array([[0.5, 1.5], [2.0, 0.5], [1.0, 1.0], [3.0, 2.5]])` compute `Z1 = X @ W1 + b1`, `A1 = np.maximum(Z1, 0)` and `Z2 = A1 @ W2 + b2`. Print every shape and the final logits rounded to 3 decimals.

**5. BCE punishes confident mistakes.** Labels `y = np.array([1.0, 0.0, 1.0])`. Detector A outputs `p_close = np.array([0.9, 0.1, 0.8])`; detector B outputs `p_confident_wrong = np.array([0.01, 0.99, 0.8])`. Implement `bce(y_true, p) = -np.mean(y * np.log(p) + (1 - y) * np.log(1 - p))`, clipping `p` into `[1e-12, 1 - 1e-12]` first so `log(0)` cannot explode. Print both losses - which detector wins?

**6. Descend the bowl.** For the bowl `f(x) = (x - 3) ** 2` with slope `g(x) = 2 * (x - 3)`: start at `x = 0.0` and take 20 gradient-descent steps with `lr = 0.3` using `x -= lr * g(x)`. Print `x` and `f(x)` every 5 steps, then the final distance from the true minimum `abs(x - 3)`.

**7. Goldilocks learning rate.** Repeat the bowl descent from `x = 0.0` for 15 steps with EACH learning rate in `[0.01, 0.9, 1.1]`. Print the final `abs(x - 3)` for each and label them in comments as too slow / converging / diverged. Which rate multiplies the error by -1.2 every step?

## Part 3 — Challenge

**8. Backprop by hand - then prove it.** With inputs `x = 1.0`, `b = 0.5`, weight `w = 2.0`, target `t = 1.0`: run the forward pass `z = w * x + b`, `a = sigmoid(z)`, `L = (a - t) ** 2`. Then go backwards: `dL_da = 2 * (a - t)`, `da_dz = a * (1 - a)`, `dz_dw = x`, `dz_db = 1.0`; multiply the chain out into `dL_dw` and `dL_db`. Finally verify `dL_dw` with a central difference `(loss_with(w + eps) - loss_with(w - eps)) / (2 * eps)` using `eps = 1e-6`, and print both values.
